## Question 1

Mini Project: Food Delivery RL Concept Simulator

Build a console-based Food Delivery Reinforcement Learning Simulator
that models the full RL loop—agent, environment, actions, states, and
rewards—for a delivery dispatch scenario. Combine the reward function,
state tracking, episode logging, and environment interaction concepts
from Session 2 into one working program.

Requirements: - Provide a menu-driven interface with at least three
options: Start New Delivery Episode, View Agent Stats, and Exit. - Each
episode must simulate a delivery agent completing a sequence of at least
five actions, updating the agent’s state and calculating reward after
every action. - Track and display cumulative statistics across multiple
episodes: total episodes run, total reward earned, and average reward
per episode. - Store the agent’s state (current location, orders
delivered, total reward) in a dictionary and update it correctly after
every action in every episode. - At the end of each episode, print a
full episode summary showing each action, reward per step, and total
episode reward.

**Answer:**

```python
import random

# Global statistics tracking
global_stats = {"total_episodes": 0, "total_reward": 0}


def run_episode():
    # Agent state tracking
    state = {"current_location": "Restaurant", "orders_delivered": 0, "total_reward": 0}

    actions = [
        "Pick Up Order",
        "Navigate Traffic",
        "Deliver Order",
        "Wait for Dispatch",
        "Return to Hub",
    ]
    episode_log = []

    for step in range(5):
        action = random.choice(actions)

        # Environment dynamics and reward function
        if action == "Pick Up Order":
            reward = 5
            state["current_location"] = "Restaurant"
        elif action == "Navigate Traffic":
            reward = -2
            state["current_location"] = "Transit"
        elif action == "Deliver Order":
            reward = 15
            state["current_location"] = "Customer Location"
            state["orders_delivered"] += 1
        elif action == "Wait for Dispatch":
            reward = -1
            state["current_location"] = "Idle Zone"
        else:  # Return to Hub
            reward = 2
            state["current_location"] = "Dispatch Hub"

        state["total_reward"] += reward
        episode_log.append(
            {
                "step": step + 1,
                "action": action,
                "reward": reward,
                "location": state["current_location"],
            }
        )

    # Update cumulative statistics
    global_stats["total_episodes"] += 1
    global_stats["total_reward"] += state["total_reward"]

    # Episode Summary
    print("\n--- EPISODE SUMMARY ---")
    for log in episode_log:
        print(
            f"Step {log['step']}: Action='{log['action']}', Reward={log['reward']}, Location='{log['location']}'"
        )
    print(f"Total Episode Reward: {state['total_reward']}")
    print(f"Final Orders Delivered: {state['orders_delivered']}\n")


def view_stats():
    episodes = global_stats["total_episodes"]
    total_reward = global_stats["total_reward"]
    avg_reward = (total_reward / episodes) if episodes > 0 else 0

    print("\n--- AGENT CUMULATIVE STATS ---")
    print(f"Total Episodes Run: {episodes}")
    print(f"Total Reward Earned: {total_reward}")
    print(f"Average Reward Per Episode: {avg_reward:.2f}\n")


def main():
    while True:
        print("=== Food Delivery RL Simulator ===")
        print("1. Start New Delivery Episode")
        print("2. View Agent Stats")
        print("3. Exit")
        choice = input("Select an option (1-3): ").strip()

        if choice == "1":
            run_episode()
        elif choice == "2":
            view_stats()
        elif choice == "3":
            print("Exiting simulator.")
            break
        else:
            print("Invalid selection. Try again.\n")


if __name__ == "__main__":
    main()
```
