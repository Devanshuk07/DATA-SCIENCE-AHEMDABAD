## Question 1

Write a Python function calculate_reward(delivery_time,
is_on_time, customer_rating) that calculates and returns a reward score
for one delivery. Award +10 points if is_on_time is True; deduct -5
points if False; add customer_rating (integer 1–5) to the total. Print a
descriptive final reward and test at least three different input
combinations showing different outcomes.

**Answer:**

```python
def calculate_reward(delivery_time, is_on_time, customer_rating):
    # Base reward calculation
    if is_on_time:
        score = 10
    else:
        score = -5

    # Add customer rating (1-5)
    score += customer_rating

    print(
        f"Delivery Time: {delivery_time} mins | On Time: {is_on_time} | Rating: {customer_rating} -> Final Reward: {score} points"
    )
    return score


# Test cases
calculate_reward(25, True, 5)
calculate_reward(45, False, 3)
calculate_reward(50, False, 1)
```

## Question 2

Create a dictionary-based Python program to track a delivery agent’s
    state across multiple time steps. Include the keys location
    (string), orders_delivered (integer), total_reward (float), and
    is_available (boolean). Define update_state(state, new_location,
    reward_earned) to update the location, increment orders_delivered by
    1, and add reward_earned to total_reward. Simulate at least four
    state transitions with different locations and rewards. After each
    update, print the complete current state clearly.

**Answer:**

```python
def update_state(state, new_location, reward_earned):
    state["location"] = new_location
    state["orders_delivered"] += 1
    state["total_reward"] += reward_earned
    return state


# Initial state
agent_state = {
    "location": "Hub",
    "orders_delivered": 0,
    "total_reward": 0.0,
    "is_available": True,
}

# Transitions: (new_location, reward_earned)
transitions = [
    ("Downtown", 15.50),
    ("Uptown", 22.00),
    ("Suburbs", 18.75),
    ("Airport", 30.00),
]

# Simulation
for step, (location, reward) in enumerate(transitions, start=1):
    agent_state = update_state(agent_state, location, reward)
    print(f"Step {step}: {agent_state}")
```

## Question 3

Write a Python program that simulates one episode of a delivery RL
    agent executing a fixed sequence of actions and logs cumulative
    reward at every step. Define possible actions: [‘accept_order’,
    ‘reject_order’, ‘request_directions’, ‘mark_delivered’]. Assign
    rewards: accept_order = +2, reject_order = -1, request_directions =
    0, mark_delivered = +10. Simulate a predefined sequence of at least
    six actions, accumulating reward at each step. Print each action,
    its reward, and the running cumulative reward, then print the total
    episode reward.

**Answer:**

```python
# Define reward mapping for actions
rewards = {
    "accept_order": 2,
    "reject_order": -1,
    "request_directions": 0,
    "mark_delivered": 10,
}

# Predefined sequence of actions (at least 6)
actions_sequence = [
    "reject_order",
    "accept_order",
    "request_directions",
    "request_directions",
    "mark_delivered",
    "accept_order",
    "mark_delivered",
]

cumulative_reward = 0

# Simulate episode and log rewards
for step, action in enumerate(actions_sequence, 1):
    reward = rewards[action]
    cumulative_reward += reward
    print(
        f"Step {step}: Action = '{action}', Reward = {reward}, Cumulative Reward = {cumulative_reward}"
    )

print(f"\nTotal Episode Reward: {cumulative_reward}")
```

## Question 4

Build a console-based Python simulation of the agent-environment
    interaction loop for a delivery scenario. Define a
    DeliveryEnvironment class with attributes current_state (a
    dictionary containing at least location and pending_orders),
    total_reward, and step_count. Implement step(action) to update
    state, calculate and return a reward, and print the new state and
    reward after each step. Define at least three distinct actions with
    different state-transition effects and reward values (for example,
    deliver_order, wait, navigate_to_zone). Execute a predefined
    sequence of five or more actions and print the result of each step,
    final total reward, and total step count.

**Answer:**

```python
class DeliveryEnvironment:

    def __init__(self):
        self.current_state = {"location": "Depot", "pending_orders": 3}
        self.total_reward = 0
        self.step_count = 0

    def step(self, action):
        self.step_count += 1
        reward = 0

        if action == "navigate_to_zone":
            self.current_state["location"] = "Customer_Zone"
            reward = -2
        elif action == "deliver_order":
            if (
                self.current_state["location"] == "Customer_Zone"
                and self.current_state["pending_orders"] > 0
            ):
                self.current_state["pending_orders"] -= 1
                reward = 10
            else:
                reward = -5
        elif action == "wait":
            reward = -1

        self.total_reward += reward

        print(f"Step: {self.step_count} | Action: {action}")
        print(f"New State: {self.current_state}")
        print(f"Reward: {reward}\n")

        return reward


# Simulation execution
env = DeliveryEnvironment()
action_sequence = [
    "wait",
    "navigate_to_zone",
    "deliver_order",
    "deliver_order",
    "wait",
]

for act in action_sequence:
    env.step(act)

print(f"Final Total Reward: {env.total_reward}")
print(f"Total Step Count: {env.step_count}")
```
