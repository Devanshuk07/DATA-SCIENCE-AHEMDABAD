### 1. Prompt Given to the AI Tool

```text
Write a Python program that performs K-Means clustering on a simulated food delivery dataset with features: monthly_orders, avg_spend, and avg_rating. 
Use the Elbow Method to plot inertia for finding optimal k. 
Print the centroid values and customer counts for each cluster. 
Reduce the features to 2D using PCA and plot the final clusters with distinct colors and a legend.
```

### 2. AI's Original Code

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# Generate dummy data
np.random.seed(42)
data = pd.DataFrame(
    {
        "monthly_orders": np.random.randint(1, 30, 100),
        "avg_spend": np.random.uniform(10, 100, 100),
        "avg_rating": np.random.uniform(1, 5, 100),
    }
)

# Elbow Method
inertia = []
for k in range(1, 10):
    km = KMeans(n_clusters=k)
    km.fit(data)
    inertia.append(km.inertia_)

plt.plot(range(1, 10), inertia)
plt.show()

# Final Clustering (k=3)
kmeans = KMeans(n_clusters=3)
data["cluster"] = kmeans.fit_predict(data)

# Print centroids and counts
print("Centroids:\n", kmeans.cluster_centers_)
print("Counts:\n", data["cluster"].value_counts())

# PCA Visualisation
pca = PCA(n_components=2)
coords = pca.fit_transform(data[["monthly_orders", "avg_spend", "avg_rating"]])

scatter = plt.scatter(coords[:, 0], coords[:, 1], c=data["cluster"], cmap="viridis")
plt.legend(*scatter.legend_elements(), title="Clusters")
plt.show()
```

### 3. Corrected Code

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Generate dummy data
np.random.seed(42)
features = ["monthly_orders", "avg_spend", "avg_rating"]
data = pd.DataFrame(
    {
        "monthly_orders": np.random.randint(1, 30, 100),
        "avg_spend": np.random.uniform(10, 100, 100),
        "avg_rating": np.random.uniform(1, 5, 100),
    }
)

# Fix 1: Feature scaling
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data[features])

# Elbow Method (Fix 2: added random_state)
inertia = []
for k in range(1, 10):
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(scaled_data)
    inertia.append(km.inertia_)

plt.figure(figsize=(6, 4))
plt.plot(range(1, 10), inertia, marker="o")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()

# Final Clustering (k=3)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
data["cluster"] = kmeans.fit_predict(scaled_data)

# Print centroids (unscaled for readability) and counts
centroids_unscaled = scaler.inverse_transform(kmeans.cluster_centers_)
print("Centroids (Original Scale):\n", pd.DataFrame(centroids_unscaled, columns=features))
print("\nCluster Counts:\n", data["cluster"].value_counts())

# PCA Visualisation (Fix 3: labelled axes)
pca = PCA(n_components=2)
coords = pca.fit_transform(scaled_data)

plt.figure(figsize=(6, 4))
scatter = plt.scatter(coords[:, 0], coords[:, 1], c=data["cluster"], cmap="viridis")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Customer Segments (PCA)")
plt.legend(*scatter.legend_elements(), title="Cluster")
plt.show()
```

---

### 4. Explanation of Changes

1. Added `StandardScaler` because unscaled features (`avg_spend` vs `avg_rating`) skewed distance calculations toward higher-magnitude values.
2. Added `random_state=42` and `n_init=10` to `KMeans` to guarantee reproducible cluster assignments and prevent deprecation warnings.
3. Added axis labels (`PC1`, `PC2`) and inverse-transformed centroids back to their original scales for meaningful business interpretation.
