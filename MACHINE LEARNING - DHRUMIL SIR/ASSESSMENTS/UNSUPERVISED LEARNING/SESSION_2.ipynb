## Question 1

Task 1: Delivery Time Statistical Profiler

Build a Python program that computes descriptive statistics for a list
of food delivery times and identifies the direction of skew in the data.

Requirements: - Define a hardcoded list of at least 15 delivery time
values (in minutes) that includes realistic values between 20–40 minutes
and at least 3 outliers above 75 minutes. - Calculate and print the
mean, median, mode, variance, and standard deviation using Python’s
built-in statistics module. - Determine whether the distribution is
left-skewed, right-skewed, or approximately symmetrical by comparing the
mean and median, and print a clearly worded conclusion. - Display a
formatted summary report with each statistic labelled and aligned, so
the output is readable without any additional context.

**Answer:**

```python
import statistics

# List of 15 delivery times (realistic values + 3 outliers above 75)
delivery_times = [22, 25, 27, 28, 30, 30, 32, 35, 36, 38, 40, 40, 78, 85, 92]

# Calculate statistics
mean_val = statistics.mean(delivery_times)
median_val = statistics.median(delivery_times)
mode_val = statistics.mode(delivery_times)
variance_val = statistics.variance(delivery_times)
stdev_val = statistics.stdev(delivery_times)

# Determine skewness
if mean_val > median_val:
    skewness = (
        "Right-skewed (Mean > Median, pulled by high-value outliers)"
    )
elif mean_val < median_val:
    skewness = "Left-skewed (Mean < Median)"
else:
    skewness = "Approximately symmetrical (Mean == Median)"

# Formatted summary report
print("--- Delivery Time Statistical Profile ---")
print(f"{'Mean:':<20} {mean_val:.2f} minutes")
print(f"{'Median:':<20} {median_val:.2f} minutes")
print(f"{'Mode:':<20} {mode_val:.2f} minutes")
print(f"{'Variance:':<20} {variance_val:.2f} minutes²")
print(f"{'Standard Deviation:':<20} {stdev_val:.2f} minutes")
print("-" * 42)
print(f"Distribution Shape:  {skewness}")
```

features, and scales numeric columns.

containing columns: order_id (int), restaurant_rating (float with at
least 2 NaN values), payment_method (Cash / Card / Wallet), cuisine_type
(Indian / Chinese / Italian / Fast Food), and order_value (float). -
Fill missing restaurant_rating values with the column mean and verify no
NaN values remain using isnull().sum(). - Apply one-hot encoding to both
payment_method and cuisine_type using pd.get_dummies() and drop the
original categorical columns from the DataFrame. - Apply MinMaxScaler
from sklearn.preprocessing to the restaurant_rating and order_value
columns only, then print the final preprocessed DataFrame.

**Answer:**

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# 1. Create DataFrame
data = {
    "order_id": range(101, 113),
    "restaurant_rating": [
        4.2,
        np.nan,
        3.8,
        4.5,
        np.nan,
        4.0,
        3.5,
        4.8,
        4.1,
        3.9,
        np.nan,
        4.6,
    ],
    "payment_method": [
        "Cash",
        "Card",
        "Wallet",
        "Card",
        "Cash",
        "Wallet",
        "Card",
        "Cash",
        "Wallet",
        "Card",
        "Cash",
        "Wallet",
    ],
    "cuisine_type": [
        "Indian",
        "Chinese",
        "Italian",
        "Fast Food",
        "Indian",
        "Chinese",
        "Italian",
        "Fast Food",
        "Indian",
        "Chinese",
        "Italian",
        "Fast Food",
    ],
    "order_value": [
        250.0,
        450.5,
        150.0,
        800.0,
        320.0,
        600.0,
        220.5,
        950.0,
        180.0,
        510.0,
        340.0,
        720.0,
    ],
}
df = pd.DataFrame(data)

# 2. Fill missing restaurant_rating values and verify
df["restaurant_rating"] = df["restaurant_rating"].fillna(
    df["restaurant_rating"].mean()
)
print("Missing values count:\n", df.isnull().sum(), "\n")

# 3. Apply one-hot encoding and drop originals
df = pd.get_dummies(
    df, columns=["payment_method", "cuisine_type"], dtype=float
)

# 4. Scale numeric columns
scaler = MinMaxScaler()
df[["restaurant_rating", "order_value"]] = scaler.fit_transform(
    df[["restaurant_rating", "order_value"]]
)

# Print final DataFrame
print("Final Preprocessed DataFrame:")
print(df)
```

## Question 3

Task 3: Order Cancellation Classifier with Evaluation

Train a Random Forest classifier to predict whether a food delivery
order will be cancelled and evaluate the model using metrics suited for
an imbalanced target class.

Requirements: - Create or use a sample dataset of at least 120 records
with features: order_value (float), delivery_distance_km (float),
hour_of_day (int 0–23), and restaurant_rating (float), with a target
column cancelled containing 1 (cancelled) and 0 (not cancelled) where
approximately 20–25% are cancelled. - Split the data into 80% training
and 20% test sets using train_test_split with stratify=cancelled and
random_state=42. - Train a RandomForestClassifier with random_state=42,
generate predictions on the test set, and print the full
classification_report showing precision, recall, and F1-score for both
classes. - Compute and print the confusion matrix as a labelled 2×2
table identifying TP, TN, FP, and FN values explicitly in the output.

**Answer:**

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.model_selection import train_test_split

# 1. Create dataset (150 records, ~20% cancelled)
np.random.seed(42)
n_samples = 150

data = {
    "order_value": np.random.uniform(10.0, 100.0, n_samples).round(2),
    "delivery_distance_km": np.random.uniform(1.0, 20.0, n_samples).round(2),
    "hour_of_day": np.random.randint(0, 24, n_samples),
    "restaurant_rating": np.random.uniform(2.5, 5.0, n_samples).round(1),
    "cancelled": np.random.choice([0, 1], size=n_samples, p=[0.78, 0.22]),
}
df = pd.DataFrame(data)

# 2. Train-test split
X = df[
    [
        "order_value",
        "delivery_distance_km",
        "hour_of_day",
        "restaurant_rating",
    ]
]
y = df["cancelled"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# 3. Train RandomForestClassifier and predict
clf = RandomForestClassifier(random_state=42)
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)

# 4. Classification report
print("Classification Report:")
print(classification_report(y_test, y_pred))

# 5. Labelled confusion matrix
cm = confusion_matrix(y_test, y_pred)
tn, fp, fn, tp = cm.ravel()

cm_table = pd.DataFrame(
    [[f"TN: {tn}", f"FP: {fp}"], [f"FN: {fn}", f"TP: {tp}"]],
    index=["Actual: Not Cancelled (0)", "Actual: Cancelled (1)"],
    columns=["Predicted: Not Cancelled (0)", "Predicted: Cancelled (1)"],
)

print("Confusion Matrix:")
print(cm_table)
```

Task 4: Customer Segmentation with K-Means and PCA Visualisation

Apply K-Means clustering to food delivery customer data, determine the
optimal k using the Elbow Method, and produce a 2D cluster visualisation
using PCA.

Requirements: - Create or load a dataset of at least 150 customers with
three features: monthly_orders (int), avg_order_value (float), and
avg_delivery_rating (float 1.0–5.0). Apply StandardScaler to all three
features before clustering. - Run K-Means for k values from 2 to 9,
record the inertia for each k, and plot the Elbow Method graph with k on
the x-axis, inertia on the y-axis, and clear axis labels and a title. -
Train a final K-Means model using the optimal k identified from the
elbow graph and assign cluster labels to all customer records. - Apply
PCA to reduce the scaled data to 2 components, then produce a scatter
plot with each cluster in a distinct colour, a legend identifying
cluster numbers, and axis labels showing PC1 and PC2.

**Answer:**

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# 1. Create dataset (150 customers)
np.random.seed(42)
n_samples = 150

data = pd.DataFrame(
    {
        "monthly_orders": np.random.randint(1, 30, size=n_samples),
        "avg_order_value": np.random.uniform(10.0, 100.0, size=n_samples).round(
            2
        ),
        "avg_delivery_rating": np.random.uniform(1.0, 5.0, size=n_samples).round(
            1
        ),
    }
)

# Scale features
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)

# 2. Elbow Method (k from 2 to 9)
k_values = range(2, 10)
inertias = []

for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(scaled_data)
    inertias.append(kmeans.inertia_)

plt.figure(figsize=(8, 4))
plt.plot(k_values, inertias, marker="o")
plt.xlabel("k")
plt.ylabel("Inertia")
plt.title("Elbow Method for Optimal k")
plt.show()

# 3. Train final K-Means model (optimal k = 3)
optimal_k = 3
final_kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
data["cluster"] = final_kmeans.fit_predict(scaled_data)

# 4. PCA to 2 components and visualisation
pca = PCA(n_components=2)
pca_data = pca.fit_transform(scaled_data)

plt.figure(figsize=(8, 6))
scatter = plt.scatter(
    pca_data[:, 0], pca_data[:, 1], c=data["cluster"], cmap="viridis"
)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Customer Segmentation Visualisation (PCA)")
plt.legend(*scatter.legend_elements(), title="Clusters")
plt.show()
```
