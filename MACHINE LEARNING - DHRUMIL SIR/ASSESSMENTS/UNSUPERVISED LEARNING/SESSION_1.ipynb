## Question 1

SCENARIO: You are analysing delivery time data for a
food delivery platform. You find that most deliveries complete in 25–30
minutes, but a small number of orders take over 90 minutes due to
extreme traffic conditions. When the operations head asks for a single
number to represent the ‘typical delivery time’ to display on the app,
you notice that the mean is significantly higher than the value most
customers actually experience.

Question: Explain why the mean is a misleading measure of central
tendency for this delivery time dataset. Which alternative measure would
you recommend, and how does the concept of skewness justify your choice?
Describe what the shape of the distribution looks like.

**Answer:**

* **Why the mean is misleading:** The mean is sensitive to extreme values (outliers). The rare deliveries taking over 90 minutes pull the average upward, making it unrepresentative of the typical 25–30 minute delivery experience.

* **Recommended alternative:** The **median**. It is resistant to outliers because it represents the middle value of the data, accurately reflecting what most customers experience.

* **Justification via skewness:** The dataset is **positively skewed (right-skewed)**. In a right-skewed distribution, the mean is pulled to the right of the peak ($Mean > Median$), making the median the statistically appropriate measure of central tendency.

* **Shape of the distribution:** The distribution has a single tall peak around 25–30 minutes with a long, thin tail stretching far to the right toward the 90+ minute marks.

## Question 2

SCENARIO: You are building a fraud detection system for a food
    delivery platform. Historical data shows that 5% of all orders are
    fraudulent. You have trained a model that, when an order is
    fraudulent, correctly flags it 90% of the time. However, when an
    order is legitimate, the model incorrectly flags it as fraud 8% of
    the time. A flagged order arrives, and your manager asks you to
    estimate the actual probability that it is genuinely fraudulent.

Question: Apply Bayes’ theorem to calculate the probability that a
flagged order is truly fraudulent. Clearly identify the prior
probability, likelihood, and how you arrive at the posterior. Explain
why the result might surprise a non-technical stakeholder.

**Answer:**

**1. Identify Components:**
*   **Prior Probability, $P(\text{Fraud})$:** $0.05$ (5% base fraud rate).
*   **Likelihood, $P(\text{Flag} \mid \text{Fraud})$:** $0.90$ (90% true positive rate).
*   **False Positive Rate, $P(\text{Flag} \mid \text{Legitimate})$:** $0.08$ (8% false flag rate).
*   **$P(\text{Legitimate})$:** $1 - 0.05 = 0.95$.

**2. Calculation (Posterior Probability):**
*   **Total Probability of being Flagged, $P(\text{Flag})$:**
    $$P(\text{Flag}) = (P(\text{Flag} \mid \text{Fraud}) \times P(\text{Fraud})) + (P(\text{Flag} \mid \text{Legitimate}) \times P(\text{Legitimate}))$$
    $$P(\text{Flag}) = (0.90 \times 0.05) + (0.08 \times 0.95) = 0.045 + 0.076 = 0.121$$

*   **Posterior Probability, $P(\text{Fraud} \mid \text{Flag})$:**
    $$P(\text{Fraud} \mid \text{Flag}) = \frac{P(\text{Flag} \mid \text{Fraud}) \times P(\text{Fraud})}{P(\text{Flag})} = \frac{0.045}{0.121} \approx 0.3719 \text{ (or } 37.19\%)$$

**3. Why it Surprises Non-Technical Stakeholders:**
Stakeholders expect a model with 90% accuracy to be correct most of the time when it flags an alert. However, because fraud is rare (5%) and legitimate orders are vast (95%), the small 8% error rate generates more false alarms ($0.076$) than true fraud detections ($0.045$), meaning a flagged order is actually more likely to be legitimate than fraudulent.

## Question 3

SCENARIO: You are training a machine learning model to predict
    whether a food order will be cancelled before delivery. Your dataset
    has features including restaurant rating (numeric, with some missing
    values), payment method (Cash, Card, or Wallet as text), cuisine
    type (Indian, Chinese, Italian, Fast Food as text), and order value
    (numeric). Your model achieves 97% training accuracy but only 61%
    test accuracy.

Question: Identify two distinct preprocessing problems in this dataset
that are directly responsible for the large gap between training and
test accuracy. For each problem, name the specific technique you would
apply to fix it and explain the mechanism by which the unresolved issue
causes overfitting.

**Answer:**

**Problem 1: Improper Missing Value Handling in Restaurant Rating**
* **Fix Technique:** Median Imputation (or Mean Imputation) calculated strictly on the training set.
* **Mechanism:** If missing ratings are dropped only in training, imputed using the whole dataset (data leakage), or replaced with arbitrary extreme values (e.g., -1 or 0), the model learns artificial, unrepresentative patterns to split the training data. At test time, these rules fail on real unseen distributions, leading to overfitting.

**Problem 2: Flawed Encoding of Categorical Variables (Payment Method and Cuisine Type)**
* **Fix Technique:** One-Hot Encoding (applied to low-cardinality nominal text data).
* **Mechanism:** If these nominal text categories are converted using Arbitrary Integer/Label Encoding (e.g., Indian=1, Chinese=2), the model incorrectly assumes a mathematical order and magnitude between cuisines. It fits spurious linear or threshold-based patterns to these artificial numeric ranks in the training set that do not generalize to the test set.

## Question 4

SCENARIO: You are evaluating two classifiers built to detect
    fraudulent food orders. The dataset contains 50,000 orders of which
    only 500 (1%) are fraudulent. Model A achieves 99.1% overall
    accuracy. Model B achieves 94% overall accuracy but correctly
    identifies 420 of the 500 fraudulent orders. The finance team’s
    primary concern is minimising undetected fraud, not minimising false
    alarms.

Question: Explain why overall accuracy is a misleading primary metric
for this problem. Which model should the finance team adopt? Name the
specific evaluation metric that best captures the finance team’s
concern, define it mathematically, and justify your model choice using
that metric.

**Answer:**

**Why Accuracy is Misleading**
Overall accuracy is misleading because the dataset has an extreme class imbalance (99% non-fraudulent vs. 1% fraudulent). A naive model that classifies every single order as non-fraudulent would achieve 99% accuracy while failing to detect any fraud at all.

**Recommended Model**
The finance team should adopt **Model B**.

**Specific Evaluation Metric**
The metric is **Recall** (also known as Sensitivity or the True Positive Rate).

**Mathematical Definition**
$$\text{Recall} = \frac{\text{True Positives (TP)}}{\text{True Positives (TP)} + \text{False Negatives (FN)}}$$

**Justification**
Recall directly measures the proportion of actual fraudulent orders successfully detected, minimizing undetected fraud (False Negatives). 

* **Model B Recall:** $\frac{420}{500} = 84\%$
* **Model A Maximum Recall:** Even if Model A correctly classified all remaining 450 misclassifications as fraud ($50,000 \times (1 - 0.991) = 450$ errors), its recall could not exceed $\frac{450}{500} = 90\%$, and could realistically be as low as $0\%$. 

Because Model B guarantees capturing 84% of fraud cases, it best satisfies the finance team’s requirement to minimize undetected fraud.

## Question 5

SCENARIO: You are a data scientist tasked with segmenting 80,000
    food delivery customers into meaningful groups based on their order
    frequency, average spend, and preferred delivery window. You are
    deciding between K-Means and DBSCAN. Your exploratory analysis shows
    that customers naturally fall into 3–4 distinct groups with similar
    density, and there are very few genuine outlier customers.

Question: Compare K-Means and DBSCAN specifically for this use case.
Recommend one algorithm and justify your choice based on the stated data
characteristics. Then describe one meaningful limitation of your chosen
algorithm that the marketing team should understand before acting on the
segments.

**Answer:**

### Comparison

| Feature | K-Means | DBSCAN |
| :--- | :--- | :--- |
| **Cluster Shape & Density** | Assumes spherical clusters of similar density. | Finds arbitrarily shaped clusters; struggles with uniform density. |
| **Number of Clusters** | Requires specifying $k$ (matches the known 3–4 groups). | Determines clusters automatically via density thresholds ($\epsilon$, MinPts). |
| **Handling of Outliers** | Forces all points into clusters. | Flags sparse points as noise/outliers. |
| **Scalability (80k rows)** | Fast and computationally efficient ($O(n)$). | Slower and more memory-intensive ($O(n^2)$ or $O(n \log n)$). |

---

### Recommendation and Justification

**Recommendation:** K-Means

**Justification:** 
1. **Cluster Count:** You already know there are naturally 3–4 groups, making it straightforward to set $k=3$ or $k=4$.
2. **Cluster Density:** The data exhibits similar density across groups, which directly satisfies K-Means' core assumption.
3. **Outlier Profile:** With very few genuine outliers, K-Means will not suffer significantly from its sensitivity to extreme values.
4. **Data Scale:** K-Means handles 80,000 records much faster and with fewer computational resources than DBSCAN.

---

### Limitation for the Marketing Team

**Rigid Boundaries (Forced Assignment):** K-Means assigns every customer to a cluster, even those who sit near the borderline between two segments. Customers near these boundaries may not strongly represent the target segment's behavior, leading to mixed or less effective campaign results for those specific individuals.

## Question 6

SCENARIO: You are preparing a customer behaviour dataset with 35
    features — including order history, app usage patterns, location
    data, discount usage, and demographic attributes — for a K-Means
    clustering pipeline. Many of these features are highly correlated
    with each other. Your team reports that training is slow, clusters
    are hard to interpret, and visualising the results is impossible in
    the original feature space.

Question: Explain the purpose of applying PCA before K-Means in this
scenario. Describe how PCA reduces dimensionality while retaining the
most important variance in the data. What trade-off must you consciously
make when deciding how many principal components to retain, and how
would you use the explained variance ratio to guide that decision?

**Answer:**

**Purpose of PCA in this Scenario**
* **Speeds up training:** Reduces the 35 features to a smaller set of dimensions, lowering the computational burden for K-Means distance calculations.
* **Removes multicollinearity:** Transforms correlated features into linearly uncorrelated principal components, preventing redundant features from artificially dominating cluster distances.
* **Enables visualization:** Compresses the data into 2 or 3 dimensions, making it possible to plot and visually evaluate the clusters.

**How PCA Reduces Dimensionality**
PCA calculates the eigenvectors (directions of maximum spread) and eigenvalues (magnitude of spread) of the data's covariance matrix. It projects the original features onto these orthogonal axes, ordering them so the first few components capture the highest possible variance while dropping the lower-variance components that mostly represent noise.

**The Trade-off**
The trade-off is **information retention versus simplicity and computational efficiency**. Retaining more components preserves original data patterns and clustering accuracy, while retaining fewer components reduces noise, speeds up computation, and simplifies analysis at the cost of losing subtle data variations.

**Using Explained Variance Ratio to Guide the Decision**
* **Cumulative Variance Threshold:** Sum the individual explained variance ratios of each component until reaching a target threshold (typically 80% to 90% of total variance), retaining only the components needed to hit that mark.
* **Elbow Method (Scree Plot):** Plot the explained variance ratio for each component and identify the "elbow" point where the marginal variance explained by adding another component drops off significantly.
