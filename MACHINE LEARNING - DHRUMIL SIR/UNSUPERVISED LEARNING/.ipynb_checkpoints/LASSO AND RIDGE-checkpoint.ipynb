```python
'''WHAT IS LASSO ?
-Lasso is mainly used for two things:

1) REDUCE OVERFITTING IN MODEL...
-Lasso adds a penalty to large coefficients.So instead of allowing the model to make coefficients unnecessarily large, it says: "Keep the coefficients under control."
This can help the model generalize better to new data.

WHAT IS CO-EFFICIENT ?
-WHAT IS CO-EFFICIENT, MODEL WHEN TRAINING IT WILL TAKE SOME VALUES FOR EVERY COLUMN(FEATURES) AND TRY TO BEST FIT IT WITH THE 
COLUMN SO WE CAN GET ACCURATE PREDICTION = Y(OUTPUT) LIKE THIS IT WILL LEARN AND USE THAT VALUE FOR NEW DATA PREDICTION...

2) SELECT IMPORTANT FEATURES...
-Lasso decides which input columns are useful enough to keep contributing to the prediction.

THEIR IS SOMETHINGS SPECIAL ABOUT LASSO... ---> Lasso can make a coefficient exactly 0.THE COLUMNS WITH CO-EFFICIENT TAKEN 0 CONTRIBUTE NOTHING FOR THE
PREDICTION SO IT WILL REMOVE THAT COLUMN AND ONLY TAKES COLUMN THAT CONTRIBUTE FOR THE PREDICTION...

For example:
alcohol          →  0.35
pH               → -0.12
density          →  0
sulphates        →  0.18
------------------------------------------------------------------------------------------------------------------
CODE :---
------------------------------------------------------------------------------------------------------------------
from sklearn.linear_model import Lasso

model=Lasso(alpha=0.1)                 # alpha controls how strongly we want to keep the coefficients small while still keeping predictions close to actual y.BCZ BIG CO-EFFICIENT CAN LET TO BIG CHANGE AND MAKES MODEL OVERFITTING...
model.fit(x,y)
pre=model.predict(x)

for features,coefficient in zip(x.columns,model.coef_):
    print(features,":",coefficient)
------------------------------------------------------------------------------------------------------------------'''
```

```python
'''WHAT IS RIDGE ?
-Ridge is another type of regularized linear regression, just like Lasso.
-The easiest way to understand it is to compare it with Lasso.

RIDGE MAIN PURPOSE :-
-Keep the coefficients from becoming too large.

For example, suppose normal Linear Regression learns:
alcohol   →  5.2
pH        → -4.8
density   →  7.1
These coefficients may be larger than necessary.

Ridge applies a penalty and might shrink them to:
alcohol   →  3.2
pH        → -2.1
density   →  3.8

HOW IT DIFFER FROM LASSO .... --> Ridge usually does NOT make them exactly 0.
------------------------------------------------------------------------------------------------------------------
CODE :---
------------------------------------------------------------------------------------------------------------------
from sklearn.linear_model import Ridge

ridge_model = Ridge(alpha=0.1)

ridge_model.fit(x, y)

for feature, coefficient in zip(x.columns, ridge_model.coef_):
    print(feature, ":", coefficient)
------------------------------------------------------------------------------------------------------------------
```

```python
'''SO WHEN YOU DONT WANNA TO TAKE UNNECESSARY FEATURES YOU CAN USE LASSO....
AND IF YOU DON'T WANNA TO DROP FEATURES THEN USE RIDGE...

BIGGEST DIFFERNECE :-
|                                | Lasso | Ridge     |
| ------------------------------ | ----- | --------- |
| Controls large coefficients    | ✅     | ✅         |
| Helps with overfitting         | ✅     | ✅         |
| Can make coefficient exactly 0 | ✅     | ❌         |
| Performs feature selection     | ✅     | ❌         |
| Keeps all features             | ❌     | ✅         |
```
