```python
'''Imagine you have these customers:

A = (2, 3)
B = (8, 9)
C = (6, 7)
D = (100, 500)
E = (200, 600)
F = (5, 5)

You say K = 3.
That means: "I want 3 groups."

What K-Means does

Step 1: It puts 3 random centroids on the graph.
Think of a centroid as a group center.
For example, it might initially choose:

Centroid 1 → A (2,3)
Centroid 2 → D (100,500)
Centroid 3 → E (200,600)


Step 2: It looks at every customer. SO HERE DISTANCE FORMULA IS USED TO CHECK WHICH CENTROID IS CLOSEST TO WHICH CUSTOMER ?
DISTANCE FOR :- (X1-X2)2 + (Y1-Y2)2
For example:

A is closest to Centroid 1 → A goes to Group 1.
B is closest to Centroid 1 → B goes to Group 1.
D is closest to Centroid 2 → D goes to Group 2. And so on...

Step 3: Now it looks at each group and calculates the average.
That average becomes the new centroid.

For example, if Group 1 contains:
A = (2,3)
B = (8,9)
C = (6,7)
F = (5,5)

Then:
X average = (2+8+6+5)/4 = 5.25
Y average = (3+9+7+5)/4 = 6
So the new centroid is:  (5.25, 6) --> this is new centroid because The old centroid might have been somewhere else.
After seeing where the group's points actually are, K-Means says:"The average position of these points is (5.33, 6.33), so let's move the centroid there."

Then it repeats....It checks again:  "Are these customers still closest to this centroid?"
If yes → keep them.
If not → move them to another group.

----------------------------------------------------------------------------------------------------------------------------
After calculating the new centroid, K-Means checks again and again...

The process is:

1. Choose centroids
↓
2. Put each customer into the nearest centroid's group
↓
3. Calculate the average of each group
↓
4. Move the centroid to that average
↓
5. 🔄 Check again: Which centroid is now closest to each customer?
↓
6. Some customers may change groups
↓
7. Calculate average again
↓
8. Repeat

It keeps doing this until the groups/centroids stop changing.
K-Means keeps checking and moving the centroids until the clusters become stable. ✅
```

```python
'''WHAT IS K-MEANS++  ?
"K-MEANS++ chooses better starting centroids than completely random selection LIKE KMEANS.
Then both use the same K-Means process:
Assign → Average → Move centroid → Check again → Repeat

model = KMeans(init='k-means++', n_clusters=2, random_state=42)

What each part means:
KMeans() → creates the K-Means model
init='k-means++' → chooses better starting centroids
n_clusters=2 → tells the model to make 2 clusters(groups)
random_state=42 → makes the starting selection reproducible 

K-Means++ does NOT decide K for you. ✅
```

```python
'''
------------------------------------------------------------------------------------------------------------------------------------------------
UNDER THINGS WE HAVE TO USE ONLY WHEN WE DONT KNOW WHICH VALUE FOR K IS TO BE TAKEN ? SO AT THAT TIME USE UNDER THINGS...
------------------------------------------------------------------------------------------------------------------------------------------------
WCSS — Within-Cluster Sum of Squares :-
WCSS tells us how close the data points are to their cluster's centroid.

Small WCSS → points are close to centroid → good/tight cluster
Large WCSS → points are far from centroid → spread-out cluster
------------------------------------------------------------------------------------------------------------------------------------------------
Step 1: We don't know K

Suppose we don't know whether we should make:

2 clusters
3 clusters
4 clusters
etc.

So we try different K values.

For example:
K = 2
K = 3
K = 4
K = 5     We calculate WCSS for each K.
------------------------------------------------------------------------------------------------------------------------------------------------
Step 2: Let's say K = 1
Imagine our data has:
A = (2,3)
B = (4,7)
C = (6,11)
Because K = 1, we want only one cluster. So all 3 points are in one cluster.

Find the centroid:-
X average: (2 + 4 + 6) / 3 = 4
Y average: (3 + 7 + 11) / 3 = 7
So:
Centroid = (4,7)
------------------------------------------------------------------------------------------------------------------------------------------------
Step 3: Calculate distance from centroid
Now we ask:

How far is A from the centroid?
A = (2,3)
Centroid = (4,7)

For WCSS, we calculate: (2 - 4)² + (3 - 7)²   =   (-2)² + (-4)²   =   4 + 16    =    20 ------So A contributes 20.
For B: (4 - 4)² + (7 - 7)² =  0
For C: (6 - 4)² + (11 - 7)²  = 4 + 16  = 20

Now add everything: So WCSS = 40

Why is it called "Sum of Squares"?
Because we: Find difference from centroid and square it...

so like this we take k=2 then it will make 2 clustors...
cluster 1 :- A & B
cluster 2 :- C
let say each cluster has suppose wcss value,
WCSS Cluster 1 = 10
WCSS Cluster 2 = 0       total = 10 .....

like this for k=3 if total wcss = 0 then you will think that more k value means wcss becomes smaller and thats good but...
we cannot simply say "lowest WCSS is best."

Instead, we use the: Elbow Method
We calculate WCSS for different K values:
K = 1 → WCSS = 500
K = 2 → WCSS = 250
K = 3 → WCSS = 120
K = 4 → WCSS = 110
K = 5 → WCSS = 105

Then make a graph: K on X-axis  &&  WCSS on Y-axis....
We look for the point where the WCSS starts decreasing much more slowly.

see the graph...
WCSS
 |
 |\
 | \
 |  \
 |   \
 |    \__
 |       \___
 |____________
   1  2  3  4  5
             K

now here at k=3 is the elbow... so we choose k=3 value...

CONCLUSION :-- WCSS is used with the Elbow Method to help decide which value of K is suitable.
```
