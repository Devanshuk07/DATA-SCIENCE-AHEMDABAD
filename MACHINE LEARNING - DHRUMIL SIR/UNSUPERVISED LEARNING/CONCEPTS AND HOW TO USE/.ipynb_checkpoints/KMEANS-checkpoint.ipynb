```python
'''DIFFERNECE BTW SUPERVISED AND UNSUPERVISED...
SUPER--> WE ALREADY HAVE LABEL IN OUR DATA FROM IT OUR MODEL WAS BEEN TRAINED...
UNSUPER--> WE DONT HAVE LABEL IN OUR DATA OUR MODEL WILL BY ITS OWN UNDERSATND THE FEATURES FROM DATA AND WILL DIVIDE IT INTO DIFFERENT GROUPS...
```

```python
'''ALGO FOR UNSUPERVISED LEARNING:--
1) K-MEANS
2) DBSCAN
3) HEIRARCHICAL CLUSTERING'''
```

```python
'''WE HAVE TO IMPLEMENT 3 THINGS INDIVIDUALLY IN EVERY ALGORITHM :-
1) CLUSTERING :- WILL CREATE GROUP 
2) DIMENSIONALLY REDUCTION(TECNIQUE=PCA) :- WE WILL REDUCE OUR COULMNS(FEATURES)/ DROP WHICH COLUMN DONT MAKE ANY DIFFERNECE/UNNECESSARY COLUMNS..(METHOD = PCA)
3) ASSOCIATION :- (USES APRI ORI) :- THE THINGS WITH BEST ASSOCIATION WILL BE GIVEN AS SUGGETIONS. SUGGESTIONS AS PER YOUR LIKES/CHOICES'''
```

```python
'''WHEN TO USE UNSUPERVISED :-
WHEN YOU WANT TO MAKE CLUSTER(GROUPS)AND DO ASSOCIATION RELATION-SHIP...

EX:- CLUSTER FOR CUSTOMERS WHO ORDERED ICE-CREAM 7-10 TIMES IN A MONTH THAT ALL CUSTOMER IN ONE CLUSTER.
NOW ASSOCIATION WILL GIVE SUGGESTIONS ACCORDING TO ICE-CREAM LIKE CHOCOBAR,CANDY LIKE THIS... TO IMPROVE SELLS...
BUT IT WILL NOT SUGGEST WAFFERS IN ICE-CREAM AS THEY BOTH HAVE DIFFERNET TASTE...'''
```

```python
'''1) K-MEANS :- CLUSTER K=3(3GROUPS)

CUSTOMER     SPENDING    INCOME
 A            2            3
 B            8            9
 C            6            7
 D           100          500
 E           200          600
 F            5            5
 
YOU HAVE TO FIRST DEFINE K...(SO ACCORDING TO K VALUE IT WILL CREATE GROUP)
YOU CANNOT GIVE CONDITION HERE TO CREATE CLUSTER - MODEL WILL UNDERSTAND IT BY ITS OWN AND MAKE CLUSTER...)

A(2,3) --> 2=X-AXIS 3=Y-AXIS
B(8,9) --> 8=X-AXIS 9=Y-AXIS
C(6,7) --> 6=X-AXIS 7=Y-AXIS

SO , X-AXIS = 2+8+6 = 16 /3 =6 (LET)
SO , Y-AXIS = 3+9+7 = 19/3 =7(LET)
SO THIS (6,7) IS THE CENTROID OF THIS CLUSTOR...CENTROID IS THE AVG...
THEN ALL VALUES LIKE SPENDING(2,8,6,5)FOR X-AXIS AND INCOME(3,9,7,5)FOR Y-AXIS ACCORDING TO (6,7) IS THE CENTROID IS TAKEN IN ONE CLUSTER...
AND THE VALUES SPENDING(100,200) AND INCOME(500,600) ARE TAKEN IN DIFFERENT CLUSTER...

DISTANCE FORMULA :-  (X2-X1)2 + (Y2-Y1)2   --> CENTROID USE THIS FORMULA TO CHOOSE WHICH VALUE WILL BE IN WHICH CLUSTOR AFTER TAKING (6,7)AS CENTROID.
LET WE HAVE ,,
K=2  A(2,3) B(4,5) C(7,8)  D(9,10)  --> HERE A,B,C,D IS NAMES AND 2,3,4,5,6,7,8,9,10 ARE SPENDING AND INCOME....
LET SAY AVG AND X-AXIS AND Y-AXIS IT HAS...
CENTROID ---> C1(2,3) -->(X1,Y1)
         ---> C2(7,8) -->(X1,Y1)

DISTANCE --> (X2-X1)2 + (Y2-Y1)2
         --> A= (2-2)2 =0  + (3-3)2 = 0
         --> B= (4-2)2 = 4 +  (5-3)2 = 4  = 8
         --> C= (7-7)2=0  + (8-8)2 =  0  = 0
         --> D= (9-7)2=4  + (10-8)2 = 4 = 8

SO AFTER THIS CENTROID IS UPDATED AGAIN...
WE WILL TAKE AVG OF A AND B AGAIN...C1(2,3) --> A(2,3),B(4,5)   ---> AFTER AVG VALUE IS (2+4)/2 = 3 AND (3+5)/2 = 4 SO
UPDATED CENTROID IS (3,4)--> AGAIN IT WILL DIVIDE DATA BY DISTANCE FORMULA SAME AS BEFORE...
THIS WILL BE DONE TILL OUR C1 AND UPDATED CENTROID WILL HAVE AT A TIME SAME A AND B VALUES IN THEM ,, AT THAT TIME IT WILL STOP.. AS IT WILL BE REPEATED..

WHAT IS K-MEANS++ ==> IT WILL ALSO GENERATE RANDOM CLUSTOR LIKE K-MEANS BUT IT WILL BE MORE ACCURATE.
CODE :- MODEL = K_MEANS(INIT=K_MEANS++, K=2 ,RANDOM_STATE=42)   # DONT RANDOMLY USE K=2 AS WILL GIVE OUTPUT BUT ACCORDING TO YOUR DATASET YOUR 2,3,4,
CLUSTOR IS OK WILL BE BEST DECIDE BY DIFFERENT METHOD...
------------------------------------------------------------------------------------------------------------------------------------------------
UNDER THINGS WE HAVE TO USE ONLY WHEN WE DONT KNOW WHICH VALUE FOR K IS TO BE TAKEN ? SO AT THAT TIME USE UNDER THINGS...
------------------------------------------------------------------------------------------------------------------------------------------------
1) WCSS(WITHIN CLUSTER SUM OF SQUARE) :- FIRST WE HAVE TO RANDOMLY DECIDE HOW MANY CLUSTER YOU WANNA TO TAKE LOWEST IS BETTER CLUSTOR...
  A=2RS, B=4RS, C=6RS
CENTROID = 2+4+6/3 = 4,7(LET) HERE 4=X-AXIS AND 7=Y-AXIS,,NOW WE HAVE TO FIND WCSS... FOR EVERY CLUSTOR YOU HAVE SELECTED YOU WILL GET CENTROID... AND ON THAT WCSS WILL BE DONE...
THEN ,  4-2 = 2    4 FOR X-AXIS
        4-4 = 0 
        4-6 = 2
NOW SUM... 2+0+2 = 4(SQUARE OF THIS VALUE) = 16 --> THIS IS WCSS
FOR EVERY CENTROID WE HAVE TWO WCSS... ONE FOR X-AXIS AND ONE FOR Y-AXIS...
NOW WHEN YOU TOOK K=CLUSTER =2 , 3 ,4 THEN FOR EVERY CLUSTER YOU WILL GET WCSS FOR INDIVIDUAL CLUSTER...
THEN YOU WILL DECIDE FROM THAT WCSS WHICH K VALUE YOU SHOULD TAKE...LOWEST IS BETTER CLUSTOR...WCSS WORK HERE OVER...
------------------------------------------------------------------------------------------------------------------------------------------------
2) INERTIA :-NOW THAT INDIVIDUAL WCSS CLUSTOR VALUE WILL BE ADDED THAT IS INERTIA....
------------------------------------------------------------------------------------------------------------------------------------------------
3) ELBOW METHOD :- ELBOW METHOD IS PURELY DEPEND ON INERTIA...IN WHICH CLUSTER YOUR 
LET WE TOOK K=1 AND WE GET CENTROID AND THEN WE DONE WCSS THEN ITS ALL VALUES GIVEN TO INERTIA THEN IT WILL SUM OF ALL WCSS VALUES.. AND WE GET LET(150)
LIKE THIS K=2 (INERTIA=50), K=3(INERTIA=25) , K=4(INERTIA=23) THEN :- HERE WE WIL NOT TAKE LOWEST....WE WILL SELECT THAT VALUE WHICH MAKES OUR INTERIA STABLE.THE VALUE AT WHICH OUR
INERTIA GETS DECREASE SLOWLY IS STABLE...

EX:-
K-VALUE  INETRIA
K=1        150
K=2         50
K=3         25   ---> HERE IT STARTS TO DECREASE SLOWLY SO HERE IT IS STABLE THAT IS DECIDE BY ELBOW...WE CAN SEE IT IN GRAPHS AND ALSO IN MODEL._INERTIA....
K=4         23
K=5         21
```
