```python
'''What is Hierarchical Clustering in Unsupervised Learning?

Hierarchical Clustering is an unsupervised machine learning algorithm that groups similar data points into clusters and 
creates a tree-like structure called a Dendrogram.It does not require labeled data.

EXAMPLE:-

Suppose we have 5 customers with different spending habits. We want to group similar customers together.

-Customers who spend similarly are grouped together.
-Different groups are formed based on similarity.
-We can visualize these groups using a dendrogram.

Types of Hierarchical Clustering :-

1. Agglomerative Clustering (Bottom-Up):-
Initially, every data point is a separate cluster.
The closest clusters are merged step by step.
Finally, all points can form one large cluster.
Most commonly used

2. Divisive Clustering (Top-Down):-
Initially, all data points belong to one cluster.
The cluster is split into smaller clusters step by step.
Finally, individual data points can become separate clusters.

What is a Dendrogram?
-A dendrogram is a tree-like diagram that shows how data points or clusters are joined or split at different distances.
```
