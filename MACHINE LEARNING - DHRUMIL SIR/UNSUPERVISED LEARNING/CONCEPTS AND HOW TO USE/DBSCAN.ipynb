```python
'''WHAT IS DBSCAN :-
DBSCAN = Density-Based Spatial Clustering of Applications with Noise
It is a clustering algorithm, just like K-Means.

Main difference: 
IN K-MEANS YOU TELL IT VALUE OF K=3  MEANS MAKE 3 CLUSTERS. IN DBSCAN YOU DO NOT TELL IT VALUE OF K... INSTEAD YOU GIVE EPS AND MIN_SAMPLE...

DBSCAN looks at the density of points and decides how many clusters it can form.WITH THE HELP OF EPS AND MIN_SAMPLE.
---------------------------------------------------------------------------------------------------------------------------------------------------
WHAT IS EPS ?
-Think of eps as a circle around a point.
For example: eps=2 then look within 2 km around the point. If another point is within that distance, it is considered a neighbor.
K-distance graph / k-distance plot can be used to help choose a suitable eps value.It doesn't automatically decide the perfect EPS for you.
---------------------------------------------------------------------------------------------------------------------------------------------------
What is min_samples?
-min_samples tells DBSCAN: How many points should be present inside that eps neighborhood to consider it a dense area.

DBSCAN(eps=2, min_samples=3)
means :- Look within distance 2. If there are at least 3 points there, this is considered a dense region.

EXAMPLE:-
A = (2,3)
B = (3,2)
C = (4,4)
D = (3,4)
E = (20,30)

So DBSCAN can potentially make:  Cluster 1 → A, B, C, D     Noise → E

DBSCAN identifies points that don't belong to any dense cluster and labels them as noise/outliers.

---------------------------------------------------------------------------------------------------------------------------------------------------
DBCAN WORK ON 3 POINTS:-

1) CORE POINT :- 
- EVERY CONDITION THAT WILL BE FULFILLED TO MAKE A CLUSTER...
EX:- IF MARKS > 40 ... HERE IT WILL BE MORE THAN 40 ONLY...
ABOVE EX:- 

2) BORDER POINT:-
-CONDITION WILL NOT BE FULFILLED WILL BE AT THE BORDER OF OUR CONDITION.
EX:- IF MARKS > 40 ... HERE IT WILL BE 38,39 MEANS LESS THAN 40 BUT NOT TOO MUCH LESS LIKE THIS...

3) NOISE POINT:-
-VALUES WILL BE FAR FROM CONDITION
EX:- IF MARKS > 40 ... HERE IT WILL BE FAR AWAY 2,3,5,10 LIKE THIS
```
