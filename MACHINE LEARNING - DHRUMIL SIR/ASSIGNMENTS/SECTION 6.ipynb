```python
'''Download a dataset of mobile phone prices (with features like RAM, storage, camera MP, battery, brand, etc.) and 
fit a multiple linear regression model to predict the price using at least three features of your choice.'''
import numpy as np
import pandas as pd

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

data=pd.read_csv('mobile_phone.csv')
data.head()
```

```python
x=data[['RAM_GB','Brand','Storage_GB']]

y=data['Price_INR']

x=pd.get_dummies(x, columns=['Brand'])     # this is used to convert string to float like indicators... easy for ml to understand... you can also give dtype=int to convert each brand to 0,1 category...

x_train,x_test,y_train,y_test=train_test_split(
    x,y,train_size=0.80,random_state=42
)

model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y=model.predict(x_test)
y
```

```python
result=pd.DataFrame({
    'Original':y_test.values,
    'Prediction':y
})
result
```

```python
scr=r2_score(y_test,y)
scr

# Your model explains about 88% of the variation in mobile phone prices in your test data using RAM, Storage, and Brand.
```
