## Question 1

List 3 differences between Reinforcement Learning, Supervised Learning, and Unsupervised Learning in a table, using real examples from apps like Instagram (feed ranking), Spotify (music recommendations), and chess game AI.

**Answer:**

Here is a comparison table showing 3 key differences:

| Difference | Supervised Learning | Unsupervised Learning | Reinforcement Learning |
| :--- | :--- | :--- | :--- |
| **1. Data Type** | Uses **labeled data** (inputs matched with correct answers). | Uses **unlabeled data** (inputs without any target answers). | Uses **trial-and-error data** (interactions with an environment). |
| **2. Learning Goal** | Learns to predict the correct label for new data. | Learns to find hidden patterns, groupings, or clusters. | Learns to take actions that maximize rewards over time. |
| **3. Real App Example** | **Instagram (Feed Ranking):** Predicts if you will like a post based on your past labeled actions (liked vs. skipped). | **Spotify (Recommendations):** Clusters songs with similar tempos and tones together without being told the genres. | **Chess Game AI:** Learns the best moves by playing games, receiving positive rewards for winning and penalties for losing. |

## Question 2

Pick any one real-world application of Reinforcement Learning (such as game AI, robotics, self-driving cars, or recommendation systems) and explain step-by-step how an agent, environment, actions, and rewards would work in that scenario.

Hint: Use your own words and pick an example you find interesting, like how a Zomato recommendation engine could use RL to suggest restaurants.

**Answer:**

**Application: Restaurant Recommendation System (like Zomato)**

* **Agent:** The recommendation algorithm that decides which restaurant to suggest.
* **Environment:** The user using the app, along with their context like location, time of day, and past orders.
* **Actions:** Showing a specific restaurant or dish on the user's home screen.
* **Rewards:** The agent gets a positive reward (+1) if the user clicks and places an order, and a negative or zero reward if the user scrolls past without ordering. 

Over time, the agent learns from these rewards to suggest restaurants that the user is most likely to order from.

## Question 3

Imagine you are designing a simple RL agent for a cricket game app (like IPL fantasy league). Describe what could be the agent, environment, possible actions, and how rewards would be assigned for the agent's choices.


**Answer:**

Here is how you can define the Reinforcement Learning components for a fantasy cricket app:

* **Agent:** The automated decision-maker (the bot) that selects and manages your fantasy team.
* **Environment:** The fantasy game platform, which includes real-world player statistics, match rules, budget limits, and pitch conditions.
* **Actions:** Choosing the 11 players within budget, picking the captain and vice-captain, and making substitutions before matches.
* **Rewards:** The agent receives positive points (rewards) when selected players perform well in real matches (scoring runs, taking wickets, or catches) and zero or negative points if players underperform or do not play.

## Question 4

Think of a feature in any app you use (like YouTube autoplay, Flipkart product suggestions, or Spotify playlist generation) that could be improved using Reinforcement Learning. Briefly describe how RL could make the feature smarter, and what the reward signal might be.


**Answer:**

**App Feature:** YouTube "Up Next" Autoplay.

YouTube's autoplay feature can be improved using Reinforcement Learning (RL) to optimize for a user's entire viewing session rather than just their next immediate click. An RL agent can learn the best sequence of videos to play over time, figuring out when to show similar content and when to introduce fresh topics to prevent boredom. The **reward signal** would be positive (+1) when a user watches a recommended video for a long duration or gives it a "thumbs up," and negative (-1) if the user skips the video within a few seconds or exits the app. This feedback allows the agent to continuously adapt its choices to maximize long-term user satisfaction and watch time.
