## Question 1

Load the Iris dataset using sklearn, apply PCA to reduce it to 2 principal components, and print the explained variance ratio for both components.

**Answer:**

Here is the Python code to do this:

```python
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA

# 1. Load the Iris dataset
iris = load_iris()
X = iris.data

# 2. Apply PCA to reduce to 2 components
pca = PCA(n_components=2)
pca.fit(X)

# 3. Print the explained variance ratio
print("Explained variance ratio:", pca.explained_variance_ratio_)
```

### Explanation:
1. We load the Iris dataset features into `X`.
2. We create a `PCA` object with `n_components=2` and fit it to our data to find the top two principal components.
3. We print `explained_variance_ratio_`, which shows the percentage of the dataset's total variance (information) captured by each of the two components.

## Question 2

Plot the 2D projection of the Iris dataset after PCA transformation, using a different color for each species, similar to how Spotify visualizes genre clusters.

Hint: Use matplotlib's scatter() and color by target label.

**Answer:**

Here is the Python code to reduce the Iris dataset to 2 dimensions using PCA and plot the result:

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA

# 1. Load the built-in Iris dataset
iris = load_iris()
X = iris.data
y = iris.target

# 2. Reduce data from 4 features to 2 principal components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# 3. Create the scatter plot
plt.figure(figsize=(7, 5))
scatter = plt.scatter(
    X_pca[:, 0], X_pca[:, 1], c=y, cmap="viridis", edgecolor="k"
)

# 4. Add labels, title, and legend
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("2D PCA Projection of Iris Dataset")
plt.legend(
    handles=scatter.legend_elements()[0],
    labels=list(iris.target_names),
    title="Species",
)

plt.show()
```

### Explanation:
1. **PCA Transformation:** `PCA(n_components=2)` compresses the 4 original measurements into 2 summary features (Principal Components) while keeping as much pattern information as possible.
2. **Plotting:** `plt.scatter()` graphs the two new components on the X and Y axes. 
3. **Coloring:** Setting `c=y` colors each dot based on its species label, showing how the different species group into distinct clusters.

## Question 3

Given a dataset of Zomato restaurant ratings with 10 features, use PCA from sklearn to reduce the data to 3 components and visualize the cumulative explained variance as a line plot.

**Answer:**

Here is the Python code to perform PCA and plot the cumulative explained variance.

### Python Code

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# 1. Create sample data (Zomato dataset with 10 features)
np.random.seed(42)
sample_data = pd.DataFrame(
    np.random.rand(50, 10), columns=[f"feature_{i+1}" for i in range(10)]
)

# 2. Scale the data (important for PCA)
scaler = StandardScaler()
scaled_data = scaler.fit_transform(sample_data)

# 3. Apply PCA to reduce to 3 components
pca = PCA(n_components=3)
pca.fit(scaled_data)

# 4. Calculate cumulative explained variance
cum_variance = np.cumsum(pca.explained_variance_ratio_)

# 5. Plot the cumulative explained variance
plt.figure(figsize=(6, 4))
plt.plot([1, 2, 3], cum_variance, marker="o", color="b")
plt.title("Cumulative Explained Variance")
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Variance Ratio")
plt.xticks([1, 2, 3])
plt.grid(True)
plt.show()
```

### Explanation
First, we create a sample dataset with 10 features and standardize it using `StandardScaler`. Next, we fit `PCA` with 3 components to compress the data. Finally, we use `np.cumsum()` on `pca.explained_variance_ratio_` to find the total variance captured by the components and display it using a line plot.

## Question 4

Take the PCA-transformed Iris data (2 components) and run KMeans clustering with 3 clusters. Plot the clusters and centroids on the 2D PCA plot.

**Answer:**

Here is the code to apply PCA, run KMeans, and plot the results:

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA

# 1. Load Iris data
iris = load_iris()
X = iris.data

# 2. Reduce to 2 dimensions with PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# 3. Run KMeans with 3 clusters
kmeans = KMeans(n_clusters=3, random_state=42)
labels = kmeans.fit_predict(X_pca)
centroids = kmeans.cluster_centers_

# 4. Plot the clusters and centroids
plt.scatter(
    X_pca[:, 0], X_pca[:, 1], c=labels, cmap="viridis", label="Data Points"
)
plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    c="red",
    marker="X",
    s=200,
    label="Centroids",
)

plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.title("KMeans Clustering on Iris (2D PCA)")
plt.legend()
plt.show()
```

### Explanation:
1. **PCA**: Reduces the original 4-feature Iris data down to 2 principal components so we can plot it in 2D.
2. **KMeans**: Groups the 2D points into 3 clusters and calculates the center point (centroid) of each group.
3. **Plot**: Displays each data point colored by its assigned cluster, with the centroids marked as large red **X** symbols.

## Question 5

Use the silhouette_score function from sklearn.metrics to evaluate the quality of your KMeans clusters on the PCA-reduced Iris data. Print the silhouette score and explain what it indicates about your clustering result in 2-3 lines.

**Answer:**

Here is the Python code using the Iris dataset:

```python
from sklearn.cluster import KMeans
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

# 1. Load Iris data and reduce to 2 dimensions using PCA
iris = load_iris()
pca = PCA(n_components=2)
X_pca = pca.fit_transform(iris.data)

# 2. Fit KMeans clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init="auto")
labels = kmeans.fit_predict(X_pca)

# 3. Calculate and print the silhouette score
score = silhouette_score(X_pca, labels)
print(f"Silhouette Score: {score:.3f}")
```

**Output:**
```text
Silhouette Score: 0.598
```

**What the score indicates:**
The silhouette score ranges from -1 to 1, where higher values mean better-defined clusters. A score of around 0.60 indicates that the clusters are well-separated and data points are closer to their assigned cluster than to neighboring clusters.
