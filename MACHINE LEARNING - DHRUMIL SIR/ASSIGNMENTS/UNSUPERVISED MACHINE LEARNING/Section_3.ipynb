## Question 1

Load the 'Mall Customers' dataset (or any small customer dataset) and apply the DBSCAN algorithm using scikit-learn to cluster users based on their annual income and spending score. Print the number of clusters found and the number of noise points detected.

**Answer:**

Here is the Python code using a small sample dataset to demonstrate DBSCAN clustering.

### Python Code

```python
import pandas as pd
from sklearn.cluster import DBSCAN

# --- Sample Data (Invented customer data) ---
data = {
    "Annual_Income": [15, 16, 17, 29, 30, 70, 72, 75, 80, 20],
    "Spending_Score": [39, 81, 77, 40, 42, 75, 79, 85, 12, 10],
}
df = pd.DataFrame(data)

# Select the features for clustering
X = df[["Annual_Income", "Spending_Score"]]

# Apply DBSCAN
# eps is the maximum distance between two points to be considered neighbors
# min_samples is the minimum number of points required to form a cluster
dbscan = DBSCAN(eps=15, min_samples=2)
labels = dbscan.fit_predict(X)

# Count clusters and noise points
# DBSCAN labels noise points as -1
n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
n_noise = list(labels).count(-1)

print(f"Number of clusters found: {n_clusters}")
print(f"Number of noise points detected: {n_noise}")
```

### Explanation
1. We create a small sample DataFrame with annual income and spending score values.
2. We initialize `DBSCAN` by setting `eps` (distance threshold) and `min_samples` (minimum points to make a dense region).
3. The model assigns each point a cluster label, using `-1` for noise points.
4. Finally, we count the unique cluster labels (ignoring `-1`) and count how many points are labeled as `-1` (noise).

## Question 2

Visualize the DBSCAN clustering results from your previous task using matplotlib — plot each cluster in a different color, and mark noise points in black. Add axis labels and a legend for clarity.

**Answer:**

Here is the code to visualize DBSCAN clustering results.

### Python Code

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.cluster import DBSCAN

# Sample data (created for demonstration)
X = np.array(
    [
        [1, 2],
        [2, 2],
        [2, 3],  # Cluster 0
        [8, 7],
        [8, 8],
        [7, 8],  # Cluster 1
        [20, 20],  # Noise point
    ]
)

# Apply DBSCAN
dbscan = DBSCAN(eps=2, min_samples=2)
labels = dbscan.fit_predict(X)

# Plot each cluster
unique_labels = set(labels)
colors = plt.cm.tab10(np.linspace(0, 1, len(unique_labels)))

for label, color in zip(unique_labels, colors):
    if label == -1:
        # Black color for noise points
        plt.scatter(
            X[labels == label, 0],
            X[labels == label, 1],
            c="black",
            marker="x",
            s=100,
            label="Noise",
        )
    else:
        # Different colors for normal clusters
        plt.scatter(
            X[labels == label, 0],
            X[labels == label, 1],
            color=color,
            s=50,
            label=f"Cluster {label}",
        )

# Add labels and legend
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("DBSCAN Clustering Results")
plt.legend()
plt.show()
```

### Explanation
- **Data & Model:** We create a small 2D dataset and run `DBSCAN` to get cluster labels.
- **Plotting:** We loop through each unique label and use `plt.scatter()` to plot the points.
- **Noise Points:** In DBSCAN, noise is assigned the label `-1`. The code checks for this label and colors those points black with an `'x'` marker.
- **Labels & Legend:** `plt.xlabel()`, `plt.ylabel()`, and `plt.legend()` are used to make the plot clear and readable.

## Question 3

Change the DBSCAN parameters (eps and min_samples) to see how the number of clusters and noise points change. Write a short summary (3-4 lines) explaining how parameter choices affect the clustering outcome for your dataset.

Hint: Try at least two different values for eps and min_samples, and compare results.

**Answer:**

### Python Code

Here is a simple example using a small sample dataset:

```python
import pandas as pd
from sklearn.cluster import DBSCAN

# Sample data (clearly labeled invented data)
data = pd.DataFrame({"x": [1, 2, 2, 8, 8, 9, 25], "y": [2, 1, 2, 8, 9, 8, 25]})

# Testing two different sets of parameters
params = [
    {"eps": 2.0, "min_samples": 2},  # Setting 1: Larger radius
    {"eps": 0.5, "min_samples": 3},  # Setting 2: Smaller radius, more points
]

for p in params:
    model = DBSCAN(eps=p["eps"], min_samples=p["min_samples"]).fit(data)
    labels = model.labels_

    # Count clusters (ignoring noise labeled as -1) and noise points
    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
    n_noise = list(labels).count(-1)

    print(
        f"eps={p['eps']}, min_samples={p['min_samples']} -> Clusters: {n_clusters}, Noise points: {n_noise}"
    )
```

**Output:**
```text
eps=2.0, min_samples=2 -> Clusters: 2, Noise points: 1
eps=0.5, min_samples=3 -> Clusters: 0, Noise points: 7
```

---

### Summary

1. Increasing `eps` makes the search radius larger, which connects more points together and reduces the number of noise points. 
2. In contrast, increasing `min_samples` or decreasing `eps` makes the rules for forming a cluster stricter. 
3. If `eps` is too small or `min_samples` is too high, most points fail to form clusters and are marked as noise.

## Question 4

Compare DBSCAN and K-Means on the same dataset: run both algorithms, plot their cluster assignments side by side, and briefly note which method better handles outliers or oddly-shaped clusters.

Constraint: Use the same features and number of clusters (for K-Means) as you observed with DBSCAN.

**Answer:**

### Python Code

Here is the code using sample data to compare both models:

```python
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN, KMeans
from sklearn.datasets import make_moons

# 1. Create sample data (two crescent shapes with some noise)
X, _ = make_moons(n_samples=200, noise=0.1, random_state=42)

# 2. Run DBSCAN
dbscan = DBSCAN(eps=0.2, min_samples=5)
db_labels = dbscan.fit_predict(X)

# Count clusters from DBSCAN (ignoring noise labeled as -1)
n_clusters_found = len(set(db_labels)) - (1 if -1 in db_labels else 0)

# 3. Run K-Means using the same number of clusters found by DBSCAN
kmeans = KMeans(n_clusters=n_clusters_found, random_state=42)
km_labels = kmeans.fit_predict(X)

# 4. Plot results side by side
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# DBSCAN Plot
axes[0].scatter(X[:, 0], X[:, 1], c=db_labels, cmap="viridis")
axes[0].set_title(f"DBSCAN (Clusters: {n_clusters_found})")

# K-Means Plot
axes[1].scatter(X[:, 0], X[:, 1], c=km_labels, cmap="viridis")
axes[1].set_title(f"K-Means (k={n_clusters_found})")

plt.show()
```

### Explanation of the Code
* We generate non-linear sample data shaped like two crescents.
* We apply **DBSCAN** first to discover the natural density-based clusters and count them (excluding noise).
* We pass that same cluster count into **K-Means** and plot both results side by side to compare how they group the data points.

### Comparison
DBSCAN handles oddly-shaped clusters and outliers much better than K-Means. K-Means assumes clusters are spherical, so it incorrectly splits the crescent shapes down the middle. In contrast, DBSCAN follows the density of the shapes to group them correctly and marks stray points as noise (outliers).

## Question 5

Pick a dataset with non-spherical clusters (e.g., make_moons or make_circles from sklearn.datasets), and use DBSCAN to cluster the data. Identify and print the indices of all points labeled as noise (outliers).

**Answer:**

Here is the Python code using the `make_moons` dataset:

```python
import numpy as np
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons

# 1. Generate sample data with non-spherical clusters
X, _ = make_moons(n_samples=100, noise=0.1, random_state=42)

# 2. Apply DBSCAN clustering
# eps is the maximum distance between points, min_samples is the minimum points to form a dense region
dbscan = DBSCAN(eps=0.2, min_samples=5)
labels = dbscan.fit_predict(X)

# 3. Find indices where the label is -1 (noise)
noise_indices = np.where(labels == -1)[0]

# 4. Print the result
print("Indices of points labeled as noise (outliers):")
print(noise_indices)
```

### Explanation:
DBSCAN automatically identifies points in low-density areas as noise and gives them a cluster label of `-1`. We use `np.where(labels == -1)[0]` to find and extract the exact row numbers (indices) of these noise points from the dataset.
