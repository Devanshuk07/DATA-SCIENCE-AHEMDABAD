## Question 1

Given a small dataset of 6 food delivery orders (with features like delivery_time and distance), manually create the initial distance matrix using the Euclidean distance formula between each pair of orders.

Hint: Use the formula sqrt((x2 - x1)^2 + (y2 - y1)^2) for each pair.

**Answer:**

Here is the Python code to create the sample data and calculate the Euclidean distance matrix manually.

### Python Code

```python
import math
import pandas as pd

# 1. Create sample data for 6 food delivery orders
# delivery_time in minutes, distance in miles
sample_data = {
    "Order": ["O1", "O2", "O3", "O4", "O5", "O6"],
    "delivery_time": [10, 20, 15, 25, 30, 12],
    "distance": [2, 4, 3, 5, 7, 1],
}

df = pd.DataFrame(sample_data)
print("--- Sample Data ---")
print(df)

# 2. Compute the Euclidean distance matrix manually
orders = df["Order"].tolist()
times = df["delivery_time"].tolist()
distances = df["distance"].tolist()
n = len(orders)

# Initialize an empty matrix
matrix = []

for i in range(n):
    row = []
    for j in range(n):
        # Euclidean formula: sqrt((x2 - x1)^2 + (y2 - y1)^2)
        dist = math.sqrt((times[i] - times[j]) ** 2 + (distances[i] - distances[j]) ** 2)
        row.append(round(dist, 2))
    matrix.append(row)

# Convert to DataFrame for easy reading
distance_matrix = pd.DataFrame(matrix, index=orders, columns=orders)
print("\n--- Euclidean Distance Matrix ---")
print(distance_matrix)
```

### Explanation
We create a small dataset of 6 orders with two features: delivery time and distance. Using nested loops, we compare every pair of orders and apply the formula $\sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$. The result is a $6 \times 6$ table where the diagonal is $0.0$ because the distance from an order to itself is zero.

## Question 2

Using Python and sklearn, perform agglomerative hierarchical clustering on a dataset of 8 songs (with features like tempo and energy), and plot the resulting dendrogram using scipy's dendrogram function.

**Answer:**

Here is the Python code using sample data:

```python
import matplotlib.pyplot as plt
import pandas as pd
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering

# --- Sample Data: 8 songs with Tempo and Energy ---
data = {
    "Song": [f"Song {i+1}" for i in range(8)],
    "Tempo": [120, 125, 70, 75, 130, 65, 140, 80],
    "Energy": [0.8, 0.85, 0.3, 0.25, 0.9, 0.2, 0.95, 0.35],
}
df = pd.DataFrame(data)
X = df[["Tempo", "Energy"]]

# 1. Perform Agglomerative Clustering with sklearn
model = AgglomerativeClustering(n_clusters=2, linkage="ward")
df["Cluster"] = model.fit_predict(X)

# 2. Create the linkage matrix and plot the dendrogram using scipy
linkage_matrix = linkage(X, method="ward")

plt.figure(figsize=(8, 4))
dendrogram(linkage_matrix, labels=df["Song"].values)
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Song")
plt.ylabel("Distance")
plt.show()
```

### Explanation
- **Data:** We create a sample DataFrame with 8 songs based on their tempo and energy.
- **Clustering:** `AgglomerativeClustering` from `sklearn` groups similar songs into clusters based on their feature distances.
- **Dendrogram:** `scipy`'s `linkage()` calculates the step-by-step merges, and `dendrogram()` visualizes these connections as a tree diagram.

## Question 3

Change the linkage method in your clustering code from 'single' to 'complete' and 'average', and observe how the dendrogram structure changes for your songs dataset.

Hint: Use the linkage parameter in AgglomerativeClustering or scipy's linkage function.

**Answer:**

Here is the Python code using sample song data to compare different linkage methods.

### Python Code

```python
import matplotlib.pyplot as plt
import pandas as pd
from scipy.cluster.hierarchy import dendrogram, linkage

# Sample data: Song features (Tempo in BPM, Energy score 1-10)
data = {
    "Tempo": [120, 125, 80, 85, 130],
    "Energy": [8, 9, 3, 2, 8],
}
df = pd.DataFrame(data, index=["Song A", "Song B", "Song C", "Song D", "Song E"])

# Linkage methods to compare
methods = ["single", "complete", "average"]

# Plot dendrograms
plt.figure(figsize=(12, 4))
for i, method in enumerate(methods, 1):
    plt.subplot(1, 3, i)
    linked = linkage(df, method=method)
    dendrogram(linked, labels=df.index)
    plt.title(f"{method.capitalize()} Linkage")

plt.tight_layout()
plt.show()
```

### Observation

Different linkage methods change how clusters are merged in the dendrogram:
- **'Single' linkage** measures the shortest distance between points, which often causes a "chaining" effect where points join one by one at low heights.
- **'Complete' linkage** measures the maximum distance, resulting in more compact, well-separated clusters with taller vertical branches.
- **'Average' linkage** uses the mean distance between all points, creating a balanced tree structure that sits between single and complete linkage.

## Question 4

Write a function that, given two points (x1, y1) and (x2, y2), returns both their Euclidean and Manhattan distances. Test it on three pairs of coordinates from your dataset.

**Answer:**

Here is the Python function and the test using a small sample dataset.

### Python Code

```python
import math


def calculate_distances(p1, p2):
    x1, y1 = p1
    x2, y2 = p2

    # Euclidean distance formula
    euclidean = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

    # Manhattan distance formula
    manhattan = abs(x2 - x1) + abs(y2 - y1)

    return euclidean, manhattan


# --- Sample Data (Invented dataset of 3 coordinate pairs) ---
sample_pairs = [((1, 2), (4, 6)), ((0, 0), (3, 3)), ((-1, -1), (2, 3))]

# Testing the function
for i, (point_a, point_b) in enumerate(sample_pairs, start=1):
    euc, man = calculate_distances(point_a, point_b)
    print(f"Pair {i}: {point_a} and {point_b}")
    print(f"  Euclidean: {euc:.2f}")
    print(f"  Manhattan: {man}\n")
```

### Explanation
- **Euclidean distance** calculates the straight-line distance between two points using the Pythagorean theorem (`sqrt((x2 - x1)^2 + (y2 - y1)^2)`).
- **Manhattan distance** calculates the grid-like distance by summing the absolute differences of the coordinates (`|x2 - x1| + |y2 - y1|`). 
- The function unpacks the points, calculates both values, and returns them as a pair.

## Question 5

Imagine you are building a 'similar movies' feature for a movie app. Briefly describe which linkage method (single, complete, average, or Ward) you would choose for grouping movies by user ratings and why.

**Answer:**

I would choose **Ward's linkage**. 

Ward's method minimizes the variance (spread) inside each cluster, which creates compact and evenly sized groups of movies. Other methods, like single linkage, often cause a "chaining effect" where very different movies end up in the same group. By keeping within-cluster variance low, Ward's method ensures that movies grouped together have genuinely similar rating patterns, leading to better recommendations for users.
