## Question 2

Use pandas to check for missing values in the Age, Annual Income, and Spending Score columns, and print the total number of missing values for each.

**Answer:**

Here is the Python code using a small sample dataset:

```python
import pandas as pd

# Sample data
data = {
    "Age": [19, 21, None, 23, 31],
    "Annual Income": [15, None, 16, 17, None],
    "Spending Score": [39, 81, 6, None, 77],
}

df = pd.DataFrame(data)

# Select the columns and count missing values
columns_to_check = ["Age", "Annual Income", "Spending Score"]
missing_counts = df[columns_to_check].isnull().sum()

print(missing_counts)
```

**Explanation:**
The `isnull()` function checks each cell for missing (null) values and returns `True` if it is empty. The `sum()` function then adds up all the `True` values for each specified column to give you the total count of missing values.

## Question 3

Create a scatterplot of Annual Income vs Spending Score using matplotlib or seaborn to visualize how customers are distributed.

Hint: Use seaborn.scatterplot() or plt.scatter().

**Answer:**

Here is the Python code to create the scatterplot using `seaborn`:

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Sample data
data = {
    "Annual_Income": [15, 16, 25, 30, 45, 55, 60, 70, 80, 90],
    "Spending_Score": [39, 81, 6, 77, 40, 50, 42, 90, 15, 85],
}
df = pd.DataFrame(data)

# Create the scatterplot
sns.scatterplot(data=df, x="Annual_Income", y="Spending_Score")

# Add labels and a title
plt.title("Annual Income vs Spending Score")
plt.xlabel("Annual Income ($k)")
plt.ylabel("Spending Score (1-100)")

# Display the plot
plt.show()
```

### Explanation:
1. We create a sample DataFrame containing **Annual_Income** and **Spending_Score**.
2. `sns.scatterplot()` places income on the x-axis and spending score on the y-axis to show each customer as a single point.
3. We add labels and a title using `matplotlib.pyplot` so the chart is easy to read.

## Question 4

Generate a heatmap of the correlation matrix for the numerical features in the dataset using seaborn, and interpret which two features are most strongly correlated.

**Answer:**

Here is the code using **sample data** to create the heatmap:

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# --- Sample Data ---
data = {
    'Study_Hours': [2, 4, 6, 8, 10],
    'Exam_Score': [50, 65, 70, 85, 95],
    'Sleep_Hours': [9, 8, 7, 6, 5]
}
df = pd.DataFrame(data)

# 1. Calculate correlation matrix
corr_matrix = df.corr()

# 2. Plot heatmap
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix Heatmap")
plt.show()
```

### Explanation & Interpretation:
1. **The Code:** `df.corr()` calculates the correlation between numerical columns, and `sns.heatmap(..., annot=True)` draws the plot and displays the correlation numbers inside the boxes.
2. **Interpretation:** The two most strongly correlated features are **Study_Hours** and **Exam_Score** with a correlation score of approximately **0.99**. This means that as study hours increase, exam scores almost always increase as well.

## Question 5

Apply K-Means clustering (from sklearn.cluster) to segment customers into 3 groups based on Annual Income and Spending Score, then add a new column 'Cluster' to your DataFrame with the assigned cluster for each customer.

Constraint: Use only the two features for clustering, and set random_state=42 for reproducibility.

**Answer:**

Here is the Python code using sample data:

```python
import pandas as pd
from sklearn.cluster import KMeans

# Sample data
data = {
    "CustomerID": [1, 2, 3, 4, 5, 6],
    "Annual Income": [15, 16, 60, 65, 110, 115],
    "Spending Score": [39, 81, 50, 55, 20, 90],
}
df = pd.DataFrame(data)

# 1. Select only the two required features
features = df[["Annual Income", "Spending Score"]]

# 2. Create and fit the K-Means model
kmeans = KMeans(n_clusters=3, random_state=42)

# 3. Predict clusters and add them as a new column
df["Cluster"] = kmeans.fit_predict(features)

print(df)
```

### Explanation:
- We create a `KMeans` model with `n_clusters=3` to form three groups and set `random_state=42` so the results are repeatable.
- We select only `'Annual Income'` and `'Spending Score'` to train the model.
- The `fit_predict()` method learns the groups and returns the cluster labels (0, 1, or 2), which we store in the new `'Cluster'` column.
