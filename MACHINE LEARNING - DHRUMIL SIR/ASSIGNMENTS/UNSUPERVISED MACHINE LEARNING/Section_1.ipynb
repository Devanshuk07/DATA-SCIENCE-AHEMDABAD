## Question 1

Write a short Python script that generates a random dataset of 10 users, each with two features: number of daily app opens and average session duration (in minutes). Print the dataset as a list of dictionaries.

**Answer:**

Here is the Python script:

```python
import random

# Sample data: Generating random metrics for 10 users
dataset = []

for user_id in range(1, 11):
    user = {
        "user_id": user_id,
        "daily_app_opens": random.randint(1, 20),
        "avg_session_duration": round(random.uniform(2.0, 30.0), 1),
    }
    dataset.append(user)

# Print the list of dictionaries
print(dataset)
```

**Explanation:**
This script uses the `random` module to create sample data for 10 users. A `for` loop generates each user's data as a dictionary with a random number of app opens and a random duration rounded to one decimal place. Finally, each dictionary is added to a list and printed.
