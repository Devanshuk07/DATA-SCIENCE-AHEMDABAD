## Question 2

Write a Python function assign_clusters(points, centroids) that takes a list of points and a list of centroids, and returns a list showing which cluster each point belongs to using Euclidean distance.

**Answer:**

Here is the Python function and an example of how to use it:

```python
import math


def assign_clusters(points, centroids):
    cluster_assignments = []

    for point in points:
        distances = []
        for centroid in centroids:
            # Calculate Euclidean distance between the point and centroid
            dist = math.sqrt(
                sum((p - c) ** 2 for p, c in zip(point, centroid))
            )
            distances.append(dist)

        # Find the index of the closest centroid
        closest_centroid_index = distances.index(min(distances))
        cluster_assignments.append(closest_centroid_index)

    return cluster_assignments


# --- Sample Data ---
sample_points = [(1, 2), (2, 1), (8, 9), (9, 8)]
sample_centroids = [(1, 1), (10, 10)]

# Run the function
result = assign_clusters(sample_points, sample_centroids)
print("Cluster assignments:", result)
```

### Explanation:
1. For each point, the function calculates the Euclidean distance to every centroid.
2. It uses `min(distances)` to find the smallest distance.
3. It finds the index of that closest centroid using `distances.index()` and adds it to the output list.

## Question 3

Update your assign_clusters code to work with Manhattan distance as well. Allow the user to choose between 'euclidean' and 'manhattan' as a parameter.

**Answer:**

Here is the updated function that supports both Euclidean and Manhattan distances.

### Python Code

```python
import numpy as np


def assign_clusters(data, centroids, distance_metric="euclidean"):
    clusters = []

    for point in data:
        distances = []
        for centroid in centroids:
            # Calculate distance based on the chosen metric
            if distance_metric == "euclidean":
                dist = np.sqrt(np.sum((point - centroid) ** 2))
            elif distance_metric == "manhattan":
                dist = np.sum(np.abs(point - centroid))
            else:
                raise ValueError(
                    "Invalid metric. Choose 'euclidean' or 'manhattan'."
                )

            distances.append(dist)

        # Assign point to the nearest centroid
        clusters.append(np.argmin(distances))

    return clusters


# --- Sample Data ---
sample_data = np.array([[1, 2], [2, 1], [8, 9], [9, 8]])
sample_centroids = np.array([[1, 1], [10, 10]])

# Example usage
print("Euclidean:", assign_clusters(sample_data, sample_centroids, "euclidean"))
print("Manhattan:", assign_clusters(sample_data, sample_centroids, "manhattan"))
```

### Explanation
* We added a `distance_metric` parameter that defaults to `'euclidean'`.
* **Euclidean distance** calculates the straight-line distance using the square root of squared differences.
* **Manhattan distance** calculates distance by summing the absolute differences along each axis (grid-like distance).
* The function compares the calculated distances and assigns each point to the index of the closest centroid using `np.argmin()`.

## Question 4

Implement a function update_centroids(points, assignments, k) that calculates the new centroid for each cluster as the mean of the assigned points. Test it with a sample dataset of your choice.

**Answer:**

Here is the Python function and a test using sample data.

### Code

```python
import numpy as np


def update_centroids(points, assignments, k):
    points = np.array(points)
    assignments = np.array(assignments)
    new_centroids = []

    for i in range(k):
        # Get all points assigned to cluster i
        cluster_points = points[assignments == i]
        # Calculate the mean position of these points
        centroid = cluster_points.mean(axis=0)
        new_centroids.append(centroid)

    return np.array(new_centroids)


# --- Test with Sample Data (Invented) ---
sample_points = [[1, 2], [2, 2], [8, 8], [9, 10]]
sample_assignments = [0, 0, 1, 1]  # First two points in cluster 0, last two in cluster 1
k = 2

# Calculate new centroids
centroids = update_centroids(sample_points, sample_assignments, k)
print("New Centroids:\n", centroids)
```

### Explanation
The function loops through each cluster ID from `0` to `k-1`. For each cluster, it filters the points that belong to it and calculates their average coordinates using `mean(axis=0)`. Finally, it returns an array containing the new center for each cluster.

## Question 5

Use the Elbow Method to determine the optimal number of clusters (k) for a dataset of your choice (e.g., favorite songs by tempo and danceability, or Zomato restaurant ratings and price). Plot the within-cluster sum of squares for k=1 to k=6 and identify the 'elbow' point.

Hint: Use sklearn's KMeans and matplotlib for plotting.

**Answer:**

### Sample Data and Code

Here is the Python code using a sample dataset of restaurant ratings and average price for two:

```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.cluster import KMeans

# Sample Data: Restaurant Ratings (1-5) and Average Price for two ($)
data = {
    "Rating": [4.5, 4.2, 4.8, 2.1, 2.5, 2.3, 3.8, 3.5, 3.9],
    "Price": [80, 75, 90, 20, 25, 15, 50, 45, 55],
}
df = pd.DataFrame(data)

# Calculate WCSS for k = 1 to 6
wcss = []
k_values = range(1, 7)

for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(df)
    wcss.append(kmeans.inertia_)

# Plot the Elbow graph
plt.plot(k_values, wcss, marker="o")
plt.title("Elbow Method for Optimal k")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("WCSS (Inertia)")
plt.show()
```

### Explanation

* **WCSS (Within-Cluster Sum of Squares):** Measures how close the data points are to their cluster centers. A smaller value means tighter clusters.
* **The 'Elbow' Point:** When you look at the plot, the line drops sharply and then starts flattening out. For this dataset, the bend (or "elbow") happens at **$k = 3$**, meaning 3 is the optimal number of clusters.
