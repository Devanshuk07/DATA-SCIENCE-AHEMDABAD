## Question 1

Load the Iris dataset using Python's sklearn, perform hierarchical clustering with the 'ward' linkage, and plot the resulting dendrogram using scipy's dendrogram() function.

**Answer:**

Here is the code to load the Iris dataset, perform hierarchical clustering, and plot the dendrogram:

```python
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.datasets import load_iris

# 1. Load the Iris dataset
iris = load_iris()
X = iris.data

# 2. Perform hierarchical clustering using 'ward' linkage
linked = linkage(X, method="ward")

# 3. Plot the dendrogram
plt.figure(figsize=(10, 6))
dendrogram(linked)

plt.title("Iris Hierarchical Clustering Dendrogram (Ward)")
plt.xlabel("Sample Index")
plt.ylabel("Distance")
plt.show()
```

### Explanation:
* **`load_iris()`**: Loads the built-in Iris dataset from scikit-learn.
* **`linkage(X, method='ward')`**: Calculates the hierarchical clustering using Ward’s method, which minimizes the variance inside each cluster.
* **`dendrogram()`**: Visualizes the tree-like diagram showing how data points are grouped together step by step.

## Question 2

Visually inspect the dendrogram you plotted and cut it at a height that results in 3 distinct clusters. Assign each data point to its cluster and print the number of samples in each cluster.

Hint: Use scipy's fcluster() with the 'criterion' set to 'distance' and your chosen height.

**Answer:**

Here is the Python code to cut the dendrogram and count the samples in each cluster using sample data.

### Python Code

```python
import numpy as np
import pandas as pd
from scipy.cluster.hierarchy import fcluster, linkage

# --- Sample Data (6 data points) ---
data = np.array(
    [[1, 2], [1, 4], [8, 8], [9, 8], [15, 15], [16, 16]]  # Cluster 1  # Cluster 2  # Cluster 3
)

# 1. Create the linkage matrix
Z = linkage(data, method="ward")

# 2. Cut the dendrogram at your chosen height (e.g., height = 5)
cut_height = 5
cluster_labels = fcluster(Z, t=cut_height, criterion="distance")

# 3. Print the number of samples in each cluster
cluster_counts = pd.Series(cluster_labels).value_counts().sort_index()
print("Sample counts per cluster:")
print(cluster_counts)
```

### Explanation
1. **`linkage()`**: Calculates the hierarchical distances between your data points.
2. **`fcluster()`**: Cuts the tree horizontally at `t=cut_height`. The `criterion='distance'` tells Scipy to use that specific height to form clusters.
3. **`value_counts()`**: Counts how many data points were assigned to each cluster ID.

## Question 3

Repeat the dendrogram plotting for the same data using 'single' and 'complete' linkage methods, and compare how the cluster shapes and heights differ from 'ward' linkage.

Hint: Place all three dendrograms side by side for visual comparison.

**Answer:**

### Python Code

Here is the code using a small sample dataset:

```python
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage

# --- Sample Data (Invented) ---
data = [[1, 2], [2, 3], [3, 3], [8, 7], [8, 8], [9, 8]]

# Calculate linkages
link_single = linkage(data, method="single")
link_complete = linkage(data, method="complete")
link_ward = linkage(data, method="ward")

# Plot side by side
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

dendrogram(link_single, ax=axes[0])
axes[0].set_title("Single Linkage")

dendrogram(link_complete, ax=axes[1])
axes[1].set_title("Complete Linkage")

dendrogram(link_ward, ax=axes[2])
axes[2].set_title("Ward Linkage")

plt.show()
```

### Explanation of the Code
We create sample 2D data points and use `scipy`'s `linkage()` function to calculate the clusters using the three different methods. Then, we use `plt.subplots(1, 3)` to display all three dendrograms side by side for an easy visual comparison.

---

### Comparison of Shapes and Heights

* **Single linkage** uses the minimum distance between points, which often causes a "chaining" effect where points join one by one, resulting in flatter shapes and the lowest merge heights.
* **Complete linkage** uses the maximum distance between points, creating more compact, balanced clusters with higher merge heights.
* **Ward linkage** minimizes the total variance within clusters, which results in the tallest dendrogram heights and very distinct, evenly sized cluster shapes.

Interpret the dendrogram you created: Write a short explanation (3-4 lines) describing how you can tell which clusters are most similar and which are most different, based on the heights and structure of the branches.

**Answer:**

In a dendrogram, the vertical height of the horizontal line connecting two branches represents the distance between those groups. 

Clusters that merge near the bottom at a **lower height** are the **most similar** because the distance between them is very small. In contrast, clusters that merge near the top at a **greater height** are the **most different** because they are separated by a much larger distance. Therefore, you can judge similarity directly by how early and low the branches connect.
