## Question 2

Create a POST endpoint /predict-likes in your FastAPI app that accepts a JSON payload with 'current_likes' (int) and 'new_likes' (int), and returns the updated total likes as JSON.

**Answer:**

Here is the code to create the endpoint using FastAPI:

```python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()


# Define the expected JSON input structure
class LikeData(BaseModel):
    current_likes: int
    new_likes: int


# Create the POST endpoint
@app.post("/predict-likes")
def predict_likes(data: LikeData):
    total = data.current_likes + data.new_likes
    return {"total_likes": total}
```

### Explanation:
1. **`LikeData`**: A Pydantic model that ensures the incoming JSON data has both `current_likes` and `new_likes` as integers.
2. **`@app.post("/predict-likes")`**: Defines a POST route at `/predict-likes`.
3. **`predict_likes` function**: Adds the two numbers together and automatically returns the result as a JSON response.

## Question 4

Modify your FastAPI endpoint to handle the case where either 'current_likes' or 'new_likes' is missing in the request, and return an error message with status code 400.

Hint: Use Pydantic models and FastAPI's validation features to enforce required fields.

**Answer:**

Here is the code to validate the fields and return a 400 status code if any are missing:

```python
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
from typing import Optional

app = FastAPI()

# 1. Define the Pydantic model
class LikesRequest(BaseModel):
    current_likes: Optional[int] = None
    new_likes: Optional[int] = None

# 2. Create the endpoint
@app.post("/update-likes")
def update_likes(data: LikesRequest):
    # Check if either field is missing
    if data.current_likes is None or data.new_likes is None:
        raise HTTPException(
            status_code=400, 
            detail="Error: Both 'current_likes' and 'new_likes' are required."
        )
    
    total = data.current_likes + data.new_likes
    return {"message": "Success", "total_likes": total}
```

### Explanation:
* **`Optional[int] = None`**: Allows the incoming request to be accepted even if a field is omitted, so we can handle it manually.
* **`if` condition**: Checks if either `current_likes` or `new_likes` is not provided (`None`).
* **`HTTPException(status_code=400, ...)`**: Stops execution and sends a 400 Bad Request error response with your custom message to the user.

For an IPL fantasy points prediction API, I would choose **FastAPI** over Flask.
