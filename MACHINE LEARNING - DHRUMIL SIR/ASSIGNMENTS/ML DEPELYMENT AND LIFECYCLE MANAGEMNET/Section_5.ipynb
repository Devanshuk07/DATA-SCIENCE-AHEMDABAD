## Question 1

Create a simple Streamlit app in a file called app.py that displays your favorite IPL team's name and logo, then deploy it to Streamlit Cloud and share your live app link.

**Answer:**

Here is the Python code for your **`app.py`** file.

### Python Code (`app.py`)

```python
import streamlit as st

# Set the page title
st.title("My Favorite IPL Team")

# Team name
st.header("Chennai Super Kings (CSK)")

# Team logo (using a sample online image URL)
logo_url = "https://upload.wikimedia.org/wikipedia/en/2/2b/Chennai_Super_Kings_Logo.svg"
st.image(logo_url, caption="Whistle Podu!", width=300)

# Short description
st.write("Chennai Super Kings is one of the most successful teams in the IPL!")
```

### Explanation
* `st.title()` and `st.header()` display large text headers.
* `st.image()` displays the team logo using an image link and sets its width.
* `st.write()` prints simple text on the screen.

---

### Steps to Deploy and Get Your Link

1. **Create `requirements.txt`:** In the same folder as `app.py`, create a file named `requirements.txt` and write:
   ```text
   streamlit
   ```
2. **Push to GitHub:** Upload both `app.py` and `requirements.txt` to a new GitHub repository.
3. **Deploy:** Go to [share.streamlit.io](https://share.streamlit.io/), sign in with GitHub, click **"New app"**, select your repository and `app.py`, then click **"Deploy"**.
4. **Copy Link:** Once deployed, copy the URL from your browser (e.g., `https://your-username-app-name.streamlit.app`) to share your live app link.

## Question 2

Build a mini movie recommendation app using Streamlit that shows 5 trending Bollywood movies (hardcoded list) with their posters and a short description, then upload your project files to Hugging Face Spaces and deploy your app online.

Hint: Use st.image() and st.write() in Streamlit to display images and text.

**Answer:**

Here is the complete solution to build and deploy your mini movie app.

---

### Step 1: Create the Python Code (`app.py`)

Create a file named `app.py` and add the following code:

```python
import streamlit as st

# Sample data: Hardcoded list of 5 Bollywood movies
movies = [
    {
        "title": "Jawan",
        "description": "A man is driven by a personal vendetta to rectify the evils in society.",
        "poster": "https://image.tmdb.org/t/p/w500/jYW34f7b6qYdE0gG211f44eF567.jpg",
    },
    {
        "title": "Pathaan",
        "description": "An Indian agent races against a ruthless mercenary planning a deadly attack.",
        "poster": "https://image.tmdb.org/t/p/w500/m1b9ToOi5uh4JpZ3f7g4C3r4sQk.jpg",
    },
    {
        "title": "3 Idiots",
        "description": "Two friends search for their long-lost college buddy and revisit college memories.",
        "poster": "https://image.tmdb.org/t/p/w500/66A9MqXOyVFCssoloscw79z8Tew.jpg",
    },
    {
        "title": "Dangal",
        "description": "A former wrestler trains his daughters to become world-class wrestlers.",
        "poster": "https://image.tmdb.org/t/p/w500/5m5bO8z0zT2G5iX8iUuF0qQx8vW.jpg",
    },
    {
        "title": "RRR",
        "description": "A fearless revolutionary and a legendary cop unite against British forces.",
        "poster": "https://image.tmdb.org/t/p/w500/nEufeZlyAOLqO2brrs0yeMu1Q2P.jpg",
    },
]

# App Title
st.title("Top 5 Bollywood Movies")
st.write("Here are 5 trending Bollywood movies you should watch:")

# Display each movie
for movie in movies:
    st.header(movie["title"])
    st.image(movie["poster"], width=200)
    st.write(movie["description"])
    st.write("---")
```

**Explanation:**  
We store the movie details in a list of dictionaries (sample data). We then use a simple `for` loop, along with `st.image()` and `st.write()`, to show each movie's title, poster, and description on the page.

---
