## Question 1

Train a machine learning model to predict whether a Flipkart product review is positive or negative using a sample dataset, and save the trained model as review_sentiment_model.pkl.

**Answer:**

Here is the simple Python code to create sample review data, train a model, and save it.

### Python Code

```python
import pickle
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline

# 1. Sample Data (labeled as 1 for Positive, 0 for Negative)
reviews = [
    "This product is amazing and works great!",
    "Terrible quality, completely broke in one day.",
    "Very good value for money, I loved it.",
    "Worst purchase ever, do not buy this.",
    "Fast delivery and excellent packaging.",
    "Very disappointed with the item, poor quality.",
]
labels = [1, 0, 1, 0, 1, 0]

# 2. Build and train a simple text classification pipeline
model = make_pipeline(CountVectorizer(), MultinomialNB())
model.fit(reviews, labels)

# 3. Save the trained model to a file
with open("review_sentiment_model.pkl", "wb") as file:
    pickle.dump(model, file)

print("Model successfully trained and saved as review_sentiment_model.pkl")
```

### Explanation
1. **Sample Data:** We created a small list of positive and negative reviews with matching labels (`1` for positive, `0` for negative).
2. **Pipeline:** We used `CountVectorizer` to convert words into numbers and `MultinomialNB` (Naive Bayes) to classify the sentiment.
3. **Save Model:** Using Python's `pickle` library, we saved the trained pipeline directly into `review_sentiment_model.pkl` so it can be reused later.

## Question 2

Build a Flask-based prediction API (predict_review.py) that loads your saved review_sentiment_model.pkl and returns 'Positive' or 'Negative' for a given review text sent via POST request.

Hint: Use Flask's request.get_json() to read the input review from the API call.

**Answer:**

Here is the code for **`predict_review.py`**:

```python
import pickle
from flask import Flask, jsonify, request

app = Flask(__name__)

# Load the saved model
with open("review_sentiment_model.pkl", "rb") as file:
    model = pickle.load(file)


@app.route("/predict", methods=["POST"])
def predict():
    # Read JSON data from the POST request
    data = request.get_json()
    review_text = data.get("review", "")

    # Make prediction (assuming model outputs 1 for Positive, 0 for Negative)
    prediction = model.predict([review_text])[0]

    # Map the output to Positive or Negative
    sentiment = "Positive" if prediction == 1 else "Negative"

    return jsonify({"sentiment": sentiment})


if __name__ == "__main__":
    app.run(debug=True)
```

### Explanation:
1. **Load Model**: It uses `pickle.load()` to load your trained model into memory when the app starts.
2. **POST Route**: The `/predict` endpoint receives review text sent in a JSON body (for example: `{"review": "This movie was great!"}`).
3. **Prediction**: The model predicts the sentiment, and the API returns `{"sentiment": "Positive"}` or `{"sentiment": "Negative"}`.

## Question 3

Create a simple HTML frontend (review_predictor.html) with a text box for the user to enter a product review and a 'Predict' button that calls your Flask API and displays the prediction result without reloading the page.

Hint: Use JavaScript fetch() to send the review to your API and update the result in the DOM.

**Answer:**

Here is the simple HTML file (`review_predictor.html`):

```html
<!DOCTYPE html>
<html>
<head>
    <title>Review Sentiment Predictor</title>
</head>
<body>
    <h2>Product Review Predictor</h2>
    
    <!-- Input text area -->
    <textarea id="reviewText" rows="4" cols="50" placeholder="Type your review here..."></textarea><br><br>
    
    <!-- Predict button -->
    <button onclick="predictSentiment()">Predict</button>

    <!-- Area to display the result -->
    <h3 id="result"></h3>

    <script>
        function predictSentiment() {
            // 1. Get the text from the textarea
            const review = document.getElementById("reviewText").value;

            // 2. Send the text to the Flask API using fetch
            fetch('/predict', {
                method: 'POST',
                headers: {
                    'Content-Type': 'application/json'
                },
                body: JSON.stringify({ review: review })
            })
            .then(response => response.json())
            .then(data => {
                // 3. Display the prediction result on the page
                document.getElementById("result").innerText = "Prediction: " + data.prediction;
            })
            .catch(error => {
                console.error('Error:', error);
            });
        }
    </script>
</body>
</html>
```

### Explanation:
* **HTML Elements:** A `<textarea>` lets the user type a review, a `<button>` triggers the prediction, and an `<h3>` holds the result.
* **`fetch()`:** Sends the review text as JSON to your Flask `/predict` route in the background.
* **DOM Update:** When the Flask API responds, `document.getElementById("result").innerText` displays the returned prediction without refreshing the web page.
