## Question 1

Train a simple scikit-learn DecisionTreeClassifier on a small dataset of your choice (for example, a list of songs with genres from your Spotify playlist), then use Pickle to save the trained model to a file named song_genre_model.pkl.

**Answer:**

Here is the Python code to create sample song data, train the decision tree model, and save it using Pickle.

```python
import pickle
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

# 1. Create a small sample dataset (invented data)
sample_data = {
    "tempo_bpm": [120, 140, 60, 128, 75, 130],
    "energy": [0.8, 0.9, 0.2, 0.85, 0.3, 0.75],
    "genre": ["Pop", "Rock", "Classical", "Pop", "Classical", "Rock"],
}

df = pd.DataFrame(sample_data)

# 2. Separate features (X) and target (y)
X = df[["tempo_bpm", "energy"]]
y = df["genre"]

# 3. Train the Decision Tree Classifier
model = DecisionTreeClassifier()
model.fit(X, y)

# 4. Save the trained model to a file using pickle
with open("song_genre_model.pkl", "wb") as file:
    pickle.dump(model, file)

print("Model successfully saved as song_genre_model.pkl")
```

### Explanation:
1. **Sample Data:** We created an artificial dataset containing two features (`tempo_bpm` and `energy`) and the target label (`genre`).
2. **Model Training:** We initialized `DecisionTreeClassifier()` and used `.fit()` to learn patterns from the features and targets.
3. **Pickle Saving:** We opened a new file named `song_genre_model.pkl` in write-binary mode (`"wb"`) and used `pickle.dump()` to save the trained model object for future use.

## Question 2

Load the song_genre_model.pkl file you saved earlier and use it to predict the genre for a new song entry, printing the predicted result.

**Answer:**

```python
import pandas as pd
import pickle

# 1. Load the saved model file
with open("song_genre_model.pkl", "rb") as file:
    model = pickle.load(file)

# 2. Define sample data for a new song (invented features: tempo, energy, danceability)
new_song_sample = pd.DataFrame(
    {"tempo": [128], "energy": [0.85], "danceability": [0.72]}
)

# 3. Predict the genre
predicted_genre = model.predict(new_song_sample)

# 4. Print the result
print(f"Predicted Genre: {predicted_genre[0]}")
```

### Explanation
- **`pickle.load()`** opens and reads the saved model from the `.pkl` file into memory.
- **`new_song_sample`** holds invented sample feature values matching the format the model was trained on.
- **`model.predict()`** runs the new sample through the loaded model, and `print()` displays the predicted genre name.

## Question 3

Train a scikit-learn KNeighborsClassifier to predict whether a Flipkart product is 'electronics' or 'fashion' based on sample features, then use Joblib to save the model as product_category_model.joblib.

Hint: Use joblib.dump() for saving and joblib.load() for loading.

**Answer:**

Here is the Python code to create sample data, train the `KNeighborsClassifier`, and save the model:

```python
import joblib
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier

# 1. Create a sample dataset (invented data for demonstration)
sample_data = {
    "weight_grams": [150, 200, 1500, 300, 50, 800],
    "price_inr": [15000, 800, 45000, 1200, 500, 25000],
    "category": [
        "electronics",
        "fashion",
        "electronics",
        "fashion",
        "fashion",
        "electronics",
    ],
}

df = pd.DataFrame(sample_data)

# 2. Separate features (X) and target label (y)
X = df[["weight_grams", "price_inr"]]
y = df["category"]

# 3. Train the K-Nearest Neighbors model
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)

# 4. Save the trained model to a file using joblib
joblib.dump(model, "product_category_model.joblib")
print("Model saved successfully as product_category_model.joblib")
```

### Explanation:
1. **Sample Data**: We create a small DataFrame with two features (`weight_grams` and `price_inr`) and the target label (`category`).
2. **Model Training**: We initialize `KNeighborsClassifier(n_neighbors=3)` and train it on the features using `.fit()`.
3. **Saving the Model**: `joblib.dump()` serializes and saves the trained model to a file named `product_category_model.joblib`. You can later reload it anytime using `joblib.load('product_category_model.joblib')`.

## Question 5

List three best practices for storing and managing machine learning model files in a real-world app like BookMyShow, and briefly explain why each is important.

**Answer:**

1. **Use a Model Registry with Versioning:** Store models in a dedicated tool (like MLflow) or cloud storage (like AWS S3) with version tags. This ensures you can track changes and quickly roll back to an older version if a newly deployed model fails.
2. **Keep Model Files Out of Git:** Avoid pushing large binary files (like `.pkl` or `.h5`) directly into Git, which slows down development. Instead, store the actual files in cloud storage and use tools like DVC (Data Version Control) to track lightweight pointers to them.
3. **Track Metadata and Lineage:** Always save the model file alongside the code version, hyperparameters, and training dataset used to build it. This makes your work reproducible and helps you debug why a model made a specific prediction in production.
