## Question 4

Given a list of 10 user reviews for a restaurant (as text), write a Python function that simulates batch prediction by categorizing each review as 'positive' or 'negative' based on whether it contains the word 'good'. Print the results for all reviews.

Constraint: Do not use any external ML libraries; just use basic string operations.

**Answer:**

Here is the Python code using a sample dataset of 10 invented reviews:

```python
# Sample data: 10 invented restaurant reviews
sample_reviews = [
    "The food was really good and fresh.",
    "Terrible service, will not return.",
    "A good place for a quick lunch.",
    "The soup tasted bland and cold.",
    "Everything was very good, especially the dessert.",
    "Worst dining experience of my life.",
    "Staff was polite and the pasta was good.",
    "Too noisy and the seats were uncomfortable.",
    "Always a good meal here.",
    "The pizza was completely burnt.",
]


def batch_predict(reviews):
    results = []
    for review in reviews:
        # Convert text to lowercase to catch "Good", "GOOD", etc.
        if "good" in review.lower():
            sentiment = "positive"
        else:
            sentiment = "negative"
        results.append((review, sentiment))
    return results


# Run the function and print the results
predictions = batch_predict(sample_reviews)

for review, sentiment in predictions:
    print(f"Review: '{review}' | Sentiment: {sentiment}")
```

### Explanation
1. **Lowercase Conversion:** We use `.lower()` so the check finds the word "good" regardless of capitalization (like "Good" or "GOOD").
2. **Batch Processing:** The function loops through the entire list of reviews, assigns a sentiment based on the presence of the keyword, and stores each result in a new list.

## Question 5

Think of a common deployment challenge (such as model version mismatch, slow predictions, or scaling to many users) that could happen if you were deploying an ML model in a Flipkart-like app. Briefly describe the challenge and suggest one solution to handle it.

**Answer:**

**Challenge: High Latency During Traffic Spikes**  
In an e-commerce app, a major challenge is slow prediction times when millions of users shop simultaneously during events like flash sales. If the machine learning model has to calculate recommendations in real time for every click, the servers can become overloaded, causing slow page loads or app crashes. 

**Solution: Precomputing and Caching**  
A practical solution is to generate recommendations offline in batches (for example, every night) and store the results in a fast in-memory cache, such as Redis. When a user opens the app, the system fetches the precomputed results instantly from the cache instead of running the heavy model in real time.
