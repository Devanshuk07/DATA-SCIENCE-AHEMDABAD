## Question 4

Compare the responses you get from zero-shot, one-shot, and few-shot prompts when asking ChatGPT to write a Spotify playlist description for 'Monsoon Vibes'. List which method gave the most relevant result and why.

**Answer:**

### Comparison of Prompting Methods

| Prompting Method | Typical Response Characteristics |
| :--- | :--- |
| **Zero-shot** | Generic, broad descriptions using common clichés (e.g., rain, cozy vibes, hot tea); lacks specific Spotify formatting, tone, or taglines. |
| **One-shot** | Better aligned with Spotify's style (e.g., character count, punchy tone) by mimicking the single provided example, but may overfit to that specific sample. |
| **Few-shot** | Highly tailored, captures the exact tone, structure, and brevity of real Spotify descriptions by learning patterns across multiple examples. |

---

### Most Relevant Result and Why

* **Most Relevant Method:** Few-shot prompting
* **Why:** Providing multiple varied examples allows the model to learn the distinct nuances, constraints, and platform-specific style (short, emotive, artist-referencing) of Spotify copy without overfitting to a single template.
