## Question 1

Write two prompts for ChatGPT: one asking for a list of trending songs in India (unclear), and one asking for a list of trending Bollywood songs released after 2022 with artist names (clear). Compare the responses and note the difference in detail and relevance.

**Answer:**

### Prompts

* **Unclear Prompt:** "Give me a list of trending songs in India."
* **Clear Prompt:** "Provide a list of trending Bollywood songs released after 2022, including the artist names for each song."

---

### Comparison of Responses

| Feature | Response to Unclear Prompt | Response to Clear Prompt |
| :--- | :--- | :--- |
| **Relevance** | Low; mixes multiple languages (Hindi, Punjabi, Tamil, English) and various release years. | High; strictly focuses on Bollywood tracks released after 2022. |
| **Detail** | Minimal; often lists only song titles without specifying artists or release dates. | High; explicitly includes song titles, release years, and contributing artists. |

---

### Key Difference
The clear prompt narrows the genre, sets a specific timeframe, and mandates metadata (artists), resulting in a directly usable and focused output. The unclear prompt lacks constraints, forcing the AI to guess the user's intent and producing generic, inconsistent results.

## Question 2

List 3 common mistakes people make when writing prompts for AI tools, and rewrite each mistake as a 'Do' and 'Don't' example using a scenario like ordering food on Zomato or booking a movie on BookMyShow.

**Answer:**

**1. Being too vague or lacking specific details**
*   **Don't:** "Order some food from Zomato."
*   **Do:** "Order one paneer butter masala and two butter naans from a restaurant rated 4+ stars on Zomato."

**2. Omitting constraints or preferences**
*   **Don't:** "Book movie tickets on BookMyShow."
*   **Do:** "Book two evening tickets for the latest Batman movie on BookMyShow, keeping the budget under ₹500 per ticket and avoiding front-row seats."

**3. Providing conflicting or overloaded instructions**
*   **Don't:** "Find a cheap, high-end luxury sushi place on Zomato that delivers in 10 minutes and is 20 km away."
*   **Do:** "Find top-rated sushi restaurants within a 5 km radius on Zomato that deliver in under 45 minutes."

## Question 3

Use ChatGPT or any AI chatbot to generate a restaurant review for a new café. First, give a vague prompt and copy the output. Then, give a detailed prompt specifying cuisine, ambience, and price range, and copy that output. Submit both outputs and explain which is better and why.

Hint: Focus on how the extra details in your prompt affect the quality of the review.

**Answer:**

**Vague Prompt:**
"Write a review for a new café."

**Output 1:**
> "I recently visited the new café in town, and it was great. The food was tasty, and the coffee was really good. The place has nice seating and the staff were friendly. I would definitely recommend checking it out if you are in the area."

***

**Detailed Prompt:**
"Write a review for a new café called 'The Daily Roast' specializing in French pastries and specialty pour-over coffee. Mention the cozy, rustic ambience with soft jazz music, and note that it has a moderate price range."

**Output 2:**
> "I stopped by The Daily Roast this weekend, and it is a fantastic addition to the neighborhood. The rustic wooden decor and low jazz music create an incredibly cozy atmosphere to relax or work. Their French pastries are outstanding—the almond croissant was flaky and fresh—and the single-origin pour-over coffee was brewed to perfection. Prices are moderate at around $5–$8 per item, offering great value for the quality. Highly recommended!"

***

**Which is better and why:**

Output 2 is better. 

* **Specificity:** It includes real menu items (almond croissant, pour-over coffee) rather than generic terms like "food" and "coffee."
* **Accurate Tone:** The sensory details about the rustic decor and jazz music give the reader a realistic sense of the café's actual atmosphere.
* **Helpful Context:** Including the price range gives potential customers practical information needed to make a dining decision.

## Question 4

Pick any feature from an app you use (e.g., Instagram's Explore page or Flipkart's product recommendations) and write a prompt that would help an AI generate content for that feature. Make sure your prompt follows at least three 'Dos' of good prompting discussed in class.

**Answer:**

**App Feature:** Spotify’s "Discover Weekly" personalized playlist descriptions.

**Prompt:**
> Act as a music curator for Spotify. Write a personalized, two-sentence playlist description for a user who listens to 80s synth-pop and indie rock. Keep the tone energetic and friendly, and do not use generic clichés like "tunes for your soul."

**Dos applied:**
1. **Define a persona:** Assigned the role of a "music curator for Spotify."
2. **Provide context and specifics:** Included the user's specific genres ("80s synth-pop and indie rock").
3. **Set clear constraints:** Specified the length ("two-sentence"), tone ("energetic and friendly"), and negative constraints ("do not use generic clichés").
