## Question 1

Open ChatGPT or any AI chatbot and ask it to list the top 5 trending dishes on Zomato in Ahmedabad. Check if the AI gives real, verifiable answers or makes up information, and note any hallucinations you spot.

**Answer:**

When asked for real-time data like "top 5 trending dishes on Zomato in Ahmedabad," a standard AI chatbot cannot access live proprietary platform metrics and typically responds in one of two ways:

1. **Hallucination / Made-up Information:** Without active web browsing or direct API access to Zomato, the AI invents plausible-sounding trends (e.g., specific dishes like "Cheese Angoori from Manek Chowk" or "Truffle Khichdi") based on general knowledge of Ahmedabad's food culture rather than verifiable data.
2. **Refusal / Disclaimers:** With real-time browsing enabled, the AI cannot log in to Zomato's live internal tracking, so it relies on public food blogs and often notes that it cannot verify real-time platform rankings.

**Identified Hallucinations:**
* **Fabricated Metrics:** Claiming a dish is ranked "#1" or "trending this week" without live Zomato backend data.
* **Invented Specifics:** Pairing real Ahmedabad restaurants with specific dishes they may not actually serve or that are not officially trending.

## Question 4

Refactor this prompt to reduce hallucination: 'Write a review of the latest iPhone.' Make it more specific and instruct the AI to use only official Apple website data. Write your improved prompt and explain in one line why it should reduce hallucination.

**Answer:**

**Improved Prompt:**
"Write a factual review of the latest iPhone using only the official specifications and feature details provided on Apple.com. Do not include external rumors, assumptions, or unverified claims; if a specific detail is not found on the official Apple website, state that it is unavailable."

**Why it reduces hallucination:**
It restricts the AI's knowledge boundary to a single, authoritative source and explicitly instructs it to acknowledge missing information rather than inventing facts.
