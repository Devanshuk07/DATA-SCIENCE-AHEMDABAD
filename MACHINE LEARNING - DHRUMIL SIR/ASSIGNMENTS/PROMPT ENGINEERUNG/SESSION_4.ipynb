## Question 3

Design a chain-of-thought prompt for ChatGPT to solve this problem: 'You have ₹2000 in your Paytm wallet. You want to buy 3 items costing ₹499, ₹799, and ₹899. Will you have enough balance? Show step-by-step reasoning.' Copy the AI's response.

**Answer:**

**Chain-of-Thought Prompt:**
"You have ₹2000 in your Paytm wallet. You want to buy 3 items costing ₹499, ₹799, and ₹899. Will you have enough balance? Think step-by-step: first, calculate the total cost of all three items; then, compare this total to your wallet balance; finally, state whether you have enough balance."

***

**AI's Response:**
"**Step 1: Calculate the total cost of the items**
* Item 1: ₹499
* Item 2: ₹799
* Item 3: ₹899
Total Cost = 499 + 799 + 899 = ₹2197

**Step 2: Compare the total cost with the wallet balance**
* Wallet balance = ₹2000
* Total cost = ₹2197
* Difference = 2000 - 2197 = -₹197

**Step 3: Conclusion**
No, you will not have enough balance. You are short by ₹197."

## Question 4

Use ChatGPT or any AI tool to generate a step-by-step reasoning (chain-of-thought) prompt for planning a one-day trip to Ahmedabad, including food, travel, and entertainment. Paste the prompt you used and the AI's detailed response.

**Answer:**

**PROMPT USED:**

```text
Plan a one-day trip to Ahmedabad covering food, travel, and entertainment. Use step-by-step chain-of-thought reasoning to explain your choices before providing the final schedule. Follow these steps:
1. Analyze the constraints (time, budget, logistics).
2. Select key locations based on geographical proximity to minimize travel time.
3. Integrate famous local food stops corresponding to meal times.
4. Finalize a logical, time-blocked itinerary.
```

***

**AI RESPONSE:**

### Step-by-Step Reasoning (Chain-of-Thought)

1. **Constraints Analysis:** A one-day trip requires maximizing daylight hours, minimizing intra-city travel fatigue, and keeping travel affordable via auto-rickshaws, cabs, or the metro.
2. **Geographical Clustering:** 
   - Morning: Old City (East Ahmedabad) for heritage, culture, and architecture.
   - Afternoon: Central area near the Sabarmati River for indoor/peaceful activities during peak heat.
   - Evening: Western Ahmedabad for modern entertainment, scenic views, and dinner.
3. **Food Integration:**
   - Morning: Traditional Gujarati breakfast (Fafda, Jalebi, Khaman) in the Old City.
   - Afternoon: Authentic Gujarati Thali for lunch.
   - Evening/Night: Renowned street food (Manek Chowk or Law Garden).

---

### Final One-Day Ahmedabad Itinerary

* **08:00 AM – 09:30 AM: Breakfast & Heritage Walk (Food & Culture)**
  * **Activity:** Explore the Old City starting at Kalupur Swaminarayan Temple.
  * **Food:** Eat fresh Fafda, Jalebi, and Khaman at *Chandravilas Restaurant*.

* **09:30 AM – 11:30 AM: Historic Sightseeing (Entertainment/Culture)**
  * **Activity:** Visit the historic Sidi Saiyyed Mosque (famous for the stone lattice "Tree of Life") and Bhadra Fort.
  * **Travel:** Auto-rickshaw (10 mins).

* **12:00 PM – 02:00 PM: Sabarmati Ashram & Lunch (History & Food)**
  * **Activity:** Tour the Sabarmati Ashram to learn about Mahatma Gandhi's life.
  * **Travel:** Cab or auto-rickshaw (20 mins).
  * **Food:** Gujarati Thali at *Agashiye* (The House of MG) or *Gordhan Thal*.

* **02:30 PM – 04:30 PM: Science City / Calico Museum Alternative (Entertainment)**
  * **Activity:** Visit the Auto World Vintage Car Museum or explore the air-conditioned pavilions at Gujarat Science City.
  * **Travel:** Cab (25–30 mins).

* **05:00 PM – 07:00 PM: Sabarmati Riverfront & Atal Bridge (Leisure & Views)**
  * **Activity:** Walk across the Atal Pedestrian Bridge, enjoy the flower garden, and take an evening boat ride along the Sabarmati Riverfront.
  * **Travel:** Cab (20 mins).

* **07:30 PM – 09:30 PM: Shopping & Dinner (Food & Entertainment)**
  * **Activity:** Shop for traditional handicrafts at Law Garden Night Market, then head to Manek Chowk night market.
  * **Food:** Street food dinner at Manek Chowk (Gwalior Dosa, Pav Bhaji, Chocolate-Cheese Sandwich, and Kulfi).
  * **Travel:** Auto-rickshaw (15 mins).
