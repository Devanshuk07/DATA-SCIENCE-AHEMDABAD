## Question 1

Install Git on your system, open a terminal or command prompt, and run git --version to confirm the installation.

**Answer:**

Here are the steps to install and check Git:

1. **Install Git:** 
   - **Windows/Mac:** Download and run the installer from the official Git website (git-scm.com) using default settings. 
   - **Linux (Ubuntu/Debian):** Run `sudo apt install git` in your terminal.

2. **Open Terminal:** 
   - On Windows, open **Command Prompt** or **Git Bash**. 
   - On Mac or Linux, open **Terminal**.

3. **Check Installation:** 
   Type this command and press Enter:
   ```bash
   git --version
   ```

**Expected Result:**
You will see output similar to `git version 2.40.0`. This confirms Git is successfully installed on your system.

## Question 2

Initialize a new local Git repository named music-recommendation in a folder, create a README.md file inside it, and make your first commit with the message 'Initial commit for music-recommendation project'.

**Answer:**

Here are the Git commands to complete this task in your terminal:

```bash
# 1. Create a new folder and move into it
mkdir music-recommendation
cd music-recommendation

# 2. Initialize the Git repository
git init

# 3. Create a README.md file
echo "# Music Recommendation" > README.md

# 4. Stage the file for commit
git add README.md

# 5. Make the first commit
git commit -m "Initial commit for music-recommendation project"
```

**Explanation:**
* `mkdir` and `cd` create and open the new project folder.
* `git init` turns the folder into a Git repository.
* `git add` prepares the new `README.md` file to be saved, and `git commit` saves your changes with a message.

## Question 3

Create a new GitHub repository called foodie-favorites, connect your local music-recommendation repository to this remote using git remote add origin, and push your commits to GitHub.

Hint: Use git push -u origin main (or master) after connecting the remote.

**Answer:**

Here are the steps to link and push your project to GitHub:

### Step 1: Create the repository on GitHub
1. Log in to GitHub and click the **+** icon in the top-right corner, then select **New repository**.
2. Set the **Repository name** to `foodie-favorites`.
3. Leave it empty (do not check README, .gitignore, or license) and click **Create repository**.
4. Copy the repository URL (it looks like: `https://github.com/your-username/foodie-favorites.git`).

---

### Step 2: Connect and push from your terminal
Open your terminal, make sure you are inside your local `music-recommendation` folder, and run:

```bash
# Connect your local repository to GitHub (replace with your actual URL)
git remote add origin https://github.com/your-username/foodie-favorites.git

# Set your branch name to main
git branch -M main

# Push your commits to GitHub
git push -u origin main
```

**Brief Explanation:**
* `git remote add origin` creates a link named "origin" between your local computer and the GitHub repository.
* `git push -u origin main` uploads your files to GitHub and remembers the connection for future pushes.

## Question 5

Write a short note (3-4 lines) explaining how using Git and GitHub can help you avoid the problem of having multiple versions of the same Jupyter notebook file (like analysis_v1.ipynb, analysis_final2.ipynb, etc.) when working on a data science project.

**Answer:**

Git tracks every change you make to a single notebook through "commits," creating a complete timeline of your work over time. Instead of manually saving new files like `analysis_v2.ipynb`, you keep just one file and use Git to view or restore previous versions whenever needed. GitHub stores this history safely online, which keeps your project folder clean, organized, and free of duplicate files.
