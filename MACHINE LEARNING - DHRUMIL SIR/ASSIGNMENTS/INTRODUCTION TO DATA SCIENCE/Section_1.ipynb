## Question 3

Imagine you work for a music streaming app like Spotify. Write a short scenario (3-4 sentences) describing how data-driven decision making could help the app recommend new songs to users.

**Answer:**

The app tracks user data, such as which songs a person listens to repeatedly, saves to playlists, or skips. Data scientists analyze this behavior to identify listening patterns and group together users with similar music tastes. Using these insights, the algorithm decides to recommend songs that other users with matching tastes have enjoyed. This data-driven approach keeps users happy and engaged by consistently introducing them to new music they are likely to love.

## Question 4

Pick one industry from this list: e-commerce, sports, or transportation. Research and describe one innovative way data science is being used in that industry today.

Hint: Search for terms like 'data science in e-commerce' or 'data analytics in sports' to find real examples.

**Answer:**

**Industry:** E-commerce  
**Innovative Use:** Dynamic Pricing

In e-commerce, companies use data science for dynamic pricing, which automatically adjusts product prices in real time based on market demand. Machine learning algorithms analyze competitor prices, customer browsing behavior, time of day, and current inventory levels to calculate the best price. For example, if demand for an item suddenly spikes, the system can slightly raise the price to maximize profits, or lower it to clear out unsold stock. This helps online stores stay competitive while maximizing revenue without needing manual updates from human managers.
