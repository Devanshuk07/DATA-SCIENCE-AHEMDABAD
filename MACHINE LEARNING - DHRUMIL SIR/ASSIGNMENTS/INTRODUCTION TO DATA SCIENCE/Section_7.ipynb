## Question 1

Create a new GitHub repository named 'my-movie-ratings' and push a local Jupyter notebook file (movie_ratings.ipynb) to it using git commands (init, add, commit, remote add, push).

**Answer:**

First, go to [GitHub](https://github.com) in your browser and click **New** to create a repository named **`my-movie-ratings`**.

Then, open your terminal in the folder containing `movie_ratings.ipynb` and run these commands:

```bash
# 1. Initialize a new local Git repository
git init

# 2. Add your notebook file to the staging area
git add movie_ratings.ipynb

# 3. Save your changes with a message
git commit -m "Add movie ratings notebook"

# 4. Rename the default branch to main
git branch -M main

# 5. Connect your local repository to GitHub (replace YOUR-USERNAME with your actual username)
git remote add origin https://github.com/YOUR-USERNAME/my-movie-ratings.git

# 6. Upload your file to GitHub
git push -u origin main
```

### Brief Explanation:
* **`git init`** starts tracking your folder with Git.
* **`git add`** prepares your notebook to be saved.
* **`git commit`** takes a snapshot of your file.
* **`git remote add`** links your computer to your online GitHub repository.
* **`git push`** uploads your committed file to GitHub.

## Question 2

Add a large sample CSV file (over 100MB, e.g., fake Spotify playlist data) to your repo using Git LFS instead of regular git, and explain in a README.md why raw data files should not be uploaded directly to GitHub repos.

Hint: Use 'git lfs track' and mention privacy, storage, and bandwidth issues in your explanation.

**Answer:**

### Step 1: Create the Large Sample CSV (Python)

Run this Python script to generate a fake CSV file over 100MB.

```python
import pandas as pd

# Sample data: fake Spotify songs
sample_data = {
    "track_name": ["Song A", "Song B", "Song C"] * 1_000_000,
    "artist": ["Artist X", "Artist Y", "Artist Z"] * 1_000_000,
    "streams": [1200, 35000, 99000] * 1_000_000,
}

df = pd.DataFrame(sample_data)
# This creates a file around 100MB+
df.to_csv("fake_spotify_data.csv", index=False)
print("Sample data file created!")
```

---

### Step 2: Track and Push with Git LFS (Terminal Commands)

Run these terminal commands in your repository folder:

```bash
# 1. Initialize Git LFS in your repo
git lfs install

# 2. Tell Git LFS to track large CSV files
git lfs track "*.csv"

# 3. Add the tracking configuration and the large file
git add .gitattributes
git add fake_spotify_data.csv

# 4. Commit and push
git commit -m "Add large fake Spotify data using Git LFS"
git push origin main
```

---

### Step 3: `README.md` Content

Add this explanation to your `README.md` file:

```markdown
## Why Raw Data Belongs in Git LFS, Not Regular Git

Raw data files should not be committed directly to regular Git repositories. First, GitHub blocks files larger than 100 MB to prevent repository storage bloat. Second, pushing and pulling large files wastes network bandwidth and makes cloning the repository very slow for other contributors. Finally, raw datasets often contain private, sensitive, or copyrighted user information that must be kept out of public version control history. Using tools like Git LFS solves this by storing the actual data on a separate server while keeping your Git history lightweight.
```

## Question 4

Showcase your data science work by adding to your GitHub repo: a Jupyter notebook (analysis.ipynb), a processed dataset (processed_data.csv), a plot image (plot.png), and a Python script (helper_functions.py).

Constraint: Do not include raw or unprocessed data files.

**Answer:**

Here are the files you need to create and add to your GitHub repository.

---

### 1. Python Script (`helper_functions.py`)
This script contains reusable functions to clean and process your data.

```python
# helper_functions.py


def clean_salary(salary):
    """Fills missing salaries with the average salary."""
    return salary.fillna(salary.mean())
```

---

### 2. Jupyter Notebook (`analysis.ipynb`)
Run this code inside your notebook. It creates sample data, processes it using your helper function, and saves both the processed data and the plot.

```python
import matplotlib.pyplot as plt
import pandas as pd
from helper_functions import clean_salary

# Sample Data (Raw)
raw_data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Salary": [50000, None, 70000, 65000],
}
df = pd.DataFrame(raw_data)

# Process Data (Constraint: only save the processed version)
df["Salary"] = clean_salary(df["Salary"])
df.to_csv("processed_data.csv", index=False)

# Create and save plot
plt.figure()
plt.bar(df["Name"], df["Salary"], color="skyblue")
plt.title("Employee Salaries")
plt.xlabel("Name")
plt.ylabel("Salary")
plt.savefig("plot.png")
plt.show()
```

**Explanation:**  
This code defines sample data, uses the function from `helper_functions.py` to fix missing values, exports `processed_data.csv`, and exports `plot.png`.

---

### 3. Add to GitHub (Terminal Commands)
Run these commands in your project folder to push only the required files and avoid including raw data:

```bash
# Add only the required files
git add analysis.ipynb processed_data.csv plot.png helper_functions.py

# Commit the changes
git commit -m "Add analysis notebook, processed data, plot, and helper script"

# Push to your GitHub repository
git push origin main
```
