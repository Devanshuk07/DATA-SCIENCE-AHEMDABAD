## Question 1

Install Git on your operating system (Windows, Mac, or Linux), open a terminal, and run git --version to confirm the installation.

**Answer:**

Here are the simple steps to install and check Git based on your operating system:

### 1. Install Git

* **Windows:** Download and run the installer from the official website (git-scm.com). Use the default settings during setup.
* **Mac:** Open your terminal and run `xcode-select --install`, or install it using Homebrew with `brew install git`.
* **Linux (Ubuntu/Debian):** Open your terminal and run:
  ```bash
  sudo apt update
  sudo apt install git
  ```

---

### 2. Confirm Installation

Open your terminal (or Git Bash on Windows) and type:

```bash
git --version
```

**Expected Output:**
You should see a message showing the installed version, similar to this:
```text
git version 2.43.0
```

## Question 2

Configure your Git username and email globally using git config --global user.name and git config --global user.email with your actual details.

**Answer:**

Run the following commands in your terminal to set your Git username and email globally. 

Replace `"Your Name"` and `"your.email@example.com"` with your actual details:

```bash
git config --global user.name "Your Name"
git config --global user.email "your.email@example.com"
```

To verify that your details were saved correctly, run:

```bash
git config --list
```

## Question 3

Create a new folder called InstaCloneRepo, initialize it as a Git repository using git init, and observe the .git folder that gets created.

**Answer:**

Here are the commands to run in your terminal:

```bash
# 1. Create the new folder
mkdir InstaCloneRepo

# 2. Move into the new folder
cd InstaCloneRepo

# 3. Initialize the Git repository
git init

# 4. View the hidden .git folder
ls -a
```
*(Note: If you are using Windows Command Prompt, use `dir /a` instead of `ls -a` to view hidden files).*

**Observation:**
Running `git init` creates a hidden folder named `.git` inside your project directory. This folder is where Git stores all the tracking data, settings, and version history for your project. It is hidden by default so you do not accidentally edit or delete it.

## Question 4

Clone the public repository https://github.com/public-apis/public-apis into your system using git clone, then run git status inside the cloned folder and describe what you see.

**Answer:**

Run these commands in your terminal:

```bash
git clone https://github.com/public-apis/public-apis
cd public-apis
git status
```

**What you will see:**

After running `git status`, Git will show:
1. `On branch master` – This tells you the current active branch.
2. `Your branch is up to date with 'origin/master'` – This means your local copy matches the online repository.
3. `nothing to commit, working tree clean` – This confirms you have not made, modified, or deleted any files yet.

## Question 5

Create a file called playlist.txt inside your InstaCloneRepo folder, add three song names, stage and commit the file using git add and git commit, and write a meaningful commit message following best practices (e.g., 'Add initial playlist with 3 songs').

Hint: Make sure your commit message is clear, concise, and describes the change as if you were working on a real music app project.

**Answer:**

Here are the terminal commands to complete this task:

### Commands

```bash
# 1. Navigate into your repository folder
cd InstaCloneRepo

# 2. Create the file and add three song names
echo -e "Bohemian Rhapsody\nShape of You\nBlinding Lights" > playlist.txt

# 3. Stage the new file
git add playlist.txt

# 4. Commit the file with a meaningful message
git commit -m "Add initial playlist with 3 songs"
```

### Explanation
First, we use `echo` to create `playlist.txt` and write three song names into it. Next, `git add playlist.txt` stages the file so Git knows to track it. Finally, `git commit` saves the changes permanently with a clear, descriptive message written in the present tense.
