## Question 1

Create a new Git branch called feature-playlist in your local clone of any existing project folder, switch to it, and add a file called playlist.md describing your top 3 favorite Spotify playlists.

**Answer:**

Here are the steps to do this using your terminal:

## Question 2

Simulate a team workflow by merging your feature-playlist branch into main using a merge commit (not fast-forward), then intentionally create a merge conflict by editing the same line in playlist.md on both branches, and resolve the conflict using the VS Code merge tool.

Hint: Make sure to save different changes to the same line in both main and feature-playlist before merging.

**Answer:**

Here is a step-by-step guide to complete this workflow:

### Step 1: Merge with a merge commit (no fast-forward)
Switch to the `main` branch and merge `feature-playlist` without fast-forwarding:

```bash
git checkout main
git merge --no-ff feature-playlist -m "Merge feature-playlist into main"
```

---

### Step 2: Create a merge conflict
Make different changes to line 1 of `playlist.md` on both branches.

1. **Edit on `main`:**
   * Open `playlist.md` and change line 1 to: `# Main Playlist`
   * Commit the change:
     ```bash
     git commit -am "Update title on main"
     ```

2. **Edit on `feature-playlist`:**
   * Switch branches:
     ```bash
     git checkout feature-playlist
     ```
   * Open `playlist.md` and change line 1 to: `# Feature Playlist`
   * Commit the change:
     ```bash
     git commit -am "Update title on feature branch"
     ```

3. **Trigger the conflict:**
   * Switch back to `main` and try to merge:
     ```bash
     git checkout main
     git merge feature-playlist
     ```
   * Git will stop and warn you that there is a merge conflict in `playlist.md`.

---

### Step 3: Resolve using the VS Code Merge Tool
1. Open `playlist.md` in **VS Code**.
2. Click the **"Resolve in Merge Editor"** button in the bottom-right corner of the editor.
3. Compare the incoming changes (feature branch) and current changes (`main`). Click the checkboxes to accept the version you want (or click **"Accept Current"** / **"Accept Incoming"** at the top of the conflict block).
4. Click **"Complete Merge"** in the bottom-right corner.
5. Finalize the merge in your terminal:
   ```bash
   git add playlist.md
   git commit -m "Resolve merge conflict in playlist.md"
   ```

## Question 3

Fork a public GitHub repository related to data science (for example, a simple dataset repo), clone it to your machine, create a new branch called feature-zomato-analysis, and add a file zomato_analysis.txt with a one-line description of how you would analyze Zomato order data.

**Answer:**

Here are the steps and Git commands to complete this task:

**Step 1: Fork the repository**
Go to GitHub, find a public data science repository (for example, a dataset repo), and click the **Fork** button at the top-right of the page to create a copy under your account.

**Step 2: Clone the repository to your machine**
Open your terminal or command prompt and run:
```bash
git clone https://github.com/<your-username>/<repo-name>.git
cd <repo-name>
```

**Step 3: Create and switch to the new branch**
```bash
git checkout -b feature-zomato-analysis
```

**Step 4: Create the file with the description**
Run this command to create the file and add your one-line analysis plan:
```bash
echo "I will analyze customer ratings and delivery times across different cuisines using Pandas." > zomato_analysis.txt
```

**Step 5: Save your changes**
```bash
git add zomato_analysis.txt
git commit -m "Add Zomato analysis plan"
```

## Question 4

Push your feature-zomato-analysis branch to your fork on GitHub and open a Pull Request (PR) to the original repository. In the PR, request a review from a classmate and add a comment explaining what your change does.

**Answer:**

Follow these simple steps to push your branch and create the Pull Request (PR):

### Step 1: Push your branch using Git
Open your terminal and run this command to push your branch to your GitHub fork:
```bash
git push origin feature-zomato-analysis
```

---

### Step 2: Open the Pull Request on GitHub
1. Go to your GitHub profile and open your forked repository.
2. Click the yellow **"Compare & pull request"** button that appears at the top.
3. Ensure the **base repository** is the original repository and the **head repository** is your fork with `feature-zomato-analysis`.

---

### Step 3: Add your comment and request a review
1. **Add a comment:** In the description box, briefly explain your work.  
   *Example:*  
   > "This PR adds data cleaning and exploratory analysis for the Zomato dataset, focusing on average ratings and price ranges."
2. **Request a review:** On the right sidebar, click on **Reviewers** (gear icon) and search for your classmate's GitHub username.
3. Click **"Create pull request"**.
