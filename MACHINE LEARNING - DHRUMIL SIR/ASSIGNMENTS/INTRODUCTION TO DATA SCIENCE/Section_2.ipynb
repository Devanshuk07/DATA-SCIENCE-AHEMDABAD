## Question 2

Use the requests library in Python to fetch the latest 5 posts from the JSONPlaceholder API (https://jsonplaceholder.typicode.com/posts) and print the title of each post.

Hint: Use requests.get() and response.json().

**Answer:**

Here is the code to fetch and print the titles of the first 5 posts:

```python
import requests

# 1. Fetch data from the API
url = "https://jsonplaceholder.typicode.com/posts"
response = requests.get(url)

# 2. Convert the response into a Python list of dictionaries
posts = response.json()

# 3. Loop through the first 5 posts and print their titles
for post in posts[:5]:
    print(post["title"])
```

### Explanation:
* **`requests.get(url)`** sends a request to the website to retrieve the data.
* **`response.json()`** transforms the raw API data into a standard Python list of dictionaries.
* **`posts[:5]`** uses Python slicing to take only the first 5 posts, and `post["title"]` accesses the title of each post.

## Question 3

Given a CSV file with columns: song_name, artist, streams, and genre (simulate a mini Spotify dataset with at least 10 rows), write Python code to load the data using pandas and display the number of songs in each genre.

**Answer:**

Here is the Python code to load the simulated data and count the songs in each genre.

```python
import io
import pandas as pd

# --- Sample Data (Simulated CSV file) ---
csv_data = """song_name,artist,streams,genre
Blinding Lights,The Weeknd,3500,Pop
Shape of You,Ed Sheeran,3300,Pop
Someone Like You,Adele,1500,Pop
Bohemian Rhapsody,Queen,2100,Rock
Smells Like Teen Spirit,Nirvana,1700,Rock
Hotel California,Eagles,1400,Rock
God's Plan,Drake,2200,Hip-Hop
HUMBLE.,Kendrick Lamar,1900,Hip-Hop
Lose Yourself,Eminem,1800,Hip-Hop
Take Five,Dave Brubeck,500,Jazz
"""

# Load the data using pandas
df = pd.read_csv(io.StringIO(csv_data))

# Count and display the number of songs in each genre
genre_counts = df["genre"].value_counts()
print(genre_counts)
```

**Explanation:**
1. We use `io.StringIO` to read the sample CSV text directly into a pandas DataFrame without needing an external file.
2. The `value_counts()` function automatically counts how many times each genre appears in the `genre` column.

## Question 4

Create a simple bar chart using matplotlib to visualize the number of movies per genre from a dataset (you can make up a small dataset with at least 3 genres and 10 movies, like BookMyShow's movie listings).

**Answer:**

Here is the Python code using sample data to create the bar chart:

```python
import matplotlib.pyplot as plt
import pandas as pd

# Sample Data: 10 movies with genres
data = {
    "Movie": ["M1", "M2", "M3", "M4", "M5", "M6", "M7", "M8", "M9", "M10"],
    "Genre": [
        "Action",
        "Action",
        "Action",
        "Action",
        "Comedy",
        "Comedy",
        "Comedy",
        "Drama",
        "Drama",
        "Drama",
    ],
}

df = pd.DataFrame(data)

# Count the number of movies per genre
genre_counts = df["Genre"].value_counts()

# Create the bar chart
plt.bar(genre_counts.index, genre_counts.values, color="skyblue")

# Add labels and title
plt.xlabel("Genre")
plt.ylabel("Number of Movies")
plt.title("Number of Movies per Genre")

# Display the chart
plt.show()
```

### Brief Explanation:
1. We create a sample DataFrame with 10 movies divided into three genres (Action, Comedy, and Drama).
2. The `value_counts()` function counts how many movies belong to each genre.
3. Finally, `plt.bar()` plots these counts as a bar chart, and we add simple labels to make the chart easy to read.
