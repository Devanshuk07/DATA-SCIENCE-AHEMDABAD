## Question 1

Write a Python function called perceive_input() that takes a user's text message (like 'Show me trending songs on Spotify') and returns a structured dictionary with intent and keywords extracted from the input.

Hint: Use basic string methods or regular expressions to identify intent words (like 'show', 'play', 'book') and key app names.

**Answer:**

```python
import re


def perceive_input(text):
    text_lower = text.lower()

    # Identify intent
    intent_match = re.search(r"\b(show|play|book)\b", text_lower)
    intent = intent_match.group(0) if intent_match else "unknown"

    # Identify app name
    app_match = re.search(r"\b(spotify|netflix|uber)\b", text_lower)
    app = app_match.group(0) if app_match else None

    # Extract general keywords by removing common stopwords
    stopwords = {"me", "on", "the", "a", "an", "in", "to", "for"}
    words = re.findall(r"\b\w+\b", text_lower)
    keywords = [
        w
        for w in words
        if w not in stopwords and w != intent and (app is None or w != app)
    ]

    return {"intent": intent, "app": app, "keywords": keywords}
```

## Question 2

Implement a simple reasoning function called decide_next_action(perceived_data) that takes the output of perceive_input() and returns the next step as a string, such as 'fetch_spotify_trending' or 'search_movie_bookmyshow'.

**Answer:**

```python
def decide_next_action(perceived_data):
    intent = perceived_data.get("intent", "").lower()
    entities = perceived_data.get("entities", [])

    if "music" in intent or "song" in entities:
        return "fetch_spotify_trending"
    elif "movie" in intent or "cinema" in entities:
        return "search_movie_bookmyshow"
    else:
        return "ask_clarification"
```

## Question 3

Simulate the 'act' phase by creating an execute_action(action) function that prints a mock response for each action, e.g., 'Fetching trending songs from Spotify...' or 'Searching movies on BookMyShow...'. Connect this with your previous functions to complete a basic perceive → think → act loop.

**Answer:**

```python
def perceive(user_input):
    return user_input.lower()


def think(perceived_data):
    if "song" in perceived_data or "music" in perceived_data:
        return "fetch_songs"
    elif "movie" in perceived_data:
        return "search_movies"
    else:
        return "unknown"


def execute_action(action):
    if action == "fetch_songs":
        print("Fetching trending songs from Spotify...")
    elif action == "search_movies":
        print("Searching movies on BookMyShow...")
    else:
        print("Action not recognized.")


# Perceive -> Think -> Act loop
user_request = "Play some trending songs"

perceived = perceive(user_request)  # Perceive
action = think(perceived)  # Think
execute_action(action)  # Act
```
