## Question 1

Use the Open-Meteo Weather API to fetch and display the current temperature for your city in a Python script.

Hint: Refer to the Open-Meteo API docs for the correct endpoint and parameters.

**Answer:**

```python
import requests

# Coordinates for London (replace with your city's latitude and longitude)
latitude = 51.5074
longitude = -0.1278

url = f"https://api.open-meteo.com/v1/forecast?latitude={latitude}&longitude={longitude}&current=temperature_2m"

response = requests.get(url).json()
current_temp = response["current"]["temperature_2m"]
temp_unit = response["current_units"]["temperature_2m"]

print(f"Current temperature: {current_temp}{temp_unit}")
```

## Question 2

Write a Python function that reads a CSV file named 'my_expenses.csv' (with columns: date, amount, category) and prints the total amount spent in the 'Food' category.

**Answer:**

```python
import csv


def total_food_expenses():
    total = 0.0
    with open("my_expenses.csv", mode="r") as file:
        reader = csv.DictReader(file)
        for row in reader:
            if row["category"].strip() == "Food":
                total += float(row["amount"])
    print(total)
```

## Question 3

Integrate the GNews API in a Python script to fetch and display the top 3 latest technology news headlines.

**Answer:**

```python
import requests

API_KEY = "YOUR_API_KEY"
url = f"https://gnews.io/api/v4/top-headlines?category=technology&lang=en&max=3&apikey={API_KEY}"

response = requests.get(url)
data = response.json()

for i, article in enumerate(data.get("articles", []), 1):
    print(f"{i}. {article['title']}")
```

## Question 4

Build a calculator tool in Python that takes a mathematical expression as input (like '23+7*2') and outputs the result, handling basic operations (+, -, *, /).

Constraint: Do not use Python's eval() function for this task.

**Answer:**

```python
import operator
import re


def calculate(expression: str) -> float:
    # Tokenize the expression into numbers and operators
    tokens = re.findall(r"\d+\.?\d*|[+\-*/]", expression.replace(" ", ""))

    ops = {
        "+": operator.add,
        "-": operator.sub,
        "*": operator.mul,
        "/": operator.truediv,
    }

    # First pass: Handle multiplication and division
    i = 0
    while i < len(tokens):
        if tokens[i] in ("*", "/"):
            left = float(tokens[i - 1])
            right = float(tokens[i + 1])
            res = ops[tokens[i]](left, right)
            tokens[i - 1 : i + 2] = [res]
            i -= 1
        else:
            i += 1

    # Second pass: Handle addition and subtraction
    i = 0
    while i < len(tokens):
        if tokens[i] in ("+", "-"):
            left = float(tokens[i - 1])
            right = float(tokens[i + 1])
            res = ops[tokens[i]](left, right)
            tokens[i - 1 : i + 2] = [res]
            i -= 1
        else:
            i += 1

    return float(tokens[0])
```
