## Question 1

Use the Gemini API to fetch a summary of the latest IPL cricket match and print the summary in your Python script.

Hint: Use the requests library to call the Gemini API endpoint and handle the JSON response.

**Answer:**

```python
import os
import requests

API_KEY = os.getenv("GEMINI_API_KEY")
url = f"https://generativelanguage.googleapis.com/v1beta/models/gemini-1.5-flash:generateContent?key={API_KEY}"

headers = {"Content-Type": "application/json"}
payload = {
    "contents": [
        {
            "parts": [
                {"text": "Provide a brief summary of the latest IPL cricket match."}
            ]
        }
    ]
}

response = requests.post(url, headers=headers, json=payload)
data = response.json()

summary = data["candidates"][0]["content"]["parts"][0]["text"]
print(summary)
```

## Question 2

Integrate the Hugging Face Transformers library in Python to perform sentiment analysis on 5 recent Zomato food reviews (you can use any sample review texts). Display whether each review is positive, negative, or neutral.

**Answer:**

```python
from transformers import pipeline

# Load a sentiment analysis pipeline supporting positive, negative, and neutral labels
sentiment_analyzer = pipeline(
    "sentiment-analysis",
    model="cardiffnlp/twitter-roberta-base-sentiment-latest",
)

# 5 sample Zomato food reviews
reviews = [
    "The biryani was absolutely delicious, hot, and delivered on time!",
    "Worst experience ever. The food was cold and the packaging was completely damaged.",
    "The pizza was okay, nothing special, but average for the price.",
    "Loved the creamy butter chicken and garlic naan. Will definitely order again!",
    "The delivery was made, but the taste was neither good nor bad.",
]

# Analyze and display results
results = sentiment_analyzer(reviews)

for review, result in zip(reviews, results):
    print(f"Review: {review}")
    print(f"Sentiment: {result['label']}\n")
```

## Question 3

Write a simple Python agent loop that takes a list of WhatsApp-style messages, checks if any message contains the word 'urgent', and prints 'Alert: Urgent message received!' if found, otherwise prints 'No urgent messages.'

**Answer:**

```python
messages = [
    "Hey, are you free today?",
    "Please call me, this is urgent!",
    "See you tomorrow.",
]

urgent_found = False

for msg in messages:
    if "urgent" in msg.lower():
        urgent_found = True
        break

if urgent_found:
    print("Alert: Urgent message received!")
else:
    print("No urgent messages.")
```

## Question 4

Modify your agent logic to use a local inference model (such as a small spaCy NER model) to detect if any message in a given list mentions a restaurant name (like 'Barbeque Nation' or 'Domino's'). Print all detected restaurant names.

**Answer:**

```python
import spacy

# Load local small spaCy NER model
nlp = spacy.load("en_core_web_sm")

messages = [
    "Let's grab lunch at Domino's today.",
    "I have a meeting at 3 PM.",
    "We celebrated the party at Barbeque Nation.",
    "The weather is nice.",
]

detected_restaurants = []

for message in messages:
    doc = nlp(message)
    for ent in doc.ents:
        # Check for named entities typically representing business names
        if ent.label_ in ["ORG", "FAC"]:
            detected_restaurants.append(ent.text)

print("Detected restaurant names:", detected_restaurants)
```
