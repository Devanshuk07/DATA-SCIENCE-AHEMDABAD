## Question 1

Write a Python function that simulates an LLM agent's 'internal thought' and 'external action' process for a food delivery chatbot. The function should print its reasoning (internal thought) before deciding whether to suggest a restaurant, show the menu, or confirm an order (external action) based on a user's input.

**Answer:**

```python
def food_delivery_agent(user_input: str):
    user_input = user_input.lower()

    # Internal Thought: Analyze intent based on input
    if any(word in user_input for word in ["order", "buy", "checkout"]):
        thought = "User wants to finalize a purchase. I should confirm the order details."
        action = "Action: Confirming your order now. Please provide your delivery address."
    elif any(word in user_input for word in ["menu", "food", "dishes", "items"]):
        thought = (
            "User is looking for specific options. I should display the menu."
        )
        action = "Action: Here is the menu: 1. Burger, 2. Pizza, 3. Pasta."
    else:
        thought = "User's request is general or they are looking for a place to eat. I should recommend a restaurant."
        action = "Action: I suggest ordering from 'The Italian Place'. Would you like to see their menu?"

    # Output Thought (Internal) then Action (External)
    print(f"Thought: {thought}")
    print(action)
```

## Question 3

Implement a simple ReAct Pattern (Reason + Action) in code for a virtual cricket assistant. The agent should reason about a user's question like 'Who won the last IPL match?' and then take an action to fetch or display the answer.

Hint: Use print statements to show the 'Reason' step before the 'Action' step.

**Answer:**

```python
# Simulated tool to fetch match details
def fetch_cricket_data(query):
    database = {
        "last_ipl_winner": "Kolkata Knight Riders won the last IPL final."
    }
    return database.get(query, "Information not found.")


def cricket_react_agent(user_query):
    # Step 1: Reason
    print("User Question:", user_query)
    print(
        "Thought (Reason): The user is asking for the winner of the recent IPL match. "
        "I need to call the database tool with the key 'last_ipl_winner' to retrieve the result."
    )

    # Step 2: Action
    print("Action: Calling tool `fetch_cricket_data('last_ipl_winner')`...")
    result = fetch_cricket_data("last_ipl_winner")

    # Final Output
    print("Final Answer:", result)


# Run the agent
cricket_react_agent("Who won the last IPL match?")
```

Use ChatGPT or another LLM-based AI tool to generate a step-by-step reasoning (internal thought) and action plan for answering the question: 'How do I transfer money using PhonePe?'. Paste the AI's output and highlight how it separates reasoning from action.

**Answer:**

### Generated AI Output

**Internal Thought (Reasoning):**
> The user wants to know how to transfer money using PhonePe. The core flow requires opening the app, selecting a transfer mode (contact, account/IFSC, or UPI ID), entering recipient details, specifying the amount, and authenticating via UPI PIN. I should structure this into clear, sequential steps for a standard mobile-to-mobile transfer as it is the most common method.

**Action Plan (User-Facing Response):**
1. **Open PhonePe:** Launch the app on your mobile device.
2. **Select Transfer Option:** Under the "Transfer Money" section on the home screen, tap **To Mobile Number** (or *To Bank/UPI ID*).
3. **Enter Recipient Details:** Search for the recipient's name/number or manually enter their mobile number/UPI ID.
4. **Enter Amount:** Type the amount you wish to transfer.
5. **Select Bank Account:** Choose the linked bank account to debit the money from.
6. **Authenticate:** Tap **Pay**, enter your secret **UPI PIN**, and confirm the transaction.

---

### How Reasoning is Separated from Action
* **Structural Isolation:** Reasoning is contained within an internal block (the "Thought" phase) that remains hidden from or marked distinct from the output, while the "Action" is the user-facing instruction.
* **Objective Difference:** The reasoning focuses on **intent recognition and planning** (evaluating user goals and organizing steps), whereas the action focuses purely on **execution and delivery** (direct, step-by-step instructions).
