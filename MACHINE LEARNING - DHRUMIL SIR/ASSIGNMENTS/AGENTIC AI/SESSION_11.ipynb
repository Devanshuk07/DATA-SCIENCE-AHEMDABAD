## Question 1

Simulate a multi-agent system for a food delivery scenario: create three Python functions—fetch_menu(), analyze_orders(), and generate_delivery_report()—where each function represents an agent handling one part of the process. Call them in sequence and print the final delivery report.

**Answer:**

Here is a complete, working Python simulation of the multi-agent food delivery system. 

In this implementation:
1. **`fetch_menu()` (Menu Agent)**: Simulates retrieving available restaurant items and their prices.
2. **`analyze_orders()` (Order Processing Agent)**: Validates incoming customer orders against the menu, calculates sub-totals, delivery fees, and taxes.
3. **`generate_delivery_report()` (Dispatch/Reporting Agent)**: Formats the processed order details into a clean, human-readable delivery manifest.

### Python Code

```python
import json


def fetch_menu():
    """Agent 1: Menu Agent

    Retrieves and provides the current available restaurant menu.
    """
    menu = {
        "BURGER": 9.99,
        "PIZZA": 14.50,
        "FRIES": 3.50,
        "SODA": 1.99,
        "SALAD": 7.25,
    }
    return menu


def analyze_orders(raw_orders, menu):
    """Agent 2: Order Processing Agent

    Validates items, calculates costs, and handles delivery logistics.
    """
    processed_orders = []
    delivery_fee = 3.99
    tax_rate = 0.08  # 8%

    for order in raw_orders:
        customer = order["customer_name"]
        address = order["delivery_address"]
        items = order["items"]

        subtotal = 0.0
        valid_items = []
        unavailable_items = []

        # Validate items and sum prices
        for item in items:
            item_upper = item.upper()
            if item_upper in menu:
                valid_items.append(item_upper)
                subtotal += menu[item_upper]
            else:
                unavailable_items.append(item)

        tax = subtotal * tax_rate
        total = subtotal + tax + delivery_fee

        processed_orders.append(
            {
                "customer": customer,
                "address": address,
                "fulfilled_items": valid_items,
                "unavailable_items": unavailable_items,
                "subtotal": round(subtotal, 2),
                "tax": round(tax, 2),
                "delivery_fee": delivery_fee,
                "total": round(total, 2),
            }
        )

    return processed_orders


def generate_delivery_report(processed_orders):
    """Agent 3: Delivery Reporting Agent

    Generates a structured delivery manifest for the fleet.
    """
    report = []
    report.append("=" * 45)
    report.append("          FINAL DELIVERY MANIFEST            ")
    report.append("=" * 45)

    grand_total_revenue = 0.0

    for idx, order in enumerate(processed_orders, 1):
        report.append(f"\n[Order #{idx}] Customer: {order['customer']}")
        report.append(f"Deliver To : {order['address']}")
        report.append(f"Items      : {', '.join(order['fulfilled_items'])}")

        if order["unavailable_items"]:
            report.append(
                f"Out-of-Stock/Skipped: {', '.join(order['unavailable_items'])}"
            )

        report.append(f"Subtotal   : ${order['subtotal']:.2f}")
        report.append(f"Tax (8%)   : ${order['tax']:.2f}")
        report.append(f"Delivery   : ${order['delivery_fee']:.2f}")
        report.append(f"Total Due  : ${order['total']:.2f}")
        report.append("-" * 30)

        grand_total_revenue += order["total"]

    report.append(f"\nTotal Orders Scheduled : {len(processed_orders)}")
    report.append(f"Grand Total Revenue    : ${grand_total_revenue:.2f}")
    report.append("=" * 45)

    return "\n".join(report)


# --- Sequential Execution Pipeline ---
if __name__ == "__main__":
    # Sample incoming orders to be handled by the system
    incoming_customer_orders = [
        {
            "customer_name": "Alice Smith",
            "delivery_address": "123 Maple St",
            "items": ["Burger", "Fries", "Soda"],
        },
        {
            "customer_name": "Bob Jones",
            "delivery_address": "456 Oak Ave",
            "items": ["Pizza", "Pasta", "Soda"],  # "Pasta" is not on the menu
        },
    ]

    # Step 1: Agent 1 fetches the menu
    active_menu = fetch_menu()

    # Step 2: Agent 2 analyzes and prices the orders using the menu
    analyzed_order_data = analyze_orders(incoming_customer_orders, active_menu)

    # Step 3: Agent 3 generates the final output report
    final_report = generate_delivery_report(analyzed_order_data)

    # Output the result
    print(final_report)
```

---

### Output Generated

```text
=============================================
          FINAL DELIVERY MANIFEST            
=============================================

[Order #1] Customer: Alice Smith
Deliver To : 123 Maple St
Items      : BURGER, FRIES, SODA
Subtotal   : $15.48
Tax (8%)   : $1.24
Delivery   : $3.99
Total Due  : $20.71
------------------------------

[Order #2] Customer: Bob Jones
Deliver To : 456 Oak Ave
Items      : PIZZA, SODA
Out-of-Stock/Skipped: Pasta
Subtotal   : $16.49
Tax (8%)   : $1.32
Delivery   : $3.99
Total Due  : $21.80
------------------------------

Total Orders Scheduled : 2
Grand Total Revenue    : $42.51
=============================================
```

## Question 2

Implement cooperative agent behavior by having two agents (functions) in Python: one fetches trending songs from a mock Spotify playlist (as a list), and the second analyzes the list to count how many songs are from the 'Pop' genre. Display the result.

**Answer:**

Here is a clear and complete Python implementation demonstrating cooperative agent behavior. 

In this design:
1. **Agent 1 (`playlist_fetcher_agent`)** simulates retrieving data from an external environment (a mock Spotify playlist) and hands the output over.
2. **Agent 2 (`genre_analyzer_agent`)** receives that data, performs a specific analytical task (filtering and counting 'Pop' songs), and produces the final metric.
3. A **coordinator workflow** links them together, showing the cooperative flow of information.

---

### Python Implementation

```python
from typing import List, Dict

# --- Agent 1: Data Retrieval Agent ---
def playlist_fetcher_agent() -> List[Dict[str, str]]:
    """
    Simulates fetching a trending playlist from Spotify.
    Returns a list of song dictionaries containing title, artist, and genre.
    """
    print("[Agent 1: Fetcher] Accessing mock Spotify API...")
    
    # Mock data representing the retrieved playlist
    trending_playlist = [
        {"title": "Espresso", "artist": "Sabrina Carpenter", "genre": "Pop"},
        {"title": "Not Like Us", "artist": "Kendrick Lamar", "genre": "Hip-Hop"},
        {"title": "Birds of a Feather", "artist": "Billie Eilish", "genre": "Pop"},
        {"title": "A Bar Song (Tipsy)", "artist": "Shaboozey", "genre": "Country"},
        {"title": "Good Luck, Babe!", "artist": "Chappell Roan", "genre": "Pop"},
        {"title": "Houdini", "artist": "Eminem", "genre": "Hip-Hop"},
        {"title": "Cruel Summer", "artist": "Taylor Swift", "genre": "Pop"},
    ]
    
    print(f"[Agent 1: Fetcher] Successfully retrieved {len(trending_playlist)} trending songs.\n")
    return trending_playlist


# --- Agent 2: Analytical Agent ---
def genre_analyzer_agent(playlist: List[Dict[str, str]], target_genre: str = "Pop") -> int:
    """
    Analyzes a given playlist and counts how many songs match the target genre.
    """
    print(f"[Agent 2: Analyzer] Receiving playlist data...")
    print(f"[Agent 2: Analyzer] Scanning for genre: '{target_genre}'...")
    
    # Case-insensitive comparison to ensure accuracy
    pop_count = sum(
        1 for track in playlist 
        if track.get("genre", "").strip().lower() == target_genre.lower()
    )
    
    return pop_count


# --- Cooperative Execution ---
def main():
    print("=== Starting Cooperative Agent Workflow ===\n")
    
    # Step 1: Agent 1 performs its task
    trending_songs = playlist_fetcher_agent()
    
    # Step 2: Agent 1 passes the result directly to Agent 2
    pop_song_count = genre_analyzer_agent(playlist=trending_songs, target_genre="Pop")
    
    # Display the final cooperative result
    print("\n=== Workflow Results ===")
    print(f"Total 'Pop' songs found: {pop_song_count} out of {len(trending_songs)} songs.")

if __name__ == "__main__":
    main()
```

---

### Output

```text
=== Starting Cooperative Agent Workflow ===

[Agent 1: Fetcher] Accessing mock Spotify API...
[Agent 1: Fetcher] Successfully retrieved 7 trending songs.

[Agent 2: Analyzer] Receiving playlist data...
[Agent 2: Analyzer] Scanning for genre: 'Pop'...

=== Workflow Results ===
Total 'Pop' songs found: 4 out of 7 songs.
```

### Key Elements of Cooperation Demonstrated:
* **Separation of Concerns:** Each agent has a single, well-defined responsibility (fetching vs. analyzing).
* **Inter-Agent Communication:** Agent 1's output schema (`List[Dict]`) serves directly as the input contract for Agent 2.
* **Modularity:** Either agent can be updated independently (e.g., swapping the mock list with real Spotify Web API calls in Agent 1) without breaking Agent 2.

## Question 3

Build a simple task delegation system: create a Python script where Agent 1 fetches recent IPL match scores (use a predefined dictionary), Agent 2 calculates the highest team score, and Agent 3 prints a formatted summary report for the user.

**Answer:**

Here is a complete Python script implementing the task delegation system. 

In this design, tasks are divided among three separate agent functions, and an orchestrator function coordinates the flow of data from one agent to the next.

```python
"""
Simple Task Delegation System: IPL Match Report
Agent 1: Fetches raw match data.
Agent 2: Analyzes data to find the highest team score.
Agent 3: Formats and prints the final report.
"""

# ==========================================
# Agent 1: Data Retrieval Agent
# ==========================================
def agent_1_fetch_scores() -> list[dict]:
    """
    Fetches recent IPL match scores from a predefined data source.
    """
    # Predefined dictionary-based dataset of recent matches
    ipl_matches = [
        {"match": "CSK vs RCB", "team1": "CSK", "score1": 218, "team2": "RCB", "score2": 214},
        {"match": "MI vs KKR",  "team1": "MI",  "score1": 157, "team2": "KKR", "score2": 160},
        {"match": "SRH vs DC",  "team1": "SRH", "score1": 266, "team2": "DC",  "score2": 199},
        {"match": "RR vs GT",   "team1": "RR",  "score1": 196, "team2": "GT",  "score2": 199},
    ]
    print("[Agent 1] Successfully fetched recent IPL match scores.")
    return ipl_matches


# ==========================================
# Agent 2: Analytical Agent
# ==========================================
def agent_2_calculate_highest_score(matches: list[dict]) -> dict:
    """
    Processes the match data to find the highest team score recorded.
    """
    highest_score = -1
    top_team = None
    top_match = None

    for match in matches:
        # Check team 1
        if match["score1"] > highest_score:
            highest_score = match["score1"]
            top_team = match["team1"]
            top_match = match["match"]

        # Check team 2
        if match["score2"] > highest_score:
            highest_score = match["score2"]
            top_team = match["team2"]
            top_match = match["match"]

    print("[Agent 2] Analysis complete. Identified the highest team score.")
    return {
        "team": top_team,
        "score": highest_score,
        "match": top_match
    }


# ==========================================
# Agent 3: Reporting Agent
# ==========================================
def agent_3_generate_report(matches: list[dict], top_performance: dict) -> None:
    """
    Formats the raw match data and analytical findings into a user-friendly report.
    """
    print("\n" + "=" * 45)
    print("         IPL RECENT MATCHES SUMMARY         ")
    print("=" * 45)
    
    print("\nRecent Results:")
    print("-" * 45)
    for m in matches:
        print(f" • {m['team1']} ({m['score1']}) vs {m['team2']} ({m['score2']})")
    
    print("-" * 45)
    print("Peak Performance of the Round:")
    print(f" • Top Team  : {top_performance['team']}")
    print(f" • Score     : {top_performance['score']} runs")
    print(f" • Match     : {top_performance['match']}")
    print("=" * 45 + "\n")


# ==========================================
# Orchestrator: Task Delegation Controller
# ==========================================
def run_task_delegation_system():
    """
    Coordinates the pipeline by delegating tasks sequentially to Agents 1, 2, and 3.
    """
    print("Starting Task Delegation System...\n")
    
    # Task 1: Delegated to Agent 1
    match_data = agent_1_fetch_scores()
    
    # Task 2: Delegated to Agent 2 (passing output of Agent 1)
    highest_score_info = agent_2_calculate_highest_score(match_data)
    
    # Task 3: Delegated to Agent 3 (passing outputs of Agents 1 & 2)
    agent_3_generate_report(match_data, highest_score_info)


if __name__ == "__main__":
    run_task_delegation_system()
```

### How the System Works:
1. **Agent 1 (`agent_1_fetch_scores`)**: Acts as the data provider. It holds the predefined data structure (a list of match dictionaries) and returns it cleanly.
2. **Agent 2 (`agent_2_calculate_highest_score`)**: Acts as the data processor. It receives the dataset, iterates through both teams' scores in every match, and extracts the record for the highest score.
3. **Agent 3 (`agent_3_generate_report`)**: Acts as the presenter. It receives the outputs from both upstream agents, formats the data visually using dividers and indentation, and prints the final summary to the console.
4. **`run_task_delegation_system()`**: Serves as the delegation pipeline/manager, ensuring clean separation of concerns and data passing between the agents.

## Question 4

Refactor your multi-agent code so that each agent (function) receives its input only from the previous agent’s output, not from global variables.

Hint: Pass data explicitly between functions to mimic agent-to-agent communication.

**Answer:**

Here is a clear and accurate guide on how to refactor your code to eliminate global variables and establish explicit agent-to-agent communication.

---

### The Problem: Relying on Global Variables
In poorly structured multi-agent systems, agents often read from and write to shared, global variables. This creates hidden dependencies, makes the code hard to debug, and breaks the concept of isolated, autonomous agents.

**Example of the Anti-Pattern (Before Refactoring):**
```python
# GLOBAL VARIABLES (Avoid this)
raw_data = ""
cleaned_data = ""
analyzed_data = ""


def data_collector_agent():
    global raw_data
    raw_data = "  Agent Output: System Normal  "


def data_cleaner_agent():
    global cleaned_data
    # Accessing global variable directly
    cleaned_data = raw_data.strip().lower()


def summarizer_agent():
    global analyzed_data
    # Accessing global variable directly
    analyzed_data = f"Summary: {cleaned_data}"


# Execution
data_collector_agent()
data_cleaner_agent()
summarizer_agent()
print(analyzed_data)
```

---

### The Solution: Explicit Data Passing (Agent-to-Agent Communication)
To refactor this, each agent (function) should:
1. **Accept its required input as a parameter.**
2. **Perform its specific task.**
3. **Return the result explicitly** so it can be passed to the next agent in the pipeline.

**Refactored Code (After Refactoring):**

```python
# Agent 1: Collects or initiates data
def data_collector_agent() -> str:
    # Generates initial data without depending on outside state
    return "  Agent Output: System Normal  "


# Agent 2: Receives data strictly from Agent 1, processes it, and returns it
def data_cleaner_agent(input_text: str) -> str:
    return input_text.strip().lower()


# Agent 3: Receives data strictly from Agent 2, generates the final output
def summarizer_agent(cleaned_text: str) -> str:
    return f"Summary: {cleaned_text}"


# Pipeline / Orchestrator: Chains agents explicitly
def main():
    # Agent 1 passes output to step 1
    step1_output = data_collector_agent()

    # Agent 1's output is passed directly as Agent 2's input
    step2_output = data_cleaner_agent(step1_output)

    # Agent 2's output is passed directly as Agent 3's input
    final_output = summarizer_agent(step2_output)

    print(final_output)


if __name__ == "__main__":
    main()
```

---

### Key Takeaways for Your Assignment:
1. **Purity & Isolation:** Functions are now pure-like units. They only know about the data provided through their arguments (`input_text`, `cleaned_text`).
2. **Sequential Flow (Message Passing):** Instead of sharing memory, Agent A *sends a message* (returns data) to Agent B by passing it as an argument:  
   $$\text{Agent A}() \rightarrow \text{Agent B}(\text{output}_A) \rightarrow \text{Agent C}(\text{output}_B)$$
3. **Alternative Chaining Syntax:** You can also chain them directly in a single readable line:
   ```python
   final_result = summarizer_agent(
       data_cleaner_agent(data_collector_agent())
   )
```

## Question 5

Use ChatGPT or any LLM to help you design a new multi-agent workflow for a Zomato-style restaurant review aggregator: describe the roles of at least three agents and outline the data flow between them. Submit both your prompt and the AI’s response.

**Answer:**

Here is a complete, assignment-ready submission featuring the prompt used and the resulting LLM response.

---

### Part 1: The Prompt

> **Prompt entered into the LLM:**
> 
> "Act as a Senior AI Architect. Design a production-ready, multi-agent AI workflow for a food delivery and restaurant aggregator platform like Zomato. 
> 
> The system's objective is to ingest unstructured customer reviews, detect spam/fake content, extract actionable food/service insights, and generate automated summaries and merchant responses.
> 
> Please provide:
> 1. Clear names and defined roles for at least three specialized agents (including their inputs, core responsibilities, and outputs).
> 2. A step-by-step description of the data flow between them (including a text-based architecture diagram).
> 3. An explanation of how edge cases or anomalies are handled in this pipeline."

---

### Part 2: The LLM Response

Here is the architectural design generated by the model:

#### 1. System Overview
The proposed system, named **ReviewPulse AI**, is a multi-agent framework designed to automate review moderation, attribute-level sentiment extraction, and automated merchant engagement.

---

#### 2. Agent Definitions

##### **Agent 1: The Gatekeeper (Integrity & Spam Triage Agent)**
* **Role:** Acts as the primary firewall to filter out malicious, unhelpful, or fraudulent submissions before downstream processing.
* **Inputs:** Raw review text, user metadata (account age, order history, device fingerprint), and order details (items ordered, delivery timestamp).
* **Core Responsibilities:**
  * Cross-references review timestamp with delivery confirmation to verify real interaction.
  * Employs toxicity classification to catch hate speech, profanity, and harassment.
  * Uses adversarial pattern matching to flag promotional spam, competitor sabotage, and repetitive bot-generated reviews.
* **Outputs:** 
  * A boolean `is_verified` flag.
  * A `moderation_status` (`APPROVED`, `FLAGGED_SPAM`, `FLAGGED_TOXIC`).
  * Clean, normalized text passed downstream (if approved).

##### **Agent 2: The Critic (Aspect-Based Sentiment & Feature Extraction Agent)**
* **Role:** Deconstructs approved, unstructured review text into structured operational intelligence.
* **Inputs:** Verified review text, restaurant catalog (menu, category tags).
* **Core Responsibilities:**
  * Performs **Aspect-Based Sentiment Analysis (ABSA)** across four core dimensions: *Food Quality/Taste*, *Portion Size*, *Packaging*, and *Delivery Speed/Rider Professionalism*.
  * Extracts specific dish entities mentioned (e.g., mapping "the butter chicken was too sweet" directly to Item ID `#BC-101`).
  * Normalizes sentiment into a standardized rating schema (-1.0 to +1.0) per aspect.
* **Outputs:** A structured JSON object containing:
  * Dish-level sentiment mappings.
  * Overall operational aspect ratings.
  * Extracted dietary/allergy flags (e.g., "undercooked chicken").

##### **Agent 3: The Diplomat (Public Summarization & Merchant Engagement Agent)**
* **Role:** Converts structured review metrics into customer-facing insights and drafts contextual merchant responses.
* **Inputs:** Structured JSON from Agent 2, restaurant owner brand voice settings (e.g., "Formal", "Trendy/Casual"), customer tier status (e.g., Zomato Gold).
* **Core Responsibilities:**
  * Updates the rolling "AI Dish Summary" (e.g., *"85% of reviewers recommend the Butter Chicken; note that gravy tends to be sweet"*).
  * Automatically generates personalized, context-aware draft responses for the restaurant manager.
  * If sentiment is critically negative, formats an escalation ticket with recommended compensatory actions (e.g., partial refund, discount code).
* **Outputs:**
  * Updated public-facing dish/restaurant summary blurb.
  * Draft response ready for one-click merchant approval or auto-reply.

---

#### 3. Data Flow Architecture

##### **Step-by-Step Flow:**
1. **Ingestion:** A customer submits a rating and review on the order confirmation screen.
2. **Phase 1 (Verification):** **The Gatekeeper** validates the review. 
   * If flagged for severe abuse or high probability of bot activity, the review is routed to an async human audit queue. 
   * If valid, it is sanitized and forwarded.
3. **Phase 2 (Information Extraction):** **The Critic** consumes the sanitized review. It parses named entities against the restaurant’s active menu database, scores attributes, and writes structured records to an analytics data store.
4. **Phase 3 (Synthesis & Action):** **The Diplomat** ingests the Critic’s structured output. It immediately updates dynamic aggregate tags on the restaurant page and generates an appropriate response draft in the Merchant Dashboard.

##### **Text-Based Architecture Diagram:**

```text
[User Submits Review] 
          │
          ▼
┌──────────────────────────────────────────────┐
│        Agent 1: The Gatekeeper               │
│  (Validates Order, Flags Spam/Toxicity)      │
└──────┬───────────────────────────────────────┘
       │
       ├──[Failed / Spam] ───► [Discard or Human Escalation Queue]
       │
       ▼ [Approved]
┌──────────────────────────────────────────────┐
│        Agent 2: The Critic                   │
│  (ABSA: Extracts Dishes, Taste, Portions)    │
└──────┬───────────────────────────────────────┘
       │
       ▼ [Structured JSON]
┌──────────────────────────────────────────────┐
│        Agent 3: The Diplomat                 │
│  (Generates Responses & Dynamic Summaries)   │
└──────┬───────────────────────────────────────┘
       │
       ├──► Public Storefront: Updated Aggregate Insights ("Crispy Crust: 94%")
       └──► Merchant Portal: Auto-Drafted Apology/Gratitude Response
```
