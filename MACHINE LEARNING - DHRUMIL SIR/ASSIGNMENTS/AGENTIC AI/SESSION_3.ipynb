## Question 1

Simulate the 'Perceive → Think → Act → Learn' loop by writing a simple Python script that takes a user's food preference as input (Perceive), decides on a restaurant type (Think), prints a suggested restaurant (Act), and then asks for feedback to improve the next suggestion (Learn).

**Answer:**

```python
# Knowledge base for suggestions
recommendations = {
    "italian": "Pasta Palace",
    "mexican": "Taco Town",
    "japanese": "Sushi Central",
}

# 1. Perceive: Gather input from the user
preference = (
    input("What kind of food do you like? (italian/mexican/japanese): ")
    .strip()
    .lower()
)

# 2. Think: Decide based on the preference and knowledge base
suggested_restaurant = recommendations.get(preference, "The Daily Diner")

# 3. Act: Output the recommendation
print(f"You should try: {suggested_restaurant}")

# 4. Learn: Receive feedback and update knowledge
feedback = input("Did you like this suggestion? (yes/no): ").strip().lower()
if feedback == "no":
    new_restaurant = input(
        f"What is a better restaurant for {preference}? "
    ).strip()
    recommendations[preference] = new_restaurant
    print(f"Learned! Next time I will suggest {new_restaurant}.")
else:
    print("Great! Keeping that recommendation.")
```

## Question 2

Implement a Short-Term Memory for a chatbot that remembers the last 3 user messages and can repeat them back when asked.

Hint: Use a list and append each new message, removing the oldest if there are more than 3.

**Answer:**

```python
class ShortTermMemory:

    def __init__(self):
        self.memory = []

    def add_message(self, message):
        self.memory.append(message)
        if len(self.memory) > 3:
            self.memory.pop(0)

    def get_messages(self):
        return self.memory


# Example usage
bot_memory = ShortTermMemory()


def chat(user_input):
    if user_input.lower() == "repeat":
        return bot_memory.get_messages()
    else:
        bot_memory.add_message(user_input)
        return "Message remembered."
```

## Question 3

Create a Long-Term Memory simulation for a playlist recommender: store every song a user has played in a text file, and write code to load and display the full play history each time the script runs.

**Answer:**

```python
FILE_NAME = "play_history.txt"


def log_song(song_name):
    with open(FILE_NAME, "a") as file:
        file.write(song_name + "\n")


def display_history():
    try:
        with open(FILE_NAME, "r") as file:
            history = file.read().splitlines()
            print("Full Play History:")
            for idx, song in enumerate(history, 1):
                print(f"{idx}. {song}")
    except FileNotFoundError:
        print("No play history found.")


# Example usage:
log_song("Bohemian Rhapsody - Queen")
log_song("Shape of You - Ed Sheeran")

display_history()
```

## Question 4

Design a Contextual Memory feature for a movie recommendation agent: when a user asks for 'comedy movies like the one I watched last week', your code should retrieve the genre and title of their last watched movie and suggest similar ones.

**Answer:**

```python
from datetime import datetime, timedelta

# Mock database
user_watch_history = [
    {
        "title": "Superbad",
        "genre": "Comedy",
        "watched_at": datetime.now() - timedelta(days=7),
    },
    {
        "title": "Inception",
        "genre": "Sci-Fi",
        "watched_at": datetime.now() - timedelta(days=30),
    },
]

movie_catalog = [
    {"title": "Step Brothers", "genre": "Comedy"},
    {"title": "The Hangover", "genre": "Comedy"},
    {"title": "Interstellar", "genre": "Sci-Fi"},
]


class ContextualMemoryAgent:

    def __init__(self, history, catalog):
        self.history = history
        self.catalog = catalog

    def get_last_watched(self, days_ago=7):
        """Retrieves title and genre of the movie watched within the specified timeframe."""
        target_date = datetime.now() - timedelta(days=days_ago)
        # Find the movie closest to the target timeframe
        last_movie = min(
            self.history,
            key=lambda x: abs((x["watched_at"] - target_date).total_seconds()),
        )
        return last_movie["title"], last_movie["genre"]

    def recommend_similar(self, query):
        if "last week" in query:
            title, genre = self.get_last_watched(days_ago=7)

            # Recommend movies with the same genre, excluding the watched one
            recommendations = [
                m["title"]
                for m in self.catalog
                if m["genre"].lower() == genre.lower() and m["title"] != title
            ]

            return {
                "context_used": {"last_watched": title, "genre": genre},
                "recommendations": recommendations,
            }
        return "Query not understood."


# Execution
agent = ContextualMemoryAgent(user_watch_history, movie_catalog)
result = agent.recommend_similar(
    "comedy movies like the one I watched last week"
)
print(result)
```
