## Question 2

Implement a simple state machine in Python that models a multi-step IPL ticket booking process: states should include 'Start', 'Select Match', 'Choose Seats', 'Payment', and 'Confirmation'. Print the current state after each transition.

Hint: Use a dictionary to map transitions and a loop to simulate user actions.

**Answer:**

```python
transitions = {
    "Start": {"next": "Select Match"},
    "Select Match": {"next": "Choose Seats"},
    "Choose Seats": {"next": "Payment"},
    "Payment": {"next": "Confirmation"},
}

current_state = "Start"
print(f"Current state: {current_state}")

actions = ["next", "next", "next", "next"]

for action in actions:
    if current_state in transitions and action in transitions[current_state]:
        current_state = transitions[current_state][action]
        print(f"Current state: {current_state}")
    else:
        break
```

## Question 3

Write a Python function that simulates an agent performing a multi-hop reasoning task: given a movie name, first fetch its release year from a dictionary, then use that year to look up the top trending song of that year from another dictionary, and print both results.

**Answer:**

```python
def movie_to_song_agent(movie_name):
    # Knowledge bases
    movie_years = {
        "Inception": 2010,
        "Titanic": 1997,
        "Interstellar": 2014,
    }
    trending_songs = {
        2010: "Tik Tok by Kesha",
        1997: "Candle in the Wind by Elton John",
        2014: "Happy by Pharrell Williams",
    }

    # Hop 1: Get release year
    release_year = movie_years.get(movie_name)

    # Hop 2: Get trending song using the retrieved year
    song = trending_songs.get(release_year)

    # Print results
    print(f"Movie: {movie_name} | Release Year: {release_year}")
    print(f"Trending Song ({release_year}): {song}")
```

## Question 4

Modify the following agent loop code so it safely stops execution after 5 steps to prevent infinite loops:

while not done:
  action = agent.next_step()
  done = action['done']

Constraint: Add a step counter and break the loop if it exceeds 5 iterations.

**Answer:**

```python
step_count = 0
while not done:
  if step_count >= 5:
    break
  action = agent.next_step()
  done = action['done']
  step_count += 1
```

## Question 5

Use ChatGPT to generate Python code for an agent that handles multiple steps in ordering food on Swiggy (e.g., select restaurant, choose dishes, confirm order). Paste the generated code and describe one safety check you would add to prevent incorrect tool calls.

**Answer:**

### Python Code

```python
import json


# Mock Swiggy Tools
def select_restaurant(name: str):
    return f"Restaurant selected: {name}"


def choose_dishes(dishes: list):
    return f"Dishes added: {', '.join(dishes)}"


def confirm_order(payment_method: str):
    return f"Order placed successfully using {payment_method}."


# Tool Mapping
tools = {
    "select_restaurant": select_restaurant,
    "choose_dishes": choose_dishes,
    "confirm_order": confirm_order,
}


def run_swiggy_agent(plan: list):
    """Executes a multi-step food ordering plan."""
    for step in plan:
        tool_name = step.get("tool")
        arguments = step.get("args", {})

        if tool_name in tools:
            result = tools[tool_name](**arguments)
            print(result)
        else:
            print(f"Error: Unknown tool {tool_name}")


# Example Execution
order_plan = [
    {"tool": "select_restaurant", "args": {"name": "A2B Veg"}},
    {"tool": "choose_dishes", "args": {"dishes": ["Masala Dosa", "Filter Coffee"]}},
    {"tool": "confirm_order", "args": {"payment_method": "UPI"}},
]

run_swiggy_agent(order_plan)
```
