## Question 1

Write a Python function that takes a user's search query (like a movie name) and calls the OMDb API (http://www.omdbapi.com/) to fetch and display the movie's title, year, and IMDb rating. Show the result in the console.

Hint: Use the requests library to make the API call and handle the JSON response.

**Answer:**

```python
import requests


def get_movie_details(movie_title, api_key):
    url = f"http://www.omdbapi.com/?t={movie_title}&apikey={api_key}"
    response = requests.get(url).json()

    if response.get("Response") == "True":
        print(f"Title: {response.get('Title')}")
        print(f"Year: {response.get('Year')}")
        print(f"IMDb Rating: {response.get('imdbRating')}")
    else:
        print("Movie not found.")
```

## Question 2

Simulate a Zomato-style menu search: Create a script that reads a local CSV file named 'menu.csv' containing columns 'dish', 'cuisine', and 'price'. Ask the user to enter a cuisine type (e.g., 'Italian') and display all matching dishes with their prices.

**Answer:**

```python
import csv

cuisine_input = input("Enter cuisine type: ").strip().lower()

found = False
with open("menu.csv", mode="r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        if row["cuisine"].strip().lower() == cuisine_input:
            print(f"{row['dish']} - ${row['price']}")
            found = True

if not found:
    print("No dishes found for this cuisine.")
```

## Question 3

Build a simple agent workflow that takes user input for a song name, checks if the song exists in a local SQLite database called 'music.db' (table: songs), and prints 'Song found' or 'Song not found'.

Hint: Use Python's sqlite3 module to interact with the database.

**Answer:**

```python
import sqlite3


def check_song():
    song_name = input("Enter song name: ")

    conn = sqlite3.connect("music.db")
    cursor = conn.cursor()

    query = "SELECT 1 FROM songs WHERE title = ? LIMIT 1"
    cursor.execute(query, (song_name,))
    result = cursor.fetchone()

    conn.close()

    if result:
        print("Song found")
    else:
        print("Song not found")


if __name__ == "__main__":
    check_song()
```

## Question 4

Implement prompt chaining: Write a script where the agent first asks the user for a movie genre, then uses that genre to call the OMDb API and fetch 3 movies, and finally asks the user to pick one for detailed info. Display the full details of the selected movie.

**Answer:**

```python
import json
import requests

API_KEY = "YOUR_OMDB_API_KEY"  # Replace with your OMDb API key
BASE_URL = "http://www.omdbapi.com/"

# Step 1: Prompt the user for a genre/keyword
genre = input("Enter a movie genre: ").strip()

# Step 2: Fetch 3 movies matching the genre via OMDb API
search_params = {"apikey": API_KEY, "s": genre, "type": "movie"}
search_response = requests.get(BASE_URL, params=search_params).json()

movies = search_response.get("Search", [])[:3]

if not movies:
    print("No movies found.")
    exit()

print(f"\nTop 3 movies for '{genre}':")
for i, movie in enumerate(movies, 1):
    print(f"{i}. {movie['Title']} ({movie['Year']})")

# Step 3: Prompt the user to pick one movie
choice = int(input("\nSelect a movie number for detailed info (1-3): ")) - 1
selected_imdb_id = movies[choice]["imdbID"]

# Step 4: Fetch and display full details of the selected movie
detail_params = {"apikey": API_KEY, "i": selected_imdb_id, "plot": "full"}
detail_response = requests.get(BASE_URL, params=detail_params).json()

print("\n--- Movie Details ---")
print(json.dumps(detail_response, indent=2))
```
