## Question 1

Create a Python function called call_weather_api() that takes a city name as input, calls a mock weather API (you can just return a fake temperature), and returns the result as a dictionary. Simulate how an LLM agent could use this function as a tool.

**Answer:**

### Python Code

```python
import json


# 1. Define the mock tool
def call_weather_api(city_name):
    # Fake database of temperatures
    mock_database = {"tokyo": "15°C", "new york": "22°C", "london": "18°C"}

    # Get temperature or return a default value
    temp = mock_database.get(city_name.lower(), "20°C")

    return {"city": city_name, "temperature": temp, "condition": "Sunny"}


# 2. Simulate an LLM Agent using the tool
user_prompt = "What is the weather like in Tokyo?"
print(f"User: {user_prompt}")

# Step A: The LLM "decides" it needs to call a tool and extracts the argument
tool_call = {"tool_name": "call_weather_api", "arguments": {"city_name": "Tokyo"}}
print(f"LLM Agent Action: Calling {tool_call['tool_name']}...")

# Step B: The system executes the function
tool_result = call_weather_api(**tool_call["arguments"])
print(f"Tool Output: {tool_result}")

# Step C: The LLM uses the tool result to form the final answer
final_response = f"The weather in {tool_result['city']} is currently {tool_result['temperature']} and {tool_result['condition']}."
print(f"Agent Final Answer: {final_response}")
```

### Explanation

The `call_weather_api()` function takes a city name, looks up a fake temperature, and returns the data inside a Python dictionary. 

To simulate an LLM agent, the code shows the 3-step tool-use process: 
1. The model detects the user wants weather data and chooses the right tool.
2. The code runs the function using the city extracted from the prompt.
3. The model takes the dictionary result and writes a natural sentence for the user.

**Answer:**

### Generated Python Code

```python
import fitz  # PyMuPDF
import openai

# 1. Extract text from the PDF
def extract_pdf_text(pdf_path):
    text = ""
    with fitz.open(pdf_path) as doc:
        for page in doc:
            text += page.get_text()
    return text

# 2. Ask question to the LLM agent using the extracted text
def ask_ipl_agent(pdf_path, question):
    context = extract_pdf_text(pdf_path)
    client = openai.OpenAI(api_key="YOUR_API_KEY")

    response = client.chat.completions.create(
        model="gpt-4o",
        messages=[
            {"role": "system", "content": "You are a cricket stats assistant. Answer questions based only on the provided context."},
            {"role": "user", "content": f"Context:\n{context}\n\nQuestion: {question}"}
        ]
    )
    return response.choices[0].message.content

# Usage
# answer = ask_ipl_agent("ipl_stats.pdf", "Who scored the most runs?")
# print(answer)
```

### One Change to Improve Accuracy

**Use a table extraction tool instead of plain text parsing:**  
IPL statistics are stored in tables. Standard text extractors like `fitz.get_text()` scramble table rows and columns into unformatted text, causing the LLM to misread numbers. Replacing it with **`pdfplumber`** or **Camelot** to extract and convert tables directly into Markdown or Pandas DataFrames preserves the column-row relationships, drastically improving data retrieval accuracy.

## Question 5

Extend an existing agent with RAG (Retrieval-Augmented Generation) by adding a function that extracts key information from a BookMyShow-style PDF ticket (movie name, seat number, showtime) and returns it as a dictionary.

Hint: You may use PyPDF2 or a similar library to parse the PDF.

**Answer:**

```python
import re
from pypdf import PdfReader


def extract_ticket_info(pdf_path: str) -> dict:
    reader = PdfReader(pdf_path)
    text = ""
    for page in reader.pages:
        text += page.extract_text() or ""

    movie = re.search(r"Movie:\s*(.*)", text)
    seats = re.search(r"Seats?:\s*(.*)", text)
    showtime = re.search(r"Time:\s*(.*)", text)

    return {
        "movie_name": movie.group(1).strip() if movie else None,
        "seat_number": seats.group(1).strip() if seats else None,
        "showtime": showtime.group(1).strip() if showtime else None,
    }
```
