## Question 3

Write a Python function fetch_song_info(song_name) that uses the Spotify Web API to retrieve and print the artist and album name for a given song.

Hint: Use the requests library and Spotify's search endpoint. You can use a demo or test token.

**Answer:**

```python
import requests


def fetch_song_info(song_name):
    # Replace with a valid Spotify access token
    access_token = "YOUR_SPOTIFY_ACCESS_TOKEN"

    url = "https://api.spotify.com/v1/search"
    headers = {"Authorization": f"Bearer {access_token}"}
    params = {"q": song_name, "type": "track", "limit": 1}

    response = requests.get(url, headers=headers, params=params)
    data = response.json()

    tracks = data.get("tracks", {}).get("items", [])
    if tracks:
        track = tracks[0]
        artist_name = track["artists"][0]["name"]
        album_name = track["album"]["name"]
        print(f"Artist: {artist_name}")
        print(f"Album: {album_name}")
    else:
        print("Song not found.")
```
