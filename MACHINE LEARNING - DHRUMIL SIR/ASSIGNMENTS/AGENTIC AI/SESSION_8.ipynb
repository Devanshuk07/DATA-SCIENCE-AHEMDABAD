## Question 2

Use the requests library to fetch current weather data for a given city from the OpenWeatherMap API (https://api.openweathermap.org/data/2.5/weather) and print the temperature in Celsius.

Hint: Sign up for a free API key and use the 'main.temp' field from the JSON response.

**Answer:**

```python
import requests

API_KEY = "YOUR_API_KEY"
city = "London"
url = f"https://api.openweathermap.org/data/2.5/weather?q={city}&appid={API_KEY}&units=metric"

response = requests.get(url)
data = response.json()

temperature = data["main"]["temp"]
print(f"Temperature: {temperature}°C")
```

## Question 3

Write a function weather_advisory(temp_celsius) that takes the temperature in Celsius and returns an advisory message: 'Carry an umbrella' if below 20°C, 'Stay hydrated' if above 30°C, or 'Weather is pleasant' otherwise.

**Answer:**

```python
def weather_advisory(temp_celsius):
    if temp_celsius < 20:
        return "Carry an umbrella"
    elif temp_celsius > 30:
        return "Stay hydrated"
    else:
        return "Weather is pleasant"
```

## Question 4

Combine your get_weather and weather_advisory functions to build a simple command-line Weather Assistant: input a city name, fetch and display the temperature, and show the advisory message.

**Answer:**

```python
def weather_assistant():
    city = input("Enter city name: ")
    temp = get_weather(city)
    advisory = weather_advisory(temp)

    print(f"Temperature: {temp}°C")
    print(f"Advisory: {advisory}")


if __name__ == "__main__":
    weather_assistant()
```

## Question 5

Refactor your Weather Assistant code to handle invalid city names gracefully by displaying 'City not found, please try again' if the API returns an error.

Hint: Check the 'cod' field in the API response for error handling.

**Answer:**

```python
import requests


def get_weather(city_name, api_key):
    url = f"http://api.openweathermap.org/data/2.5/weather?q={city_name}&appid={api_key}&units=metric"
    response = requests.get(url).json()

    # Check the 'cod' field for errors (OpenWeatherMap returns 200 for success)
    if response.get("cod") != 200:
        return "City not found, please try again"

    weather = response["weather"][0]["description"]
    temp = response["main"]["temp"]
    return f"Weather in {city_name}: {weather}, {temp}°C"
```
