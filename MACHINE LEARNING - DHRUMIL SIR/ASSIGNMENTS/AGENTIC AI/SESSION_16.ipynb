## Question 2

Create a new endpoint /order-status on your MCP server that accepts a POST request with a JSON payload containing an orderId and returns a simulated order status like 'Delivered', 'In Transit', or 'Cancelled'.

**Answer:**

```python
import random
from mcp.server.fastmcp import FastMCP

mcp = FastMCP("Order Service")


@mcp.tool()
def order_status(orderId: str) -> dict:
    """Returns the simulated status of a given order ID."""
    statuses = ["Delivered", "In Transit", "Cancelled"]
    return {"orderId": orderId, "status": random.choice(statuses)}
```

## Question 3

Add error handling to your MCP server so that if a required field (like orderId) is missing in the /order-status request, it returns a JSON error message with status code 400.

Hint: Use Flask's request.get_json() and check for missing keys before processing.

**Answer:**

```python
from flask import Flask, jsonify, request

app = Flask(__name__)


@app.route("/order-status", methods=["POST"])
def order_status():
    data = request.get_json()

    if not data or "orderId" not in data:
        return jsonify({"error": "Missing required field: orderId"}), 400

    order_id = data["orderId"]
    return jsonify({"orderId": order_id, "status": "Processing"}), 200
```

## Question 4

Simulate a Flipkart-style notification system by adding an endpoint /notify that accepts a JSON payload with a userId and message, then logs the notification to a file named notifications.log.

**Answer:**

```python
from flask import Flask, jsonify, request

app = Flask(__name__)


@app.route("/notify", methods=["POST"])
def notify():
    data = request.get_json()

    user_id = data.get("userId")
    message = data.get("message")

    with open("notifications.log", "a") as file:
        file.write(f"User: {user_id} | Notification: {message}\n")

    return (
        jsonify({"status": "success", "message": "Notification logged"}),
        200,
    )


if __name__ == "__main__":
    app.run(port=5000)
```

## Question 5

Use ChatGPT or Copilot to generate code for adding a new endpoint /user-profile that returns a mock user profile (name, email, phone) when given a userId, then test and integrate the generated code into your MCP server.

**Answer:**

```python
from mcp.server.fastmcp import FastMCP

# Initialize MCP server
mcp = FastMCP("UserProfileServer")

# Mock database
USER_DATA = {
    "1": {
        "name": "Jane Doe",
        "email": "jane@example.com",
        "phone": "+1-555-0199",
    },
    "2": {
        "name": "John Smith",
        "email": "john@example.com",
        "phone": "+1-555-0144",
    },
}


@mcp.tool()
def get_user_profile(user_id: str) -> dict:
    """Returns a mock user profile given a userId."""
    profile = USER_DATA.get(user_id)
    if not profile:
        return {"error": "User not found"}
    return profile


if __name__ == "__main__":
    mcp.run()
```
