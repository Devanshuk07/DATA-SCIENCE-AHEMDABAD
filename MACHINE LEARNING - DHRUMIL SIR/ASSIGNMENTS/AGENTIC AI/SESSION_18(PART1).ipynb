## Question 1

Create a Python class called FoodOrderAgent that represents an agent which can place and track orders for a food delivery app like Swiggy. Include methods to place_order(), check_status(), and cancel_order().

**Answer:**

```python
class FoodOrderAgent:

    def __init__(self):
        self.orders = {}
        self.order_counter = 1

    def place_order(self, restaurant, items):
        order_id = self.order_counter
        self.orders[order_id] = {
            "restaurant": restaurant,
            "items": items,
            "status": "Placed",
        }
        self.order_counter += 1
        return f"Order #{order_id} placed successfully."

    def check_status(self, order_id):
        if order_id in self.orders:
            return f"Order #{order_id} status: {self.orders[order_id]['status']}"
        return f"Order #{order_id} not found."

    def cancel_order(self, order_id):
        if order_id in self.orders:
            if self.orders[order_id]["status"] == "Cancelled":
                return f"Order #{order_id} is already cancelled."
            self.orders[order_id]["status"] = "Cancelled"
            return f"Order #{order_id} has been cancelled."
        return f"Order #{order_id} not found."
```

## Question 3

Implement an agent communication scenario where two agents, PlaylistAgent and SongAgent, interact: PlaylistAgent requests a song of a specific mood (e.g., 'happy'), and SongAgent responds with a song name. Use simple Python classes and method calls to simulate this exchange.

Hint: Focus on message passing between objects to mimic agent communication.

**Answer:**

```python
class SongAgent:

    def __init__(self):
        self.songs = {
            "happy": "Walking on Sunshine",
            "sad": "Someone Like You",
            "energetic": "Eye of the Tiger",
        }

    def receive_request(self, mood):
        # Process request and return response
        song = self.songs.get(mood, "Unknown Mood Song")
        return song


class PlaylistAgent:

    def request_song(self, song_agent, mood):
        # Send message to SongAgent and receive response
        print(f"PlaylistAgent: Requesting a '{mood}' song.")
        response = song_agent.receive_request(mood)
        print(f"SongAgent: Recommended '{response}'.")


# Simulate communication
song_agent = SongAgent()
playlist_agent = PlaylistAgent()

playlist_agent.request_song(song_agent, "happy")
```

## Question 4

Refactor the following code to make the agent's decision logic more modular: an agent currently has a single decide_action() method with all logic inside. Extract the logic for checking user preferences and for checking item availability into separate helper methods, then call them from decide_action().

**Answer:**

```python
class Agent:

    def check_user_preferences(self, user, item):
        return item in user.get("preferences", [])

    def check_item_availability(self, item, inventory):
        return inventory.get(item, 0) > 0

    def decide_action(self, user, item, inventory):
        prefers_item = self.check_user_preferences(user, item)
        item_available = self.check_item_availability(item, inventory)

        if prefers_item and item_available:
            return "recommend"
        return "do_not_recommend"
```

## Question 5

Use ChatGPT or Copilot to generate Python code for an agent that manages a shopping cart (like Flipkart), with methods to add_item(), remove_item(), and get_total(). Paste the generated code, then test each method with sample data and note any corrections you had to make.

**Answer:**

### Generated Code

```python
class ShoppingCartAgent:

    def __init__(self):
        # Stores items as {item_name: {"price": float, "quantity": int}}
        self.cart = {}

    def add_item(self, item_name, price, quantity=1):
        if item_name in self.cart:
            self.cart[item_name]["quantity"] += quantity
        else:
            self.cart[item_name] = {"price": price, "quantity": quantity}
        print(f"Added {quantity} x {item_name} to cart.")

    def remove_item(self, item_name, quantity=1):
        if item_name not in self.cart:
            print(f"{item_name} is not in the cart.")
            return

        if self.cart[item_name]["quantity"] <= quantity:
            del self.cart[item_name]
            print(f"Removed all {item_name} from cart.")
        else:
            self.cart[item_name]["quantity"] -= quantity
            print(f"Removed {quantity} x {item_name}. Remaining: {self.cart[item_name]['quantity']}")

    def get_total(self):
        total = sum(
            details["price"] * details["quantity"]
            for details in self.cart.values()
        )
        return total
```

---

### Testing with Sample Data

```python
# Initialize agent
agent = ShoppingCartAgent()

# 1. Test add_item()
agent.add_item("Wireless Mouse", 500.0, 2)
agent.add_item("Keyboard", 1200.0, 1)

# 2. Test get_total()
print("Total Cost:", agent.get_total())  # Expected: (500*2) + (1200*1) = 2200.0

# 3. Test remove_item()
agent.remove_item("Wireless Mouse", 1)  # Partial removal
agent.remove_item("Keyboard", 2)  # Full removal (quantity >= existing)
agent.remove_item("Headphones", 1)  # Non-existent item removal

# 4. Final total check
print("Updated Total:", agent.get_total())  # Expected: 500.0
```

### Corrections Made

1. **Handling Existing Items on Addition:** The initial generated code simply overwrote the item price and quantity instead of incrementing `quantity` when adding an item already present in the cart. Added an `if item_name in self.cart` condition to increment count.
2. **Partial Quantity Removal:** The original code deleted the item entirely regardless of the requested removal count. Updated `remove_item()` to subtract quantity first and only `del` the item when the requested removal count meets or exceeds the current stock.
3. **Missing Item Exception Handling:** Added a guard check in `remove_item()` to prevent a `KeyError` if attempting to delete an item not present in the cart.
