## Question 1

Create two simple agent classes in Python, SearchAgent and RecommendationAgent, where SearchAgent takes a query and returns a list of matching products (from a hardcoded list), and RecommendationAgent takes a user ID and returns three recommended products. Simulate a Flipkart-style shopping assistant where both agents work independently.

**Answer:**

```python
class SearchAgent:

    def __init__(self):
        self.catalog = [
            "Apple iPhone 15",
            "Samsung Galaxy S23",
            "Sony WH-1000XM5 Headphones",
            "Apple MacBook Air",
            "Nike Running Shoes",
        ]

    def search(self, query):
        query = query.lower()
        return [item for item in self.catalog if query in item.lower()]


class RecommendationAgent:

    def __init__(self):
        self.user_recommendations = {
            "user_1": [
                "Apple iPhone 15",
                "Apple MacBook Air",
                "AirPods Pro",
            ],
            "user_2": [
                "Nike Running Shoes",
                "Adidas Trackpants",
                "Puma Backpack",
            ],
        }
        self.default_recommendations = [
            "Smartwatch",
            "Wireless Mouse",
            "Power Bank",
        ]

    def recommend(self, user_id):
        return self.user_recommendations.get(
            user_id, self.default_recommendations
        )[:3]


# Simulation
search_agent = SearchAgent()
rec_agent = RecommendationAgent()

user_id = "user_1"
search_query = "Apple"

print(f"Recommendations for {user_id}:", rec_agent.recommend(user_id))
print(f"Search results for '{search_query}':", search_agent.search(search_query))
```

## Question 2

Build a MultiAgentManager class that coordinates communication between a ChatAgent (handles user queries) and a PlaylistAgent (creates playlists) to simulate a Spotify-like assistant. When the user asks for a playlist, the ChatAgent should pass the request to PlaylistAgent and return the playlist result.

**Answer:**

```python
class PlaylistAgent:

    def create_playlist(self, query):
        # Generates a sample playlist based on the request
        return [
            f"{query.title()} Track 1",
            f"{query.title()} Track 2",
            f"{query.title()} Track 3",
        ]


class ChatAgent:

    def __init__(self, playlist_agent):
        self.playlist_agent = playlist_agent

    def process_query(self, user_input):
        if "playlist" in user_input.lower():
            # Delegate to PlaylistAgent if a playlist is requested
            genre = (
                user_input.lower()
                .replace("playlist", "")
                .replace("create", "")
                .strip()
            )
            playlist = self.playlist_agent.create_playlist(genre or "Top Hits")
            return f"Here is your playlist: {', '.join(playlist)}"
        return "I can help you create music playlists. What would you like to listen to?"


class MultiAgentManager:

    def __init__(self):
        self.playlist_agent = PlaylistAgent()
        self.chat_agent = ChatAgent(self.playlist_agent)

    def route_request(self, user_input):
        return self.chat_agent.process_query(user_input)
```

## Question 3

Modify your MultiAgentManager to allow agents to share information: for example, if the ChatAgent receives a request for a 'top trending playlist', it should first ask the PlaylistAgent for trending data, then respond to the user. Log all agent-to-agent messages in a list for debugging.

Hint: Use method calls or message passing between agent objects to simulate communication.

**Answer:**

```python
class PlaylistAgent:

    def get_trending(self):
        return ["Song A", "Song B", "Song C"]


class ChatAgent:

    def __init__(self, manager):
        self.manager = manager

    def handle_request(self, query):
        if "top trending playlist" in query:
            # Send message to PlaylistAgent via Manager
            data = self.manager.send_message(
                sender="ChatAgent",
                recipient="PlaylistAgent",
                message="get_trending",
            )
            return f"Here is the top trending playlist: {', '.join(data)}"
        return "I can only help with trending playlists right now."


class MultiAgentManager:

    def __init__(self):
        self.message_log = []
        self.playlist_agent = PlaylistAgent()
        self.chat_agent = ChatAgent(self)

    def send_message(self, sender, recipient, message):
        # Log communication for debugging
        self.message_log.append(
            {"from": sender, "to": recipient, "message": message}
        )

        # Route message to the target agent
        if recipient == "PlaylistAgent" and message == "get_trending":
            return self.playlist_agent.get_trending()
        return None

    def process_user_query(self, query):
        return self.chat_agent.handle_request(query)
```

## Question 4

Use ChatGPT or another LLM-based AI tool to generate Python code for a simple multi-agent system where one agent fetches cricket scores (simulate IPL scores) and another agent summarizes the scores. Test the generated code and note any changes you had to make to get it working.

**Answer:**

### Python Code

```python
import random


class ScoreFetcherAgent:
    """Agent that simulates fetching IPL scores."""

    def fetch_scores(self):
        matches = [
            {"match": "CSK vs MI", "scores": "CSK: 178/4 (20), MI: 180/5 (19.2)"},
            {"match": "RCB vs KKR", "scores": "RCB: 205/2 (20), KKR: 165/9 (20)"},
            {"match": "GT vs RR", "scores": "GT: 192/6 (20), RR: 155/8 (20)"},
        ]
        return random.sample(matches, 2)


class ScoreSummarizerAgent:
    """Agent that summarizes match results."""

    def summarize(self, match_data):
        summary = "IPL Match Summary:\n"
        for game in match_data:
            summary += f"- {game['match']}: {game['scores']}\n"
        return summary


# Running the Multi-Agent System
fetcher = ScoreFetcherAgent()
summarizer = ScoreSummarizerAgent()

raw_scores = fetcher.fetch_scores()
final_summary = summarizer.summarize(raw_scores)

print(final_summary)
```

### Changes Made to Get It Working

1. **Replaced External API Dependencies**: The original LLM output attempted to use a live Cricbuzz API wrapper (`pycricbuzz`) which failed due to deprecated endpoints. Replaced it with a mock data generator using Python's built-in `random` module.
2. **Removed Outdated Framework Imports**: The LLM initially imported non-existent modules from an older version of `crewai`. Replaced the framework logic with standard Python classes to ensure standalone execution without dependency errors.
3. **Fixed Data Passing**: Corrected the pipeline so the `ScoreSummarizerAgent` directly accepts the dictionary output format generated by the `ScoreFetcherAgent`.
