## Question 1

Create a tool schema for a 'Song Search' tool that could be used by an AI agent in a Spotify-style music app — include the tool's name, a short description, and at least two input parameters (such as song name and artist).

**Answer:**

```json
{
  "name": "search_song",
  "description": "Searches for tracks in the music catalog based on track title and artist name.",
  "parameters": {
    "type": "object",
    "properties": {
      "song_name": {
        "type": "string",
        "description": "The title of the song to search for."
      },
      "artist": {
        "type": "string",
        "description": "The name of the artist or band."
      }
    },
    "required": ["song_name"]
  }
}
```

## Question 2

Design a tool schema for a 'BookMyShow Movie Ticket Finder' agent tool — specify the tool name, a description, and three input parameters (such as city, movie name, and date).

Hint: Think about what information a user would provide to search for available movie tickets.

**Answer:**

## Question 3

Write a JSON object that represents the function calling format for a 'GetWeather' tool, following the OpenAI function calling schema — include the function name, description, and an object for parameters with location and date as required fields.

**Answer:**

```json
{
  "name": "GetWeather",
  "description": "Retrieves the weather forecast for a specified location and date.",
  "parameters": {
    "type": "object",
    "properties": {
      "location": {
        "type": "string",
        "description": "The city and state or country, e.g., Paris, France"
      },
      "date": {
        "type": "string",
        "description": "The date for the weather forecast, e.g., YYYY-MM-DD"
      }
    },
    "required": ["location", "date"]
  }
}
```

## Question 4

Given the following tool schema for a 'Food Delivery Tracker' agent, identify and fix any issues in the input parameters section so it matches the Gemini function calling format: 

{ "name": "trackOrder", "description": "Tracks the status of a food delivery order", "parameters": [ { "order_id": "string" }, { "user_phone": "number" } ] }

Hint: Check how parameters should be structured in the Gemini schema.

**Answer:**

**Issues:**
1. The `parameters` field must be an `OBJECT` defining an OpenAPI schema, not a JSON list (array).
2. Parameter types must be specified inside a nested `properties` object.
3. Data types must be uppercase valid schema types (e.g., `STRING`, `INTEGER`/`NUMBER`).
4. Required arguments should be specified in a `required` list.

**Fixed Schema:**

```json
{
  "name": "trackOrder",
  "description": "Tracks the status of a food delivery order",
  "parameters": {
    "type": "OBJECT",
    "properties": {
      "order_id": {
        "type": "STRING",
        "description": "The unique identifier of the order"
      },
      "user_phone": {
        "type": "NUMBER",
        "description": "The phone number of the user"
      }
    },
    "required": ["order_id", "user_phone"]
  }
}
```
