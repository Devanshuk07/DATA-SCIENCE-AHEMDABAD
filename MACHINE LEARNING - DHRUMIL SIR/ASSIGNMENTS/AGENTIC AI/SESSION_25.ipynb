## Question 1

In a new Streamlit app file called assistant_ui.py, design a simple chat-like interface where users can type a request (for example, 'Show me my upcoming movie bookings') and see the agent's response displayed below.

**Answer:**

```python
# assistant_ui.py
import streamlit as st

st.title("AI Assistant")

# Initialize chat history
if "messages" not in st.session_state:
    st.session_state.messages = []

# Display chat messages from history
for message in st.session_state.messages:
    with st.chat_message(message["role"]):
        st.markdown(message["content"])

# React to user input
if prompt := st.chat_input("How can I help you?"):
    # Display and store user message
    st.chat_message("user").markdown(prompt)
    st.session_state.messages.append({"role": "user", "content": prompt})

    # Generate and store agent response (placeholder logic)
    response = f"Agent response to: {prompt}"
    with st.chat_message("assistant"):
        st.markdown(response)
    st.session_state.messages.append({"role": "assistant", "content": response})
```

## Question 2

Write a Python function called detect_intent(user_input) that takes a user's message as input and returns the detected intent as a string (e.g., 'show_bookings', 'play_music', 'get_weather').

Hint: Use basic keyword matching for at least three different intents.

**Answer:**

```python
def detect_intent(user_input):
    user_input = user_input.lower()

    if any(word in user_input for word in ["book", "reservation", "booking"]):
        return "show_bookings"
    elif any(word in user_input for word in ["song", "music", "play"]):
        return "play_music"
    elif any(word in user_input for word in ["weather", "forecast", "rain", "temperature"]):
        return "get_weather"
    else:
        return "unknown_intent"
```

## Question 3

Create a function called select_tool(intent) that decides which tool or API to use based on the detected intent (for example, if intent is 'play_music', return 'SpotifyAPI').

**Answer:**

```python
def select_tool(intent):
    tool_mapping = {
        "play_music": "SpotifyAPI",
        "get_weather": "WeatherAPI",
        "send_email": "GmailAPI",
        "search_web": "GoogleSearchAPI",
    }
    return tool_mapping.get(intent, "DefaultSearchAPI")
```

## Question 4

Implement a memory feature in your agent: save each user request and the agent's response to a list called conversation_history, and display the last three exchanges in the Streamlit UI.

**Answer:**

```python
import streamlit as st

# Initialize conversation history in session state
if "conversation_history" not in st.session_state:
    st.session_state.conversation_history = []

user_input = st.chat_input("Enter your message:")

if user_input:
    # Generate agent response (placeholder logic)
    agent_response = f"Echo: {user_input}"

    # Save request and response to conversation_history
    st.session_state.conversation_history.append(
        {"user": user_input, "agent": agent_response}
    )

# Display the last three exchanges
st.subheader("Last 3 Exchanges")
for exchange in st.session_state.conversation_history[-3:]:
    st.write(f"**User:** {exchange['user']}")
    st.write(f"**Agent:** {exchange['agent']}")
```

## Question 5

Map agent actions to UI feedback: when the agent is 'gathering information' or 'executing a task', show a loading spinner or status message in Streamlit so the user knows something is happening.

Hint: Use st.spinner or st.info components in Streamlit for this.

**Answer:**

```python
import time
import streamlit as st

agent_action = "gathering information"  # or "executing a task"

if agent_action == "gathering information":
    with st.spinner("Gathering information..."):
        time.sleep(2)  # Simulating agent search
    st.info("Information gathered successfully.")

elif agent_action == "executing a task":
    with st.spinner("Executing task..."):
        time.sleep(2)  # Simulating agent execution
    st.success("Task completed.")
```
