## Question 1

Design an agent that recommends trending music playlists to users based on their recent listening history, similar to Spotify's Discover Weekly. Define the agent's perception, reasoning, and action modules in a Python script.

**Answer:**

```python
class MusicRecommendationAgent:

    def __init__(self):
        # Perception Module: Gathers environment data (user history and trending playlists)
        self.user_history = []
        self.trending_playlists = {}

    def perception(self, user_history: list, trending_playlists: dict):
        """Perceives user listening history (genres) and available trending playlists."""
        self.user_history = user_history
        self.trending_playlists = trending_playlists

    def reasoning(self) -> str:
        """Analyzes history to find top genre and selects matching trending playlist."""
        if not self.user_history or not self.trending_playlists:
            return "Default Hits"

        # Find the most frequent genre in user history
        top_genre = max(
            set(self.user_history), key=self.user_history.count, default=None
        )

        # Match with trending playlists, fallback to first available
        return self.trending_playlists.get(
            top_genre, list(self.trending_playlists.values())[0]
        )

    def action(self, recommended_playlist: str):
        """Delivers the recommendation to the user."""
        return f"Recommended Playlist: {recommended_playlist}"


# Example Usage
agent = MusicRecommendationAgent()

# 1. Perception
user_recent_genres = ["Rock", "Pop", "Rock", "Indie"]
current_trending = {
    "Rock": "Trending Rock Weekly",
    "Pop": "Top 50 Pop Hits",
    "Hip-Hop": "Rap Caviar",
}
agent.perception(user_recent_genres, current_trending)

# 2. Reasoning
decision = agent.reasoning()

# 3. Action
result = agent.action(decision)
print(result)
```

## Question 2

Implement a multi-agent system where one agent fetches new movie releases (simulate with a static list), and another agent suggests which movies to watch based on the user's preferred genres. Demonstrate the communication between agents using function calls.

**Answer:**

```python
class MovieFetcherAgent:

    def __init__(self):
        self.movies = [
            {"title": "Dune: Part Two", "genre": "Sci-Fi"},
            {"title": "Kung Fu Panda 4", "genre": "Animation"},
            {"title": "Civil War", "genre": "Action"},
            {"title": "Inside Out 2", "genre": "Animation"},
        ]

    def get_new_releases(self):
        return self.movies


class RecommendationAgent:

    def __init__(self, fetcher_agent):
        self.fetcher = fetcher_agent

    def recommend_movies(self, preferred_genres):
        # Communication via function call to the fetcher agent
        available_movies = self.fetcher.get_new_releases()

        recommendations = [
            movie["title"]
            for movie in available_movies
            if movie["genre"] in preferred_genres
        ]
        return recommendations


# Execution
fetcher = MovieFetcherAgent()
recommender = RecommendationAgent(fetcher)

user_genres = ["Animation", "Sci-Fi"]
results = recommender.recommend_movies(user_genres)

print("Recommended Movies:", results)
```

## Question 3

Refactor an existing single-agent code that suggests restaurants based on cuisine (using a simple Python function) to a modular agent class structure with clear separation of perception, reasoning, and action methods.

Hint: Create an AgentRestaurantRecommender class and split the logic into three methods for each module.

**Answer:**

```python
class AgentRestaurantRecommender:

    def __init__(self, restaurant_database):
        self.database = restaurant_database
        self.user_cuisine = None
        self.selected_restaurants = []

    def perceive(self, user_input):
        """Extract and store the user's cuisine preference."""
        self.user_cuisine = user_input.strip().lower()

    def reason(self):
        """Filter restaurants matching the perceived cuisine."""
        self.selected_restaurants = [
            r["name"]
            for r in self.database
            if r["cuisine"].lower() == self.user_cuisine
        ]

    def act(self):
        """Return the recommendation output to the user."""
        if self.selected_restaurants:
            return f"Recommended restaurants: {', '.join(self.selected_restaurants)}"
        return f"No restaurants found for {self.user_cuisine} cuisine."
```
