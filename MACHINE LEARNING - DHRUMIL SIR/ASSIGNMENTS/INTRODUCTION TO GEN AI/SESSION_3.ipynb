## Question 1

Write a Python function called simple_generator() that takes a random vector as input and returns a 28x28 array of grayscale pixel values between 0 and 1, simulating a fake handwritten digit image like those used in GAN demos.

**Answer:**

```python
import numpy as np


def simple_generator(noise_vector):
    # Set seed based on input vector to generate consistent pseudo-features
    seed = int(np.abs(np.sum(noise_vector)) * 1e6) % (2**32)
    rng = np.random.default_rng(seed)

    # Generate a 28x28 array of pixel values between 0 and 1
    fake_image = rng.uniform(0.0, 1.0, (28, 28))

    return fake_image
```

Create a Python class Discriminator with a method discriminate(image) that takes a 28x28 image array and returns a probability score between 0 and 1 indicating if the image is 'real' or 'fake'.

Hint: For this exercise, use a simple rule (e.g., mean pixel value threshold) instead of a neural network.

**Answer:**

```python
import numpy as np


class Discriminator:

    def discriminate(self, image):
        # Normalize the mean pixel value (assuming 0-255 range) to a 0-1 probability
        mean_val = np.mean(image)
        probability = float(np.clip(mean_val / 255.0, 0.0, 1.0))
        return probability
```
