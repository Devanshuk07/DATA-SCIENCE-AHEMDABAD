## Question 1

Write a short paragraph in your own words explaining what Generative AI (GenAI) is, and give two real-world examples of GenAI applications you have seen or used (such as ChatGPT, DALL-E, or Instagram filters).

**Answer:**

Generative AI (GenAI) is a type of artificial intelligence that creates new, original content rather than just analyzing existing data. It learns patterns, structures, and relationships from massive training datasets to produce human-like outputs. Based on user prompts, it can generate text, realistic images, synthetic audio, and functional computer code.

Two real-world examples:
1. **ChatGPT**: An AI chatbot that generates human-like text responses, writes essays, and debugs code.
2. **DALL-E**: An AI system that creates original, realistic images and art from natural language text descriptions.

## Question 2

Create a table comparing GenAI with traditional AI by listing at least three key differences in how they work and the types of outputs they produce.

Hint: Consider aspects like data input, creativity, and output format (fixed vs. new content).

**Answer:**

| Feature | Traditional AI | Generative AI (GenAI) |
| :--- | :--- | :--- |
| **How it Works** | Analyzes existing data using rules or patterns to classify, predict, or optimize. | Uses deep learning models to learn underlying patterns to create new content. |
| **Creativity** | Non-creative; operates strictly within predefined logic and historical constraints. | Creative; generates novel outputs that did not exist in the training data. |

## Question 3

List 5 popular apps or features you use daily (like Instagram Reels, Spotify recommendations, Zomato chatbots, YouTube video summaries, or Flipkart product images) and identify which ones likely use GenAI. For each, briefly explain how GenAI might be working behind the scenes.

**Answer:**

1. **YouTube Video Summaries**
   * **Uses GenAI:** Yes
   * **Behind the scenes:** A Large Language Model (LLM) reads the video transcript, identifies the main topics, and generates a concise, readable text summary for the user.

2. **Spotify Recommendations**
   * **Uses GenAI:** Mostly No (Traditional AI/ML)
   * **Behind the scenes:** It primarily uses collaborative filtering and predictive algorithms based on past listening history; however, features like the "AI DJ" use generative voice AI and LLMs to create personalized commentary between tracks.

3. **Zomato Customer Support Chatbot**
   * **Uses GenAI:** Yes
   * **Behind the scenes:** It uses natural language understanding and generation to comprehend user queries, check real-time order status via APIs, and generate conversational, human-like responses.

4. **Instagram Reels (Recommendation Feed)**
   * **Uses GenAI:** No (Traditional AI/ML)
   * **Behind the scenes:** It relies on predictive ranking algorithms that analyze watch time, likes, and shares to rank existing content, rather than generating new media or text.

5. **Flipkart Product Summaries and Reviews**
   * **Uses GenAI:** Yes
   * **Behind the scenes:** An LLM scans hundreds of user-submitted reviews for an item and writes a synthesized summary highlighting the common pros and cons.
