## Question 2

Create a simple Q&A chatbot for Zomato-style restaurant info using any language or platform you know (Python, JavaScript, etc.): given a restaurant name and cuisine, return a canned response like 'Yes, Pizza Palace serves Italian cuisine.'

**Answer:**

```python
def restaurant_qa_bot(restaurant_name, cuisine):
    # Dictionary storing restaurant and their available cuisines
    database = {
        "Pizza Palace": ["Italian", "Fast Food"],
        "Dragon House": ["Chinese", "Asian"],
        "Taco Fiesta": ["Mexican"],
        "Spice Hub": ["Indian"],
    }

    # Fetch cuisines for the restaurant (case-insensitive check)
    matched_restaurant = next(
        (name for name in database if name.lower() == restaurant_name.lower()), None
    )

    if matched_restaurant:
        available_cuisines = [c.lower() for c in database[matched_restaurant]]
        if cuisine.lower() in available_cuisines:
            return f"Yes, {matched_restaurant} serves {cuisine} cuisine."
        else:
            return f"No, {matched_restaurant} does not serve {cuisine} cuisine."
    else:
        return f"Sorry, {restaurant_name} is not in our database."


# Example usage:
response = restaurant_qa_bot("Pizza Palace", "Italian")
print(response)
```

## Question 4

Use GitHub Copilot or ChatGPT to help you write a function that takes a list of song names and returns only those that start with the letter 'S', like a Spotify playlist filter.

Constraint: You must show the prompt you used and the code Copilot or ChatGPT generated.

**Answer:**

**Prompt:**
"Write a Python function that takes a list of song names and returns a new list containing only the songs that start with the letter 'S' (case-insensitive)."

**Generated Code:**
```python
def filter_s_songs(songs):
    return [song for song in songs if song.lower().startswith("s")]
```

## Question 5

List two production challenges companies face when using generative AI models for chatbots in apps like BookMyShow or Myntra, and suggest one solution for each challenge.

**Answer:**

**1. Hallucinations (Incorrect Information)**
* **Challenge:** The model may invent fake movie timings, wrong seat availability, or incorrect product return policies.
* **Solution:** Implement Retrieval-Augmented Generation (RAG) to ground the model's responses strictly in real-time, verified database records.

**2. High Latency and Cost**
* **Challenge:** Processing long customer queries via large models causes slow response times and expensive API/compute bills during peak shopping or booking hours.
* **Solution:** Use smaller, task-specific fine-tuned models combined with semantic caching to serve common queries instantly without repeatedly calling the main model.
