## Question 1

Simulate the forward diffusion process by writing a Python function that takes a grayscale image (use any small PNG or JPEG) and adds Gaussian noise to it for 10 steps, saving each noisy image to disk.

**Answer:**

```python
import numpy as np
from PIL import Image


def forward_diffusion(image_path, num_steps=10):
    # Load grayscale image and normalize to [0, 1]
    img = Image.open(image_path).convert("L")
    x = np.array(img, dtype=np.float32) / 255.0

    # Define linear variance schedule
    betas = np.linspace(0.01, 0.2, num_steps)
    alphas = 1.0 - betas
    alphas_cumprod = np.cumprod(alphas)

    # Add noise step-by-step using the closed-form formula
    for t in range(num_steps):
        noise = np.random.normal(size=x.shape)
        noisy_x = np.sqrt(alphas_cumprod[t]) * x + np.sqrt(1.0 - alphas_cumprod[t]) * noise

        # Clip values to [0, 1], scale to [0, 255], and save
        noisy_x_clipped = np.clip(noisy_x * 255.0, 0, 255).astype(np.uint8)
        Image.fromarray(noisy_x_clipped).save(f"step_{t+1}.png")


# Example usage:
# forward_diffusion("input.png")
```

## Question 2

Write a Python script that takes a noisy image and applies a simple denoising filter (like median blur or Gaussian blur) to approximate the reverse diffusion process, then compare the denoised output visually with the original image.

Hint: Use OpenCV's cv2.medianBlur() or cv2.GaussianBlur() for denoising.

**Answer:**

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Load original image
original = cv2.imread("input.jpg")
original = cv2.cvtColor(original, cv2.COLOR_BGR2RGB)

# Add Gaussian noise (simulating forward diffusion)
noise = np.random.normal(0, 25, original.shape).astype(np.uint8)
noisy = cv2.add(original, noise)

# Apply Gaussian blur to approximate reverse diffusion
denoised = cv2.GaussianBlur(noisy, (5, 5), 0)

# Visually compare the results
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
titles = ["Original", "Noisy", "Denoised"]
images = [original, noisy, denoised]

for ax, img, title in zip(axes, images, titles):
    ax.imshow(img)
    ax.set_title(title)
    ax.axis("off")

plt.tight_layout()
plt.show()
```

## Question 5

List two possible risks or ethical concerns when using text-to-image models like Stable Diffusion for social media content, and suggest one way developers can reduce each risk.

**Answer:**

1. **Spread of Misinformation and Deepfakes**
* **Risk:** Users can generate hyper-realistic, fabricated images of real people or events to deceive the public.
* **Mitigation:** Developers can embed invisible, cryptographic watermarks into generated images to verify their synthetic origin.

2. **Amplification of Social Biases**
* **Risk:** Models trained on uncurated internet data often reinforce harmful gender, racial, or cultural stereotypes in generated visuals.
* **Mitigation:** Developers can filter training datasets to ensure diverse representation and use system-level prompt re-weighting to balance outputs.
