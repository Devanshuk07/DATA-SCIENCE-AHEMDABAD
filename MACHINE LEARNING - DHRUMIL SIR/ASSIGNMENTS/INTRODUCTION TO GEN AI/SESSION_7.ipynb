## Question 1

Create a table comparing traditional AI/ML and Gen AI across at least five aspects such as data requirements, output type, human interaction, adaptability, and example use cases. Use real-world apps you know (like Instagram, Zomato, or Spotify) as examples for each.

**Answer:**

| Aspect | Traditional AI / ML | Generative AI |
| :--- | :--- | :--- |
| **Data Requirements** | Large volumes of structured, task-specific, labeled data. | Massive, diverse, unstructured datasets for pre-training. |
| **Output Type** | Predictions, classifications, labels, or numerical scores. | Brand new content (text, images, audio, code). |
| **Human Interaction** | Passive; user inputs filters or clicks, system responds automatically. | Active; interactive, conversational prompting in natural language. |
| **Adaptability** | Rigid; limited strictly to the single task it was trained on. | Flexible; easily adapts to diverse tasks via prompt adjustments. |
| **Example Use Case** | **Spotify**: Song recommendation algorithm based on past listening history. | **Spotify**: "AI DJ" generating personalized, spoken commentary between tracks. |

## Question 2

Given a scenario where Zomato recommends restaurants based on your previous orders, explain whether this uses traditional AI/ML or Gen AI, and justify your answer in 3-4 sentences.

**Answer:**

This scenario uses **traditional AI/ML**. 

Recommendation systems of this type rely on predictive and discriminative algorithms, such as collaborative filtering and classification, to rank existing data based on past user behavior. Unlike Generative AI, they do not create entirely new content, images, or text. Instead, traditional ML analyzes patterns in historical order data to predict and match relevant restaurants already present in the database.

## Question 3

Pick any feature from an app you use daily (such as Instagram's photo filters or Spotify's song recommendations) and describe how it could be improved or transformed using Gen AI instead of traditional AI/ML. Write 4-5 lines explaining the change.

**Answer:**

**Feature:** Spotify’s playlist generation.

* **Traditional AI/ML:** Recommends fixed lists of existing tracks based on static listening history and collaborative filtering.
* **Gen AI Transformation:** Users can type complex natural language prompts (e.g., "songs for a rainy drive feeling nostalgic about 2010s indie rock"), and the system instantly generates tailored playlists with dynamic, AI-voiced transition commentary between tracks. 
* **Key Improvement:** It moves from passive pattern-matching to an interactive, contextual co-creator that understands nuanced human moods.

## Question 4

Imagine you are building a chatbot for Flipkart customer support. List two ways a traditional AI/ML approach would handle user queries, and two ways a Gen AI approach would differ in handling the same queries.

Hint: Focus on how responses are generated and the flexibility of the answers.

**Answer:**

**Traditional AI/ML Approach:**
1. **Predefined Template Retrieval:** Classifies user intent (e.g., "track order") using standard NLP and fills predefined, hardcoded templates with database values.
2. **Rigid Query Matching:** Fails or redirects to a human agent when user queries contain typos, complex phrasing, or fall outside predefined rule sets.

**Gen AI Approach:**
1. **Dynamic Text Generation:** Uses Large Language Models (LLMs) to generate natural, context-aware responses on the fly rather than relying on fixed templates.
2. **Flexible and Multi-Turn Understanding:** Handles ambiguous inputs, slang, and complex multi-part questions while maintaining conversational memory across multiple turns.
