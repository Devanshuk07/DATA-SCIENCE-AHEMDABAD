## Question 1

List 3 apps you use daily (like Instagram, Zomato, Spotify) and describe one feature in each that could be powered by a generative AI model instead of traditional rule-based logic.

**Answer:**

1. **Spotify - AI DJ / Dynamic Playlist Generation**: Instead of rule-based genre or tempo filters, a generative AI model uses natural language prompts and listening history to dynamically generate custom track transitions, conversational commentary, and personalized playlists.

## Question 2

Create a table comparing discriminative models and generative models across at least 4 aspects (definition, main use, example task, example model name).

**Answer:**

| Aspect | Discriminative Models | Generative Models |
| :--- | :--- | :--- |
| **Definition** | Learn the boundary between classes by modeling the conditional probability $P(Y \mid X)$. | Learn the underlying distribution of the data by modeling the joint probability $P(X, Y)$ or $P(X)$. |
| **Main Use** | Classifying or predicting labels for given input data. | Generating new, synthetic data instances similar to the training data. |
| **Example Task** | Spam email detection | Creating realistic human face images |
| **Example Model** | Support Vector Machine (SVM) | Generative Adversarial Network (GAN) |

## Question 4

Pick one generative AI application (text generation, image synthesis, music/audio generation, or code generation) and describe a new feature you would add to an app you use, explaining how generative AI would power it.

**Answer:**

**Application:** Text generation  
**App:** Notion  
**New Feature:** Context-Aware Meeting Synthesizer
