## Question 2

Take a long WhatsApp chat exported as a PDF (or any e-book PDF), extract the text, and split it into chunks of 500 characters each, storing each chunk in a list.

Hint: Use simple string slicing or a loop to create the chunks after extracting the text.

**Answer:**

```python
from pypdf import PdfReader


def extract_and_chunk_pdf(pdf_path, chunk_size=500):
    # Extract text from PDF
    reader = PdfReader(pdf_path)
    full_text = ""
    for page in reader.pages:
        text = page.extract_text()
        if text:
            full_text += text

    # Split into 500-character chunks
    chunks = [
        full_text[i : i + chunk_size]
        for i in range(0, len(full_text), chunk_size)
    ]

    return chunks


# Example usage
# chunks_list = extract_and_chunk_pdf("chat.pdf")
```

## Question 4

Write a function that takes a user question (e.g., 'Who messaged the most?') and finds the most relevant chunk from your WhatsApp chat chunks using cosine similarity between the question's embedding and your chunk embeddings.

Constraint: Return the chunk with the highest similarity score.

**Answer:**

```python
import numpy as np


def find_most_relevant_chunk(
    question, chunks, chunk_embeddings, embedding_model
):
    # Generate embedding for the question
    question_embedding = embedding_model.encode(question)

    # Compute cosine similarity
    dot_products = np.dot(chunk_embeddings, question_embedding)
    norms = np.linalg.norm(chunk_embeddings, axis=1) * np.linalg.norm(
        question_embedding
    )
    similarities = dot_products / norms

    # Get the index of the highest similarity score
    best_index = np.argmax(similarities)

    return chunks[best_index]
```
