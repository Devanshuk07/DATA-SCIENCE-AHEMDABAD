## Question 1

Write a Python function chunk_text(text, chunk_size, overlap) that splits a long string into chunks of the given size with the specified overlap. Test your function with a 500-word sample from a Wikipedia article.

**Answer:**

```python
def chunk_text(text, chunk_size, overlap):
    words = text.split()
    step = chunk_size - overlap
    chunks = []
    for i in range(0, len(words), step):
        chunk = " ".join(words[i : i + chunk_size])
        chunks.append(chunk)
        if i + chunk_size >= len(words):
            break
    return chunks


# Test with a 500-word sample from Wikipedia (Article: Artificial intelligence)
sample_text = """
Artificial intelligence is the intelligence of machines or software, as opposed to the intelligence
of humans or other animals. It is also the field of study in computer science that develops and studies
intelligent machines. Such machines may be called AIs. AI technology is widely used throughout industry,
government, and science. Some high-profile applications include advanced web search engines, recommendation
systems, understanding human speech, self-driving cars, generative creativity, and superhuman play in games.
Alan Turing was the first person to conduct substantial research in AI. The field of AI research was
founded at a workshop at Dartmouth College in 1956. The attendees became the leaders of AI research for
decades. Many of them predicted that a machine as intelligent as a human being would exist in no more
than a generation, and millions of dollars were invested to make this vision come true. Eventually, it
became obvious that researchers had severely underestimated the difficulty of the project. In response,
funding was cut and the field entered a period known as an AI winter. In the 1980s, AI research was
revived by the commercial success of expert systems. By 1985, the market for AI had reached over a billion
dollars. At the same time, the connectionism revival was led by John Hopfield and David Rumelhart. In the
late 1990s and early 21st century, AI technology began to be used widely as part of larger systems, though
the field was rarely credited for these successes. Faster computers, algorithmic improvements, and access
to large amounts of data enabled advances in machine learning and perception. Data-hungry deep learning
methods started to dominate accuracy benchmarks around 2012. The term artificial intelligence had previously
been criticized for implying that true machine intelligence was possible. The various sub-fields of AI
research are centered around particular goals and the use of particular tools. The traditional goals of
AI research include reasoning, knowledge representation, planning, learning, natural language processing,
perception, and support for robotics. General intelligence is among the field's long-term goals. To reach
these goals, AI researchers have adapted and integrated a wide range of techniques, including search and
mathematical optimization, formal logic, artificial neural networks, and methods based on statistics,
operations research, and economics. AI also draws upon psychology, linguistics, philosophy, neuroscience,
and other fields. Machine learning is a fundamental concept of AI research since its inception. In 1959,
Arthur Samuel defined machine learning as a field of study that gives computers the ability to learn without
being explicitly programmed. A neural network is a network or circuit of biological neurons, or, in a
modern sense, an artificial neural network, composed of artificial neurons or nodes. Deep learning is part
of a broader family of machine learning methods based on artificial neural networks with representation
learning. Deep learning has drastically improved the performance of programs in many domains including
computer vision, speech recognition, image synthesis, natural language processing, audio recognition,
social network filtering, machine translation, bioinformatics, drug design, medical image analysis, climate
science, material inspection, and board game programs.
""" * 2  # Scaled to reach ~500 words

# Truncate strictly to 500 words
sample_500_words = " ".join(sample_text.split()[:500])

# Function execution
chunks = chunk_text(sample_500_words, chunk_size=100, overlap=20)

# Display results
print(f"Total words: {len(sample_500_words.split())}")
print(f"Total chunks generated: {len(chunks)}")
for index, chunk in enumerate(chunks, 1):
    print(f"Chunk {index} word count: {len(chunk.split())}")
```

## Question 3

Use the Sentence Transformers library (sentence-transformers) in Python to generate embeddings for three different food review sentences (e.g., 'The pizza was amazing', 'Service was slow', 'Loved the ambience'). Print the embedding vector for each sentence.

Hint: Use the 'all-MiniLM-L6-v2' model from sentence-transformers.

**Answer:**

```python
from sentence_transformers import SentenceTransformer

# Load the model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Define the sentences
sentences = ["The pizza was amazing", "Service was slow", "Loved the ambience"]

# Generate embeddings
embeddings = model.encode(sentences)

# Print the embedding vector for each sentence
for sentence, embedding in zip(sentences, embeddings):
    print(f"Sentence: {sentence}")
    print(f"Embedding: {embedding}\n")
```

## Question 5

Use ChatGPT or any AI tool to explain in your own words why chunk overlap is important in document retrieval, and give an example scenario where too little overlap could cause problems in a Zomato-style restaurant review search.

**Answer:**

**Importance of Chunk Overlap:**
Chunk overlap is important because it prevents critical context from being cut in half when large documents are split into pieces. Without overlap, words or sentences that depend on each other can end up in separate chunks, causing the search system to misunderstand the meaning. Retaining a shared boundary between consecutive chunks ensures semantic continuity and improves retrieval accuracy.

**Zomato-Style Example Scenario:**
* **Review Text:** "The chicken biryani was absolute perfection, bursting with authentic spices. However, the mutton curry was completely raw and ruined our night."
* **The Problem:** If split without overlap, Chunk 1 gets "The chicken biryani was absolute perfection...", and Chunk 2 gets "...However, the mutton curry was completely raw...".
* **Failure:** A user searching for "best mutton curry" might retrieve Chunk 1 based on nearby positive restaurant sentiment, or retrieve Chunk 2 without realizing "However" contrasts it with the biryani, potentially misinterpreting the overall context and returning a restaurant with raw food as a top recommendation.
