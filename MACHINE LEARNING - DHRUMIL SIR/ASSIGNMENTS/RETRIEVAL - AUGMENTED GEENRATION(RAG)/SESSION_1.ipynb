## Question 1

In your own words, explain what Retrieval-Augmented Generation (RAG) is and give one example of how it could improve a feature in an app you use daily, like Zomato, Flipkart, or Instagram.

**Answer:**

**Retrieval-Augmented Generation (RAG)** is an AI technique that combines an external search system with a generative language model. Instead of relying only on its pre-trained data, the model first retrieves relevant, up-to-date information from a specific knowledge base or database. It then uses this retrieved context to generate an accurate, grounded, and reliable response.

**Example (Zomato):**
RAG can improve Zomato's search assistant. When a user asks, *"Which nearby open place has the best creamy white sauce pasta based on recent reviews?"*, the system retrieves live restaurant menus, operating hours, and customer reviews from the last 24 hours. The model then generates a personalized, highly accurate recommendation without hallucinating outdated menu items or closed restaurants.

## Question 2

Imagine you are chatting with a travel assistant bot that uses a Large Language Model (LLM) trained in 2022. List 3 questions about IRCTC train schedules or IPL 2024 that the bot would struggle to answer, and explain why outdated LLM knowledge is a problem in this scenario.

**Answer:**

**Three Questions the Bot Would Struggle to Answer:**
1. "What is the departure platform and current delay for the Vande Bharat Express from New Delhi to Varanasi today?" (Requires real-time railway schedule and platform data).
2. "Who won the final match of IPL 2024, and what was the scorecard?" (Requires information on an event that occurred after the 2022 training cutoff).
3. "Are there any ticket cancellations or route diversions on IRCTC due to track maintenance this week?" (Requires live, dynamic operational updates).

**Why Outdated LLM Knowledge is a Problem:**
Outdated knowledge leads to "hallucinations" or completely incorrect information because the model relies solely on past training data without knowing recent events. For travel and live sports, users need accurate, real-time facts; relying on old data can cause missed trains, invalid travel bookings, and a complete loss of user trust.
