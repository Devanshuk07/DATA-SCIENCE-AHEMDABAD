## Question 3

Write a function that takes a user's search query (e.g., 'action movie') and a list of movie description embeddings, then uses FAISS to find and display the top 2 most similar movie titles.

**Answer:**

```python
import faiss
import numpy as np


def find_top_movies(
    query_embedding, movie_embeddings, movie_titles, embedding_dim=1536
):
    # Initialize FAISS index
    index = faiss.IndexFlatL2(embedding_dim)

    # Convert embeddings to float32 numpy arrays and add to index
    movie_embeddings_np = np.array(movie_embeddings).astype("float32")
    index.add(movie_embeddings_np)

    # Search for top 2 closest movies
    query_np = np.array([query_embedding]).astype("float32")
    distances, indices = index.search(query_np, k=2)

    # Display the results
    print("Top 2 similar movies:")
    for idx in indices[0]:
        print(movie_titles[idx])
```

## Question 4

Suppose you have a database of Instagram captions stored as embeddings in ChromaDB. Simulate a scenario where a user wants to find captions similar to 'Weekend vibes'. Write code to retrieve and print the 3 most similar captions, and explain in one line how retrieval works in this example.

Hint: Use ChromaDB's similarity search function and random vectors for embeddings.

**Answer:**

```python
import chromadb
import numpy as np

# Initialize ChromaDB client and collection
client = chromadb.Client()
collection = client.create_collection(name="instagram_captions")

# Generate random vectors (3-dimensional) as dummy embeddings
np.random.seed(42)
doc_embeddings = np.random.rand(5, 3).tolist()
query_embedding = np.random.rand(1, 3).tolist()

# Add dummy caption data
collection.add(
    ids=["1", "2", "3", "4", "5"],
    embeddings=doc_embeddings,
    documents=[
        "Chilling on a Saturday",
        "Monday morning blues",
        "Sunday brunch fun",
        "Office work all day",
        "Lazy Sunday afternoon",
    ],
)

# Retrieve the 3 most similar captions to the query vector
results = collection.query(query_embeddings=query_embedding, n_results=3)

# Print results
for caption in results["documents"][0]:
    print(caption)
```

Retrieval works by calculating the distance between the query vector and stored caption vectors, returning the documents with the smallest distances.

## Question 5

Use ChatGPT to generate 5 sample product descriptions for a Flipkart-style app, then create embeddings for each (using any embedding API or random vectors). Store them in FAISS and demonstrate how to query for the most similar product to 'wireless headphones'.

Hint: Paste your ChatGPT prompt and results in your submission along with the code.

**Answer:**

### 1. ChatGPT Prompt and Output

**Prompt:**
> "Generate 5 short e-commerce product descriptions suitable for a Flipkart-style catalog across different categories."

**Generated Descriptions:**
1. "boAt Rockerz 450 Bluetooth On-Ear Headphone with Mic, up to 15 hours playback, adaptive ear cups, and 40mm dynamic drivers."
2. "Puma Men's Running Shoes with breathable mesh upper, lightweight EVA midsole, and durable rubber outsole for daily workouts."
3. "Noise ColorFit Pulse 2 Smartwatch with 1.85-inch display, 550 nits brightness, Bluetooth calling, and 10-day battery life."
4. "Philips Cordless Beard Trimmer with self-sharpening stainless steel blades, 60 minutes run time, and 20 length settings."
5. "OnePlus Nord Buds 2 True Wireless in-Ear Earbuds with 25dB active noise cancellation, BassWave algorithm, and fast charging."

---

### 2. Python Code (Embeddings, FAISS Storage, and Search)

```python
import faiss
from sentence_transformers import SentenceTransformer

# 1. Product data
products = [
    (
        "boAt Rockerz 450 Bluetooth On-Ear Headphone with Mic, up to 15 hours"
        " playback, adaptive ear cups, and 40mm dynamic drivers."
    ),
    (
        "Puma Men's Running Shoes with breathable mesh upper, lightweight EVA"
        " midsole, and durable rubber outsole for daily workouts."
    ),
    (
        "Noise ColorFit Pulse 2 Smartwatch with 1.85-inch display, 550 nits"
        " brightness, Bluetooth calling, and 10-day battery life."
    ),
    (
        "Philips Cordless Beard Trimmer with self-sharpening stainless steel"
        " blades, 60 minutes run time, and 20 length settings."
    ),
    (
        "OnePlus Nord Buds 2 True Wireless in-Ear Earbuds with 25dB active noise"
        " cancellation, BassWave algorithm, and fast charging."
    ),
]

# 2. Generate embeddings
model = SentenceTransformer("all-MiniLM-L6-v2")
product_embeddings = model.encode(products, convert_to_numpy=True)

# 3. Store in FAISS
dimension = product_embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)
index.add(product_embeddings)

# 4. Query for 'wireless headphones'
query = "wireless headphones"
query_embedding = model.encode([query], convert_to_numpy=True)

# Search top 1 most similar product
distances, indices = index.search(query_embedding, k=1)

# Result
print(f"Query: {query}")
print(f"Top Match: {products[indices[0][0]]}")
```
