```python
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import root_mean_squared_error,mean_squared_error,r2_score
```

```python
df=pd.read_csv('emp.csv')
df.head()
```

```python
x=df[['Years_With_Company','Post_Name']]
x=pd.get_dummies(x,columns=['Post_Name'],dtype='int')
y=df['Salary']
```

```python
x_train,x_test,y_train,y_test=train_test_split(
    x,
    y,
    train_size=0.80,
    random_state=42
)
```

```python
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
model=KNeighborsRegressor(n_neighbors=3)
model.fit(x_train,y_train)
pred=model.predict(x_test)
pred
```

```python
rmse=root_mean_squared_error(y_test,pred)
print(rmse)
mse=mean_squared_error(y_test,pred)
print(mse)
rscor=r2_score(y_test,pred)
print(rscor)
```

```python
a=int(input('ENTER THE YEARS WITH COMPANY :-'))
b=input('ENTER YOUR POST :-').strip().title()   # strip() is a Python string function that removes extra spaces from the beginning and end of a string..title() helps the new input match the exact category format used when get_dummies() created the training columns.
result=pd.DataFrame({
    'Years_With_Company':[a],
    'Post_Name':[b]
})
result=pd.get_dummies(result,columns=['Post_Name'],dtype=int)
result=result.reindex(columns=x.columns,fill_value=0)     #reindex() = "Make my new data structure exactly like the data my model was trained on",fill_value=0 means missing dummy columns are filled with 0.
rst=scaler.transform(result)
prediction=model.predict(rst)
print(prediction)
```
