```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.tools import tool
from langchain.agents import create_agent
```

```python
load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",
    project="project-d24e3f00-1d87-4332-818",
    location="global",
    vertexai=True,
    temperature=0.7
)
```

```python
import mysql.connector
db=mysql.connector.connect(
    host="localhost",
    user="root",
    password="",
    database="database1",
    port=3306
)
```

```python
@tool
def dbsql(query):
    """  Execute a SQL query on database1.
    Available tables:
    - student (id, name)
    - hobbies (id, student_id, hobby)
    - paying_for_hobbies (id, student_id, amount)
    """
    cursor=db.cursor()
    cursor.execute(query)
    data=cursor.fetchall()
    return data
```

```python
agent=create_agent(
    model=llm,
    tools=[dbsql]
)
```

```python
a=input('ENTER THE QUESTION :-')
res=agent.invoke({
    "messages":[
        ("user",a)
    ]
})
print(res["messages"][-1].content[0]["text"])
```
