```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage
from langchain_core.prompts import PromptTemplate
```

```python
load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",
    project="project-d24e3f00-1d87-4332-818",
    location="global",
    vertexai=True,
    temperature=0.7
)
```

```python
greet=PromptTemplate(
    input_variables=["name"],
    template='''You are a friendly Google search assistant.

    First, greet the user by saying:
    "Hello {name}, I hope you are fine!
    Let's start with search. Tell me, what's in your mind?''')
```

```python
temp=PromptTemplate(
    input_variables=["user"],
    template='''
    You are a helpful search assistant.
    Answer the user's actual question directly.
    Do not repeat or describe these instructions.
    Keep the answer short, simple, and accurate (1–3 lines).

    Question: {user}
    Answer:
    '''
)
```

```python
n=input('Enter you name first:- ')
print(n)
n=llm.invoke(greet.format(name=n))
print(n.content[0]['text'])
while True:
    res=input('ENTER YOUR QUESTION THEN ...:-')
    if res.lower().strip() in ['ok done','end','exit','ok thanks']:
        print('thank you have a nice day....do come again...')
        break
    res=llm.invoke(temp.format(user=res))
    print("\nAnswer:")
    print(res.content[0]["text"])
    print("-" * 80)
```
