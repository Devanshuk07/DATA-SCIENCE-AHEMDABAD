```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage
from langchain_core.prompts import PromptTemplate
```

```python
load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=0.7
)
```

```python
greet=PromptTemplate(
    input_variables=["name"],
    template='''You are a friendly Google search assistant.

    First, greet the user by saying:
    "Hello {name}, I hope you are fine!
    Let's start with search. Tell me, what's in your mind?''')
```

```python
temp=PromptTemplate(
    input_variables=["user"],
    template='''
    when user ask you his question{user},
    you will just give answer in short simple way,
    if answer need to be given in 3 lines then 3 but if it is
    able to answer in 2-1 line then make it short simple and sweet only...

    '''
)
```

```python
n=input('Enter you name first:- ')
print(n)
res=llm.invoke(greet.format(name=n))
print(res.content[0]['text'])
while True:
    res=input('ENTER YOUR QUESTION THEN ...:-')
    if res.lower().strip() in ['ok done','end','exit','ok thanks']:
        print('thank you have a nice day....do come again')
        break
    res=llm.invoke(temp.format(user=res))
    print("\nAnswer:")
    print(res.content[0]["text"])
    print("-" * 80)
```
