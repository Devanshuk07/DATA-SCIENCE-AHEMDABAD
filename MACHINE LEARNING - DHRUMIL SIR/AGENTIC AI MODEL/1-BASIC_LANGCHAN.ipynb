```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI     # langchain-google-genai is the library that connects LangChain to Google's Gemini AI models.ChatGoogleGenerativeAI — The class used to create a chat model powered by Gemini.
from langchain_core.messages import HumanMessage          #to generate human understand language
```

```python
load_dotenv()        # we load this here and dont change env variable in windows so our program can even be used when taking it live and on github.
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=0.7
)          # temp = from 0 to 1 it will responce according to this temp

#llm = ChatGoogleGenerativeAI(...) to create an LLM object that our Python code can use to communicate with Gemini.
```

```python
question = input('ask your question')
```

```python
response = llm.invoke([HumanMessage(content=question)])
```

```python
print('GEMINI :-',response.content[0]["text"])
```
