```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage
from langchain_core.prompts import PromptTemplate
import requests
from langchain_core.tools import tool
from langchain.agents import create_agent
```

```python
load_dotenv()
```

```python
llm=ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",
    vertexai=True,
    project="project-d24e3f00-1d87-4332-818",
    location="global",
    temperature=0.9
)
```

```python
api="https://careercenter.tops-int.com/dashboard"
```

```python
@tool
def tops():
    """you just give me website data and as it is nothing else"""
    res=requests.get(api)
    return res.text
```

```python
result=tops.invoke({})
print(result)
print(result.text)
```

```python
agent=create_agent(
    model=llm,
    tools=[tops],
    system_prompt="""
Fetch data from the given website using the available tool.
Return the fetched data as it is.
Do not add explanations, summaries, or any extra text.
"""
)
```

```python
question = input('ask replated product')


res=agent.invoke({
    "messages":[                     # why use message: --> our data isin json==key and value format...so to access it like that direct answer we write this.
        ("user",question)
    ]})
print(res["messages"][-1].content)
```
