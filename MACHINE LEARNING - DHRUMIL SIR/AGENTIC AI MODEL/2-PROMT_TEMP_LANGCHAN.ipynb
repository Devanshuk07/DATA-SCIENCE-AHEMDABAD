```python
from dotenv import load_dotenv
from langchain_core.messages import HumanMessage
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
```

```python
load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=0.7
)
```

```python
temp = PromptTemplate(                # PromptTemplate — A LangChain class used to create prompt templates.
    input_variables=['input'],         # Every placeholder{input} you use in your template must be declared in input_variables.
    template=
"""
You are a helpful Data Science tutor bot.

The user will ask a question about a Data Science topic. Provide a clear, accurate, and helpful answer to the user's question.

User's Question: {input}

Requirements:
1. Answer the user's question directly.
2. Explain the concept at an intermediate level, suitable for a Data Science student.
3. Include practical examples when they help explain the concept.
4. Use simple, easy-to-understand language.
5. Organize the answer with headings or bullet points when appropriate.
6. If the question involves Python, SQL, or Machine Learning, include relevant code examples when useful.
7. Avoid unnecessary information and stay focused on the user's question.
8. If the user's question is unclear, ask for clarification.

Return only the answer to the user's question.
"""

)
```

```python
while True:
    responce = input('ASK ME ANYTHING ABOUT DATA SCIENCE :- ')

    if responce.lower().strip() in ['all done','exit','end here']:
        print('GOOD-BYEE ... HAVE A NICE DAY...')
        break
    result = llm.invoke(temp.format(input=responce))    # here invoke --> take input from responce and then git it temp(template) to llm and when llm generate answer it will store that in variable=result.
    print(responce)
    print("\nAnswer:")
    print(result.content[0]["text"])         #We use .content because llm.invoke() returns an AI message object and potentially other metadata., and .content gives us the actual response generated by the LLM.
    print("-" * 80)                         # .content extracts the actual answer from the response.

'''remember here we used temp.format this means it will replace {input} to response means actual question. itself.
in humanmessage(content=question) --> content represnt a message sent by the user.'''
```
