```python
from dotenv import load_dotenv
import os          
from langchain_google_genai import ChatGoogleGenerativeAI      
from langchain_core.messages import HumanMessage
from langchain_core.prompts import PromptTemplate
import requests
from langchain_core.tools import tool 
from langchain.agents import create_agent
```

```python
load_dotenv()
a = os.getenv("GEMINI_API_KEY")      # this will print you api key but it should be private...
print(a is not None)
```

```python
import mysql.connector
db=mysql.connector.connect(
    host="localhost",
    user="root",
    password="",
    database="database1",
    port=3306
)
```

```python
from httpx2 import query
@tool
def mydb(query):
    """  Execute a SQL query on database1.
    Available tables:
    - student (id, name)
    - hobbies (id, student_id, hobby)
    - paying_for_hobbies (id, student_id, amount)
    """
    cursor=db.cursor()
    cursor.execute(query)
    data=cursor.fetchall()
    return data
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    temperature=0.7

)
```

```python
agent = create_agent(
    model=llm,
    tools=[mydb],
    
)
```

```python
a=input('ENTER WHAT YOU WANT TO FETCH FROM DATABASE :- ')
print(agent)
response=agent.invoke({
    "messages":[(
        "user",a
    )]
})

print(response["messages"][-1].content[0]["text"])
```
