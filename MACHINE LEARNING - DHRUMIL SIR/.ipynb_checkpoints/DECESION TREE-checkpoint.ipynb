```python
'''DECESION TREE IS USED IN CLASSIFICATION GROUP....IT IS USED TO DIVIDE DATA INTO DIFFERNT GROUP(CLUSTERS).
IT WILL GENERATE QUESTION TO DIVIDE INTO GROUP...THAT IN WHICH GROUP SHOULD IT PUT DATA THIS IS DONE INTERNALLY...
ITS MAIN WORK IS WHAT IT DIVIDES INTO GROUP SHOULD BE PURE...MEANS MIX DATA MUST NOT BE THEIR...

EX:--
SNAME             HRS      RESULT
 A                1         FAIL
 B                2         FAIL
 C                3         PASS
 D                4         PASS
 E                5         PASS
 F                6         PASS

NOW HERE IT WILL CREATE TWO GROUPS HERE AS BINARY(PASS/FAIL) ONLY....
SO TO MAKE THIS GROUP IT WILL CREATE MULTI QUESTIONS...TO ITS OWN...
LET 1ST QUESTION BE LIKE STD HRS > 3.5  ---> THIS IS ROOT NODE(1ST QUESTION)
WHEN TREE GENERATE 2ND QUESTION ---> DECESION NODE(2ND QUESTION)-- WHEN MULTI DATASET...(OTHER CATEGORY)
THEN AFTER QUESTION ASKED IT HAS TWO THING YES OR NO...
THEN AFTER IT TAKE DECESION IT WILL STOP MEANS WHEN IT WILL PUT DATA IN PERTICULAR GROUP IT WILL STOP THAT IS ---> LEAF NODE...
THE LAST AFTER NO QUESTION IS GENERATED THEN IT IS LEAF NODE...
IT WILL CREATE MORE AND MORE QUESTIONS TILL IT WILL NOT GET PURE GROUP(WHICH HAS BOTH DIFFER PASS-GROUP1 AND FAIL-GROUP2)
THEN AT LAST IT WILL CREATE PURE GROUP AND IT HAS 70% PURE...THEN IT WILL TAKE THE MORE PURITY IT WILL GIVE MORE PRIORITY TO IT...

1 GINI - IN OUR GROUP IT WILL FIND IMPURITY....IT WILL HELP TO MAKE PURE GROUP... BY CHANGING VALUE...THE QUESTION DECESION TREE ASK IS CALLED SPLIT AND GINI HELP FOR BEST SPLIT...
LET WE CREATE 4 GROUPS THEN TO WHCIH GROUP WE HAVE TO GIVE MORE PRIORITY...
LET IF WE GOT ANY GROUP GINI VALUE == 0.89 AND OTHER QUESTION IF WE GET GINI VALUE == 0.45 THEN OTHER 0.23 THEN,,,
IF THE 1ST QUESTION 0.89 ROOT NODE IS 3.5 THEN 2ND 0.45 ROOT NODE=3.5  THEN 3RD O.23 ROOT NODE > 2.5...
THEN FROM THIS...WE GET BEST GINI SCORE IS ALWAYS LOW...IF WE GET GINI VALUE =0 THEN IT IS BEST BUT HERE IS 0.23 THAT IS 3RD ONE 2.5....
FINAL DECESION MAKING IS ON GINI VALUE...THEN YOUR FULL DATA WILL BE DIVIDED INTO GROUP1 0.23 STDY HRS MORE THAN 2.5 AND 2ND GROUP STDY HRS LESS THAN 2.5
SO,,
MORE THAN 2.5           LESS THAN 2.5
3,4,5,6                     0,1,2

THIS IS STILL 90% PURE... IF WE GET GINI==0 THEN IT WILL BE 100% PURE...

IT WORK ON:-
1) ROOT NODE
2) DECESION NODE
3) LEAF NODE
BRANCH :- THIS CONNECT ALL THIS 3.
```

```python
'''let we have this data
EX:--
SNAME             HRS      RESULT
 A                1         FAIL
 B                2         FAIL
 C                3         PASS
 D                4         PASS
 E                5         PASS
 F                6         PASS

THEN OUR DECESION TREE WILL DO :-
ROOT NODE :- QUESTION 1 WILL BE ASKED AND IT WILL CREATE 2 GROUP...
STUDY HRS > 3.5    AND     STUDY HRS <=3.5
FOUND-RECODS=3              FOUND-RECORDS=3
PASS=2                         PASS=1
FAIL=1                         FAIL=2

THEN IT WILL GENERATE GINI FROM GROUP ONE ONLY THAT IS STUDY HRS > 3.5
GINI FORMULA :-- 1-(PROBABILITY-OF-PASS/TOTAL-RECORD)SQR)  +  PROBABILITY-OF-FAIL/TOTAL-RECORD)SQR)
                 1-((2/3)SQR      +  (1/3)SQR)
                 1-(4/9 + 1/9)
                 1+1/3
                 4/3 
                 1.33  == GINI

THEN AGAIN IT WILL CREATE GINI NOW FOR GROUP 2ND THAT IS STUDY HRS <=3.5
GINI FORMULA:-- 1-(PROBABILITY(PASS)SQR)  +  PROBABILITY(FAIL)SQR)
                 1-((1/3)SQR      +  (2/3)SQR)
                 1-(1/9 + 4/9)
                 1-1/3
                 2/3
                 0.66  == GINI

NOW TILL THIS WE ASKED ONLY 1 QUESTION AND WE GOT 0.66 AND 1.33 GINI SCORE

LIKE THIS IT WILL ASK MANY QUESTION WITH STUDY HRS :- CONDITION WILL BE CHANGED
AND FOR EVERY QUESTION IT WILL HAVE 2 GINI VALUE...
FROM THAT ALL GINIS THE LOWEST SCORE AROUND 0 WILL IT BE CONSIDERED BEST PURE SCORE...
FIRST IT WILL CHOOSE BEST SPLIT ONE GROUP...
AND THAT CHOSEN GROUP WILL ALSO HAVE 2 GROUP... AND IT WILL MERGE THAT TWO GROUPS.

MERGE :-(LET TAKE AS WE HAVE EXAMPLE IS THE BEST SPLIT AND IT TOOK THAT)
THEN MERGE WILL BE LIKE
3/6 * 1.33  -  3/6 * 0.66            HERE 3/6 IS INDIVIDUAL QUESTION HAS 3 RECORDS AND ALL 1 GROUP HAS TOTAL 6 RECORDS SO 3/6...
THIS IS OUR PERFECT GINI...

AND IF WE GIVEN MAX_DEPTH =2 THEN IT WILL CREATE ONLY 2 QUESTION... AND EVERY 1 QUESTION HAS 2 GROUPS.
WE USE IN CODE:-
DECESIONTREECLASSFIE(MAX_DEPTH=2) LIKE WE USED IN KNN MODEL N_NEIGHBOUR=3,5,7,11...
'''
```

```python
'''WHAT IS LEAF NODE :--

In a Decision Tree, a leaf node is the final node where the model gives its prediction/output.

Leaf Node = The end point of a decision tree that contains the final prediction/result. 

              Age > 25?          ← Root Node
              /      \
            Yes       No
             |         |
       Income > 50K?   Reject    ← Leaf Node
         /      \
       Yes       No
        |         |
     Approve    Reject           ← Leaf Nodes'''
```

```python
'''WE WILL USE IN CODE FOR DECESION TREE...
ENCODING :-- WE HAVE TEXT AND THAT WE WILL CONVERT TO NUMBER TO UNDERSTAND BY MACHINE PROPERLY...
LET,,
NAME     CITY    SALARY
A        SURAT    10,000
B        GOA      12,000
C        DUI      15,000
D        GOA      20,000

HERE WE WANT ENCODING IN CITY COLUMN BUT IT IS NOW IN TEXT THEN WE HAVE 2 METHOD :- FIT TRANSFER AND OTHER IS ONE HOT ENCODING....
HERE IS TYPE OF ENCODING:--
1) LABEL ENCODING :---EXISTING COLUMN WORKS..EXSISTING CITY WILL BE GIVEN SAME NO. AND NEW CITY WILL BE GIVEN NEW NO.
2) ONE HOT ENCODING :- WILL DROP EXISTING COLUMN AND WILL CHECK HOW MANY CITIES WE HAVE LIKE SURAT,GOA,DUI THEN IT WILL CREATE NEW COLUMN FOR EVERY CITY...
BEST OPTION WILL BE ONE HOT ENCODING...

BIAS :-- OUR MODEL IS NOT LEARNING FROM DATA BUT IT IS GIVING PRIORITY TO SOME COLUMN OR VALUES..


CROSS-VALIDATION :--LET WE TOOK SPLIT OF 80-20 TRAIN AND TEST I WILL GET RESULT LIKE THIS FOR 40-60 THEN FOR 30-70 AND FOR ALL I WILL GET RESULT...
THEN TAKING THAT ALL RESULT AVERAGE IS CROSS-VALIDATION...
```

```python
'''HOW TO CALCULATE GINI (BY CHAT GPT)
FORMULA :- 1-E(PI)SQR   --> THIS IS SAME AS WE LEARNED ...1-(PROBABILITY(PASS)SQR)  +  PROBABILITY(FAIL)SQR)'''
```

```python
'''WHAT IS DIFFERNECE BTW BAGGING AND BOOSTING.... OR WHAT IS DIFF BTW RANDOM FOREST AND XGBOOST...
RANDOM FOREST FULL PROCESS IS BAGGING... AND BOOSTING IS...


A= PASS
B= PASS
C= FAIL  

YOU HAVE XGBOOST(xtreme gradient boosting)(ALGORITHM) --> MULTIPLE TREEE IT WILL GENERATE 3 TREE AND OUTPUT WILL BE...

TREE1 = A=PASS B=FAIL C=PASS
TREE2 = THIS WILL BE DEPEND TOTALLY ON TREE1.  IN RANDOM/DECESION TREE IT WAS INDEPENDENT..
TREE2 = IT WILL UNDERSTAND THAT TREE1 HAS GIVEN 2 RECORDS WRONG THEN IT WILL UNDERSTADN IT AND CORRECT IT...
TREE2 = A=PASS B=PASS C=PASS
TREE3 IT WILL BE DEPENDED ON TREE2 AND ITS TARGET IS TO SOLVE THE PROBLEM IN TREE 2..
TREE3 = A=PASS B=PASS C=FAIL....


LIKE THIS YOU WILL GET LAST TREE AND ITS WORK IS TO SOLVE THE PROBLEM DONE BY PREVIOUS TREE...
WE WILL GET PREDICTION FROM LAST TREE ONLY....AND THAT WILL BE OUR OUTPUT...


WHAT IS GRADIENT DECENT ? 
-LIKE WE HAVE 10 DATA AND TREE 1 PREDICT 7 ACCURATE AND 3 INCORRECT...THEN IT IS 70% ACCUARTE THEN TREE2 WILL GIVE 80% AND TREE3 IF 90% THEN 90% IS TAKEN AS LAST AND IT IS CALLED PREDICTION...
```
