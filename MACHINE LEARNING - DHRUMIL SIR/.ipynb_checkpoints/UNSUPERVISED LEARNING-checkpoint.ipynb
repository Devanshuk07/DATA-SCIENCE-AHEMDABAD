```python
'''DIFFERNECE BTW SUPERVISED AND UNSUPERVISED...
SUPER--> WE ALREADY HAVE LABEL IN OUR DATA FROM IT OUR MODEL WAS BEEN TRAINED...
UNSUPER--> WE DONT HAVE LABEL IN OUR DATA OUR MODEL WILL BY ITS OWN UNDERSATND THE FEATURES FROM DATA AND WILL DIVIDE IT INTO DIFFERENT GROUPS...
```

```python
'''ALGO FOR UNSUPERVISED LEARNING:--
1) K-MEANS
2) DBSCAN
3) HEIRARCHICAL CLUSTERING'''
```

```python
'''WE HAVE TO IMPLEMENT 3 THINGS INDIVIDUALLY IN EVERY ALGORITHM :-
1) CLUSTERING :- WILL CREATE GROUP 
2) DIMENSIONALLY REDUCTION(TECNIQUE=PCA) :- WE WILL REDUCE OUR COULMNS(FEATURES)/ DROP WHICH COLUMN DONT MAKE ANY DIFFERNECE/UNNECESSARY COLUMNS..(METHOD = PCA)
3) ASSOCIATION :- (USES APRI ORI) :- THE THINGS WITH BEST ASSOCIATION WILL BE GIVEN AS SUGGETIONS. SUGGESTIONS AS PER YOUR LIKES/CHOICES'''
```

```python
'''WHEN TO USE UNSUPERVISED :-
WHEN YOU WANT TO MAKE CLUSTER(GROUPS)AND DO ASSOCIATION RELATION-SHIP...

EX:- CLUSTER FOR CUSTOMERS WHO ORDERED ICE-CREAM 7-10 TIMES IN A MONTH THAT ALL CUSTOMER IN ONE CLUSTER.
NOW ASSOCIATION WILL GIVE SUGGESTIONS ACCORDING TO ICE-CREAM LIKE CHOCOBAR,CANDY LIKE THIS... TO IMPROVE SELLS...
BUT IT WILL NOT SUGGEST WAFFERS IN ICE-CREAM AS THEY BOTH HAVE DIFFERNET TASTE...'''
```

```python
'''1) K-MEANS :- CLUSTER K=3(3GROUPS)

CUSTOMER     SPENDING    INCOME
 A            2            3
 B            8            9
 C            6            7
 D           100          500
 E           200          600
 F            5            5
 
YOU HAVE TO FIRST DEFINE K...(SO ACCORDING TO K VALUE IT WILL CREATE GROUP)
YOU CANNOT GIVE CONDITION HERE TO CREATE CLUSTER - MODEL WILL UNDERSTAND IT BY ITS OWN AND MAKE CLUSTER...)

A(2,3) --> 2=X-AXIS 3=Y-AXIS
B(8,9) --> 8=X-AXIS 9=Y-AXIS
C(6,7) --> 6=X-AXIS 7=Y-AXIS

SO , X-AXIS = 2+8+6 = 16 /3 =6 (LET)
SO , Y-AXIS = 3+9+7 = 19/3 =7(LET)
SO THIS (6,7) IS THE CENTROID OF THIS CLUSTOR...CENTROID IS THE AVG...
THEN ALL VALUES LIKE SPENDING(2,8,6,5)FOR X-AXIS AND INCOME(3,9,7,5)FOR Y-AXIS ACCORDING TO (6,7) IS THE CENTROID IS TAKEN IN ONE CLUSTER...
AND THE VALUES SPENDING(100,200) AND INCOME(500,600) ARE TAKEN IN DIFFERENT CLUSTER...'''
```
