```python
import pandas as pd 
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LinearRegression
```

```python
df=pd.read_csv('iphone.csv')
df.head()
```

```python
x=df[['MODEL','BASIC','BASIC PLUS','PLUS','PRO','PRO MAX','YEAR']]
y=df['PRIZE']
```

```python
model=LinearRegression()
```

```python
scores=cross_val_score(model,x,y,cv=5,scoring='r2')
scores
```

```python
# The data is divided into 5 equal parts, so each fold has 20 records.
# Run 1 → Test: Fold 1 | Train: 2,3,4,5
# Run 2 → Test: Fold 2 | Train: 1,3,4,5
# Run 3 → Test: Fold 3 | Train: 1,2,4,5
# Run 4 → Test: Fold 4 | Train: 1,2,3,5
# Run 5 → Test: Fold 5 | Train: 1,2,3,4

# So every record gets used for testing exactly once, while also being used for training in the other runs.
```

```python
print(scores.mean())
print(scores.std())
```

```python
# cross validation is used for it will divide the data into equal parts and cv=5 then it will train 80% 
# data for 1st fold then different 80% data for 2nd fold like this every fold it will 
# create different data so we have 5 different r2 scores and so our model 
# don't get accusumed only by just similar that train test data...
```
