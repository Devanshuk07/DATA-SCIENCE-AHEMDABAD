```python
'''**CI/CD pipeline is an automated process used to build, test, and deploy an application whenever developers make changes to the code.**

**CI** stands for **Continuous Integration** — whenever developers push code to a repository like GitHub, 
then github action automatically built and test the code to check whether everything is working correctly.

**CD** stands for **Continuous Delivery/Deployment** — after the code passes the tests,
it can be automatically delivered or deployed to the production server.

For example:

**Developer writes code → Pushes to GitHub → CI runs tests → Build is created → CD deploys the application.**

This helps developers release changes faster and reduces manual errors.
'''
```

```python
'''CI / CD PIPELINE:--(CONTINUOUS INTEGRATION AND CONT DEPLOYMENT- PIPELINE)
WHAT IS CI/CD PIPELINE :--
Suppose you have your project on GitHub. You change main.py and push it:
You change code
      ↓
git push
      ↓
GitHub
      ↓
Automatically test/build code

Instead of manually checking/building everything every time, the CI pipeline does it automatically.
If something fails ❌, deployment can be stopped.

If everything passes ✅, it can continue to CD.
---------------------------------------------------------------------------------------------------------------------
1. WITHOUT CI/CD.
MLOPS - UPLOADING  TO SERVER FOR AI USEING.
DEVOPS - USED TO UPLOAD TO SERVER IN MOBILE.

- FULL CODE PUSH TO REPO THAT IS GITHUB REPO - THAT REPO IS PASTE TO INTEGRATION SERVER.THIS NORMAL UPLOAD IS CALLED WITHOUT CI/CD.

PROBLEM WITH WITHOUT - IF WE WROTE CODE AND MINOR ERROR IN IT THEN IT WILL WENT TO SERVER.SO NO TESTING IN BTW.
----------------------------------------------------------------------------------------------------------------------
2. WITH CI/CD PIPELINE :-
PC-PUSH-GITHUB-GITHUB REPO-CI/CD 1JENKINS 2-GITHUB ACTION - INTEGRATED SERVER.

WHY TO USE ADVANTAGE :- NEW CODE WILL BE TESTED BY PIPELINE THEN AFTER TESTING WILL CREATE A BUILD.
IT WILL MEARGE ALL SMALL FILE THEN IT WILL CREATE FROM IT A SINGLE-WHOLE FILE.
TILL YOUR CODE IS NOT ERROR FREE IT WILL NOT SEND TO SERVER.


NOW WE HAVE TO DEPLOY THE MODEL TO LIVE SERVER:-
TILL NOW - LOCAL PC.
BUT WHEN WE DEPLOY WE GET URL THAT WE CAN SHARE ANYWHERE.

THE FULL PROCESS OF DEPLOYMENT IS CALLED CI/CD PIPELINE.
WE NEED THIS THINGS:-

1. REQUIREMNET.TXT - LIBRARIES/ALGORITH USED IN CODE WILL BE WRITTEN IN ONE FILE.THE THING WE WRITE IN FIRST LINE LIKE IMPORT PANDAS AND ALL THEN WRITE IT IN NEW TXT FILE LIKE PANDAS,PYTHON,NUMPY ETC.
THEN THIS WILL INSTALL ALL THE LIBRARIES TO OUR SERVER JUST BY NAME WE WRITTEN IN TXT ---> REQUIREMENT.TXT FILE.

2. GITHUB - 
3. GITHUB-ACTION/JENKINS -    FLOW :--- CODE - TEST - BUILD(GITHUB/JENKINS) - SERVER
    1. INFRASTRUCTURE - JENKINS (YOU HAVE TO MAKE INFRASTRUCTURE)
    2. USE - GITHUB-ACTION / GITHUB (IT IS READY-MADE BY GITHUB AND THAT YOU HAVE TO USE)

4. DOCKER/KUBENETES :- THIS IS CONTAINER THAT IS USED TO STORE TOO MANY THINGS IN ONE PLACE. THIS THEY CALL IT IMAGE.
    - IF YOU WANNA TO MAKE SINGLE -SINGLE CONTAINER THEN DOCKER IS BEST.
    - IF YOU WANNA TO MAKE PROCESS SMOOTH THEN KUBENETES IS USED. 500 CONTAINER TO ONE IS KUBENETES.
    
5. SERVER // AWS,AZURE,VERCEL,NETLIFY,RENDOR,RAILWAY --> MULTI SERVER.
```
