```python
'''Download a publicly available dataset of your choice (for example, a Flipkart product ratings dataset or Spotify song features) and 
split it into features (X) and target (y) for a regression problem.'''

import pandas as pd

# Load dataset
data = pd.read_csv("winequality-white.csv", sep=";")

# See the data
print(data.head())

# Separate features and target
x = data.drop("quality", axis=1)          # only the quality column is removed from x. Everything else remains exactly as it is.axis=1 means "work with columns."

y = data["quality"]

print("X:")
print(x.head())

print("y:")
print(y.head())
```

```python
'''Fit a Lasso regression model (L1 regularization) using scikit-learn's Lasso class on your dataset, and print out the coefficients for each feature.'''

from sklearn.linear_model import Lasso
model=Lasso(alpha=0.1)                 # alpha controls how strongly we want to keep the coefficients small while still keeping predictions close to actual y.BCZ BIG CO-EFFICIENT CAN LET TO BIG CHANGE AND MAKES MODEL OVERFITTING...
model.fit(x,y)
pre=model.predict(x)
for features,coefficient in zip(x.columns,model.coef_):
    print(features,":",coefficient)

'''WHAT IS CO-EFFICIENT, MODEL WHEN TRAINING IT WILL TAKE SOME VALUES FOR EVERY COLUMN(FEATURES) AND TRY TO BEST FIT IT WITH THE 
COLUMN SO WE CAN GET ACCURATE PREDICTION = Y '''
```

```python
'''Fit a Ridge regression model (L2 regularization) using scikit-learn's Ridge class on the same dataset, and 
compare the coefficients to those from the Lasso model.<br><br><em><strong>Hint:</strong> Look for features 
where Lasso sets coefficients exactly to zero but Ridge does not.</em>'''

from sklearn.linear_model import Ridge

ridge_model = Ridge(alpha=0.1)

ridge_model.fit(x, y)

for feature, coefficient in zip(x.columns, ridge_model.coef_):
    print(feature, ":", coefficient)
```

```python
'''Suppose you want to build a feature selector for a Zomato-style restaurant rating predictor. Use Lasso regression to select 
the most important features from your dataset, and list which features are kept (non-zero coefficients) and which are dropped.'''

from sklearn.linear_model import Lasso

data=pd.read_excel('zomato.xlsx')
data.head()

x = data[["no_of_reviews"]]
y = data["restaurant_rating"]

print(x.head())
print(y.head())

model = Lasso(alpha=0.1)

model.fit(x, y)

print("Coefficient:", model.coef_[0])

if model.coef_[0] != 0:
    print("no_of_reviews → KEPT")
else:
    print("no_of_reviews → DROPPED")


#model.coef_      → [0.25]   ← array
#model.coef_[0]   →  0.25    ← actual value
```
