```python
'''Given two arrays: one with actual delivery times and one with predicted delivery times for 10 Zomato orders, 
calculate the Mean Squared Error (MSE) and Root Mean Squared Error (RMSE) using Python.'''
import pandas as pd 
import numpy as np 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,root_mean_squared_error

acc=np.array([10,20,15,22,35,17,19,22,21,50]).reshape(-1,1)      # here -1 means "NumPy, you calculate how many rows are needed." and 1 for column.
pre=np.array([10,15,12,20,40,20,20,20,20,60])

acc_train,acc_test,pre_train,pre_test=train_test_split(
    acc,
    pre,
    train_size=0.80,
    random_state=42
)
```

```python
model=LinearRegression()
model.fit(acc_train,pre_train)
pre=model.predict(acc_test)
pre
```

```python
mse=mean_squared_error(pre_test,pre)
mse
```

```python
rmse=root_mean_squared_error(pre_test,pre)
rmse
```

```python
'''Create a residual plot for a dataset of actual vs predicted movie ratings (scale 1-5) for 20 movies, using matplotlib. 
Label the axes and briefly describe what the plot tells you about your model's errors.'''

import numpy as np
import matplotlib.pyplot as plt

# Actual ratings
actual = np.array([4.5, 3.2, 4.8, 2.5, 3.9,
                   4.2, 1.8, 3.5, 4.7, 2.9,
                   3.8, 4.4, 2.2, 3.1, 4.9,
                   2.7, 3.6, 4.1, 1.9, 4.6])

# Predicted ratings
predicted = np.array([4.2, 3.5, 4.6, 2.8, 3.7,
                      4.0, 2.1, 3.3, 4.5, 3.2,
                      3.6, 4.1, 2.5, 3.0, 4.7,
                      2.9, 3.4, 4.3, 2.2, 4.4])

# Calculate residuals
residuals = actual - predicted

# Residual plot
plt.scatter(predicted, residuals)
plt.plot(predicted, residuals)

plt.xlabel("Predicted Movie Rating")
plt.ylabel("Residual (Actual - Predicted)")
plt.title("Residual Plot: Movie Ratings")

plt.show()
```

```python
'''Compare two regression models for predicting Spotify song popularity: calculate MSE, RMSE, MAE, and R² 
for both models using sklearn, then identify which model is likely overfitting based on the results.
<br><br><em><strong>Hint:</strong> Overfitting often shows very low error on training data but much higher error on test data.</em>'''

from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np


# Model A
y_train_a = np.array([72, 65, 80, 55, 90, 68, 75, 85, 60, 78])
y_train_pred_a = np.array([71, 66, 79, 56, 89, 69, 74, 84, 61, 77])

y_test_a = np.array([70, 62, 88, 58, 76])
y_test_pred_a = np.array([68, 65, 80, 70, 82])


# Model B
y_train_b = np.array([72, 65, 80, 55, 90, 68, 75, 85, 60, 78])
y_train_pred_b = np.array([72, 64, 81, 54, 89, 67, 76, 86, 59, 79])

y_test_b = np.array([70, 62, 88, 58, 76])
y_test_pred_b = np.array([69, 64, 85, 61, 74])


# ================= MODEL A =================

# Training
mse_train_a = mean_squared_error(y_train_a, y_train_pred_a)
rmse_train_a = np.sqrt(mse_train_a)
mae_train_a = mean_absolute_error(y_train_a, y_train_pred_a)
r2_train_a = r2_score(y_train_a, y_train_pred_a)

# Test
mse_test_a = mean_squared_error(y_test_a, y_test_pred_a)
rmse_test_a = np.sqrt(mse_test_a)
mae_test_a = mean_absolute_error(y_test_a, y_test_pred_a)
r2_test_a = r2_score(y_test_a, y_test_pred_a)


# ================= MODEL B =================

# Training
mse_train_b = mean_squared_error(y_train_b, y_train_pred_b)
rmse_train_b = np.sqrt(mse_train_b)
mae_train_b = mean_absolute_error(y_train_b, y_train_pred_b)
r2_train_b = r2_score(y_train_b, y_train_pred_b)

# Test
mse_test_b = mean_squared_error(y_test_b, y_test_pred_b)
rmse_test_b = np.sqrt(mse_test_b)
mae_test_b = mean_absolute_error(y_test_b, y_test_pred_b)
r2_test_b = r2_score(y_test_b, y_test_pred_b)


# ================= RESULTS =================

print("MODEL A")
print("Train MSE :", mse_train_a)
print("Train RMSE:", rmse_train_a)
print("Train MAE :", mae_train_a)
print("Train R2  :", r2_train_a)

print("\nTest MSE :", mse_test_a)
print("Test RMSE:", rmse_test_a)
print("Test MAE :", mae_test_a)
print("Test R2  :", r2_test_a)


print("\n\nMODEL B")
print("Train MSE :", mse_train_b)
print("Train RMSE:", rmse_train_b)
print("Train MAE :", mae_train_b)
print("Train R2  :", r2_train_b)

print("\nTest MSE :", mse_test_b)
print("Test RMSE:", rmse_test_b)
print("Test MAE :", mae_test_b)
print("Test R2  :", r2_test_b)

'''why we take 2 mse for model a and 2 for model b so we can determine that one is for when model is training 
and one is for when model is predicting from new data so that differnet score will tell us is our model overfitting or not....

It performs extremely well on training data, but performance becomes much worse on unseen test data.
Model A is clearly overfitting.
Model B → Generalizes much better ✅
'''
```
