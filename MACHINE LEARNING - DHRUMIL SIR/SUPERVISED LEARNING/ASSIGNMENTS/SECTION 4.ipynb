```python
'''Download a small dataset of IPL cricket player stats (e.g., runs, wickets, matches, strike rate) and apply StandardScaler 
from scikit-learn to scale all numeric features. Save the scaled data to a new CSV file named ipl_scaled.csv.'''
```

```python
import pandas as pd 
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_excel("ipl.xlsx")
df.head()
```

```python
x=df[['Matches','Runs','Wickets','Strike_Rate']]
```

```python
scaler=StandardScaler()
x_scale=scaler.fit_transform(x)
x_scale
```

```python
# now fit_transform(x) returns a NumPy array, so first convert it back to a DataFrame:

x_scale2=pd.DataFrame(
    x_scale,                           # this is for data to be taken from x_scale=scaler.fit_transform(x)
    columns=x.columns                  # you have to define the columns so it should be the same that we define in x variable upper.
)
```

```python
x_scale2.to_csv('ipl_scaled.csv', index=False)
```

```python
'''Using a dataset of Flipkart product ratings (features: price, rating, number of reviews, discount), apply MinMaxScaler 
so that all features are scaled between 0 and 1. Print the minimum and maximum value for each column before and 
after scaling.<br><br><em><strong>Hint:</strong> Use pandas to load the data and scikit-learn's MinMaxScaler for scaling.</em>'''
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
df=pd.read_csv("flipkart.csv")
print(df.head())
print(df.min())
print(df.max())
```

```python
y=df[['Price','Rating','Number_of_Reviews','Discount']]
```

```python
scaler=MinMaxScaler()
scl_data=scaler.fit_transform(y)
print(scl_data)

print(scl_data.min())
print(scl_data.max())
```
