```python
import pandas  as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np
```

```python
file=pd.read_csv('iphone.csv')
file
```

```python
X=file[["MODEL","BASIC","BASIC PLUS","PLUS","PRO","PRO MAX","YEAR"]]
y=file["PRIZE"]
```

```python
X_train,X_test,y_train,y_test=train_test_split(
    X,y,train_size=0.80,random_state=42
)
```

```python
model=LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
model.intercept_
```

```python
model.coef_
```

```python
y=model.predict(X_test)

y
```

```python
result=pd.DataFrame({
    'excel':y_test.values,
    'predict':y
})

result
```

```python
from sklearn.metrics import root_mean_squared_error
mse=mean_squared_error(y_test,y)
rmse=root_mean_squared_error(y_test,y)
rsquare=r2_score(y_test,y)
rsquare
```

```python
a=float(input('ENTER THE MODEL NUMBER HERE :- '))
b=input('ENTER THE MODEL VARIANT(BASIC / BASIC PLUS / PLUS / PRO / PRO MAX) :- ')
c=float(input('ENTER THE YEAR :- '))

basic=0
basic_plus=0
plus=0
pro=0
pro_max=0

if b.upper()=='BASIC':
    basic=1
elif b.upper()=='BASIC PLUS':
    basic_plus=1
elif b.upper()=='PLUS':
    plus=1
elif b.upper()=='PRO':
    pro=1
elif b.upper()=='PRO MAX':
    pro_max=1
else:
    print('INVALID VARIANT !!')


prediction_input=model.predict([[
    a,basic,basic_plus,plus,pro,pro_max,c
]])

print("PREDICTED PRICE:", prediction_input[0])
```
