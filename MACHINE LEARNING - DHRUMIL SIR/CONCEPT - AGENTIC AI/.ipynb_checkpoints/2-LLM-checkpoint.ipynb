```python
'''WHAT IS LLM ?
-LLM STANDS FOR LARGE LANGUAGE MODEL.
-In simple words, an LLM is an AI model that can understand text and generate text in response to our instructions.

EXAMPLES OF DIFFERNET LLM :-
-OPEN AI
-GOOGLE GEMINI
-CHAT GPT
-GROQ
(OpenAI and Google create language models, while Groq is primarily an inference platform that runs models created by other organizations.)
```

```python
'''HOW DOES LLM WORKS ?

1.USER INPUT
You ask: “What is Python?”

2.LLM PROCESS THE INPUT
It uses patterns learned during training to determine a suitable response.

3.GENERATED OUTPUT
It produces text in response to your question.


An LLM does not automatically know what's inside your private TXT files.
That's why our Study AI Model needs a retrieval system to find the original text.
```

```python
'''WHAT IS TEMPERATURE ?

Temperature controls how predictable or varied the model's generated text can be.
- Lower temperature: Usually more predictable responses.
- Higher temperature: Usually more varied responses.

Temperature does not directly control answer length.
----------------------------------------------------------------------------------------------------------
EXAMPLE :-
Imagine you ask an AI this question:

“Give me a name for a pet cat.”

Low temperature: the AI may suggest a common cat name, such as “Milo.”            ---> Medium temperature — 0.7

High temperature: the AI may suggest a more unusual name, such as “Captain Moonbeam.”          ---> High temperature — 1.5

----------------------------------------------------------------------------------------------------------
If we set a low temperature, what would you generally expect?
-More predictable answers
```

```python
'''WHAT IS PROMPT ?
-A prompt is the instruction or question we give to an AI model.

Examples:
- “What is Python?”
- “Explain the purpose of LangChain.”
- “Find the original study text related to this question.”

The prompt tells the AI what we want it to do.
```

```python
'''WHAT IS PROMPT TEMPLATE ?
-A Prompt Template is a reusable format for creating prompts.

EXAMPLE:-
Imagine you want to ask questions about many Python topics. Instead of writing a completely new instruction every time, you can use one template.
----------------------------------------------------------------------------------------------------------
Our template
Explain the following Python topic: {topic}

Example 1
Explain the following Python topic: Variables       (HERE VARIABLE IS PLACE HOLDER)

Example 2
Explain the following Python topic: Functions       (HERE FUNCTIONS IS PLACE HOLDER)

Example 3
Explain the following Python topic: Loops           (HERE LOOPS IS PLACE HOLDER)

Notice that only {topic} changes. The rest of the instruction stays the same.
----------------------------------------------------------------------------------------------------------
Why do we need PromptTemplate?
It saves us from writing the same instruction again and again.
- Prompt: The actual question or instruction.
- PromptTemplate: A reusable format with placeholders.
- Placeholder: The part we replace with different information, such as {topic}.

important point: a PromptTemplate does not generate an answer by itself. 
It only prepares the prompt. We need to connect it to a language model if we want an AI-generated response.

LET SAY FOR OUR STUDY-BOT AI MODEL WE USE THIS TEMPLATE...
Find the original study text related to this question: {question}        ---> HERE ONLY {QUESTION} CHANGES....
```
