```python
'''WHAT IS CHAINS ?
-A chain connects multiple steps so that the output of one step becomes the input of the next step.

EXAMPLE:-
Step 1: User input
Topic = Python loops

Step 2: PromptTemplate
Creates the instruction using the topic

Step 3: LLM
Generates an explanation

Final output
Explanation of Python loops

Here, the PromptTemplate and LLM are connected into one workflow. That is the basic idea of a chain.
SO EEVRYTIME WHEN WE ASK ANYTHING TO LLM IT WILL CREATE A PROMPT TEMPLATE AND THEN THAT WILL BE GIVEN TO LLM AND THEN OUR ANSWER WILL BE GENERATED....
```

```python
'''WHAT IS THE | OPERATOR ?
-In LangChain Expression Language (LCEL), the | operator connects runnable components into a sequence.

SEE THIS THE WORK OF CHAINS...
chain = prompt | llm         ---> WE READ IT AS “then pass the result to the next step.” SO | PASSES THE RESULTS... TO NEXT STEPS...

The pipe symbol does not mean “or” here. It connects the workflow steps.
```

```python
'''How do we connect a PromptTemplate and an LLM?

Here is a simple example using LangChain.
-----------------------------------------------------------------------------
from langchain_core.prompts import PromptTemplate                  PromptTemplate.from_template() creates a reusable prompt.

prompt = PromptTemplate.from_template(
    "Explain {topic} in simple words."                           {topic} is a placeholder.
)

chain = prompt | llm

response = chain.invoke({"topic": "Python functions"})
-----------------------------------------------------------------------------
In this example:
- PromptTemplate.from_template() creates a reusable prompt.
- {topic} is a placeholder.
- llm represents your configured language model.
- prompt | llm connects the prompt to the model.
-----------------------------------------------------------------------------
OUR STUDY AI MODEL USES THIS CODE...

from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template(
    "Find the original study text related to this question: {question}"
)

result = prompt.format(
    question="How does Python execute a program?"
)

print(result)
```

```python
'''WHAT IS INVOKE ?
-invoke() is a method used to run a runnable component or chain with an input.
----------------------------------------------------------------------------------------------------------------------------
EXAMPLE :-

Imagine you have a calculator
You want to calculate 10 + 20.
- You give the calculator an input: 10 + 20
- You press the calculate button.
- The calculator gives you the result: 30.

In LangChain, invoke() is like pressing the calculate button. It tells a component or chain to run using the input you provide.
---------------------------------------------------------------------------------------------------------------------------
responce = chain.invoke({"topic": "Python loops"})
responce = chain.invoke({"topic": "Python functions"})

It supplies the topic, runs the connected workflow, and stores the returned result in response.
---------------------------------------------------------------------------------------------------------------------------
```
