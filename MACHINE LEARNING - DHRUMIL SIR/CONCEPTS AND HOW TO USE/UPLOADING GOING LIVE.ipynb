```python
'''CRAETING THE MAIN.PY FILE :--
-THIS IS USED AS API FOR MODEL AND CLIENT(WEB-PAGE CLIENT)QUESTIONING AND ANSWERS.

*************** CODE ******************
from fastapi import FastAPI(write-as-it-is)            ---->You bring FastAPI's functionality into your Python program.
import joblib                     ----> this is used to save/load your saved model.first you save it with joblib.dump(model,'file_name.pkl') then you can access it by joblib.load('same_file_name')

app = FastAPI()     --->This creates your FastAPI application.that we store in app(variable) which we can use later...Later, Uvicorn will run this app.

model = joblib.load("iphone_model.pkl")          ---> here we load your trained model that we have saved in iphone_model.pkl

#its just normal python function used so that when in url some-one type / it will give them index.html file. MAIN THING --> url has 2 main things 1.for client  2.for ourselfs that for ourselfs we dont show that to clients and its just has /(something)with it to differentiate from client.
@app.get("/")                            ----> @app is the variable we took in fastAPI() then get is for GET request we read in python API & this (/) creates a main url,send them the index.html webpage
def home():                              ----> This creates a Python function called home.FastAPI will run this function whenever someone visits /. so if your main website is this https://yourwebsite.com/ then you '/' part is this --->yourwebsite.com/
    return FileResponse("index.html")    ----> FileResponse is used to return a file.

# Predict price ---> first i had like home so now i have index.html file in front of me and its asking for model_no,variant,year when i fill that and execute it,it will then make url of http://127.0.0.1:8000/predict which first was http://127.0.0.1:8000 this and a predict answer is generated by api(by model) then given back to html price predict page.
@app.get("/predict")                     
def predict(model_no: int, variant: str, year: int):
    basic = 0                            ----> initially all are 0.
    basic_plus = 0
    plus = 0
    pro = 0
    pro_max = 0

    variant = variant.lower().strip()      ---> strip() used to Removes unnecessary spaces from the beginning and end.

    if variant == "basic":
        basic = 1

    elif variant == "basic_plus":
        basic_plus = 1

    elif variant == "plus":
        plus = 1

    elif variant == "pro":         ---> let user input pro then,,
        pro = 1

    elif variant == "pro_max":
        pro_max = 1

    input_data = [[                ---> input_data = [[
        model_no,                                     17,
        basic,                                         0,
        basic_plus,                                    0,
        plus,                                          0,
        pro,                                           1,
        pro_max,                                       0,
        year                                           2027
    ]]                                                 ]]

    prediction = model.predict(input_data)          ---> send this data to ML model.

    return {
        "price": float(prediction[0])    ---> to get answer in float.
    }
'''
```
