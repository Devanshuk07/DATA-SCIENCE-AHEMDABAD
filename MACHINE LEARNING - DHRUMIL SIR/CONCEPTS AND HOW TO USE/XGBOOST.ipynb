````python
'''WHAT IS DIFFERENCE BETWEEN BAGGING AND BOOSTING?
OR
WHAT IS DIFFERENCE BETWEEN RANDOM FOREST AND XGBOOST?

---------------------------------------------------------------------------------------------------------------------------------

RANDOM FOREST — BAGGING

Random Forest mainly works using BAGGING (Bootstrap Aggregating).

In Random Forest, multiple Decision Trees are created and the trees are INDEPENDENT of each other.

Example:

Tree 1 → PASS
Tree 2 → PASS
Tree 3 → FAIL

Final Prediction → PASS

For classification, the final prediction is generally based on MAJORITY VOTING.

RANDOM FOREST = INDEPENDENT TREES + BAGGING

---------------------------------------------------------------------------------------------------------------------------------

BOOSTING — XGBOOST

XGBoost = Extreme Gradient Boosting

XGBoost also generates multiple trees, but unlike Random Forest, the trees are built SEQUENTIALLY.

Suppose our actual data is:

A = PASS
B = PASS
C = FAIL

TREE 1:

A = PASS
B = FAIL
C = PASS

Tree 1 made errors.

TREE 2:

Tree 2 is built to help correct the errors made by the previous prediction.

A = PASS
B = PASS
C = PASS

TREE 3:

Tree 3 again tries to correct the remaining error.

A = PASS
B = PASS
C = FAIL

So:

Tree 1
↓
Correct Errors
↓
Tree 2
↓
Correct Remaining Errors
↓
Tree 3

IMPORTANT:

We DO NOT take the prediction from only the last tree.

XGBoost combines the contribution of ALL TREES to make the final prediction.

XGBOOST = SEQUENTIAL TREES + BOOSTING

---------------------------------------------------------------------------------------------------------------------------------

WHAT IS GRADIENT DESCENT?

Suppose:

Actual Marks = 80

XGBoost starts with an INITIAL PREDICTION.

Suppose:

Initial Prediction = 60

Difference/Error:

Actual - Prediction

80 - 60 = 20

Loss function measures how wrong our prediction is.

For example:

Loss = (Actual - Prediction)²

Loss = (80 - 60)²

Loss = 400

400 is the LOSS VALUE.

It tells us how wrong the model's prediction is.

The loss value 400 itself is NOT directly added to the prediction.

---------------------------------------------------------------------------------------------------------------------------------

HOW TREES IMPROVE THE PREDICTION

Formula:

NEW PREDICTION = OLD PREDICTION + LEARNING RATE × TREE CORRECTION

Suppose:

Actual Marks = 80
Initial Prediction = 60
Learning Rate = 0.1

Suppose Tree 1 learns a correction of 20.

TREE 1:

New Prediction = 60 + 0.1 × 20

```
           = 60 + 2

           = 62
```

Now our prediction moved from:

60 → 62

So it became closer to the actual value 80.

---------------------------------------------------------------------------------------------------------------------------------

TREE 2

Previous Prediction = 62

Remaining Error:

80 - 62 = 18

New Prediction = 62 + 0.1 × 18

```
           = 62 + 1.8

           = 63.8
```

So:

Initial Prediction = 60

Tree 1 → 62

Tree 2 → 63.8

Tree 3 → Closer to 80

Tree 4 → Closer to 80

...

Each new tree tries to reduce the remaining error.

---------------------------------------------------------------------------------------------------------------------------------

GRADIENT

For squared error, a simplified gradient is:

GRADIENT = 2 × (PREDICTION - ACTUAL)

Suppose:

Prediction = 90
Actual = 80

Gradient = 2 × (90 - 80)

Gradient = +20

Positive Gradient means our prediction is TOO HIGH.

So the correction should move the prediction DOWN.

---------------------------------------------------------------------------------------------------------------------------------

Suppose:

Prediction = 60
Actual = 80

Gradient = 2 × (60 - 80)

Gradient = -40

Negative Gradient means our prediction is TOO LOW.

So the correction should move the prediction UP.

Therefore:

NEGATIVE GRADIENT → PREDICTION NEEDS TO MOVE UP

POSITIVE GRADIENT → PREDICTION NEEDS TO MOVE DOWN

The purpose is to move toward a prediction that gives LOWER LOSS.

---------------------------------------------------------------------------------------------------------------------------------

IMPORTANT PARAMETERS IN XGBOOST

1. n_estimators

It controls the number of  trees.

Example:

n_estimators = 5

Means XGBoost will build 5 boosting trees.

---------------------------------------------------------------------------------------------------------------------------------

2. max_depth

It controls the MAXIMUM DEPTH of each tree.

Example:

max_depth = 3

It means each tree can have a maximum depth of 3.

It does NOT mean every tree must have exactly 3 questions.

It means a path in the tree can have UP TO 3 split levels.

---------------------------------------------------------------------------------------------------------------------------------

3. learning_rate

It controls how much each new tree's correction contributes to the prediction.

Example:

learning_rate = 0.1

Smaller Learning Rate:

Each tree makes a SMALLER correction.

Larger Learning Rate:

Each tree makes a LARGER correction.

0.1 is a commonly used starting value, but learning_rate does NOT always have to be 0.1.

---------------------------------------------------------------------------------------------------------------------------------

LEARNING RATE VS NUMBER OF TREES

OPTION 1:

LEARNING RATE = LESS
NUMBER OF TREES = MORE

This is commonly preferred because the model learns in smaller steps.

It can help reduce the chance of overfitting compared with taking very large boosting steps.


OPTION 2:

LEARNING RATE = MORE
NUMBER OF TREES = LESS

Each tree makes a larger correction.

If the learning rate is too high, there can be a greater chance of overfitting.

---------------------------------------------------------------------------------------------------------------------------------'''
````
