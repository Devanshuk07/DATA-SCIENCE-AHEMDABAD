```python
'''
RANDOM FOREST (ALGORITHM)
------------------------------------------------------------

DECISION TREE:
→ We use ONE single decision tree.

RANDOM FOREST:
→ Random Forest creates MULTIPLE Decision Trees.
→ It combines the output of those trees to make the FINAL prediction.

Example:
Suppose we create 5 trees.

Tree 1 → PASS
Tree 2 → PASS
Tree 3 → FAIL
Tree 4 → PASS
Tree 5 → FAIL

Final Output → PASS

Because PASS got maximum votes (3 out of 5).

------------------------------------------------------------
RANDOM FOREST IS USED FOR:

1. CLASSIFICATION
2. REGRESSION

CLASSIFICATION:(VOTES)
→ Final decision is generally based on the predictions/votes
  from multiple trees.

Example:
5 Trees → Pass, Pass, Fail, Pass, Fail
Final → PASS


REGRESSION:(AVERAGE)
→ Final prediction = AVERAGE of predictions of all trees.

Example:
Tree 1 → ₹50,000
Tree 2 → ₹55,000
Tree 3 → ₹52,000

Final Prediction = Average
                 = ₹52,333

------------------------------------------------------------
WHY IS IT CALLED "RANDOM" FOREST?

There are TWO important types of randomness:

1. RANDOM ROWS → Bootstrap Sampling
2. RANDOM FEATURES → Random Feature Selection

------------------------------------------------------------
1. BOOTSTRAP SAMPLING

Suppose our original dataset has 6 rows:

| Row | Study Hours | Marks | Result |
| --- | ----------: | ----: | ------ |
| A   |           2 |    40 | Fail   |
| B   |           5 |    65 | Pass   |
| C   |           3 |    45 | Fail   |
| D   |           7 |    80 | Pass   |
| E   |           6 |    75 | Pass   |
| F   |           1 |    30 | Fail   |

You tell Random Forest:
n_estimators = 3      ,,,,  So it needs to create 3 trees.

What Bootstrap Sampling does ?

Instead of giving exactly the same 6 students to every tree, Random Forest randomly picks 6 rows separately for each tree.

For example:
Original:
A B C D E F

Tree 1 gets:
A C C D E F

Tree 2 gets:
B B C D D A

Tree 3 gets:
F A A C E E

Now your question will probably be:
"Why can C come twice? Once C is selected, shouldn't it be removed?"

Imagine a box containing:
[A] [B] [C] [D] [E] [F]

Random Forest does this:
Pick one → B
Put B BACK into the box 🔄

Pick again → B
Put B BACK 🔄

Pick again → C
Put C BACK 🔄

Pick again → D
Put D BACK 🔄

Pick again → D
Put D BACK 🔄

Pick again → A
That's why B and D can repeat, while E and F might not be selected at all.

Why does Random Forest do this?
-Because we want the trees to be different.If all 100 trees received exactly the same rows and 
behaved the same way, we could end up with many very similar trees. Random sampling helps create diversity among them.
------------------------------------------------------------
2. RANDOM FEATURE SELECTION

Suppose we have:

FEATURES--> INPUT(X):
Study_Hours
Attendance
Previous_Marks
Sleep_Hours

OUTPUT (Y):
Pass/Fail

Random Forest DOES NOT simply make:
Tree 1 = Study_Hours
Tree 2 = Attendance
Tree 3 = Previous_Marks

❌ THIS UNDERSTANDING IS WRONG.

Instead, while creating each tree, Random Forest considers
a RANDOM SUBSET OF FEATURES at each split.

Example:
                 Tree 1
                   |
          [Split/Node 1]
         Study Hours > 4?            DEPTH=0
             /        \
           Yes         No
            |           |
       [Split 2]    [Split 3]         DEPTH=1
     Marks > 60?   Attendance > 70?
        /   \          /   \
      Pass  Fail     Pass  Fail        DEPTH=2
HERE,max_depth=2 → each tree can grow to a maximum depth of 2 levels
BUT IF WE GIVE N_ESTIMATORS=3 THEN IT WILL CREATE 3 TREES AND WITH MAX_DEPTH=2 THEN LIKE UPPER DIAGRAM DEPTH WILL BE 2 OF 1 EACH TREE...

So different trees can become different from each other.

------------------------------------------------------------
NUMBER OF TREES:

n_estimators = number of trees.

Example:

RandomForestClassifier(n_estimators=100)

Means:
→ Create 100 Decision Trees.

------------------------------------------------------------
MAX_DEPTH:

Just like Decision Tree, we can control tree depth.

Example:

RandomForestClassifier(
    n_estimators=100,
    max_depth=3
)

Here:
n_estimators = 100 → 100 trees
max_depth = 3      → Maximum depth of EACH tree

------------------------------------------------------------
IMPORTANT CORRECTION:

bootstrap=False DOES NOT make Random Forest a Decision Tree.

bootstrap=True
→ Each tree gets a bootstrap sample of rows.

bootstrap=False
→ Each tree uses the whole training dataset.

BUT Random Forest still has MULTIPLE TREES and still uses
random feature selection.

So it is STILL RANDOM FOREST.

If you want only one Decision Tree, use:

DecisionTreeClassifier()

------------------------------------------------------------
OVERFITTING:

DECISION TREE:
→ Higher chance of overfitting if allowed to grow too much.
→ It may learn the training data too specifically.
  (Simple language: "Ratta marna")

RANDOM FOREST:
→ Usually reduces overfitting compared with a single
  Decision Tree because it combines many different trees.

------------------------------------------------------------
'''
```

```python
'''YOU HAVE TO IMPORT:

from sklearn.ensemble import RandomForestClassifier

OR

from sklearn.ensemble import RandomForestRegressor


WHY "ENSEMBLE"?

ENSEMBLE MEANS COMBINING MULTIPLE DECESION TREES TO MAKE ONE FINAL PREDICTION.

RANDOM FOREST CREATES MULTIPLE DECISION TREES SEPARATELY
AND COMBINES THEIR PREDICTIONS.

CLASSIFICATION → COMBINES/VOTES ON TREE PREDICTIONS
REGRESSION     → TAKES AVERAGE OF TREE PREDICTIONS

THAT IS WHY RANDOM FOREST COMES UNDER sklearn.ensemble.
'''
```
