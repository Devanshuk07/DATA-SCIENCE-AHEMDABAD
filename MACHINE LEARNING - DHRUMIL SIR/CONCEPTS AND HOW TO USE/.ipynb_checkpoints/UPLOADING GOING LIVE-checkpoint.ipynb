```python
'''5TH FILE IS DOCKER FILE:--(ALWAYS CREATE IN NOTEPAD THEN SAVE AS Dockerfile(same) then all files. DONT ADD ANYTHING BACK.)
WHAT IS DOCKER FILE AND WHY WE NEED IT ?
-Think of a Dockerfile as instructions for creating a complete package/environment in which your application can run.

-server needs to know things like: Which Python version? Which packages? Where are the files? What command starts the application?

That's what your Dockerfile describes:

IT CONTAINS CODE :--
----------------------------------------------------------------------------------
FROM python:3.12-slim                  #THIS YOU HAVE TO GIVE PYTHON VERSION ALWAYS GIVE UPPER THAN 3.10 SO IT WORKS.

WORKDIR /app                           # Here /app is simply a folder/directory inside the Docker container.ou could even call the folder something else: WORKDIR /project

COPY . .                               # Copy your project files into /app.
 
RUN pip install --no-cache-dir -r requirements.txt        # Install your Python packages from requirements.txt.

CMD ["sh", "-c", "uvicorn main:app --host 0.0.0.0 --port $PORT"]                #Start your FastAPI application.
-----------------------------------------------------------------------------------
AFTER THIS .....
"Start main.py's FastAPI app with Uvicorn"
   ↓
Docker Image is created
   ↓
Container runs that image.'''
```

```python
''' 6)  .YML FILE....
-.yml is basically the instruction file for GitHub Actions.

The Dockerfile tells Docker how to build/run your application.

The YML file tells GitHub what to do automatically when you push new code.
-THIS YML FILE HELPS TO DO - WHEN YOU CHANGE ANYTHING TO YOUR CODE FILE IT AUTOMATICALLY REFLECTS IT TO YOUR ORIGINAL WEBSITE/SERVER.
.yml is what told GitHub to do all of those steps.
WHEN YOU PUSH NEW CODE OR UPDATE PUSH TO GITHUB ,, GITHUB NOTICE THAT PUSH AND START THE YML WORK FLOW AUTOMATICALLY.

TO CREATE .YML FILE ----> IN GITHUB---> .github(folder) // workflows(folder) //file_name.yml
(GitHub Actions only detects it when it's located in this.)

CODE:-
-------------------------------------------------------------------------------------------------------
name: Build and Deploy

on:
  push:
    branches:
      - main

jobs:
  deploy:
    runs-on: ubuntu-latest

    steps:
      - name: Checkout Code
        uses: actions/checkout@v4

      - name: Login to Docker Hub
        uses: docker/login-action@v3
        with:
          username: ${{ secrets.DOCKER_USERNAME }}                      # YOU HAVE TO USE HERE GITHUB/SECRETS AND VARIABLE/NAME --THAT YOU HAVE GIVEN IN GITHUB JUST NAME NOT VALUES.
          password: ${{ secrets.DOCKER_PASSWORD }}                      # YOU HAVE TO USE HERE GITHUB/SECRETS AND VARIABLE/PASSWORD --THAT YOU HAVE GIVEN IN GITHUB JUST NAME NO VALUES.

      - name: Build Docker Image
        run: |
          docker build -t ${{ secrets.DOCKER_USERNAME }}/iphone_prediction:latest .            # SAME FOR DOCKER_USERNAME/SECRETS AND VARIABLE/USERNAME
                                                                                                # in lower case only.
      - name: Push Docker Image
        run: |
          docker push ${{ secrets.DOCKER_USERNAME }}/iphone_prediction:latest                  # SAME FOR DOCKER_USERNAME/SECRETS AND VARIABLE/USERNAME

      - name: Trigger Render Deploy
        run: |
          curl -X POST "${{ secrets.RENDOR_DEPLOY_HOOK }}"                                     # secrets.RENDOR_DEPLOY_HOOK

--------------------------------------------------------------------------------------------------------------------------
```

```python
'''DEPLOY TO RENDER USING DOCKER:---
FIRST MAKE SURE YOUR PROJECT HAS THESE FILES:

1. main.py
2. index.html
3. requirements.txt
4. model file (.pkl)
5. Dockerfile

------------------------------------------------------------

THEN UPLOAD ALL PROJECT FILES TO A NEW REPOSITORY IN GITHUB
AND SET THE REPOSITORY TO PUBLIC (DEFAULT).

------------------------------------------------------------

THEN GO TO RENDER IN GOOGLE → SIGN IN → GO TO DASHBOARD.

THEN:

NEW
 ↓
WEB SERVICE
 ↓
CONNECT YOUR GITHUB REPOSITORY

------------------------------------------------------------

WHILE CREATING THE WEB SERVICE:

LANGUAGE / RUNTIME  ---->  DOCKER

BRANCH              ---->  main

REGION              ---->  Keep default/preferred region

------------------------------------------------------------

IMPORTANT:

WHEN USING DOCKER, YOU DON'T NEED TO MANUALLY WRITE:

BUILD COMMAND  ----> pip install -r requirements.txt
START COMMAND  ----> uvicorn main:app --host 0.0.0.0 --port $PORT

BECAUSE THESE INSTRUCTIONS ARE ALREADY HANDLED BY YOUR
DOCKERFILE.
------------------------------------------------------------
THEN CLICK:

CREATE WEB SERVICE / DEPLOY


------------------------------------------------------------

FINAL FLOW:

PROJECT FILES
     ↓
GITHUB
     ↓
RENDER
     ↓
RUNTIME = DOCKER
     ↓
DOCKERFILE
     ↓
DOCKER IMAGE
     ↓
CONTAINER
     ↓
LIVE WEBSITE
'''
```
