```python
'''WHAT IS DIFFERNECE BTW BAGGING AND BOOSTING.... OR WHAT IS DIFF BTW RANDOM FOREST AND XGBOOST...
RANDOM FOREST FULL PROCESS IS BAGGING... AND BOOSTING IS...


A= PASS
B= PASS
C= FAIL  

YOU HAVE XGBOOST(xtreme gradient boosting)(ALGORITHM) --> MULTIPLE TREEE IT WILL GENERATE 3 TREE AND OUTPUT WILL BE...

TREE1 = A=PASS B=FAIL C=PASS
TREE2 = THIS WILL BE DEPEND TOTALLY ON TREE1.  IN RANDOM/DECESION TREE IT WAS INDEPENDENT..
TREE2 = IT WILL UNDERSTAND THAT TREE1 HAS GIVEN 2 RECORDS WRONG THEN IT WILL UNDERSTADN IT AND CORRECT IT...
TREE2 = A=PASS B=PASS C=PASS
TREE3 IT WILL BE DEPENDED ON TREE2 AND ITS TARGET IS TO SOLVE THE PROBLEM IN TREE 2..
TREE3 = A=PASS B=PASS C=FAIL....


LIKE THIS YOU WILL GET LAST TREE AND ITS WORK IS TO SOLVE THE PROBLEM DONE BY PREVIOUS TREE...
WE WILL GET PREDICTION FROM LAST TREE ONLY....AND THAT WILL BE OUR OUTPUT...


WHAT IS GRADIENT DECENT ? 
-LIKE WE HAVE 10 DATA AND TREE 1 PREDICT 7 ACCURATE AND 3 INCORRECT...THEN IT IS 70% ACCUARTE THEN TREE2 WILL GIVE 80% AND TREE3 IF 90% THEN 90% IS TAKEN AS LAST AND IT IS CALLED PREDICTION...
```

```python
'''THINGS BY XGBOOST :-
1) n_estimeter :-- no of tree ==5 then xgboost generate 5 tree.
2) max_depth =  no. of questions. == 3 (every tree has 3 questions)
3) learning rate=model frequeency learn == 0.1 ,0.2 ,0.3(always keep 0.1 so your tree don't overfit)

LIKE RANDOM FOREST XGBOOST ALSO GENERATE RANDOM ROW (5,7,8)
SO YOU HAVE TO SPLIT DATA...AND USE RANDOM_STATE...

FORMULA :--
NEW PREDICTION = OLD PREDICTION + LEARNING RATE * TREE CORRECTION.

LET,,,
1) STUDY HR = 6  MARKS =80

THEN XGBOOST WILL FIRST PREDICT ITS OWN == MARKS==60. (THIS WILL NOT GENERATE ANY TREE HERE)
THIS PREDICTED MARKS BY XGBOOST WILL BE GIVEN TO TREE 1.

LOSS = (ACTUAL -PREDICTION)SQR
     = (80-60)2
     = 400  THIS IS USED NO WHERE IT IS TO MEASURE HOW MUCH OUR MODEL IS PREDCITING WRONG...

TREE 1 = COMPARE WITH ORIGINAL.... THAT IS MARKS=80.
TREE 1 = OLD PRED + LEARNING RT * TREE CORRECT
TREE 1 =  60      +     0.1(THIS YOU HAVE KEPT)  *  20(THIS IS LOSS FROM ORIGINAL AND XGBOOST PREDICTION)  

THIS IS == 62 ( THIS HAS ADDED 2 INTO 60 SO HAD DONE SOME LOSS LESS FROM 60)
THEN THIS WILL BE TAKEN BY TREE 2.

TREE 2 = 62 + 0.1 * 18  == 63.8 
TREE 3  LIKE THIS .....IT WILL TAKE TO 80 ONLY...AND IF IT GOES UP LIKE 81,82 THEN GRADIENT WILL WORK THEIR....

LET OUR XGBOOST FIRST PREDICT 90... BUT OUR DATA HAS MARKS = 80.
THEN OUR GREDIANT WILL DECESIDE THAT TREE 1 WON'T HAVE TO GO FURTHER THAN 90 IT HAS TO GO TO NEGATIVE...

GRADIENT = 2(PREDICTION-ACTUAL)
         = NEGEATIVE // POSITIVE
         = IF GRADIENT IS NEG THEN OUR TREE WILL MOVE TO POSSITIVE AND IF GRADIENT IS POSI THEN OUR TREE WILL MOVE TO NEG-
-----------------------------------------------------------------------
YOU HAVE 2 THINGS....
1) LEARNING RATE LESS AND NO OF TREE MORE...THIS IS BEST
2) LEARNING RATE MORE AND NO OF TREES LESS.... THIS WILL MAKE MORE CHANCES FOR OVERFITING...

-------------------------------------------------------------------------
```

```python
'''
WHEN YOU TAKE 
SUB_SAMPLE = 0.8 ......... 
THEN ,, 1 SPLIT = 80(DATA) * 0.8 = 64 PER TRAINING 20 PER TESTING...

So if you have 1,000 training rows:

subsample = 1.0  → uses 1000 rows per tree
subsample = 0.8  → uses ~800 rows per tree
subsample = 0.5  → uses ~500 rows per tree 

Using something like 0.8 can help reduce overfitting, because every tree doesn't see exactly the same training examples.EVERY TREE USED DIFFERNET 800 DATA FROM 1000 DATA.
Because each tree sees a slightly different sample, the model is less dependent on particular training examples or noise.'''
```

```python
'''COLSAMPLE BYTREE:-
LET YOU HAVE A DATA SET WHICH HAS COLUMN = 6
AND YOU GAVE COLSAMPLE BYTREE =0.6
THEN YOU COLUMN = 6 * 0.6 = 3 IT WILL TAKE RANDOM ANY 3 COLUMN...AND WILL PASS ONLY 3 COLUMN TO TREE 1 THEN FOR TREE 2 IT WILL GIVE ANOTHER RANDOM 3 COLUMN
BUT YOUR ALL COLUMN WILL BE USED AND IT CAN ALSO BE REPEATED....BUT AT A TIME ONLY 3 COLUMN WILL BE TAKEN...
AND IF YOU HAVE GIVEN SUB_SAMPLE = 0.8 THAT IS TRAINING = 80 * 0.8 = 64
THEN COLSAMPLE BYTREE WILL USE 3 COLUMN AND 64 ROWS....(AS FROM SUB_SAMPLE)...
```
