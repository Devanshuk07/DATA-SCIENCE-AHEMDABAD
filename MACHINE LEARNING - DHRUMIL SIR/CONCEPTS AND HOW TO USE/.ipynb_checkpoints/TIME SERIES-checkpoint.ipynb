```python
'''WHAT IS TIME SERIES ...?

-Time series in Machine Learning (ML) is a type of data where observations are collected in time order, 
such as every second, minute, hour, day, or month. The main goal is to analyze patterns over time and predict future values.

TILL WE HAVE IN DATA ....
AGE   SALARY     MARKS    NAME
23     23000      92      DEVANSHU

THIS DATA HAS NO RELATION WITH TIME/ORDER....

TIME SERIES MEANS TIME AND DATA BOTH ARE RELATED TO EACH OTHER....
LET WE HAVE :-
| Day       | Temperature (°C) |
| --------- | ---------------- |
| Monday    | 25               |
| Tuesday   | 27               |
| Wednesday | 26               |
| Thursday  | 29               |
| Friday    | 30               |

-Here, the order of the data matters. Unlike regular datasets, you cannot randomly shuffle time series data 
because each value depends on previous points.

Common Examples
📈 Stock prices
🌤️ Weather data
💰 Daily sales
❤️ Heart rate monitoring
🚗 Traffic volume
⚡ Electricity consumption

Why is Time Series Different?

Traditional ML assumes that data points are independent.-->One row generally doesn't depend on the previous row's position.

Time series data has temporal dependency,
Temporal dependency = the current/future value has some relationship with previous values in time.
meaning:

-Today's stock price depends on yesterday's price.
-Today's temperature is related to yesterday's temperature.

Components of a Time Series
1)Trend
-Long-term increase or decrease.
Example: Company sales growing every year.

2)Seasonality
-Patterns that repeat at regular intervals.
Example: Ice cream sales rise every summer.

3)Cyclic Patterns
-Long-term fluctuations without a fixed period.
Example: Economic boom and recession cycles.

4)Noise
-Random variations that cannot be explained.
Example: Unexpected events affecting sales.
```

```python
''' TIME SERIES HOW IT WORKS... :--

# TIME SERIES

Time Series is data that is collected in a particular order of TIME.

Examples:
- Daily sales
- Monthly profit
- Stock prices
- Temperature every hour
- Electricity consumption every day

In Time Series, TIME and the ORDER of data are important.

Example:

JAN → FEB → MAR → APR → MAY

We cannot randomly shuffle the data like we normally do in Machine Learning.
--------------------------------------------------------------------------------------------------------------------------------------------------
Suppose we have monthly sales:

MONTH     SALES
JAN       100
FEB       120
MAR       130
APR       150
MAY       170

Here, we can use previous months' sales to predict future sales.

For example:

To predict FEB:
we can use JAN sales.

To predict MAR:
we can use FEB sales.

To predict APR:
we can use MAR sales.
--------------------------------------------------------------------------------------------------------------------------------------------------
WHAT IS LAG ?
# LAG

Lag means taking a PREVIOUS value and using it as an input
for predicting the current value.

Example:

MONTH     SALES
JAN       100
FEB       120
MAR       130
APR       150
MAY       170

If we create LAG1:

MONTH     SALES     LAG1
JAN       100        -
FEB       120       100
MAR       130       120
APR       150       130
MAY       170       150

LAG1 means:

Current month's LAG1 = Previous month's SALES.

FEB → LAG1 = JAN SALES = 100
MAR → LAG1 = FEB SALES = 120
APR → LAG1 = MAR SALES = 130
MAY → LAG1 = APR SALES = 150
--------------------------------------------------------------------------------------------------------------------------------------------------
LAG2 and LAG3:--
We can create more lags.

LAG1 = previous 1 value
LAG2 = previous 2 values
LAG3 = previous 3 values

Example:
| MONTH | SALES | LAG1 | LAG2 | LAG3 |
| ----- | ----: | ---: | ---: | ---: |
| JAN   |   100 |    - |    - |    - |
| FEB   |   120 |  100 |    - |    - |
| MAR   |   130 |  120 |  100 |    - |
| APR   |   150 |  130 |  120 |  100 |
| MAY   |   170 |  150 |  130 |  120 |

For APR:

SALES = 150

LAG1 = 130 → previous month
LAG2 = 120 → 2 months before
LAG3 = 100 → 3 months before

So the model can use:
LAG1 + LAG2 + LAG3
        ↓
      MODEL
        ↓
   CURRENT SALES
--------------------------------------------------------------------------------------------------------------------------------------------------
HOW DO WE CREATE LAGS :-(.shift())

df["Lag1"] = df["Sales"].shift(1)
df["Lag2"] = df["Sales"].shift(2)
df["Lag3"] = df["Sales"].shift(3)

EXPLANATION...
.shift(1) means:

Take the previous 1 value.

.shift(2) means:

Take the value from 2 periods before.

.shift(3) means:

Take the value from 3 periods before.
--------------------------------------------------------------------------------------------------------------------------------------------------
WHY DO NULL VALUES APPEARS ?
When we create lags, the first few rows do not have previous values.

Example:
| MONTH | SALES | LAG1 | LAG2 | LAG3 |
| ----- | ----: | ---: | ---: | ---: |
| JAN   |   100 |    - |    - |    - |
| FEB   |   120 |  100 |    - |    - |
| MAR   |   130 |  120 |  100 |    - |
| APR   |   150 |  130 |  120 |  100 |
| MAY   |   170 |  150 |  130 |  120 |

Example:
LAG3 needs 3 previous values.

JAN → No previous 3 months → NULL
FEB → No previous 3 months → NULL
MAR → No previous 3 months → NULL
APR → Has 3 previous months → 100

Therefore, NULL values are created at the beginning.
--------------------------------------------------------------------------------------------------------------------------------------------------
WHAT DO WE DO WITH THE NULL VALUES ?

If we are using lag values as input features,
rows with missing lag values cannot be directly used for training.

Therefore, we commonly remove those rows.

Example:

If we create LAG3:

JAN → removed  --> df = df.dropna()
FEB → removed
MAR → removed

Because they have missing lag values.

APR and MAY can be used.
--------------------------------------------------------------------------------------------------------------------------------------------------
X AND Y IN TIME SERIES ....
After creating lags:

X = Input Features
Y = Output / Target

Example:

MONTH     SALES     LAG1     LAG2     LAG3
APR       150       130      120      100
MAY       170       150      130      120

X = LAG1, LAG2, LAG3

Y = SALES
--------------------------------------------------------------------------------------------------------------------------------------------------
HOW DOES THE MODEL PREDICT THE NEXT MONTH ?
Suppose our latest available data is:

MONTH     SALES
APR       150
MAY       170

And we have created:

LAG1
LAG2
LAG3

To predict JUN, we create the input using the
latest available previous values.

For JUN:

LAG1 = MAY SALES = 170
LAG2 = APR SALES = 150
LAG3 = MAR SALES = 130(LET)

JUN input:

LAG1 = 170
LAG2 = 150
LAG3 = 130(LET)

        ↓
      MODEL
        ↓
   PREDICT JUN SALES

   NOTE.............
We don't simply say:

"the model predicts the next month using the previous 2 rows."

It depends on how many lag features we created.

If we use:

LAG1 → model predict from 1 previous value

LAG1 + LAG2 → model predict from 2 previous values

LAG1 + LAG2 + LAG3 → model predict from 3 previous values

So the number of previous values used depends on the lag features we choose.
--------------------------------------------------------------------------------------------------------------------------------------------------
More lags don't mean automatically better.
bcz ,,,  
1. Older values may no longer matter

Maybe sales are strongly affected by the last 2–3 months, but sales from 12 months ago have almost no relationship.

2. You lose more rows

This is a very important point.

If you have 100 observations:
Lag1  → lose 1 row
Lag3  → lose 3 rows
Lag10 → lose 10 rows
Lag50 → lose 50 rows

With Lag50, your model only has 50 usable rows. So adding too many lags can actually leave you with less training data.

3. More lags can cause overfitting

Suppose you have only 100 months of data.you created lag50. then now you have only 50 features(input) from 100 data.
The model may start learning random patterns/noise instead of the actual time-series pattern.That's called overfitting.

Yesterday's sales       → very useful
Last week's sales       → possibly useful
Last month's sales      → possibly useful
Sales 10 years ago      → probably not useful

So we want the right amount of history, not the maximum amount of history.
--------------------------------------------------------------------------------------------------------------------------------------------------
```

```python
'''STATIONARITY :-
# STATIONARITY

Stationary means that the important statistical properties
of a time series remain reasonably stable over time.

A stationary series generally has:

- Stable mean
- Stable variance
- Stable relationship/dependence over time

It does NOT simply mean:

"All values are close to each other."

Simple Stationary Example :--

100
98
102
101
99
103
100

The values are fluctuating around approximately the same level.

This MAY be stationary.

But we cannot decide stationarity just by looking at
whether the numbers are close together.

Non-stationary example:
100
150
200
250
300
350
400

Here, there is a clear increasing trend.
The average level is changing over time.
Therefore, this series is likely non-stationary.
--------------------------------------------------------------------------------------------------------------------------------------------------
WHY IS STATIONARITY IMPORTANT ?
Many traditional Time Series forecasting methods work better
when the data is stationary.
Real-life example: daily electricity demand of a company...

If the data has:

- Strong trend
- Changing variance
- Strong seasonality

then we may need to transform the data before applying
some time-series methods.

One common technique is DIFFERENCING.
--------------------------------------------------------------------------------------------------------------------------------------------------
WHAT IS DIFFERENCING/TRANSFORMATION ?
Differencing means:

Current value - Previous value

Example:

SALES:

100
120
150
180

First difference:

120 - 100 = 20
150 - 120 = 30
180 - 150 = 30
210 - 180 = 30

New series:

20
30
30
30

Differencing can help remove a trend and make a series
more stationary.

But we should check whether the transformed series
actually becomes stationary.'''
```

```python
'''
WHAT IS ADF TEST ?
# ADF TEST

ADF = Augmented Dickey-Fuller Test

ADF gives us a P-VALUE.

General interpretation:

p-value < 0.05
        ↓
Reject H0
        ↓
Evidence that the series is stationary.

p-value >= 0.05
        ↓
Fail to reject H0
        ↓
Not enough evidence to say the series is stationary.
--------------------------------------------------------------------------------------------------------------------------------------------------
```

```python
'''WHAT IS RESAMPLING ?
-# RESAMPLING

Resampling means changing the TIME FREQUENCY of our data.

For example:

Hourly data → Daily data
Daily data → Monthly data
Monthly data → Yearly data

Example:
Suppose we have:

01 JAN → 100
02 JAN → 120
03 JAN → 110
04 JAN → 130
...

We can resample this daily data into monthly data.

For example:
JAN → Total Sales = 10,000
FEB → Total Sales = 12,000
MAR → Total Sales = 15,000
--------------------------------------------------------------------------------------------------------------------------------------------------
Resampling does NOT simply mean:

"Convert years into months."

It can work in BOTH directions.

We can:

DOWNsample:
Daily → Monthly
Monthly → Yearly

UPsample:
Yearly → Monthly
Monthly → Daily

But when we go from a lower frequency to a higher frequency(UPSAMPLE),
we don't automatically get real new information.

Example:

2025 → ₹12,000

Changing it to:

Jan → ?
Feb → ?
Mar → ?
...

does not magically tell us the actual monthly sales.

We need a suitable method to fill/interpolate those values,
and that creates assumptions.
--------------------------------------------------------------------------------------------------------------------------------------------------
PANDAS RESAMPLING :-

# Example
Suppose your data looks like this:
| Date       | Sales |
| ---------- | ----: |
| 2025-01-05 |   100 |
| 2025-01-15 |   200 |
| 2025-01-25 |   150 |
| 2025-02-05 |   300 |
| 2025-02-20 |   250 |


df["Date"] = pd.to_datetime(df["Date"])          # pd.to_datetime-->Convert Date into a proper date.Sometimes pandas reads dates as normal text/string.

df = df.set_index("Date")          #Make Date the index. Because pandas' time-series functions such as resample() work based on a DatetimeIndex.

# Monthly total
monthly_sales = df["Sales"].resample("ME").sum()   'ME' --> MONTH  ,, 'YE' --> YEAR         # 100 + 200 + 150 = 450
```
