````python
'''DECISION TREE

Decision Tree is used for CLASSIFICATION.

It divides data into different groups/classes by asking questions internally.

Its main goal is to create groups that are as PURE as possible.

Pure Group means the data in that group mostly or completely belongs to the same class.

--------------------------------------------------------------------------------------------------------------------

EXAMPLE

SNAME        HRS        RESULT

A             1         FAIL
B             2         FAIL
C             3         PASS
D             4         PASS
E             5         PASS
F             6         PASS

Here our output has two classes:

PASS
FAIL

Decision Tree will internally try different questions to separate PASS and FAIL.

Example of question:

Study Hours > 3.5?

--------------------------------------------------------------------------------------------------------------------

1. ROOT NODE

--------------------------------------------------------------------------------------------------------------------

The FIRST question of the Decision Tree is called the ROOT NODE.

Example:
          Study Hours > 3.5?
               ROOT NODE
              /         \
            YES          NO


--------------------------------------------------------------------------------------------------------------------

2. DECISION NODE

--------------------------------------------------------------------------------------------------------------------

After the Root Node, if another question is required to divide the data further, it is called a DECISION NODE.

Example:
          Study Hours > 3.5?
              ROOT NODE
             /         \
           YES          NO
            ↓
     Attendance > 70?
       DECISION NODE


--------------------------------------------------------------------------------------------------------------------

3. LEAF NODE

--------------------------------------------------------------------------------------------------------------------

When no more questions are asked and the tree gives the final group/prediction, it is called a LEAF NODE.

Example:
          Attendance > 70?
            /          \
          YES           NO
           ↓             ↓
         PASS           FAIL
         LEAF           LEAF

Leaf Node = Final Output/Prediction.

--------------------------------------------------------------------------------------------------------------------

4. BRANCH

--------------------------------------------------------------------------------------------------------------------

Branches connect the nodes of the Decision Tree.

Example:
              ROOT NODE
              /       \
          BRANCH      BRANCH
            /           \
    DECISION NODE     LEAF NODE

--------------------------------------------------------------------------------------------------------------------

5. SPLIT

--------------------------------------------------------------------------------------------------------------------

The question/condition used by the Decision Tree to divide the data is called a SPLIT.

Example:

Study Hours > 3.5?

This question divides the data into two groups:

YES → Study Hours > 3.5

NO → Study Hours <= 3.5

--------------------------------------------------------------------------------------------------------------------

6. GINI

--------------------------------------------------------------------------------------------------------------------

Gini is used to find the IMPURITY of a group.

It helps the Decision Tree choose the BEST SPLIT.

Lower Gini = Purer Group

Gini = 0 → Completely Pure Group

For binary classification:

Gini = 0   → Completely Pure
Gini = 0.5 → Maximum Impurity
--------------------------------------------------------------------------------------------------------------------
Suppose the Decision Tree tries different questions:

Question 1 → Weighted Gini = 0.40

Question 2 → Weighted Gini = 0.32

Question 3 → Weighted Gini = 0.23

The LOWEST Weighted Gini is:

0.23

Therefore Question 3 gives the best split among these questions.

--------------------------------------------------------------------------------------------------------------------

7. MAIN WORKING OF DECISION TREE

--------------------------------------------------------------------------------------------------------------------

Decision Tree tries different questions/conditions.

Each question divides the data into groups.

Gini helps check the impurity of those groups.

Weighted Gini helps compare the possible splits.

The split with the LOWEST Weighted Gini is selected.

The tree can then continue asking more questions to make the groups purer.

When no more splitting is required, it reaches the LEAF NODE and gives the final prediction.
'''
````

````python
'''# DECISION TREE

Let we have this data:

SNAME        HRS        RESULT
A             1         FAIL
B             2         FAIL
C             3         PASS
D             4         PASS
E             5         PASS
F             6         PASS

--------------------------------------------------------------------------------------------------------------------

1. ROOT NODE AND SPLITTING

--------------------------------------------------------------------------------------------------------------------

Decision Tree will ask a question at the Root Node and divide the data into 2 groups.

Example:

```
            STUDY HRS > 3.5?
             /             \
           YES              NO
            ↓                ↓
         GROUP 1          GROUP 2
```

GROUP 1: HRS > 3.5
Found Records = 3
Pass = 3
Fail = 0

GROUP 2: HRS <= 3.5
Found Records = 3
Pass = 1
Fail = 2

--------------------------------------------------------------------------------------------------------------------

2. GINI

--------------------------------------------------------------------------------------------------------------------

Gini tells us how pure/impure a group is.

GINI FORMULA:

Gini = 1 - (Probability(Pass)² + Probability(Fail)²)

For GROUP 1:

Pass = 3
Fail = 0
Total = 3

Gini = 1 - ((3/3)² + (0/3)²)

```
 = 1 - (1 + 0)

 = 0            --------> THIS IS FOR A PERTICULAR GROUP....
```

For GROUP 2:

Pass = 1
Fail = 2
Total = 3

Gini = 1 - ((1/3)² + (2/3)²)

```
 = 1 - (1/9 + 4/9)

 = 1 - 5/9

 = 4/9

 = 0.44        --------> THIS IS FOR A PERTICULAR GROUP....
```

For binary classification:

Gini = 0   → Completely Pure
Gini = 0.1 → Very little mixing
Gini = 0.3 → Some mixing
Gini = 0.5 → Maximum mixing

Lower Gini = Purer Group.

--------------------------------------------------------------------------------------------------------------------
3. WEIGHTED GINI 

--------------------------------------------------------------------------------------------------------------------
NOTE :--  (Weighted Gini is used during training to decide which questions/splits should be used to build the tree
THEN THAT BEST SPLIT IS THEN USED WHEN NEW DATA IS ASKED AND FROM THAT IT WILL DO ITS FINAL PREDICTION... 
SO NEW DATA WILL ONLY USE THAT BEST QUESTION SPLIT THAT IS SELECTED FROM LOWEST WEIGHTED GINI)


Each question creates 2 groups.

Each group has its own Gini.

The two Gini values are combined to calculate ONE
WEIGHTED GINI for that particular split/question.

Example:

Suppose Study Hours > 4? splits 10 students:

```
                Study Hours > 4?
                /              \
              YES               NO
          6 students        4 students
          Gini = 0.20       Gini = 0.40
```

Step 1: Find weight of each group:

Left Group  = 6/10
Right Group = 4/10

Step 2: Multiply weight by Gini:

Weighted Gini =

(6/10 × 0.20) + (4/10 × 0.40)

= 0.12 + 0.16

= 0.28

Therefore:

WEIGHTED GINI = 0.28

This is the Weighted Gini for that PARTICULAR SPLIT.

--------------------------------------------------------------------------------------------------------------------

4. HOW BEST SPLIT IS SELECTED

--------------------------------------------------------------------------------------------------------------------

Decision Tree tries different questions/conditions.

Example:

Study Hours > 2? → Weighted Gini = 0.40

Study Hours > 3? → Weighted Gini = 0.32

Study Hours > 4? → Weighted Gini = 0.20  ← LOWEST

Study Hours > 5? → Weighted Gini = 0.35

Best Split = Split with the LOWEST WEIGHTED GINI.

Therefore:

Study Hours > 4?

will be selected.

IMPORTANT:

The Decision Tree does NOT simply select the lowest
individual Left/Right Gini.

It calculates the Weighted Gini of each possible split
and selects the split with the LOWEST Weighted Gini.

--------------------------------------------------------------------------------------------------------------------

5. AFTER SELECTING THE BEST SPLIT

--------------------------------------------------------------------------------------------------------------------

Suppose Study Hours > 4? wins:

```
             Study Hours > 4?
              /             \
            YES              NO
             ↓                ↓
          Group 1          Group 2
```

Both groups stay separate.

The tree can again ask questions inside each group:

```
              Study Hours > 4?
              /              \
            YES               NO
             ↓                 ↓
      Attendance > 70?     Marks > 50?
         /      \            /      \
       YES      NO         YES       NO
```

Each child group can again be split until the stopping
condition or a leaf node is reached.

RULE:

Every question calculates its OWN Weighted Gini using
only the records that reached that question.

--------------------------------------------------------------------------------------------------------------------

6. MAX_DEPTH

--------------------------------------------------------------------------------------------------------------------

If:

max_depth = 2

it means the tree can have maximum 2 levels of splitting.
It does NOT necessarily mean only 2 questions total.

```
            Question 1       ← Depth 1
           /          \
      Group 1          Group 2
         ↓                ↓
    Question 2        Question 3   ← Depth 2
     /    \            /    \
   G1      G2        G3      G4
```

Each question has its own Weighted Gini.

Example:

Question 1 → Weighted Gini = 0.34

Question 2 → Weighted Gini = 0.233

Question 3 → Weighted Gini = 0.15

There is NO single final Gini selected from these
different nodes.

Each Weighted Gini belongs to its own node.

--------------------------------------------------------------------------------------------------------------------

7. FINAL PREDICTION

--------------------------------------------------------------------------------------------------------------------

Weighted Gini is used during TRAINING for choosing
the best questions/splits.

After the tree is created, new data follows those questions
until it reaches a Leaf Node.

Example Tree:

```
                Question 1
              /            \
           YES              NO
            ↓                ↓
       Question 2        Question 3
        /      \          /      \
      YES      NO       YES      NO
       ↓        ↓         ↓        ↓
     PASS      FAIL     PASS      FAIL
     LEAF      LEAF     LEAF      LEAF
```

Now new student data comes:

Study Hours = 6
Attendance = 80

Question 1:

Study Hours > 4?

6 > 4 → YES
↓
Go Left

Question 2:

Attendance > 70?

80 > 70 → YES
↓
Go Left
↓
PASS

FINAL PREDICTION = PASS

Gini does NOT need to be calculated again for the new
student. The new data simply follows the already-created tree.

--------------------------------------------------------------------------------------------------------------------

IMPORTANT RULES TO REMEMBER

1. Decision Tree asks questions/conditions.

2. Each question divides data into 2 child groups.

3. Each child group has its own Gini.

4. Lower Gini = Purer Group.

5. The two child Ginis are combined using Weighted Gini.

6. Lowest Weighted Gini = Best Split at that node.

7. After selecting the best split, the child groups can
   again be split.

8. Every question/node has its OWN Weighted Gini.

9. There is NO single final Gini for the entire tree.

10. Gini/Weighted Gini is used during training to choose
    the questions.

11. New data follows the questions until it reaches a
    Leaf Node.

12. Leaf Node gives the FINAL PREDICTION.
--------------------------------------------------------------------------------------------------------------------
IN CODE  :-

DecisionTreeClassifier(max_depth=2)   Just like in KNN we use: KNeighborsClassifier(n_neighbors=3)
Here max_depth controls how deep the Decision Tree can grow.
'''
````

````python
'''WHAT IS CROSS-VALIDATION ?
-Cross-validation is used to check how well your ML model performs on different parts of the same dataset, 
instead of trusting only one train-test split.

-CROSS VALIDATION: It divides the dataset into multiple parts (folds) and trains/tests the model multiple times using different parts for testing, 
giving a more reliable estimate of model performance. ✅

Suppose you have 100 records and use 5-fold cross-validation.

The data is divided into 5 parts:
Fold 1 = 20 records
Fold 2 = 20 records
Fold 3 = 20 records
Fold 4 = 20 records
Fold 5 = 20 records

Then the model trains and tests 5 times:
Round 1:
TEST  → Fold 1
TRAIN → Fold 2,3,4,5

Round 2:
TEST  → Fold 2
TRAIN → Fold 1,3,4,5

Round 3:
TEST  → Fold 3
TRAIN → Fold 1,2,4,5

Round 4:
TEST  → Fold 4
TRAIN → Fold 1,2,3,5

Round 5:
TEST  → Fold 5
TRAIN → Fold 1,2,3,4

So every record gets a chance to be in the test set.

Suppose accuracy is:
Fold 1 → 90%
Fold 2 → 88%
Fold 3 → 92%
Fold 4 → 89%
Fold 5 → 91%

Average = 90%
That gives us a more reliable idea of the model's performance than checking just one train-test split.
--------------------------------------------------------------------------------------------------------------------
In Python:-

from sklearn.model_selection import cross_val_score

scores = cross_val_score(model, X, y, cv=5)

print(scores)
print(scores.mean())
--------------------------------------------------------------------------------------------------------------------
When is it especially useful?

It's especially useful when your dataset is small or medium-sized, when you want to compare models, 
or when you want to check whether your model's performance is stable across different splits

                    SAME DATASET
                        ↓
        ┌───────────────┼───────────────┐
        ↓               ↓               ↓
      KNN         Decision Tree    Random Forest ...
        ↓               ↓               ↓
   5-Fold CV        5-Fold CV        5-Fold CV
        ↓               ↓               ↓
 Avg = 86%         Avg = 91%         Avg = 94%

EXAMPLE IN CODE:--

from sklearn.model_selection import cross_val_score

#Classification
knn_score = cross_val_score(knn, x, y, cv=5, scoring='accuracy')                    # here cv=5 for 5 fold ,,, aslo scoring is for accuracy test... that will give accuracy and that is for classification only..
decesiontree_score = cross_val_score(dt, x, y, cv=5, scoring='accuracy')
randomforest_score = cross_val_score(rf, x, y, cv=5, scoring='accuracy')

# Regression
scores = cross_val_score(model, x, y, cv=5, scoring='r2')


print(knn_score.mean())
print(decesiontree_score.mean())
print(randomforest_score.mean())

'''
````
