```python

from pathlib import Path

folder_path = Path(
    r"F:\DATA SCIENCE-AHEMDABAD\MACHINE LEARNING - DHRUMIL SIR\STUDY AI MODEL"
)

print("Folder exists:", folder_path.exists())
```

```python

# Open the STUDY_MATERIAL subfolder
study_folder = folder_path / "STUDY_MATERIAL"

# Find all TXT files inside it
txt_files = list(study_folder.glob("*.txt"))

print("Number of TXT files:", len(txt_files))

for file in txt_files:
    print(file.name)
```

```python

# Select the Python study file
python_file = study_folder / "PYTHON.txt"

# Read its contents
python_notes = python_file.read_text(encoding="utf-8")

# Display the first 1000 characters
print(python_notes[:1000])
```

```python

# Ask what you want to search for
search_question = input("What do you want to learn? ").lower()

# Check whether the words appear in your Python notes
if search_question in python_notes.lower():
    print("Found in your Python notes!")
else:
    print("Not found in your Python notes.")
```

```python
#Show the exact matching answer
search_question = input("What do you want to learn? ").lower()

found = False

for line in python_notes.splitlines():
    if search_question in line.lower():
        print(line)
        found = True

if not found:
    print("No exact match found in your Python notes.")
```

```python

# Store the contents of all study files
study_notes = {}

for file in study_folder.glob("*.txt"):
    topic = file.stem
    content = file.read_text(encoding="utf-8")
    study_notes[topic] = content

# Show the topics we loaded
print("Study topics loaded:")

for topic in study_notes:
    print(topic)

print("Total topics:", len(study_notes))
```

```python

question = input("Ask your study question: ").lower()

found = False

for topic, content in study_notes.items():
    for line in content.splitlines():
        if question in line.lower():
            print(f"\nTopic: {topic}")
            print(line.strip())
            found = True

if not found:
    print("No exact match found in your study notes.")
```

```python

from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# Load a small text-embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Prepare pieces of text from all five files
notes = []

for topic, content in study_notes.items():
    lines = [
        line.strip()
        for line in content.splitlines()
        if line.strip()
    ]

    # Keep nearby lines together in small groups
    chunk_size = 5

    for i in range(0, len(lines), chunk_size):
        chunk = "\n".join(lines[i:i + chunk_size])

        notes.append({
            "topic": topic,
            "text": chunk
        })

# Convert the notes into embeddings
note_embeddings = model.encode(
    [note["text"] for note in notes]
)

print("Small note chunks created:", len(notes))
```

```python

question = input("Ask your study question: ")
question_lower = question.lower()

# Check if the user wants the complete study material
full_words = ["all", "full", "complete"]

if any(word in question_lower.split() for word in full_words):

    found_topic = None

    # Find the requested subject
    for topic in study_notes:
        if topic.lower() in question_lower:
            found_topic = topic
            break

    if found_topic:
        print(f"\nComplete {found_topic} Study Material:\n")
        print(study_notes[found_topic])
    else:
        print("Please mention a subject, such as Python, LLM, NumPy, Pandas or SQL.")

else:
    # Convert your question into an embedding
    question_embedding = model.encode([question])

    # Compare the question with all notes
    scores = cosine_similarity(
        question_embedding,
        note_embeddings
    )[0]

    # Find the most relevant note
    best_index = scores.argmax()
    best_score = scores[best_index]

    if best_score >= 0.35:
        print("\nTopic:", notes[best_index]["topic"])
        print("Relevant note:", notes[best_index]["text"])
        print("Similarity:", round(float(best_score), 2))
    else:
        print("I couldn't find a sufficiently relevant note.")
```
