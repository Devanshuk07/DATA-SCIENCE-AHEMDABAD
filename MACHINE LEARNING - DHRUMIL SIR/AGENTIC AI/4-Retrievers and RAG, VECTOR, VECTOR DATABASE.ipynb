```python
'''WHAT IS RETRIEVER ?
-A retriever is a component that searches a collection of documents and returns the ones most relevant to your question.
-Imagine you have a TXT file containing 100 Python questions and answers.

You ask your Study AI:
How does Python execute a program?

You don't want the AI to invent an answer. You want it to find the relevant text from your TXT file and return that original text.

MAIN POINT :- A retriever finds relevant content. It does not necessarily generate an answer itself.
```

```python
'''WHAT IS RAG ?
-RAG stands for Retrieval-Augmented Generation.

Let's break down the name:
- Retrieval: Find relevant information from your documents.
- Augmented: Give that information to the LLM.
- Generation: The LLM uses that provided context to generate a response.
--------------------------------------------------------------------------------------------------------------------------
A typical RAG workflow looks like this:

1.USER ASK A QUESTION 
2.RETRIEVERS FINDS RELEVANT DOCUMENTS
3.RETRIEVED TEXT IS GIVEN TO LLM
4.LLM GENERATE A RESPONSE.
```

```python
'''DIFFERENCE BETWEEN RETRIVER AND RAG:-
        Retriever	                                                                  RAG
Finds relevant information	                                            Combines retrieval with generation
Returns matching documents or text	                                    Uses retrieved content as context for an LLM
Doesn't require an LLM to generate its result	                        Usually includes an LLM generation step
```

```python
'''WHAT IS EMBEDDING ?
-An embedding is a list of numbers that represents the meaning of text.
Embeddings help your retriever search by meaning, not just by exact keyword matches.

For example, your system can turn these sentences into numerical vectors:
- “How does Python execute a program?”
- “Explain the execution of a Python program.”

--> Although the wording differs, their meanings are similar. An embedding model can represent them with vectors that are close together.
Your code then uses cosine similarity to estimate how closely the vectors match.


Embeddings help your retriever search by meaning, not just by exact keyword matches.

For example, consider these three sentences:
Sentence A
"I love dogs."


Sentence B
"I like puppies."


Sentence C
"The stock market fell today."

Sentence A and Sentence B have similar meanings.
A good embedding model will generally represent them as vectors that are closer together than the vectors for Sentence A and Sentence C.

For learning purposes, imagine these simplified vectors:
- Sentence A → [0.80, 0.60, 0.10]
- Sentence B → [0.78, 0.62, 0.12]
- Sentence C → [-0.30, 0.10, 0.95]

These numbers are illustrative only, not actual embeddings. Real embeddings often contain hundreds or thousands of dimensions.
important idea: similar meanings tend to produce similar vectors, although this depends on the embedding model and the type of text.
```

```python
'''WHAT IS VECTOR ?
-A vector is an ordered list of numbers.

EXAMPLE:-  vector = [0.25, 0.80, 0.14, 0.67]
This vector has four dimensions because it contains four numbers.

IT WORKS LIKE THIS....
-Text is the original information.
-An embedding model converts the text into numbers.
-The resulting vector represents information in a form that a computer can compare.
```

```python
'''WHAT IS VECTOR DATABASE ?
-A vector database stores vectors and helps retrieve vectors that are similar to a search query.

EXAMPLE :-
Imagine you are building a chatbot for your Data Science course. You store these documents:
Document 1: Python
"Python is a programming language used for data analysis and automation."

Document 2: Machine Learning
"Machine learning models learn patterns from data to make predictions."

Document 3: SQL
"SQL is used to query and manage data stored in relational databases."
---------------------------------------------------------------------------------------------------------------------
Now a student asks:
"How can I make predictions using data?"
-Even if the document does not contain those exact words, a good embedding model may identify Document 2 as semantically relevant.

---------------------------------------------------------------------------------------------------------------------
HOW DOES IT WORKS ?

1. Original document
"Machine learning learns patterns from data."

2. Embedding model
Converts text into a vector of numbers.

3. Vector database
Stores the vector alongside the document or its reference.

4. Similarity search
Finds relevant documents for a new question.

---------------------------------------------------------------------------------------------------------------------
```

```python
'''Remember the difference:-

Concept	Meaning
Embedding  --->  Converts information into a numerical representation.
Vector  ---> 	The actual list of numbers representing the information.
Vector database  --->	Stores vectors and helps find similar ones.

important point: The computer can already process numbers.
Embeddings make it possible to compare the meaning of text mathematically, rather than relying only on exact word matches.
```

```python
'''HOW DOES A VECTOR DATABASE MEASURE THE SIMILARITY ?
-One common method is cosine similarity. It measures the angle between two vectors.

Score = 1 (Same direction)
Score = 0 (Perpendicular directions)
Score = -1 (Opposite directions)
```

```python
'''VECTOR DATABASE VS NORMAL DATABASE :--

        VECTOR DATABASE                                                  NORMAL DATABASE
Store and search vector representations                         Store and query structured data
Similarity search                                               Exact matches, filters, joins
Banking, sales, student records                                 Semantic search, RAG, recommendations
```

```python
'''What is the connection between embeddings, vector databases and RAG?

-RAG stands for Retrieval-Augmented Generation. It lets an application retrieve relevant information and provide it to an LLM to 
help generate an answer.

Imagine you ask a chatbot:
"What did my teacher's notes say about LangGraph?"

The chatbot needs to find the relevant information from your notes instead of relying only on the LLM's general knowledge.
---------------------------------------------------------------------------------------------------------------------
Step 1 — User question
"What is LangGraph used for?"

Step 2 — Create query embedding
Convert the question into a vector.

Step 3 — Retrieve relevant content(RAG)
Search the vector database for similar vectors and their associated documents.

Step 4 — Give context to the LLM
Combine the question with the retrieved notes.

Step 5 — Generate the answer
The LLM answers using the available context.
---------------------------------------------------------------------------------------------------------------------
Notice that RAG uses several components together:

mbedding model: Converts text into vectors.
Vector database: Helps find similar content.
Retriever: Fetches relevant documents for a query. It may use a vector database, but retrieval can also use other search methods.
LLM: Generates the final response using the question and retrieved context.

IMP :- A vector database is useful for RAG, but RAG does not always require a vector database. 
For example, it can use keyword search(system looks for matching words) or a hybrid search system(system combines keyword search and vector search).
```
