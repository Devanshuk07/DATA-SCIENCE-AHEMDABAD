```python
'''WHAT IS LANGCHAIN ?
-LangChain is a Python framework that helps us connect different parts of an AI application.

EXAMPLE:-
-Imagine you want to build a Study AI Bot that answers questions from your Python TXT file.

Without organizing the parts, you would have to write the code to manage each step yourself.
LangChain gives us ready-made components that help connect those steps:-
--------------------------------------------------------------------------------------------------------------
Step 1: User asks a question
“What is a Python virtual machine?”

Step 2: Find relevant text
Our existing similarity-search code

Step 3: Return the original text
The matching passage from our TXT file
--------------------------------------------------------------------------------------------------------------
LangChain can help connect these parts, but our existing similarity code does the searching. 
We will keep the answer exactly as it appears in the TXT file.
```

```python
'''Why do we need LangChain?

We want to learn how to use components such as:
- PromptTemplate: Create reusable instructions.
- LLM: Connect to an AI language model.
- Chains: Connect components so one step passes its result to the next.
- Retrievers: Find relevant information from documents.
- RAG: Use information from our own documents to help answer questions.
```
