```python
'''WHAT IS STATISTICS ??
-STATISTICS IS THE SCIENCE OF COLLECTING,ORGANIZING,ANALYZING,AND INTERPRETING DATA TO MAKE INFORMED DECISIONS UNDER UNCERTAINITY.

In very simple words:

-If you have raw data, statistics helps you understand what that data is saying.
-It turns a big messy table of numbers into useful information.

IN DATA SCIENCE,STATISTICS HELPS US:-
EXAMPLE:-
Suppose you have marks of 100 students.

Without statistics, it is just a long list of numbers like:

45, 67, 89, 54, 72, 91, ...

With statistics, you can answer questions like:

What is the average mark?
Which mark appears most often?
Are marks spread out a lot or are they close to each other?
Did students perform better this year than last year?
Can we predict future performance?

So statistics is basically a tool to understand data.
```

```python
'''WHAT IS MEANS(AVERAGE) AND MEDIAN AND VARIANCE AND STANDARD VARIANCE:---
MEAN :-- TOTAL ELEMENT ADDED / TOTAL ELEMENTS
MEDIAN:-- MEANS MIDDLE VALUE IF 1,2,3,4,5 THEN MEDIAN==3 AND IF 1,2,3,4,5,6 THEN 3+4=7/2==3.5 IS MEDIAN

BUT WHEN A PLAYER HAS :--
| Player | Scores     |
| ------ | ---------- |
| A      | 49, 50, 51 |
| B      | 10, 50, 90 |

HERE PLAYER-A HAS MEAN:-50  AND MEDIAN :- 50
HERE PLAYER-B HAS MEAN:-50 AND MEDIAN :- 50

BUT PLAYER-A IS CONCISTANT AND while Player B's scores change a lot. SO:-

WE NEED VARIANCE:--
Variance tells us how spread out the data is.
Formula (Population):

Variance=  ∑(x−μ)2 / N

HOW TO FIND VARIANCE JUST REMEMBER THIS :--
AD-SAD:-- ADDSAD MEANS
ADSAD:-- AVERAGE-DIFFERNECE-SQUARE-ADDSQUARE-DIVISION

PLAYER-A VARIANCE:--AVERAGE==50
DIFFERENCE==SUBSTRACT THE MEAN FROM EACH NUMBER-->  49-50=-1
                                                    50-50= 0
                                                    51-50=1 
SQUARE== SQUARE OF ALL :--(-1)square
                          (0)square
                          (1)square
ADDSQUARE== 1+0+1=2
DIVISION BY TOTAL ELEMENTS:-- 2/3=0.67
SO THIS IS POPULATION VARIANCE :-- 0.67
WHEN WE ARE TOLD TO FIND SAMPLE VARIANCE:--2/3-1=1(SAMPLE VARIANCE) SO ALWAYS FOR SAMPLE 1- FROM TOTAL ELEMENTS.

NOW STANDARD DEVIATION IS JUST SQUARE ROOT OF VARIANCE.

Standard Deviation=square-root(Variance)
- SO STD DEVIANCE OF PLAYER-A IS square-root(0.67)==0.82.
```

```python
''' THIS IS THE SMALL PROJECT OR TASK GIVEN BY SIR ON SATURDAY TO FIND OUT WHAT SHOULD BE THE GREAT TO ORDER THE FOOD ONLINE 
MEANS SWIGGY OR ZOMATO WE HAVE 2 CSV FILE OF SWIGGY AND ZOMATO WITH DELIVERY TIME AND RATINGS OF CUSTOMER. NOW FROM THAT WE HAVE TO
DECIDE WHICH SHOULD WE CHOOSE?'''

import numpy as np
import pandas as pd

zom=pd.read_csv('zomato.csv')
swg=pd.read_csv('swiggy.csv')
#print(f'{zom},\n\n\n\n\n{swg}')

zom['ZOMATO_AVERAGE_TIME1']=zom['delivery_time_mins'].mean()
zom['ZOMATO_AVERAGE_TIME2']=zom['delivery_time_mins'].median()
zom['ZOMATO_STANDARD_DEV']=zom['delivery_time_mins'].std()
print(zom)
swg['SWIGGY_AVERAGE_TIME1']=swg['delivery_time_mins'].mean()
swg['SWIGGY_AVERAGE_TIME2']=swg['delivery_time_mins'].median()
swg['SWIGGY_STANDARD_DEV']=swg['delivery_time_mins'].std()
print(swg)

'''THEN IF WE SEE ACCORDING TO DELIVERY TIME -- SWIGGY IS BETTER 0.1 minute = 6 seconds
this is such a tiny difference that we cannot confidently say Swiggy is better based only on the mean....'''

'''Standard deviation tells you how much the delivery times vary from the average.
zomato=26 min and swiggy=31min.

Smaller standard deviation → Delivery times are more consistent.
Larger standard deviation → Delivery times are less consistent and fluctuate more.'''

'''My conclusion

For this dataset, Zomato appears better overall because:

It has nearly the same average delivery time as Swiggy.
It has lower variation, making its delivery times more reliable.'''
```

```python
'''SAMPLE:-
-----------What is a Sample?

A sample is a small part of a large group taken for study.

The whole group is called population
A small selected part of it is called sample
Example:

Suppose there are 10,000 students in a school chain, and you want to know the average marks of all students.

Checking marks of all 10,000 students is difficult.
So you select 200 students and study only them.

10,000 students = Population
200 selected students = Sample

---------SO, Sample = a subset (small part) of the population used to collect information about the whole population.

---------Why do we use a sample?

Because studying the whole population is often:

time-consuming
expensive
difficult
sometimes impossible

So instead of studying everyone, we study a sample and use it to understand the population.

---------Types of Sample

Broadly, sampling is of 2 types:

Random Sampling
-A random sample means every item/person in the population has a fair chance of being selected.
-There is no favoritism and no intentional selection.

Why random sample is good?
-Because it gives a fair representation of the population and usually produces more reliable results.

Biased Sampling:-
-A biased sample is a sample that is not selected fairly.
-Some people/items get a higher chance of being selected, while others may be ignored.
-A biased sample is a sample selected unfairly, so it does not properly represent the whole population.
'''
```

```python
'''Main types of distribution shape you should know now

For your level, focus on these 3:

Symmetric / Normal Distribution
Positive Skewed Distribution
Negative Skewed Distribution'''
```

```python
'''1) Symmetric / Normal Distribution

This is the balanced bell-shaped distribution.

left side and right side look almost same
most values are near the center
very few values are at the extreme ends.

Shape:
        /\
      /    \
    /        \
    
Example:

Heights of many students often look close to normal.'''
```

```python
'''2) Positive Skewed Distribution

This is also called:

Right-skewed distribution

because the tail is longer on the right side.

Meaning:

Most values are small or medium, but a few very large values pull the graph to the right.

Shape:
   /\
  /  \
 /    \______

The tail is on the right.

Example of positive skew
[2, 3, 4, 5, 6, 7, 8, 50]

Here:

most values are small
one big value 50 is stretching the data to the right

So this is positively skewed.'''
```

```python
'''Important relation between Mean, Median, Mode

This is often asked.

In positively skewed distribution:
Mean > Median

Why?
Because large values on the right pull the mean more.'''
```

```python
'''3) Negative Skewed Distribution

This is also called:

Left-skewed distribution

because the tail is longer on the left side.

Meaning:

Most values are large, but a few very small values pull the graph to the left.

Shape:
______ 
      \    
       \  /
        \/

The tail is on the left.

Example of negative skew
[2, 40, 42, 45, 46, 47, 48, 50]

Here:

most values are large
one very small value 2 stretches the data to the left

So this is negatively skewed.'''
```

```python
'''In negatively skewed distribution:
Mean < Median

Why?
Because very small values on the left pull the mean downward.'''
```

```python
a=[2, 3, 4, 5, 6, 7, 8, 50]
b=statistics.mean(a)
print('POSITIVE SKEWED :--\n')
print('mean:-',b)
c=statistics.median(a)
print('median:-',c)
print('Mean > Median')
pb.hist(a)
pb.show()
```

```python
a=[2, 40, 42, 45, 46, 47, 48, 50]
b=statistics.mean(a)
print('NEGATIVE SKEWED :--\n')
print('mean:-',b)
c=statistics.median(a)
print('median:-',c)
print('Mean < Median')
pb.hist(a)
pb.show()
```

```python
'''kartosis:--
-Kurtosis tells us how sharp the peak is and how heavy the tails are in a distribution.

In simple words, kurtosis tells:

-Is the graph very sharply peaked?
-Is it flatter?
-Are there more extreme values / outliers in the tails?'''
```

```python
'''3 types of Kurtosis

There are 3 main types:

-Mesokurtic
-Leptokurtic
-Platykurtic'''
```

```python
'''1) Mesokurtic

This is the normal level of kurtosis.

-similar to a normal distribution
-medium peak
-medium tails
Example:

A standard normal distribution is mesokurtic.'''
```

```python
import numpy as np
import matplotlib.pyplot as pd

np.random.seed(42)   # It is used so that every time you run the code, you get the same random numbers.
mesokurtic = np.random.normal(50, 10, 1000)    # (loc,scale,size) loc=the center at which more values will be taken,scale=how spread out the data is,size means total how much elemets/numbers you want.
leptokurtic = np.random.laplace(50, 7, 1000)   # “Create 1000 random values from a Laplace distribution centered at 50 with spread 7, and store them in leptokurtic.”
platykurtic = np.random.uniform(20, 80, 1000)   #start from=20,ends at=80 and generate total of 1000 numbers.

pd.figure(figsize=(25,5))
pd.subplot(1,3,1)
pd.hist(mesokurtic,bins=30)
pd.title('MESOKURTIC:-MEDIUM PEAK,MEDIUM TAILS,EVERYTHING NORMAL')

pd.subplot(1,3,2)
pd.hist(leptokurtic,bins=30)
pd.title('LEPTOKURTIC:-MORE SHARP PEAK,HEAVY/MORE TAILS MORE CHANCE OF EXTREME VALUES')

pd.subplot(1,3,3)
pd.hist(platykurtic,bins=30)
pd.title('PLATYKURTIC:-FLAT TOP,LIGHTER TAIL,FEWER EXTREME VALUES')

pd.show()


'''Why Laplace is used for leptokurtic

Because compared to a normal distribution, Laplace usually has:

more values close to the center
more extreme values in tails

-in NumPy, normal() and laplace() use the same kind of parameter structure:'''
```

```python
'''2) Leptokurtic

This means:

-more peaked / sharper top
-heavier tails
-more chance of extreme values / outliers

Easy idea:

The graph is taller in the center and also has fatter tails.

So:

Leptokurtic = sharp peak + more outliers'''
```

```python
'''3) Platykurtic

This means:

-flatter top
-lighter tails
-fewer extreme values

Easy idea:

The graph looks more spread out and less sharply peaked.

So:

Platykurtic = flatter distribution'''
```
