```python
'''In statistics, probability is the mathematical tool used to measure uncertainty.
It tells us how likely an event is to occur and helps statisticians make predictions and decisions based on data.
A simple way to think about it is:

-Statistics starts with data and tries to understand what probably happened or will happen.
-Probability starts with what is known and predicts what is likely to happen.'''
```

```python
'''Probability is the numerical measure of the chance that an event will occur. Its value always lies between 0 and 1.

0 → Impossible event
1 → Certain event
0.5 → 50% chance of occurring

FORMULA:--PROBABILTY==Number of favorable outcomes / Total number of possible outcomes.

-EXPERIMENT--> EVENT.
-OUTCOME.-->

-SAMPLE SET- IS THE SET OF ALL POSSOBLE TOTAL UNIQUE OUTCOME.(IT IS A SET WHICH DO NOT ALLOW REPEAT.)

-CLASSICAL PROBABILITY:-- NUMBER OF FAVRABLE OUTCOME / TOTAL NUMBER OF OUTCOMES
```

```python
'''TYPE OF EVENT:--
MUTUALLY EXCLUSUVE EVENTS:-  (EXCULSIVE MEANS THAT CANNOT HAPPEN TOGETHER)
-TWO  EVENTS  ARE MUTUALLY EXCLUSUVE IF THEY OCCUR AT THE SAME TIME.
-CHECK AT FAVRABLE OUTCOME.
-INTERSECTION OF A AND B IS 0.

NON-MUTUALLY EXCLUSUVE EVENT:-(CAN HAPPEN TOGETHER)
-EVENTS THAT OCCUR AT THE SAME TIME/TOGETHER.
-INTERSECTION OF A AND B.'''
```

```python
'''INDEPENDENT/DEPENDENT:--
WHEN BOTH HAVE SAME DENOMINATOR-DEPENDENT
WHEN BOTH HAVE DIFFERENT DENOMINATOR - INDEPENDENT.
```

```python
'''COMPLEMENT RULE:-
ADDING THE CHANCES OF HAVING AND NOT HAVING IF WE ADD THAT BOTH WE WILL GET COMPLETE 1.
```

```python
'''ADDITION OF MUTUALLY EXCLUSUVE EVENTS:---(A U B)=A + B
ADDITION OF NON-MUTUALLY EXCLUSUVE EVENT:--(A U B)=A + B - (A ∩ B)'''
```

```python
'''INDEPENDENT/DEPENDENT:-
INDEPENDENT:--P(A ∩ B) = P(A) × P(B) / P(B)
SO IN INDEPENDENT P(A ∩ B) = P(A)
AND IF P(A ∩ B) != P(A) THEN IT IS DEPENDENT.


DEPENDENT EVENT:--
P(A ∩ B) = P(A) × P(B | A)

EXAMPLE:-(DEPENDENT)
Event A: First card is Ace
P(A) = 4/52
After drawing one Ace:
Remaining cards = 51
Remaining Aces = 3
P(B | A) = 3/51
```
