```python
'''Probability is the numerical measure of the chance that an event will occur. Its value always lies between 0 and 1.

0 → Impossible event
1 → Certain event
0.5 → 50% chance of occurring

FORMULA:--PROBABILTY==Number of favorable outcomes / Total number of possible outcomes.

-EXPERIMENT--> EVENT.

-SAMPLE SET- IS THE SET OF ALL POSSOBLE TOTAL UNIQUE OUTCOME.(IT IS A SET WHICH DO NOT ALLOW REPEAT.)

-CLASSICAL PROBABILITY:-- NUMBER OF FAVRABLE OUTCOME / TOTAL NUMBER OF OUTCOMES
```

```python
'''MUTUALLY EXCLUSIVE EVENTS:---

-Mutually exclusive events are two or more events that cannot happen at the same time. If one event occurs, the other event cannot occur.

Definition:-
Two events A and B are mutually exclusive if:

-They have no common outcomes.
-Their intersection is zero:
P(A∩B)=0

This means the probability that both A and B happen together is 0.

* example:-
suppose you have dice:-
-so getting odd number is (1,3,5)
-so getting even number is(2,4,6)
 can both happen together in one roll ?
 -so a number can either be even or odd not both.

THIS MEANS A AND B IS MUTUALLY EXCLUSIVE EVENTS---> CANNOT HAPPEN TOGETHER.'''
```

```python
'''NON-MUTUALLY EXCLUSIVE EVENTS:---
-Non-mutually exclusive events are events that can happen at the same time. They have one or more common outcomes.

*example:-
suppose you draw a card then :-
-(Event A)card is red.
-(Event B)card is king.

-so this both can happen together so A and B is called non-mutually exclusive event.

THIS MEANS A AND B IS NON-MUTUALLY EXCLUSIVE EVENTS---> CAN HAPPEN TOGETHER.'''
```

```python
'''INDEPENDENT/DEPENDENT:--

Independent Events:-

-Two events are independent if the occurrence of one event does not affect the probability of the other event.
-WHEN BOTH HAVE SAME DENOMINATOR-DEPENDENT
example:-
Two Coin Tosses
A = First toss is Heads
B = Second toss is Heads

The second toss is not affected by the first.

These are independent events.
-------------------------------------------------------------------------------------------
Dependent Events:-

-Two events are dependent if the occurrence of one event changes the probability of the other event.
-WHEN BOTH HAVE DIFFERENT DENOMINATOR - INDEPENDENT.
example:-
A deck has 52 cards.

Suppose

A = First card is an Ace.
B = Second card is also an Ace.

If the first card is an Ace, how many cards remain?

Cards left = 51
Aces left = 3   so 3/51'''
```

```python
'''COMPLEMENT RULE:-
-The Complement Rule is one of the simplest and most useful rules in probability.
-If an event is called A, then its complement is written as: A'
-Probability of an event NOT happening = 1 − Probability of it happening.
ADDING THE CHANCES OF HAVING AND NOT HAVING IF WE ADD THAT BOTH WE WILL GET COMPLETE 1.

why is it 1 ?
-because the total probability of all possible outcomes is always: 1

IN SHORT:-
A COMPLEMENT IS ALWAYS THE OPPOSITE OF A MEANS IF A IS 1/20 THEN A'(A COMPLEMENT) IS ALWAYS 19/20.
-If A happens with probability 1/20
Then A' (A complement) means A does not happen, with probability 19/20.
```

```python
'''
ADDITION RULE:- 
MUTUALLY EXCLUSUVE EVENTS:
(A U B)=A + B

NON-MUTUALLY EXCLUSUVE EVENT:
(A U B)=A + B - (A ∩ B)

MULTIPLICATION RULE:-
INDEPENDENT:
P(A∩B)=P(A)P(B)

DEPENDENT:
P(A∩B)=P(A)P(B∣A)

COMPLEMENT RULE:-
P(A′)=1−P(A)
'''
```

```python
'''INDEPENDENT/DEPENDENT:-
INDEPENDENT:--P(A ∩ B) = P(A) × P(B) / P(B)
SO IN INDEPENDENT P(A ∩ B) = P(A)
AND IF P(A ∩ B) != P(A) THEN IT IS DEPENDENT.


DEPENDENT EVENT:--
P(A ∩ B) = P(A) × P(B | A)

EXAMPLE:-(DEPENDENT)
Event A: First card is Ace
P(A) = 4/52
After drawing one Ace:
Remaining cards = 51
Remaining Aces = 3
P(B | A) = 3/51
```

```python
'''Conditional Probability:--
-Conditional Probability = You are given additional information (a condition), and based on that condition, you find the probability of another event.
-It answers questions like:
"What is the probability of Event B, given that Event A has already happened?"

P(A∣B)=P(A∩B)/P(B)

example:--
Playing Cards

A standard deck has 52 cards.

Let:
A = Card is a King
B = Card is a Face card

their are total 12 face card(king,queen,jack==3  * 4 = 12)
now:
A is a king = 4/52
B is a face card = 12/52

now for conditional probabilty :-- (4/52)/(12/52)
we devide A by B why ?
-Because once we know B has happened, our sample space becomes only the outcomes inside B. We ignore everything outside B.
```
