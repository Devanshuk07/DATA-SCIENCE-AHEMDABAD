```python
'''Given the daily order counts for a Swiggy delivery partner over 10 days: [12, 15, 17, 14, 13, 16, 200, 18, 14, 15], 
plot a simple graph (hand-drawn or using any tool) and describe whether the data is left skewed, right skewed, or symmetrical.
Explain your reasoning based on the shape.'''
import numpy as np
import matplotlib.pyplot as mp

deli=[12, 15, 17, 14, 13, 16, 200, 18, 14, 15]
mp.plot(deli)
mp.xlabel('left-skewed',size='12')
mp.title('right-skewed')
mp.show()

'''The distribution is right-skewed (positively skewed) because almost all order counts are clustered between 12 and 18, 
while one extreme value (200) lies far to the right. This outlier stretches the distribution toward higher values, creating a long right tail.'''
```

```python
'''Take the following list of Instagram influencer follower counts (in thousands): [5, 7, 8, 8, 9, 10, 12, 15, 95].
Calculate the mean and median, then state which is higher and what that tells you about the skewness of the data.
<br><br><em><strong>Hint:</strong> If the mean is pulled away from the median, the data is skewed in that direction.</em>'''
import statistics
count=[5000,7000,8000,8000,9000,10000,12000,15000,95000]
a=statistics.mean(count)
print('MEAN :-',a)
b=statistics.median(count)
print('MEDIAN:-',b)

'''HERE MEAN >MEDIAN MEANS THIS IS RIGHT(POSITIVE) SKEWED DATA.THE DATA IS SKEWED IN RIGHT DIRECTION AS ONE VALUE IS 95000 THAT IS OUTLIER 
AND GREATER THAN EVERY ELSE VALUE'''
```

```python
'''Find a real-world example of right-skewed data from any app you use (like Zomato order amounts, Paytm wallet balances, or YouTube video views).
Briefly describe the dataset and explain why you think it is right skewed.'''
import pandas as pd
import matplotlib.pyplot as mp

a=pd.read_csv('zomato_skewed.csv')
print(a)
b=a['Amount (INR)']
mp.boxplot(b)
mp.xlabel('DATA')
mp.show()

'''HERE AS WE CAN SEE FROM BOXPLOT THAT MOSTLY THE ORDER PRIZE LIES BETWEEN RANGE 50-890 RS ROUGHLY BUT THAT DOTS IS OUT OF RANGE THAT MEANS THEY ARE OUTLIERS.
MEANS SOME ORDER HAS PRIZE RANGE ABOVE 900 OR 1000 ALSO SO THAT IS RIGHT SKEWED.RIGHT SKEWED MEANS WHEN THERE ARE NUMBERS IN DATA WHICH HAS MORE VALUES THEN
NORMAL RANGE OF THAT ACTUAL DATA AND IF THEY ARE LARGER THAN IT IS RIGHT SKEWED AND IF THEIR IS LOWER VALUES THEN IT IS LEFT SKEWED DATA.

RULE:-----Skewness is determined by the direction of the longer or more extreme tail.'''
```
