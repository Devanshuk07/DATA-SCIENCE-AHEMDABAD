```python
'''Given the data: Group A (users shown a new Zomato home page) had an average order value of ₹350 from 30 users, and 
Group B (old home page) had an average order value of ₹320 from 30 users. Assume both groups have a standard deviation of ₹50. 
Calculate the p-value using a two-sample t-test (you may use Python, R, or an online calculator), and 
decide whether to accept or reject the null hypothesis at a 0.05 significance level.<br><br><em><strong>Hint:</strong> Use scipy.stats.ttest_ind in Python or any online t-test calculator if you are not comfortable with code.</em>'''
import numpy as np
from scipy import stats

g1_size=30
g2_size=30
sample_size=g1_size+g2_size
g1_avg=350
g2_avg=320
sample_avg=(g1_avg+g2_avg)/2
pop_std=50
alpha=0.05

# calculate standard error   --- just when you take the sample every time it may vary so we have to calculate the sample error.so se (variation) is 12
se=np.sqrt((pop_std**2/g1_size)+(pop_std**2/g2_size))
print(se)

# calculate t-statistics   now when variation is how much bigger that means 30/12==2.32 that is t-statistics. 2.324 times larger than the expected random variation.
t=(g1_avg-g2_avg)/se
print(t)

# calculate degree of freedom to determine the correct t-distribution for calculating the p-value.
df=g1_size+g2_size -2
print(df)

# calcualte p_value
p_val = 2*(1-stats.t.cdf(abs(t),df))
print(p_val)

if p_val<alpha:
    print('reject null hypothesis')
else:
    print('fail to reject null hypothsis')
```
