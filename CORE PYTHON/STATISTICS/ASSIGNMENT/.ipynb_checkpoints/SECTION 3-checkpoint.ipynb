```python
'''Given an array of daily step counts from your fitness app for one week (e.g., [4500, 7000, 5000, 8000, 4000, 9000, 6000]),
calculate the variance by hand using the variance formula and show each calculation step.'''
import statistics
a=[4500, 7000, 5000, 8000, 4000, 9000, 6000]
mean=statistics.mean(a)
print(mean)
# average= 6214

# now:- avg - each value.
sqdadd=[]
for i in a:
    print('\n',(i-mean)**2)   # here we are squareing all the elements after substracting.
    sqdadd.append((i-mean)**2)

# now add every squareed numbers.
print('\nADDITION OF ALL SQUARED NUMBERS :--',sum(sqdadd))

# last is to divide the sum of square / by total elements of variable len(a).
print('\nVARIANCE:--',sum(sqdadd)/len(a))
```

```python
'''Write a Python function called calculate_standard_deviation(scores) that takes a list of exam scores and returns the 
standard deviation rounded to two decimal places.'''
import numpy as np
import pandas as pd
def calculate_std(scores):   # here scores is used to store input that is coming into the function.
    std=np.std(scores)       # here scores is used for calculating the std of data that is stored in scores
    return round(std,2)
    
np.random.seed(42)
marks=np.random.randint(35,100,50)
print(marks)

print(calculate_std(marks))
```

```python
'''Imagine two friends track their daily spending on Zomato over 5 days. Friend A spends [200, 200, 200, 200, 200] and 
Friend B spends [100, 300, 150, 400, 50]. Calculate the standard deviation for both and explain in 2-3 lines who has more consistent spending and why.'''

a=[200, 200, 200, 200, 200]
b=[100, 300, 150, 400, 50]

std1=np.std(a)
std2=np.std(b)
print(f'{std1}\n{std2}')

'''
a has standard deviation = 0

Because there is no spread of numbers at all.so mean also 0.

b has standard deviation ≈ 134.16

Because the numbers are much more spread around the mean.

a = [200, 200, 200, 200, 200]
every value is exactly the same
standard deviation = 0
so spending never changes

b = [100, 300, 150, 400, 50]
values keep changing a lot
standard deviation is large
so spending is not consistent'''


'''Simple rule
Lower standard deviation = more consistency
Higher standard deviation = less consistency'''
```

```python
'''You are given the monthly salary data of 6 employees in a startup: [28000, 29000, 27000, 60000, 26500, 27500]. 
Calculate the variance and standard deviation, then interpret in 2-3 lines what the high or low spread means for this team.'''

data=[28000, 29000, 27000, 60000, 26500, 27500]
var=np.var(data)
std=np.std(data)
print(var)
print(std)

'''here one person salary is 60,000 and other lie between 26,000-29,000 so it means here it has outliers present.also the mean will get high here.'''
```
