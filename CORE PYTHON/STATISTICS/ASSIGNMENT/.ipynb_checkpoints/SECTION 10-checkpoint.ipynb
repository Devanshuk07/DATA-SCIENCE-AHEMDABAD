```python
'''Write a Python function bayes_posterior(prior, likelihood, evidence) that calculates and returns the posterior probability 
using Bayes theorem formula for given values.'''

def bayes_posterior(prior, likelihood, evidence):
    posterior = (likelihood * prior) / evidence
    return posterior

prior = 0.20
likelihood = 0.80
evidence = 0.30

result = bayes_posterior(prior, likelihood, evidence)

print("Posterior Probability:", result)
```

```python
'''Given the following scenario: On Swiggy, 10% of restaurants are 'premium', and 80% of premium restaurants offer a discount, 
while only 20% of regular restaurants do. If you see a discount, use Bayes theorem in code to find the probability that 
the restaurant is premium.<br><br><em><strong>Hint:</strong> Define prior, likelihood, and evidence in your code with comments.</em>'''

'''probability that the restaurant is premium that means :-- P(premium|discount) == '''
prior=0.10
likelihood=0.80
evidence=(0.10*0.80)+(0.20*0.90)     # here 0.90 as only 10% are premium restaurant rest 90% is regular so.

result=bayes_posterior(prior, likelihood, evidence)

print("Posterior Probability:", result)


'''here that trick dont work becuase it is exceptional question so.The problem is that the question does not directly give you P(B).

USING BAYES THEOREM:--
P(A|B) = {P(B|A) * P(A)} / {P(B)}

P(A)      = Prior
P(B|A)    = Likelihood
P(B)      = Evidence
P(A|B)    = Posterior'''
```

```python
'''Simulate a basic spam detection filter: Given a list of 20 email subjects (some spam, some not), and the probability that 'free' appears in spam 
vs non-spam emails, write code to calculate the probability an email is spam if it contains 'free', using Bayes theorem.'''

'''20% of emails are spam.
If an email is spam, it contains the word "Free" 80% of the time.
30% of all emails contain the word "Free".
What is the probability that an email is spam given that it contains the word "Free"?'''

prior=0.20
likelihood=0.80
evidence=0.30

# P(spam|free)
prob=likelihood*prior/evidence
print(prob)
```

```python
'''Use ChatGPT or Copilot to generate a real-world Bayes theorem example related to any app you use (Instagram, Zomato, Flipkart, etc.), 
then write code to solve it and submit both the prompt you used and your code.'''

'''
On Flipkart:

20% of products are Flipkart Assured.
90% of Flipkart Assured products have 4+ star ratings.
40% of non-Assured products have 4+ star ratings.

If you find a product with a 4+ star rating, what is the probability that it is Flipkart Assured?'''

#P(f-assured|4-star)
prior=0.20
likelihood=0.90
evidence=(0.80*0.40)+(0.20*0.90)     # here 20% is f-assured then 80% remaining is not f-assured and last line 40% of non-assured means 80*40.

prob=likelihood*prior/evidence
print(prob)
```
