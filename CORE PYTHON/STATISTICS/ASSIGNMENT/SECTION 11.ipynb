```python
'''Imagine you have a table of 5 Flipkart products with columns for price and quantity sold. 
Represent this data as a matrix in Python, then write a function to calculate the total revenue for each product.
<br><br><em><strong>Hint:</strong> Multiply price and quantity for each row, then print the results.</em>'''

product=np.random.randint(1,6,5)
prize=np.random.randint(2000,100000,5)
sold=np.random.randint(20,50,5)

matrix=np.column_stack((product,sold,prize))

def total_rev(data):
    return data[:,1]*data[:,2]
    
print(f'PRODUCT CODE IS :-{product}\n\nPRIZE OF THE PRODUCT IS \n{prize}\n\n QUANTITY OF PRODUCT SOLD :-\n{sold}')
print('TOTAL REVENUE IS :-\n',total_rev(matrix))
```
