```python
'''Generate 100 random numbers in Python that follow a normal distribution with mean 50 and 
standard deviation 10 using numpy, then plot the bell curve using matplotlib.'''
import numpy as np
import matplotlib.pyplot as mp 

a=np.random.normal(loc=50,scale=10,size=100)
print(a)

mp.hist(a,edgecolor='black')
mp.show()
```

```python
'''Given an array of 20 daily step counts from your fitness tracker app, calculate the mean, median, and mode, and 
check if they are approximately equal as expected in a normal distribution.<br><br><em><strong>Hint:</strong> 
Use numpy.mean(), numpy.median(), and scipy.stats.mode().</em>'''
np.random.seed(1)
from scipy import stats

a=np.random.normal(loc=1200,scale=200,size=20)   # here loc=mean, scale =std deviation  ,,, Standard deviation tells you how far the data is spread out from the mean (average).
print(a)
mean=np.mean(a)
print('\nMEAN :- ',mean)
med=np.median(a)
print('MEDIAN :- ',med)
mod=stats.mode(a)
print(mod)
```

```python
'''Simulate the scores of 200 players in an online game (e.g., PUBG, Free Fire) using a normal distribution with mean 70 and 
standard deviation 15, then count how many players scored within one standard deviation (between 55 and 85) of the mean.
<br><br><em><strong>Hint:</strong> Use the 68-95-99 rule to check your answer.</em>'''
np.random.seed(1)
score=np.random.normal(70,15,200)
#print(score)

btw=score[(score>=55)&(score<=85)]
print('THIS IS THE ONLY SCORE OF PLAYERS WHO ARE BETWEEN 55-85 SCORE :-\n\n',btw)
cnt=len(btw)
print('\nHERE IS THE TOTAL NUMBERS OF PLAYERS WHO SCORED 55-85 IN GAME :-', cnt)
```
