```python
'''WHAT IS HYPOTHESIS:--
-A hypothesis is a tentative statement or prediction that can be tested through research or experimentation.
-THIS MEANS ASSEMPTION.
-THIS HAS ONLY TWO OUTPUTS - SOMETHING CHANGE OR NOTHING CHANGE.
-NO OTHER OUTPUT IS POSSIBLE.
------------------------------------------------------------------------------------------
NULL HYPOTHESIS (H0) :---NOTHING CHANGE/LOSS. FAIR THING IS TAKEN MEANS NO EFFECT,NO DIFFERENCE,NO RELATIONSHIP,JUST NORMAL RANDOM DATA TAKEN.
1) Coin example

Suppose you want to check if a coin is fair.

Null hypothesis H0:

The coin is fair.
Meaning:

Probability of head = 0.5
Probability of tail = 0.5
-------------------------------------------------------------------------------------------
ALTERNATE HYPOTHESIS (H1,Ha) :-- SOMETHING CHANGE/PROFIT.SOMETHING HAS DONE TO DATA AND THAT IS NOT NORMAL/FAIR FOR FURTHER PROCESS.
-Alternative hypothesis is the opposite claim of the null hypothesis.
If null hypothesis says:
“Nothing special is happening / no effect / no difference”

then alternative hypothesis says:
“Something is happening / there is an effect / there is a difference”
-------------------------------------------------------------------------------------------
P-VALUES:-p-value tells us:
A p-value is the probability of observing results at least as extreme as the ones you got, assuming the null hypothesis is true.
| P-value | Interpretation                                               |
| ------- | ------------------------------------------------------------ |
| > 0.05  | Weak evidence against the null hypothesis; do not reject H₀. |
| ≤ 0.05  | Moderate evidence against H₀; often reject H₀.               |
| ≤ 0.01  | Strong evidence their is something done to H₀.               | (be carefull it is just doubt not 100% sure).

-----------------------------------------------------------------------------------------
P-VALUES BIG OR SMALL:--
HERE IF H0 IS TRUE MEANS We assume the “default / no special effect / no real change / no real difference” statement is true,
and then check whether our data looks normal under that assumption.

Imagine a dartboard.
🎯 Bullseye = expected result

If your dart lands exactly in the center (5 heads),
you wouldn't think,
"Something strange happened!"

You'd think,
"Yep, that's normal."
Since your observation is not unusual, the probability of getting it (or something similarly close to expectation) is high.

So, p-value is big.
-If the result is what the null hypothesis would normally produce, then p-value is big.
-If the result is very unusual under the null hypothesis, then p-value is low.
-------------------------------------------------------------------------------------------
TYPE -1 ERROR:-Rejecting a true null hypothesis
-You rejected the null hypothesis even though the null hypothesis was actually true.
You say:
“Something is wrong / something changed”

but in reality:
"nothing was wrong".

EXAMPLE:---Suppose the coin is actually fair in real life.
But after tossing, you get a strange result and you say:
“No, this coin is not fair”

So you reject H0.

But the truth was:
coin was fair.

Then this is a Type 1 error.
--------------------------------------------------------------------------------------------
TYPE -2 ERROR:-
-You do not reject the null hypothesis even though the null hypothesis is actually false.

You say:
“Nothing is wrong / no effect”

but in reality:
something actually is wrong / there is an external-effect.
EXAMPLE:-Suppose in reality the coin is not fair.

But after testing, you say:
“Coin is fair”

or

“I do not reject H0".
Then this is Type 2 error.
```

```python
# EXAMPLE OF P-TYPE:-
import math

sample_mean = 54
population_mean = 50
sigma = 8            # Sigma means population standard deviation.It tells how spread out the data usually is.
n = 16               #  n means sample size ,,how many values are in your sample.
 
z = (sample_mean - population_mean) / (sigma / math.sqrt(n))   # z is the z-score. math.sqrt means square-root of (n).
print("z =", z)

p_value = 0.0455     #  p=2(1−Φ(∣z∣))    here that is standard deviation z-table for in (z) value.that z-table you have to search for google z-table values.
#  Φ(2.00)=0.9772 in z-table by google. 1-0.9772 = 0.0228*2 =0.0456
print("p-value =", p_value)

if p_value < 0.05:
    print("Reject H0")
else:
    print("Do not reject H0")
```

```python
'''the test depends on the problem type:-
1) Z-test

Use when:

-testing a mean
-population standard deviation(VARIANCE SQUARE-ROOT) σ is known

Example:

H₀: mean height = 170
σ known = 10

Then use z-test.
-------------------------------------------------------------------------
2) T-test

Use when:

testing a mean
but population standard deviation is not known

Example:

H₀: mean marks = 50
you only have sample data
no population SD given

Then use t-test.
--------------------------------------------------------------------------
3) Chi-square test

Use when data is counts / categories

Example:

40 people choose tea, coffee, juice
want to know if all choices are equally liked

Then chi-square.
---------------------------------------------------------------------------
4) ANOVA

Use when comparing more than 2 means

Example:

compare average marks of Class A, B, C

Then ANOVA.
---------------------------------------------------------------------------

IN SHORT:--

| Situation                         | Test           |
| --------------------------------- | -------------- |
| Mean test + population SD known   | **Z-test**     |
| Mean test + population SD unknown | **T-test**     |
| Counts / categories               | **Chi-square** |
| Compare 3+ means                  | **ANOVA**      |
'''
```
