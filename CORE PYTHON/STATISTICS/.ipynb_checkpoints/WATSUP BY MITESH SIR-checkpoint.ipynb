```python
'''HOW TO DECIDE WHICH TEST IT IS:-

| Words in the question                                         | Alternative Hypothesis (H₁) | Test Type        |
| ------------------------------------------------------------- | --------------------------- | ---------------- |
| **differs from**, **not equal to**, **is different from**     | μ ≠ μ₀                      | **Two-tailed**   |
| **greater than**, **more than**, **higher than**, **exceeds** | μ > μ₀                      | **Right-tailed** |
| **less than**, **smaller than**, **lower than**, **below**    | μ < μ₀                      | **Left-tailed**  |
'''
```

```python
'''Is a Sample Different from the Population?
Problem: A school claims that the average height of students is 170 cm. A random sample of 50 students has an average height of 165 cm,
with a population standard deviation of 10 cm. At a 5% significance level, can we conclude that the students’ height differs from the
population mean?'''
import math
from scipy import stats

pop_mean=170
sam_size=50
sam_mean=165
pop_std=10
alpha=0.05

# here sam_size = 50 so we use Z-test.

z_score=(sam_mean-pop_mean)/(pop_std/math.sqrt(sam_size))
print(z_score)

p_value=2*(1-stats.norm.cdf(abs(z_score)))     # this is for 2 tailed test.

print(p_value)
if p_value > alpha:
    print("DON'T REJECT THE NULL HYPOTHESIS")
else:
    print('REJECT THE NULL HYPOTHESIS')
```

```python
alpha=0.05
def hypothesis_value(p_value,alpha):
    if p_value > alpha:
        print("DON'T REJECT THE NULL HYPOTHESIS")
    else:
        print('REJECT THE NULL HYPOTHESIS')
```

```python
'''Is the Sample Mean Smaller?
Problem: The national average score in math is 75. A school wants to test if their students' average score is less than 75. A sample of 36
students has an average score of 72, and the population standard deviation is 8. Use ad test at a 5% significance level.'''

sam_size=36
pop_mean=75
sam_mean=72
pop_std=8
alpha=0.05

z_score=(sam_mean-pop_mean)/(pop_std/ math.sqrt(sam_size))
print(z_score)

p_value=stats.norm.cdf(z_score)   # this is for left tailed test.
print(p_value)

hypothesis_value(p_value,alpha)
```

```python
'''Before and After Treatment
Problem: A fitness trainer measures the weight of 8 people before and after a 4-week training program. Their weights (in kg) are as
follows:
Before: [80, 85, 78, 90, 95, 88, 76, 82]
After: [78, 84, 76, 88, 94, 87, 74, 81]'''

bef=[80, 85, 78, 90, 95, 88, 76, 82]
aft=[78, 84, 76, 88, 94, 87, 74, 81]

sam_size=len(bef)
alpha=0.05

t_statistics,p_value=stats.ttest_rel(bef,aft)
print(p_value)

hypothesis_value(p_value,alpha)
```

```python
'''Problem: A factory produces bolts with an average diameter of 5 mm. A sample of 10 bolts has the following diameters:
4.9, 5.1, 4.8, 5.0, 5.2, 4.9, 5.0, 5.1, 5.0, 4.8 Can we conclude at a 5% significance level that the sample mean is different from 5 mm?'''
import numpy as np

pop_mean=5
sam_size=10
bolts=[4.9, 5.1, 4.8, 5.0, 5.2, 4.9, 5.0, 5.1, 5.0, 4.8]
sam_mean=np.mean(bolts)
alpha=0.05

t_statistics,p_value=stats.ttest_1samp(bolts,pop_mean)
print(p_value)
hypothesis_value(p_value,alpha)
```
