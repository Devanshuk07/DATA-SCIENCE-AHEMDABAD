```python
'''WHAT IS CORRELATION ?

-Correlation in statistics is a measure of how strongly and in what direction two variables are related.
It tells you whether changes in one variable tend to be associated with changes in another.

There are three main types of correlation:

(1)Positive correlation: As one variable increases, the other also increases.
Example: The more hours a student studies, the higher their test score tends to be.

(2)Negative correlation: As one variable increases, the other decreases.
Example: The faster a car travels over a fixed distance, the less time it takes to complete the journey.

(3)No correlation: There is no consistent relationship between the two variables.
Example: A person's shoe size and their favorite color.

Correlation is a statistical tool used to describe the strength and direction of the relationship between two variables,
helping researchers identify patterns in data.'''
```

```python
'''WHAT IS PEARSON CORRELATION ?

-Pearson Correlation is the most commonly used method to measure the strength and direction of a linear relationship between two numerical variables.
-Its value always lies between -1 and +1.

| r value | Meaning                       |
| ------: | ----------------------------- |
|      +1 | Perfect positive correlation  |this
|    +0.8 | Strong positive correlation   |
|    +0.5 | Moderate positive correlation |
|       0 | No linear correlation         |this
|    -0.5 | Moderate negative correlation |
|    -0.8 | Strong negative correlation   |
|      -1 | Perfect negative correlation  |this

EXAMPLE:-

CASE 1 :
IF X IS 100 / IT BECOME 120
IF Y IS 100 / IT ALSO INCRESE TO 120 

THIS MEANS PERFECT POSITIVE CORRELATION, VALUE = 1.

IF X INCRESE TO 120 
Y INCRESE TO 110.

IT IS POSSITIVE CORRELATION , VALUE = 0.5

IN A POSSITIVE CORRELATION VALUE CAN BE IN RANGE OF 0 to 1.

CASE 2:
IF X IS 100 / IT BECOME 120
IF Y IS 100 / IT DECREASE TO 80

THIS MEANS PERFECT NEGATIVE CORRELATION, VALUE = -1.

IF X IS 100 / IT BECOME 120
IF Y IS 100 / IT  DECREASE TO 90.

THIS MEANS NEGATIVE CORRELATION, VALUE = -0.5.

NOTE:-- IF VALUE IS 1 OR -1 THEN AND THEN IT WILL BE TOLD PERFETC NEG- OR POSI+.
IT WILL BE IN RANGE OF -1 TO 1 ONLY AND IF IT IS -1 OR 1 IT WILL BE PERFECT.NOW IF THIER IS INCRESE OF 20 AND
SAME INCRESE OF Y IS 20 THEN PERFECT POSITIVE 1 CORRELATION.
IF X INCRESE TO 20 AND Y TO -20 THEN PERFECT NEGATIVE -1 CORRELATION.
ALSO IF X DECRESE TO -20 AND Y INCRESE TO 20 THEN PERFECT NEGATIVE -1 CORRELATION'''
```

```python
'''WHAT IS SPEARMAN CORRELATION ?
-Spearman Correlation (also called Spearman's Rank Correlation) measures the strength and direction of a monotonic relationship between two variables 
using their ranks instead of their actual values.

In simple words:

-Pearson compares the actual numbers.
-Spearman compares the ranks (positions) of those numbers.

WHY WE NEED SPEARMAN ?

Sometimes the data:

-isn't normally distributed,
-contains outliers,
-or doesn't follow a straight-line (linear) relationship.

In those cases, Pearson may not work well.
Spearman is more appropriate because it only looks at the order (rank) of the values.
```

```python
# EXAMPLE OF 

import numpy as np
import pandas as pd
import statistics
from scipy.stats import pearsonr,spearmanr
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(1)
data=pd.DataFrame({'study hours':np.random.randint(1,10,30),
                  'sleep hours':np.random.randint(4,8,30)})
data['marks']=data['study hours']*8+np.random.randint(-5,5,30)    # this is for marks study hours let be 8 then 8*8=64 then + np is for variation that may be -5 to 5.
data['rank']=data['marks'].rank(ascending=False)  # ascending=false everytime when we need rank in marks.
data.head()

corr,p_value= pearsonr(data["study hours"],data["marks"])     # pearsonr() is a function from SciPy that is used to measure the linear relationship between two numerical variables.it gives two output How strongly are the two variables related? → Correlation coefficient (r) and and Is that relationship statistically significant? → p-value
print(corr)    # this is for python
#print(data["study_hours"].corr(data["marks"],method="pearson"))  # this is for python.

# here 0.99 means +1 perfect positi correlation
#if it is -0.99 it is perfect neg- correlation 
```

```python
pearson_matrix = data.corr(method="pearson")
spearman_matrix = data.corr(method="spearman")
print(pearson_matrix)
print(spearman_matrix)
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(pearson_matrix,annot=True,cmap="Blues")
plt.title("Pearson Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(spearman_matrix,annot=True,cmap="pink")
plt.title("Spearman Correlation Heatmap")
plt.show()
```
