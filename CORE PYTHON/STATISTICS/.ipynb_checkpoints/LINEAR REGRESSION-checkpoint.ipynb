```python
'''WHAT IS LINEAR REGRESSION ?
-Linear Regression is a statistical method used to find the relationship between an independent variable (X) and a dependent variable (Y).

It answers questions like:

If study hours increase, how do marks change?
If advertising increases, how do sales change?
If house area increases, how does price change?

Why is it called "Linear"?

Because the relationship is a straight line.
------------------------------------------------------------------------------
Goal of Linear Regression

Find the line that predicts Y as accurately as possible.

It tries to make the prediction errors as small as possible.

Example:

Actual Marks = 80

Predicted Marks = 78'''
```

```python
'''Types of Linear Regression:--

1. One input variable.
Study Hours → Marks
One X and one Y.

2. Multiple Linear Regression
More than one input variable.
Marks =
f(Study Hours,
Attendance,
Sleep Hours)
```

```python
'''EXAMPLE :--'''
import numpy as np
from sklearn.linear_model import LinearRegression

hours = np.array([1, 2, 3, 4, 5]).reshape(5,1)   # here numpy works with 1-d and hours are in 1-d but LinearRegression expects a 2D array ,,reshape(5,1) means 5-rows and 1-column and to changes it into a 2D array for Linear-recurssion.
marks = np.array([40, 50, 60, 70, 80])

model = LinearRegression()           # This creates a Linear Regression object. it is empty right now. 

model.fit(hours,marks)          # so we fit in it the variables with values. model has learned the relationship between X and Y.

prediction = model.predict([[6]])       # we are asking him What if someone studies 6 hours?

print(prediction)

# The model predicts that studying 6 hours would correspond to about 90 marks.
```
