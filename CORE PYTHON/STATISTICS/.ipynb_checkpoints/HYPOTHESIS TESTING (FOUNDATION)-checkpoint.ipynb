```python
'''WHAT IS HYPOTHESIS:--
-A hypothesis is a tentative statement or prediction that can be tested through research or experimentation.
-THIS MEANS ASSEMPTION.
-THIS HAS ONLY TWO OUTPUTS - SOMETHING CHANGE OR NOTHING CHANGE.
-NO OTHER OUTPUT IS POSSIBLE.
------------------------------------------------------------------------------------------
NULL HYPOTHESIS (H0) :---NOTHING CHANGE/LOSS. FAIR THING IS TAKEN MEANS NO EFFECT,NO DIFFERENCE,NO RELATIONSHIP,JUST NORMAL RANDOM DATA TAKEN.
1) Coin example

Suppose you want to check if a coin is fair.

Null hypothesis H0:

The coin is fair.
Meaning:

Probability of head = 0.5
Probability of tail = 0.5
-------------------------------------------------------------------------------------------
ALTERNATE HYPOTHESIS (H1,Ha) :-- SOMETHING CHANGE/PROFIT.SOMETHING HAS DONE TO DATA AND THAT IS NOT NORMAL/FAIR FOR FURTHER PROCESS.
-Alternative hypothesis is the opposite claim of the null hypothesis.
If null hypothesis says:
“Nothing special is happening / no effect / no difference”

then alternative hypothesis says:
“Something is happening / there is an effect / there is a difference”
-------------------------------------------------------------------------------------------'''
```

```python
'''P-VALUES:-p-value tells us:
A p-value is the probability of observing results at least as extreme as the ones you got, assuming the null hypothesis is true.
| P-value | Interpretation                                               |
| ------- | ------------------------------------------------------------ |
| > 0.05  | Weak evidence against the null hypothesis; do not reject H₀. |
| ≤ 0.05  | Moderate evidence against H₀; often reject H₀.               |
| ≤ 0.01  | Strong evidence their is something done to H₀.               | (be carefull it is just doubt not 100% sure).

-----------------------------------------------------------------------------------------

P-VALUES BIG OR SMALL:--
HERE IF H0 IS TRUE MEANS We assume the “default / no special effect / no real change / no real difference” statement is true,
and then check whether our data looks normal under that assumption.

Imagine a dartboard.
🎯 Bullseye = expected result

If your dart lands exactly in the center (5 heads),
you wouldn't think,
"Something strange happened!"

You'd think,
"Yep, that's normal."
Since your observation is not unusual, the probability of getting it (or something similarly close to expectation) is high.

So, p-value is big.
-If the result is what the null hypothesis would normally produce, then p-value is big.
-If the result is very unusual under the null hypothesis, then p-value is low.
-------------------------------------------------------------------------------------------

How do we actually find the p-value?

-There are several methods.
Method 1: Using a Z-table
Suppose;
z = 2.00

NOW VALUE IS IF Z=2 IN Z TABLE IS 0.9772
THEN P-VALUE IN RIGHT - TAILER TEST = 1-0.9772=0.0228.

IF Z=2
Z-TABLE VALUE = 0.0228

THEN FOR LEFT TAILES TEST P-VALUE = 0.0228(SAME)

IF Z=2
Z-TABLE VALUE = 0.9772

THEN FOR 2-TAILED TEST P-VALUE = 2 * 0.0228 = 0.0456.
-------------------------------------------------------------------------------------
Why is it called a "probability value"?

Because it is literally a probability.

It always lies between:

0 and 1

Examples:

0.001
0.02
0.18
0.75
----------------------------------------------------------------------------------------
Does a smaller p-value mean stronger evidence?

Yes.

p-value	Meaning
0.50	Very weak evidence against H₀
0.20	Weak evidence
0.05	Borderline significance
0.01	Strong evidence against H₀
0.001	Very strong evidence against H₀
----------------------------------------------------------------------------------------
PERFECT BEST REAL-LIFE EXAMPLE:--

Think of it like a court trial ⚖️

Suppose:

H₀: The accused is innocent.
H₁: The accused is guilty.

The court starts by assuming innocence (just as hypothesis testing starts by assuming H₀ is true).

Now evidence is presented.

Case 1: Weak evidence

Imagine the evidence is weak.

There is a 70% chance that such evidence could appear even if the accused were innocent.

So,

p-value = 0.70

This means:

"This evidence is not unusual if the accused is innocent."

So the court says:

✅ We don't have enough evidence to reject innocence.

This is like failing to reject H₀.
----------------------------------------------------------------------------------------
ALWAYS TAKE LIKE THIS HOW MUCH PERCENT CHANCE IS IT TO BE ?
IF IT IS 30% MEANS 0.03 MEANS IT IS RARE MEANS SURPRISING SO REJECT H0.
IF IT IS 70% MEANS 0.07 MEANS IT IS LIKELY TO BE HAPPEN NOT SURPRISING SO GO WITH H0.'''
```

```python
# EXAMPLE OF P-TYPE(P-VALUE) PROBABILITY-VALUE:-
import math

sample_mean = 54
population_mean = 50
sigma = 8            # Sigma means population standard deviation.It tells how spread out the data usually is.
n = 16               #  n means sample size ,,how many values are in your sample.
 
z = (sample_mean - population_mean) / (sigma / math.sqrt(n))   # z is the z-score. math.sqrt means square-root of (n).
print("z =", z)

p_value = 0.0455     #  p=2(1−Φ(∣z∣))    here that is standard deviation z-table for in (z) value.that z-table you have to search for google z-table values.
#  Φ(2.00)=0.9772 in z-table by google. 1-0.9772 = 0.0228*2 =0.0456
print("p-value =", p_value)

if p_value < 0.05:
    print("Reject H0")
else:
    print("Do not reject H0")
```

```python
'''TYPE -1 ERROR:-Rejecting a true null hypothesis
-You rejected the null hypothesis even though the null hypothesis was actually true.
You say:
“Something is wrong / something changed”

but in reality:
"nothing was wrong".

EXAMPLE:---Suppose the coin is actually fair in real life.
But after tossing, you get a strange result and you say:
“No, this coin is not fair”

So you reject H0.

But the truth was:
coin was fair.

Then this is a Type 1 error.
--------------------------------------------------------------------------------------------
TYPE -2 ERROR:-
-You do not reject the null hypothesis even though the null hypothesis is actually false.

You say:
“Nothing is wrong / no effect”

but in reality:
something actually is wrong / there is an external-effect.
EXAMPLE:-Suppose in reality the coin is not fair.

But after testing, you say:
“Coin is fair”

or

“I do not reject H0".
Then this is Type 2 error.'''
```

```python
'''the test depends on the problem type:-
1) Z-test

Use when:

-testing a mean
-population standard deviation(VARIANCE SQUARE-ROOT) σ is known

Example:

H₀: mean height = 170
σ known = 10

Then use z-test.
-------------------------------------------------------------------------
2) T-test

Use when:

testing a mean
but population standard deviation is not known

Example:

H₀: mean marks = 50
you only have sample data
no population SD given

Then use t-test.
--------------------------------------------------------------------------
3) Chi-square test

Use when data is counts / categories

Example:

40 people choose tea, coffee, juice
want to know if all choices are equally liked

Then chi-square.
---------------------------------------------------------------------------
4) ANOVA

Use when comparing more than 2 means

Example:

compare average marks of Class A, B, C

Then ANOVA.
---------------------------------------------------------------------------

IN SHORT:--

| Situation                         | Test           |
| --------------------------------- | -------------- |
| Mean test + population SD known   | **Z-test**     |
| Mean test + population SD unknown | **T-test**     |
| Counts / categories               | **Chi-square** |
| Compare 3+ means                  | **ANOVA**      |
'''
```

```python
'''HOW TO DECIDE WHICH TEST IT IS:-

| Words in the question                                         | Alternative Hypothesis (H₁) | Test Type        |
| ------------------------------------------------------------- | --------------------------- | ---------------- |
| **differs from**, **not equal to**, **is different from**     | μ ≠ μ₀                      | **Two-tailed**   |
| **greater than**, **more than**, **higher than**, **exceeds** | μ > μ₀                      | **Right-tailed** |
| **less than**, **smaller than**, **lower than**, **below**    | μ < μ₀                      | **Left-tailed**  |
'''
```

```python
# COMPLETE Z-TEST PROGRAM PERFORMING 2-TAILED TEST USING P-VALUE APPROUCH:--

from scipy import stats
import math

sample_mean = 165
population_mean = 170
population_std = 10                # this is standard deviation ,,this tell us how-much is our data spread.
sample_size = 50
alpha = 0.05

# Z-Test  (Two Tailed)
# Calculate Z_SCORE

z_score = (sample_mean-population_mean)/(population_std/math.sqrt(sample_size))

# Calculate p-value
#In a two-tailed test, we don't care whether the test statistic is positive or negative. We only care how far it is from 0.

p_value = 2 * (1-stats.norm.cdf(abs(z_score)))              #stats.norm.cdf() always calculate (area to the left). Abs removes negative sign.cdf=Cumulative Distribution Function

print("Z_Score:",z_score)
print("p_value:",p_value)

if p_value<alpha:
  print("Reject the Null Hypothesis")
else:
  print("Fail to reject the Null Hypothesis")
```

```python
#COMPLETE Z-TEST PROGRAM PERFORMING RIGHT-TAILED TEST USING P-VALUE APPROUCH:--

sample_mean = 72
population_mean = 75
population_std = 8
sample_size = 36
alpha = 0.05


# Z-Test  (One Tailed)

# Calculate Z_SCORE

z_score = (sample_mean-population_mean)/(population_std/math.sqrt(sample_size))

# Calculate p-value

p_value = 1-stats.norm.cdf(abs(z_score))              # 1-stats.norm.cdf===Area to the right of z

print("Z_Score:",z_score)
print("p_value:",p_value)

if p_value<alpha:
  print("Reject the Null Hypothesis")
else:
  print("Fail to reject the Null Hypothesis")
```

```python
'''This is why, in hypothesis testing, we often use:

Left-tailed test: stats.norm.cdf(z)
Right-tailed test: 1 - stats.norm.cdf(z)
Two-tailed test: 2 * (1 - stats.norm.cdf(abs(z)))

These three formulas cover the most common Z-test p-value calculations.'''
```

```python
def hypothesis_result(p_value,alpha):
  if p_value<alpha:
    print("Reject Null Hypothesis")
  else:
    print("Fail to reject Null Hypothesis")
```

```python
'''T-TEST FULL EXPLANATION:---

ttest_1samp()--One Sample:---One sample vs one known population mean.

-----------------------------------------------------------------------------------
ttest_rel()--Related (Paired):----Used when the observations are related.
Examples:

Weight before and after diet
Blood pressure before and after medicine
Student marks before and after coaching
-----------------------------------------------------------------------------------
ttest_ind()--independent:--Used when the two groups are completely different.
example:-
boys = [70, 75, 80, 78]
girls = [65, 68, 70, 69]
'''
```

```python
# T-TEST FOR ONE SAMPLE:---

sample_data = [4.9,5.1,4.8,5.0,5.2,4.9,5.0,5.1,5.0,4.8]
population_mean = 5.0
sample_size = 10
alpha=0.05

# T-test (One Sample Test)

t_statistic,p_value = stats.ttest_1samp(sample_data,population_mean)   # ttest_1samp() is a built-in SciPy function that performs a One-Sample T-Test.  Stats.ttest_1samp(sample_data,population_mean) this always return two values one is t-test and other is p-value.

print(t_statistic)
print(p_value)
hypothesis_result(p_value,alpha)
```

```python
# T-TEST FOR RELATED PAIRED:--

before = [80,85,78,90,95,88,76,82]
after =  [78,84,76,88,94,87,74,81]

sample_size = 8
alpha = 0.05

t_statistic,p_value = stats.ttest_rel(before,after)

print(p_value)
hypothesis_result(p_value,alpha)
```

```python
#T-TEST FOR INDEPENDENT :---

boys = [70, 75, 80, 78]
girls = [65, 68, 70, 69]

t_statistics,p_value=stats.ttest_ind(boys,girls)

print(t_statistics)
print(p_value)
hypothesis_result(p_value,alpha)
```
