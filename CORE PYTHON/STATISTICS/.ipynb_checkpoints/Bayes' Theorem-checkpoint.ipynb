```python
'''WHAT IS BAYES THEOREM:-
-Bayes' Theorem helps us reverse a conditional probability.

P(A|B) = {P(B|A) * P(A)} / {P(B)}    here the before | is you find and after | is conditon.

Suppose:

A = Person has the disease.
B = Test result is positive.
Then:

P(A∣B) → Probability that the person has the disease given the test is positive. (This is what we want to find.)
P(B∣A) → Probability that the test is positive given the person has the disease. (Usually known from test accuracy.)
P(A) → Probability that a random person has the disease.
P(B) → Probability that a random person gets a positive test.

Why do we need Bayes' Theorem?
-Conditional probability gives:

P(A∣B)=P(A∩B) / P(B)

Sometimes we don't know P(A∩B), but we do know:
P(B∣A)
P(A)
P(B)
Bayes' Theorem lets us calculate P(A∣B) using these values.'''
```

```python
'''Simple Example

Suppose:

5% of people have a disease.
If someone has the disease, the test is positive 90% of the time.
10% of all people test positive.

find :--   P(Disease∣Positive)
Let

A = Person has the disease.
B = Test is positive.

so, 5% of people have a disease. P(A)=0.05

If someone has the disease, the test is positive 90% of the time.
The word "given" tells us this is a conditional probability.
P(B∣A)=0.90

10% of all people test positive.
P(B)=0.10

USING BAYES THEOREM:--
P(A|B) = {P(B|A) * P(A)} / {P(B)}
P(A∣B)=0.90 × 0.05 / 0.10 = 0.45

so ,      P(Disease∣Positive)=45%

This means:

Even if your test is positive, there is only a 45% chance that you actually have the disease, based on these numbers.'''
```

```python
'''ANOTHER EXAMPLE:--
A company finds:

20% of emails are spam.
If an email is spam, it contains the word "Free" 80% of the time.
30% of all emails contain the word "Free".
What is the probability that an email is spam given that it contains the word "Free"?
so :--

by trick :---
after reading question just put it as told in question...
P(spam|free)
now we have to find which word tells the figure for free that is 30%.
0.80 * 0.20 / 0.30 
```

```python
'''ANOTHER EXAMPLE:--
Suppose you're told:

60% of students passed the exam.
80% of students who studied passed the exam.
40% of students studied.
Find the probability that a student studied given that they passed.

so by trick:-
after reading question just put it as told in question...
P(study|pass)
now ask yourself which figure tells the probabilty of pass ? == 0.60 1st line.

formula:-
0.80 * 0.40 / 0.60
```

```python
'''Easy Memory Trick

Think of Bayes' Theorem as reversing the question.

Conditional Probability asks:

"If I know B happened, what is the probability of A?"

Bayes' Theorem says:

"I know the probability of B given A. Can I find the probability of A given B?'''
```
