```python
'''What is Normal Distribution?

A normal distribution is a probability distribution in which:

-Most values are near the mean.
-Fewer values occur as you move away from the mean.
-The graph is bell-shaped and perfectly symmetric.

Examples:

-Human height
-IQ scores
-Blood pressure
-Exam marks (approximately)
-Measurement errors
----------------------------------------------------------------------
SHAPE:--
              Mean
               |
               |
           /\ 
         /    \
       /        \
_____/            \_____
HEIGHT POINT = MEAN
-----------------------------------------------------------------------
Main Properties
1. Bell Shaped
Looks like a bell.

2. Symmetric
Left side = Right side
<------|------>
      Mean
      
3.Mean = Median = Mode
All three are exactly equal.

4. Total Probability = 1
------------------------------------------------------------------------
If data follows a normal distribution:

68% of values lie within ±1σ
95% of values lie within ±2σ
99.7% of values lie within ±3σ

          68%
      <--------->
   μ-1σ        μ+1σ

        95%
 <-------------------->
 μ-2σ             μ+2σ

       99.7%
<-------------------------->
μ-3σ                 μ+3σ

EXAMPLE:--
Mean = 100

Standard deviation = 10

68%
90 --------100--------110

95%
80 -------------100-------------120

97%
70 ----------------100----------------130'''
```

```python
'''A Z-score (Standard Score) tells you how far a value is from the mean, measured in standard deviations.

Z-score = Distance from the mean ÷ Standard deviation

QUESTION :-
Calculate the Z-score
Question

Average marks = 70

Standard deviation = 10

Student scored = 90

Find the Z-score ?

-90-70 /10 =2--->ANSWER'''
```

```python
'''Probability Using the Empirical Rule
Question

Mean = 50

SD = 5

Approximately what percentage of students scored between 45 and 55?

Answer

45 = Mean − 1 SD

55 = Mean + 1 SD

Using the empirical rule:

Answer = 68%

RULE:--
          68%
      <--------->
   μ-1σ        μ+1σ

        95%
 <-------------------->
 μ-2σ             μ+2σ

       99.7%
<-------------------------->
μ-3σ                 μ+3σ

'''
```

```python
'''What is the difference between a Normal Distribution and a Standard Normal Distribution?

| Normal Distribution          | Standard Normal Distribution |
| ---------------------------- | ---------------------------- |
| Mean can be any value        | Mean = 0                     |
| SD can be any positive value | SD = 1                       |
| Uses X values                | Uses Z values                |


What does a positive Z-score mean?
Answer: The value is above the mean.

What does a negative Z-score mean?
Answer: The value is below the mean.

What does a Z-score of 0 mean?
Answer: The value is exactly equal to the mean.'''
```
