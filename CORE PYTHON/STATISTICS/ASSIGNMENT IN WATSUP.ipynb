```python
'''1. The daily sales (in units) of a retail store for 10 days are:
45, 52, 48, 60, 57, 50, 49, 55, 58, 46
Calculate:
a) Mean
b) Median
c) Mode (if any)

State which measure best represents the data and justify your answer.'''


a=[45, 52, 48, 60, 57, 50, 49, 55, 58, 46]
a.sort()
print('A AFTER SORTED :--',a)

# mean
mean=sum(a)/len(a)
print('MEAN :-',mean)

# median
n=len(a)
if n%2!=0:
    median=a[n//2]
else:
    median=(a[n//2-1]+a[n//2])/2
print('MEDIAN IS :-',median)

# mode
'''HERE IT HAS NO MODE BECAUSE MODE IS WHEN ANY ELEMENT KEEPS REPETING ITSELF IS A MODE.'''


'''here best measure MEAN as the data values are balanced.Also their is no outliers in this.As the values are spread normally 
around the center the mean gives the best over view of the data.'''
```

```python
'''2. A student scored the following marks in seven subjects:

72, 81, 65, 90, 84, 76, 82

Calculate:
a) Mean
b) Sample Variance
c) Sample Standard Deviation
'''

a=[72, 81, 65, 90, 84, 76, 82]
a.sort()
print('A AFTER SORTED :--',a)

# mean
mean=round(statistics.mean(a),2)
print('MEAN :--',mean)

#variance
var=[]
for i in a:
    var.append((i-mean)**2)
print('\nWE HAVE SUBSTRACTED MEAN FROM EVERY NUMBER AND THEN SQUARE EACH NUMBER:---',var)
addsqr=sum(var)
print('\nHERE WE HAVE ADDED EVERY SQUARED NUMBERS:--',addsqr)
variance=round(addsqr/(len(a)-1),2)
print('\nLAST IS WE DIVIDED SQUARED TOTAL BY TOTAL ELEMENTS IN A $$$ THIS IS VARIANCE:--',variance)

# STANDARD VARIANCE
print('\nTHIS IS STANDARD VARIANCE :--',round((variance)**0.5,2))
```

```python
'''3. A factory recorded the number of defective items produced each day:

3, 5, 4, 6, 5, 7, 8, 4, 5, 6

Calculate:
a) Mean
b) Population Variance
c) Population Standard Deviation'''

a=[3, 5, 4, 6, 5, 7, 8, 4, 5, 6]
a.sort()
print('A AFTER SORTED :--',a)

# mean
mean=round(statistics.mean(a),2)
print('MEAN :--',mean)

#variance
var=[]
for i in a:
    var.append((i-mean)**2)
print('\nWE HAVE SUBSTRACTED MEAN FROM EVERY NUMBER AND THEN SQUARE EACH NUMBER:---',var)
addsqr=sum(var)
print('\nHERE WE HAVE ADDED EVERY SQUARED NUMBERS:--',addsqr)
variance=round(addsqr/len(a),2)
print('\nLAST IS WE DIVIDED SQUARED TOTAL BY TOTAL ELEMENTS IN A $$$ THIS IS VARIANCE:--',variance)

# STANDARD VARIANCE
print('\nTHIS IS STANDARD VARIANCE :--',round((variance)**0.5,2))
```

```python
'''Explain the purpose of the Null Hypothesis and Alternate Hypothesis in statistical analysis.
Why do we formulate both before conducting any hypothesis test?

In statistical analysis, the Null Hypothesis (H₀) and Alternative Hypothesis (H₁) are formulated to test a claim 
about a population using sample data.

Null Hypothesis (H₀) is the default assumption. It usually states that there is no effect, no difference, or 
no change in the population parameter. 
For example, if a coaching institute claims that the average marks of students are 75,the null hypothesis can be written as:
H₀= 75

Alternative Hypothesis (H₁) is the opposite of the null hypothesis.
It states that there is an effect, difference, or change.
In the same example, the alternative hypothesis may be:
H₁: μ ≠ 75

We formulate both hypotheses before conducting a hypothesis test because they provide a clear structure for statistical decision-making.
The null hypothesis acts as the starting assumption, and the alternative hypothesis represents the claim we want to check if the sample data 
provides enough evidence for it.

By comparing the sample result with what we would expect under the null hypothesis, we decide whether to reject H₀ or fail to reject H₀.
Thus, both hypotheses are necessary because they define the claim being tested, guide the type of test to be used, and help us draw a 
proper conclusion from the sample data.'''
```
