```python
'''What is requests?
-requests is a Python library used to send HTTP requests to a server/API.

For example, if an API has some data about users:

import requests
response = requests.get("https://example.com/api/users")

Here:

Python → sends request
Server/API → processes request
Server → sends response
response → contains that response

Think of it like:

You → API request → Server → API response → You.
```

```python
'''requests.get()
-GET is mainly used when you want to get/read data from a server.

import requests
response = requests.get("https://api.example.com/users")

print(response.status_code)
print(response.json())          # Take the JSON data that the server sent back and convert it into a Python object (usually a dictionary), so Python can work with it.

HERE STATUS.CODE :-- TELL YOU WEATHER THE REQUEST WAS SUCCESSFUL OR NOT ?
----------------------------------------------------------------------------------

print(response.json())
.json() :--APIs commonly send data in JSON format.

For example, server might return:
{
    "name": "Devanshu",
    "age": 29,
    "city": "Rajkot"
}
-----------------------------------------------------------------------------------
DIFFERNET STATUS_CODES :--
| Code  | Meaning              |
| ----- | -------------------- |
| `200` | Successful           |
| `201` | Created successfully |
| `400` | Bad request          |
| `401` | Unauthorized         |
| `403` | Forbidden            |
| `404` | Not found            |
| `429` | Too many requests    |
| `500` | Server error         |
```

```python
'''requests.post():---

POST is generally used when you want to send/create data on a server.

For example, imagine an API for registering users.

EXAMPLE CODE:-

import requests

data = {
    "name": "Devanshu",
    "age": 29
}

response = requests.post(
    "https://api.example.com/users",
    json=data                                  # json=data means: 👉 "Send this Python dictionary to the server as JSON."
)

print(response.status_code)
print(response.json())
```

```python
'''WHAT IS QUERY PARAMETERS :--

Now suppose an API contains 1000 users.

You don't necessarily want all 1000.

Maybe you want users from India.

You can send extra information through query parameters.

For example, the URL could look like:
-https://api.example.com/users?country=India

SO HERE :-
?country=India  is the query parameter. ? starts the query parameters.
-----------------------------------------------------------------------------------------------------------
BUT IN PYTHON WE DONT MANUALLY CREATE PARAMETERS ,, INSTEAD WE DO LIKE THIS :--
params = {                  # WE CREATE A VARIABLE INSTEAD.
    "country": "India"
}

response = requests.get(
    "https://api.example.com/users",
    params=params
)
-----------------------------------------------------------------------------------------------------------
WE CAN ALSO CREATE MULTIPLE PARAMS :--

params = {
    "country": "India",
    "age": 29
}

response = requests.get(
    "https://api.example.com/users",
    params=params
)
```
