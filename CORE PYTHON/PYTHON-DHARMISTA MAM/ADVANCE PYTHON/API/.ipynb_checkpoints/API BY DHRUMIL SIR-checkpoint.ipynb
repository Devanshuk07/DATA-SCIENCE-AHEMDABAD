```python
'''2 TYPES OF API:--VERYTHING IS USED BY API ONLY WITHOUT IT NOTHING CAN WORK.

1. REST API - ALSO CAN BE USED IN LOW / HIGH BANDWIDTH // DONT FOLLOW ANY RULE OR PROTOCOL // 
ALSO LIGHT WEIGHT:- SIZE WILL BE LEST JUST LIKE IF 10 MB.
EXAMPLE:-GENERAL USAGE.----> FOR GENERAL WEBSITES.

2. SOAP API :--CAN BE USED ONLY FOR HIGH BANDWIDTH // ALSO PROTOCOL/RULES HAS TO BE FOLLOW.
HEAVY WEIGHT :- SIZE WILL BE MORE THAN REST LIKE 15 MB.
EXAMPLE :- FOR SECUREITY CONSERN,BANKING,LOGIN PROCESS.----> FOR SENSITIVE DATA

--SOAP / REST :-- MORE IS SOAP SECURE..THAN REST ALTHOUGH API ITSELF IS SECURE.

API HAS TO STORE USER REQUEST THAT DATA SO FOR THAT IT HAS TO USE :--
JSON,XML,HTML,XHTML,TEXT.

DATA MANIPULATION METHODS:--(DATA UPDATE,INSERT,DELETE LIKE IN SQL WE DONE.) IN PLACE OF QUERY HERE WE USE METHODS LIKE :-
1. GET :- SELECT DATA
2. POST :- INSERT DATA
3. PUT :- MULTIPLE VALUE IN DATA UPDATE
4. PATCH :- SINGLE VALUE IN DATA UPDATE
5. DELETE :- TO DELETE DATA

FASTAPI (FRAMEWORK)IS USED IN PYTHON TO MAKE REST API.
ALSO FLASK AND DJANGO DO THE SAME.

API == SIMPLE URL ALL CODE IS PUT TO SIMPLE URL THEN ALL YOUR MODEL WILL BE ACCESSED BY SINGLE URL.
THE MODEL YOU MADE AND YOUR API IS BEING TESTED IN :-
POSTMAN - TO TEST MODEL/API // GRAPH QL- ALSO LIKE POSTMAN // 
AFTER MODEL API TESTED THEN GIVEN TO // FRONTEND DEVELOPER.
```
