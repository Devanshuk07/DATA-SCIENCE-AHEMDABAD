```python
'''Requests.Session():---
Normally, every request can create a new connection.
A Session allows you to reuse connections and also remember things like:

-Cookies
-Headers
-Authentication
-Connection settings

IT IS LIKE WHEN YOU ONCE LOGIN TO FACEBOOK THEN LIKE A POST THEN COMMENT TO A POST EVERYTIME YOUR RESPONSE GOES TO SERVER 
BUT NOT EVERYTIME YOU HAD TO LOGIN FOR EVERY SINGLE REQUEST ONCE YOU LOGIN YOU CAN USE IT FOR A WHILE THEN AGAIN YOU HAVE TO LOGIN...

SO The purpose of Session is mainly to reuse settings and connections across multiple API requests.

EXAMPLE:--

import requests

session = requests.Session()

session.headers.update({                      #SEE THIS IS AUTHORIZATION YOU DECLARED SO YOU DONT NEED TO AUTORIZE AGAINA ND AGAIN FOR EVERY REQUEST.
    "Authorization": "Bearer my_token"
})
response = session.get("https://api.example.com/users")
print(response.json())


HEADERS = AN HTTP REQUEST HAS A HEADERS THAT CARRY EXTRA INFO ABOUT THE REQUEST.SO WHEN YOU WANT TO GIVE EXTRA INFO WITH REQUEST WRITE HEADERS.
THAT MEANS HEY API I AM SENDING MY AUTORIZATION TOKEN PLEASE VERIFY ME AND PROCESS MY REQUEST.

INSTEAD OF DOING THIS EVERYTIME :---

requests.get(
    "https://api.example.com/users",
    headers={"Authorization": "Bearer my_token"}
)



WHEN TO USE SESSION() :--

-Use Session() when you're making multiple requests to the same API, especially when they share authentication or other settings.

LIKE THIS :--
session = requests.Session()

session.headers.update({
    "Authorization": "Bearer my_token"
})

users = session.get("/users")         # FIRST REQUEST
products = session.get("/products")   # SECOND REQUEEST
orders = session.get("/orders")       # THIRD REQUEST  '''
```

```python
'''Basic Async Concepts :--
Async = asynchronous programming.

If each request takes 2 seconds, approximately:
2 + 2 + 2 = 6 seconds

With asynchronous programming, while one request is waiting for the server, Python can work on another request.'''
```
