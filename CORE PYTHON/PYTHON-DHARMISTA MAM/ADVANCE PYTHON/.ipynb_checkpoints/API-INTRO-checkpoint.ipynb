```python
'''What is an API?

-API = Application Programming Interface
-Think of an API as a middleman between your Python program and a server/service.


EASY WAY TO UNDERSTAND :--
-LETS SAY THAT SOME-ONE HAS CREATED A SERVER AND ENTERED/INPUTED MANY INFORMATION INTO IT.
-NOW WE ARE JUST CALLING THAT INFO AND WITH THE HELP OF API.
-API WILL ACCEPT OUR REQUEST AND TAKE REQUIRED INFO FROM SERVER AND PROVIDE US BACK AGAIN.
-THIS IS THE MAIN USE OF API.

REAL LIFE EXAMPLE:--
Real-life example 🍔

Imagine you go to a restaurant:

You → customer
Kitchen → system that prepares food
Waiter → API

LIKE THIS......
Your Python Program
       ↓
      API
       ↓
Other Server / Application
       ↓
     Response
       ↓
Your Python Program'''
```

```python
'''What is JSON?

-JSON = JavaScript Object Notation
-JSON is a common format used by APIs to send and receive data.
-It looks very similar to a Python dictionary.

JSON STRUCTURE:- SIMILAR TO PYTHON DICTIONARY(KEY -> VALUES)
{
    "name": "iPhone 17",
    "price": 79900,
    "storage": 256,
    "available": true
}

Nested JSON:-
{
    "name": "iPhone 17",
    "specifications": {
        "storage": 256,
        "battery": 4000,
        "display": "120Hz"
    }
}
```

```python
'''WHAT IS REST API :-
-REST = Representational State Transfer.
-A REST API is a common way of designing APIs using HTTP.
The most important HTTP methods are:

-GET
-POST
-PUT
-DELETE

Think of them as basic actions on data.
GET, POST, PUT, DELETE — are called HTTP Methods.
They are used when communicating with a server through an API.
GET, POST, PUT, DELETE are used to tell the server WHAT you want to do with the data.
--------------------------------------------------------------------
| Method     | Meaning              | Example                         |
| ---------- | -------------------- | ------------------------------- |
| **GET**    | Get/read data        | "Give me user #10"              |
| **POST**   | Create/send new data | "Create a new user"             |
| **PUT**    | Update existing data | "Change user #10's information" |
| **DELETE** | Delete data          | "Delete user #10"               |

-------------------------------------------------------------------
UNDERSTAND IT VERY CLEARLY :--

SOME-ONE HAS CREATED A SERVER AND API CONTAINING PRODUCT INFORMATION....
SOME-ONE
    ↓
Server + Database + API
    ↓
GET /products
POST /products
PUT /products/10
DELETE /products/10

Now you, as a Python programmer, can use those API methods.
If you're only getting information
You will commonly use GET...

But what about POST/PUT/DELETE?
-That depends on what the API allows you to do and your permissions.

For example, an online shopping API might allow you to:
GET     → View products
POST    → Create an order
PUT     → Update your order
DELETE  → Cancel/delete an order

So even though you're only using someone else's API, you might still use POST/PUT/DELETE.
```

```python
'''WHAT IS Headers :--
-Used to provide additional request information, such as authentication or content type.

for example :-
GET /users
header is .....
Authorization: abc123
Content-Type: application/json

-Here headers tell the server things like:

Authorization → "Here is my API key/token. I am allowed to access this."
Content-Type → "The data I'm sending is JSON."
User-Agent → "This request is coming from this type of program."
```

```python
'''WHAT IS Params ?
-Params MEANS PARAMETERS.

it is used to give additional information along with your request.
for example:--
GET /products?category=phone

here,,
category=phone is params.'''
```

```python
'''What is Rate Limits ?
-A server doesn't want you to send unlimited requests.

For example, an API might say:
100 requests per minute

That means you can make:
100 requests

If you send request number 101 too quickly, the API might reject it.

You may receive:
429 Too Many Requests.
```
