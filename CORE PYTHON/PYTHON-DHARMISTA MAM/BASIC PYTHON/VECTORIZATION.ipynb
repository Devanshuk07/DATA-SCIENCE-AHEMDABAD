```python
'''Vectorization means performing an operation on an entire array/column at once instead of using a Python loop manually.'''
```

```python
'''WITHOUT VECTORIZATION :--'''
a = [1, 2, 3, 4]

result = []
for i in a:
    result.append(i * 2)

print(result)
```

```python
'''WITH VECTORIZATION :-- '''
import numpy as np

a = np.array([1, 2, 3, 4])

result = a * 2
print(result)
```
