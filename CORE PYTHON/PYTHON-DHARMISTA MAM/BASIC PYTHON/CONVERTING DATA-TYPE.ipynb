```python
'''astype() :- USED TO CONVERT THE DATA-TYPE OF DATA.'''
```

```python
'''EXAMPLE:--
# Change to integer
df['age'] = df['age'].astype(int)

# Change to float
df['salary'] = df['salary'].astype(float)

# Change to string
df['name'] = df['name'].astype(str)

# Change to boolean
df['status'] = df['status'].astype(bool)'''
```

```python
'''YOU CAN ALSO CHANGE MULTIPLE COLUMNS:--
df = df.astype({
    'age': int,
    'salary': float
})'''
```
