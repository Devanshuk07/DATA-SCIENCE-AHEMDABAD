```python
'''DDL COMMONDS :-- 
DDL = Data Definition Language in SQL. ✅

It is used to create, change, or remove the STRUCTURE of database objects like tables.

Main DDL commands
| Command    | Use                                   |
| ---------- | ------------------------------------- |
| `CREATE`   | Create table/database                 |
| `ALTER`    | Change table structure                |
| `DROP`     | Remove entire table/database          |
| `TRUNCATE` | Remove all rows, keep table structure |
| `RENAME`   | Rename an object                      |

'''
```

```python
'''DML COMMANDS :--

DML :- Data Manipulation Language in SQL. ✅

It is used to work with/change the data (rows) inside tables.

Main DML commands
| Command    | Purpose              |
| ---------- | -------------------- |
| **INSERT** | Add new data/rows    |
| **UPDATE** | Modify existing data |
| **DELETE** | Remove data/rows     |
'''
```
